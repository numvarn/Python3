## 7.1 โครงสร้างควบคุมแบบเลือกทำ
กระบวนการ **ตัดสินใจ (Decision)** ใช้เมื่อโปรแกรมมีทางเลือกมากกว่าหนึ่งทาง ซึ่งขึ้นอยู่กับค่าของตัวแปร เช่น **สัญญาณไฟจราจร**: ไฟเขียวขับต่อไป ไฟเหลืองลดความเร็ว ไฟแดงหยุด ซึ่งเป็นการตัดสินใจเชิงตรรกะตามค่าของสัญญาณไฟ

````python
light = "เหลือง"     # ลองเปลี่ยนเป็น "เขียว" หรือ "แดง"

if light == "เขียว":
    print("🟢 ขับต่อไป")
elif light == "เหลือง":
    print("🟡 ลดความเร็ว")
else:
    print("🔴 หยุด")
````

### 7.1.1 คำสั่ง `if`
คำสั่ง `if` ใช้ตัดสินใจว่า **จะทำหรือไม่ทำ** ชุดคำสั่ง โดยพิจารณาจากนิพจน์เงื่อนไข
- ถ้าเงื่อนไขเป็น **`True`** จะทำคำสั่งในบล็อกของ `if`
- ถ้าเงื่อนไขเป็น **`False`** จะข้ามบล็อกของ `if` ไปทำคำสั่งถัดไป

```python
if test_expression:
    statement(s)
```
> อย่าลืม **เครื่องหมาย `:`** และ **การเยื้องหน้า 4 ช่อง** (บทที่ 4)

<p align="center"><br><b>แผนภาพ</b> การทำงานของคำสั่ง if</p>

````python
# exam_7_1.py: ถ้าตัวเลขเป็นบวก ให้พิมพ์ข้อความที่กำหนด
num = 3
if num > 0:
    print(num, "is a positive number.")
print("This is always printed.")

num = -1
if num > 0:
    print(num, "is a positive number.")
print("This is also always printed.")
````

- เมื่อ `num = 3` เงื่อนไข `num > 0` เป็น True จึงทำคำสั่งในบล็อก `if`
- เมื่อ `num = -1` เงื่อนไขเป็น False จึงข้ามบล็อก `if`
- `print()` ที่อยู่ **นอกบล็อก** (ไม่ได้เยื้อง) ทำงานเสมอ ไม่ว่าเงื่อนไขจะเป็นอย่างไร

### 7.1.2 คำสั่ง `if…else…`
ใช้เมื่อต้องทำอย่างหนึ่งเมื่อเงื่อนไขเป็นจริง และ **ทำอีกอย่างหนึ่งในกรณีอื่นทั้งหมด** (เงื่อนไขเป็นเท็จ)
```python
if test_expression:
    Body_of_if
else:
    Body_of_else
```
<p align="center"><br><b>แผนภาพ</b> การทำงานของคำสั่ง if…else</p>

````python
# exam_7_2.py: ตรวจสอบว่าตัวเลขเป็นบวกหรือลบ
num = 3          # ลองเปลี่ยนเป็น -5 และ 0
if num >= 0:
    print("Positive or Zero")
else:
    print("Negative number")
````

| ค่า `num` | `num >= 0` | บล็อกที่ทำงาน |
|---|---|---|
| 3 | True | `if` |
| −5 | False | `else` |
| 0 | True | `if` |

### 7.1.3 คำสั่ง `if…elif…else`
ใช้ตรวจสอบ **หลายเงื่อนไขต่อเนื่องกัน** (`elif` ย่อมาจาก **else if**)
```python
if test_expression_1:
    Body_of_if
elif test_expression_2:
    Body_of_elif
else:
    Body_of_else
```
- ตรวจเงื่อนไข **จากบนลงล่าง** ถ้าเงื่อนไขใดเป็น True จะทำบล็อกนั้น **แล้วข้ามบล็อกที่เหลือทั้งหมด**
- ถ้าทุกเงื่อนไขเป็น False จะทำบล็อก `else` (ถ้ามี)
- **ทำงานเพียงบล็อกเดียวเท่านั้น** และมี `elif` กี่ตัวก็ได้

<p align="center"><br><b>แผนภาพ</b> การทำงานของคำสั่ง if…elif…else</p>

````python
# exam_7_3.py: ตรวจสอบว่าตัวเลขเป็นบวก ลบ หรือศูนย์
num = 3.4
if num > 0:
    print("Positive number")
elif num == 0:
    print("Zero")
else:
    print("Negative number")
````

#### ตัวอย่าง: โปรแกรมตัดเกรด
รับคะแนนจากผู้ใช้ แล้วพิจารณาว่าคะแนนอยู่ในช่วงใด เพื่อตัดเกรด 5 ระดับ ได้แก่ A, B, C, D และ F

````python
# exam_7_4.py: โปรแกรมตัดเกรด (การตัดสินใจแบบหลายเงื่อนไข)
score = int(input("Enter score : "))

if score <= 100 and score >= 80:
    print("Grade A")
elif score < 80 and score >= 70:
    print("Grade B")
elif score < 70 and score >= 60:
    print("Grade C")
elif score < 60 and score >= 50:
    print("Grade D")
else:
    print("Grade F")
````

💡 **เขียนให้กระชับขึ้น:** เพราะ `elif` ตรวจเงื่อนไขก็ต่อเมื่อเงื่อนไขก่อนหน้าเป็นเท็จ จึง **ไม่ต้องเขียนขอบบนซ้ำ** และใช้ **การเปรียบเทียบต่อเนื่อง** (บทที่ 6) ตรวจช่วงคะแนนที่ถูกต้องก่อนได้ ดังตัวอย่างที่ใส่ไว้ในฟังก์ชันเพื่อทดสอบได้หลายค่า

````python
def grade(score):
    if not 0 <= score <= 100:
        return "คะแนนไม่ถูกต้อง"
    elif score >= 80:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"

for s in [95, 80, 79, 65, 50, 49, 0, 120, -5]:
    print(f"{s:>4} -> {grade(s)}")
````

### 7.1.4 คำสั่ง Nested if
คำสั่ง `if…elif…else` **ซ้อนอยู่ภายใน** `if`, `elif` หรือ `else` อีกชั้นได้ เรียกว่า **Nested** โดยไพธอนใช้ **การเยื้องหน้า** แบ่งระดับการซ้อน การซ้อนหลายชั้นอาจทำให้สับสน จึง **ควรหลีกเลี่ยงการซ้อนที่ไม่จำเป็น** (Flat is better than nested, บทที่ 4)

````python
# exam_7_5.py: ตรวจสอบว่าตัวเลขเป็นบวก ลบ หรือศูนย์ โดยใช้ nested if
num = float(input("Enter a number: "))

if num >= 0:
    if num == 0:              # if ชั้นใน (เยื้อง 8 ช่อง)
        print("Zero")
    else:
        print("Positive number")
else:
    print("Negative number")
````

````python
# ตัวอย่าง nested if ที่เหมาะสม: ตรวจสอบสิทธิ์สมัครสอบ
age = 20
has_id_card = True
paid = False

if age >= 18:
    if has_id_card:
        if paid:
            print("✅ สมัครสอบได้")
        else:
            print("❌ ยังไม่ได้ชำระค่าสมัคร")
    else:
        print("❌ ต้องมีบัตรประชาชน")
else:
    print("❌ อายุยังไม่ถึง 18 ปี")
````

````python
# ✅ เซลล์ตรวจคำตอบ
import io as _io, contextlib as _cl
def _capture(func, *args):
    _buf = _io.StringIO()
    try:
        with _cl.redirect_stdout(_buf):
            func(*args)
    except Exception as _e:
        return f"เกิดข้อผิดพลาด: {_e}"
    return _buf.getvalue()
def _safe(f, *a):
    try:
        return f(*a)
    except Exception as _e:
        return f"เกิดข้อผิดพลาด: {_e}"
_tests = [("bmi(45, 1.70)", _safe(bmi_category, 45, 1.70), "ผอม"), ("bmi(60, 1.70)", _safe(bmi_category, 60, 1.70), "ปกติ"), ("bmi(68, 1.70)", _safe(bmi_category, 68, 1.70), "ท้วม"), ("bmi(80, 1.70)", _safe(bmi_category, 80, 1.70), "อ้วน"), ("bmi(95, 1.70)", _safe(bmi_category, 95, 1.70), "อ้วนมาก"), ("bmi(23, 1.0) ขอบเขตพอดี 23", _safe(bmi_category, 23, 1.0), "ท้วม")]
_ok = 0
for _label, _got, _exp in _tests:
    _pass = _got == _exp
    _ok += _pass
    _show = repr(_got) if len(repr(_got)) < 70 else repr(_got)[:67] + "..."
    print(f"{'✅' if _pass else '❌'} {_label} = {_show}" + ("" if _pass else f"\n      ควรได้ {_exp!r}"[:200]))
print(f"\nผ่าน {_ok}/{len(_tests)} ข้อ" + ("  🎉" if _ok == len(_tests) else ""))
````

## 7.2 โครงสร้างควบคุมแบบวนซ้ำ
**การวนซ้ำ (Iteration)** คือกระบวนการที่ชุดคำสั่งถูกประมวลผลซ้ำต่อเนื่อง ตั้งแต่ 1 ครั้งขึ้นไป **ตามจำนวนที่กำหนด** หรือ **จนกว่าเงื่อนไขจะเป็นไปตามที่วางไว้** โดยไพธอนใช้คำสั่ง `for` และ `while`

| | `for` | `while` |
|---|---|---|
| ใช้เมื่อ | **รู้จำนวนรอบ** หรือวนผ่านสมาชิกของลำดับข้อมูล | **ไม่รู้จำนวนรอบล่วงหน้า** วนจนกว่าเงื่อนไขจะเป็นเท็จ |
| ตัวอย่าง | พิมพ์สูตรคูณแม่ 2, รวมคะแนนในลิสต์ | รับข้อมูลจนกว่าผู้ใช้จะป้อน 0, เกมทายตัวเลข |

### 7.2.1 คำสั่ง `for`

#### 7.2.1.1 ไวยากรณ์ของคำสั่ง `for`
```python
for val in sequence:
    body_of_for
```
`val` คือตัวแปรที่รับค่าสมาชิกของ `sequence` ทีละตัวในแต่ละรอบ ลูปทำต่อไปจนถึงสมาชิกตัวสุดท้าย

<p align="center"><br><b>แผนภาพ</b> การทำงานของคำสั่ง for</p>

````python
# exam_7_6.py: หาผลรวมของตัวเลขทั้งหมดในลิสต์
numbers = [6, 5, 3, 8, 4, 2, 5, 4, 11]

total = 0                  # ตัวแปรเก็บผลรวม (เอกสารใช้ชื่อ sum ซึ่งซ้ำกับฟังก์ชัน sum())
for val in numbers:        # ดึงสมาชิกทีละตัวมาเก็บใน val
    total = total + val    # บวกสะสม

print("The sum is", total)
print("ตรวจด้วย sum():", sum(numbers))
````

#### 7.2.1.2 ฟังก์ชัน `range()`
`range()` สร้าง **ลำดับของตัวเลข** สำหรับใช้กับ `for` รับอาร์กิวเมนต์ได้สูงสุด 3 ตัว

| รูปแบบ | ความหมาย | ตัวอย่าง → ผลลัพธ์ |
|---|---|---|
| `range(stop)` | เริ่มที่ 0 ถึง **ก่อน** stop | `range(10)` → 0, 1, …, 9 |
| `range(start, stop)` | เริ่มที่ start ถึง **ก่อน** stop | `range(5, 10)` → 5, 6, 7, 8, 9 |
| `range(start, stop, step)` | เพิ่มทีละ step (ค่าเริ่มต้น step = 1) | `range(0, 10, 2)` → 0, 2, 4, 6, 8 |
| step ติดลบ | นับถอยหลัง | `range(10, 0, -1)` → 10, 9, …, 1 |

> 📌 **ค่า stop ไม่รวมอยู่ในลำดับเสมอ** (เหมือน slicing ในบทที่ 5) ถ้าต้องการ 1 ถึง 10 ต้องเขียน `range(1, 11)`

`range()` **ไม่ได้สร้างตัวเลขทั้งหมดเก็บไว้ในหน่วยความจำทันที** แต่สร้างทีละตัวเมื่อถูกใช้ (จึงประหยัดหน่วยความจำ) ถ้าต้องการดูหรือเก็บเป็นลิสต์ให้แปลงด้วย `list()`

````python
print(range(10))
print(range(1, 10))
print(list(range(1, 10)))
print(list(range(1, 10, 2)))
print(list(range(10, 0, -1)))
````

### 7.2.2 คำสั่ง `while`
`while` **ตรวจสอบเงื่อนไขก่อนทุกรอบ** และวนซ้ำตราบที่เงื่อนไขเป็น `True` จะหยุดเมื่อเงื่อนไขเป็น `False` นิยมใช้เมื่อ **ไม่ทราบจำนวนรอบล่วงหน้า**
```python
while test_expression:
    body_of_while
```
> ⚠️ ภายในลูปต้องมีคำสั่งที่ **ทำให้เงื่อนไขเปลี่ยนเป็น False ได้ในที่สุด** มิฉะนั้นจะเกิดลูปไม่สิ้นสุด

<p align="center"><br><b>แผนภาพ</b> การทำงานของคำสั่ง while</p>

````python
# exam_7_12.py
count = 0
while count <= 10:
    print('The index is: {}'.format(count))
    count += 1          # ทำให้ count เพิ่มขึ้นจนเงื่อนไขเป็น False
````

เมื่อ `count` ถูกเพิ่มจนมีค่าเกิน 10 เงื่อนไข `count <= 10` จะเป็น False และลูปหยุดทำงาน

#### ตัวอย่างที่เหมาะกับ `while`: ไม่รู้จำนวนรอบล่วงหน้า

````python
# เงินฝาก 10,000 บาท ดอกเบี้ย 5% ต่อปี (ทบต้น) ต้องฝากกี่ปีจึงจะได้ 20,000 บาท
balance = 10000
years = 0
while balance < 20000:
    balance = balance * 1.05
    years += 1
print(f"ต้องฝาก {years} ปี จะมีเงิน {balance:,.2f} บาท")
````

````python
# เกมทายตัวเลข 1-20
import random
secret = random.randint(1, 20)
guess = 0
tries = 0
while guess != secret:
    guess = int(input("ทายตัวเลข 1-20: "))
    tries += 1
    if guess < secret:
        print("  น้อยไป ⬆")
    elif guess > secret:
        print("  มากไป ⬇")
print(f"🎉 ถูกต้อง! ตัวเลขคือ {secret} ใช้ไป {tries} ครั้ง")
````

### 7.2.3 ลูปซ้อน (Nested Loop)
ลูปที่อยู่ภายในลูปอื่นเรียกว่า **ลูปซ้อน** ซึ่งพบบ่อยเมื่อแก้ปัญหาที่ซับซ้อน แต่ **ลูปซ้อนที่ไม่จำเป็นทำให้โปรแกรมช้าลงมาก** ถ้าลูปนอกทำ n รอบและลูปในทำ m รอบ คำสั่งในลูปในจะทำงาน **n × m ครั้ง**

#### ตารางสูตรคูณ 2 มิติ
ลูปนอกกำหนด **ตัวคูณ (แนวตั้ง)** และลูปในกำหนด **ตัวถูกคูณ (แนวนอน)** โดยลูปนอก 1 รอบ ลูปในจะทำ 12 รอบ รวม 12 × 12 = **144 รอบ**

````python
# exam_7_13.py
for i in range(1, 13):
    print("{0:>5} : ".format(i), end="")
    for j in range(1, 13):
        print("{0:>5}".format(i * j), end="")
    print("")
````

#### จำนวนรอบของลูปในแปรผันตามลูปนอก
ลูปในทำ `i` รอบ ดังนั้นจำนวนรอบเพิ่มขึ้นตามค่า `i` ของลูปนอก (**แปรผันตรง**)

````python
# exam_7_14.py
for i in range(1, 11):
    for j in range(1, i + 1):
        print("{0:>3}".format(j), end="")
    print("")
````

ลูปในทำ `len(lt) - i` รอบ ดังนั้นยิ่งลูปนอกทำงานมากรอบ ลูปในจะทำงานน้อยลง (**แปรผกผัน**)

````python
# exam_7_15.py
lt = ['a', 'b', 'c', 'd', 'e', 'f', 'g']
for i in range(len(lt)):
    for j in range(len(lt) - i):
        print("{0:>2}".format(lt[j]), end="")
    print("")
````

ลูปซ้อนด้วย `while` ก็ทำได้ แต่ **ต้องจัดการตัวแปรควบคุมและเงื่อนไขของแต่ละลูปเอง** (กำหนดค่าเริ่มต้นและเพิ่มค่า)

````python
# exam_7_16.py
i = 1
while i < 11:
    j = 0                      # ต้องรีเซ็ต j ทุกครั้งที่เริ่มรอบใหม่ของลูปนอก
    while j < i:
        print('* ', end='')
        j = j + 1
    print("")
    i = i + 1
print("Rest of the program")
````

### 7.2.4 ลูปแบบไม่สิ้นสุด (Infinite Loop)
**ลูปไม่สิ้นสุด** คือลูปที่วนทำงานต่อเนื่องไม่มีที่สิ้นสุด เพราะไม่มีเงื่อนไขหยุด หรือการทำงานผิดพลาดจนเงื่อนไขไม่มีวันเป็นเท็จ
- ถ้า **ไม่ได้ตั้งใจ** ถือเป็น **ข้อผิดพลาด** ของโปรแกรม
- ถ้า **ตั้งใจ** ก็มีประโยชน์ เช่น โปรแกรมเซิร์ฟเวอร์ที่ต้องรอรับการเชื่อมต่อจากไคลเอนต์ตลอดเวลา โปรแกรมเมนู หรือเกม แต่ต้องมี **ทางออกจากลูป** เช่น `break`

#### ลูปไม่สิ้นสุดที่เกิดจากความผิดพลาด
```python
# exam_7_17.py  ⚠️ ไม่ต้องรัน: เป็นลูปไม่สิ้นสุด
var = 1
while var == 1:             # var ไม่เคยเปลี่ยนค่า เงื่อนไขจึงเป็น True ตลอดไป
    num = input("Enter a number :")
    print("You entered: ", num)
print("Good bye!")          # ไม่มีวันมาถึงบรรทัดนี้
```
ในบล็อกของ `while` ไม่มีคำสั่งใดเปลี่ยนค่า `var` เลย ลูปจึงไม่มีวันหยุด (ต้องกด Interrupt เพื่อหยุดโปรแกรม) **วิธีแก้** คือนำค่าที่รับเข้ามาไปเก็บในตัวแปรที่ใช้เป็นเงื่อนไข และแปลงเป็นจำนวนเต็มก่อนเปรียบเทียบ (เพราะ `input()` คืนค่าเป็นสตริงเสมอ)

````python
# exam_7_18.py (ฉบับแก้ไข): ป้อน 0 หรือค่าติดลบเพื่อออกจากลูป
var = 1
while var > 0:
    var = int(input("Enter a number (0 = exit): "))
    print("You entered: ", var)
print("Good bye!")
````

````python
# exam_7_21.py: หารากที่สองไปเรื่อย ๆ จนกว่าจะป้อนค่า <= 0
import math

while True:
    inp = int(input("Enter integer value: "))
    if inp <= 0:
        print("Exit program")
        break
    else:
        print("Square root of %s is %.3f\n" % (inp, math.sqrt(inp)))
````

> 💡 `"%s ... %.3f" % (...)` เป็นการจัดรูปแบบสตริง **แบบเก่า** (ยืมมาจากภาษา C) ซึ่งยังพบในโค้ดทั่วไป ปัจจุบันแนะนำให้ใช้ f-string แทน: `f"Square root of {inp} is {math.sqrt(inp):.3f}"`

````python
# break ในลูปซ้อน ออกจากลูปชั้นในเท่านั้น
for i in range(1, 4):
    for j in range(1, 4):
        if j == 3:
            break               # ออกจากลูป j เท่านั้น ลูป i ยังทำต่อ
        print(f"i={i}, j={j}")
````

#### 7.2.5.2 คำสั่ง `continue`
ข้ามโค้ดที่เหลือในรอบปัจจุบัน แล้ววนรอบถัดไปทันที (ลูปไม่สิ้นสุด)

````python
# exam_7_22.py: ใช้คำสั่ง continue ภายในลูป
for val in "string":
    if val == "i":
        continue        # ข้าม print ของรอบนี้ ตัว i จึงไม่ถูกพิมพ์
    print(val)
print("The end")
````

## 7.3 สรุป
โครงสร้างการควบคุมทำให้โปรแกรม **วิเคราะห์และเลือกทางประมวลผลได้โดยอัตโนมัติ** คล้ายการตัดสินใจของมนุษย์ และใช้ประโยชน์จากความสามารถของคอมพิวเตอร์ในการ **ประมวลผลเชิงตรรกะ** และ **ทำงานซ้ำ** กับข้อมูลจำนวนมากได้อย่างรวดเร็ว

- **โครงสร้างควบคุมแบบเลือกทำ (Selection)** กำหนดทางเลือกให้โปรแกรม โดยมีเงื่อนไขควบคุมว่าจะทำงานตามทางเลือกใด
- คำสั่ง **`if`**, **`if…else`** และ **`if…elif…else`** ร่วมกับนิพจน์เงื่อนไข ใช้ควบคุมทางเลือกทั้ง 3 แบบ และซ้อนกันได้ (Nested if) โดยใช้การเยื้องหน้าแบ่งระดับ
- การวนซ้ำในไพธอนใช้คำสั่ง **`while`** หรือ **`for`**
- **`for`** นิยมใช้เมื่อ **รู้จำนวนรอบที่แน่นอน** ทำงานร่วมกับข้อมูลแบบลำดับหรือฟังก์ชัน **`range(start, stop, step)`** (ไม่รวม stop)
- **`while`** ตรวจสอบเงื่อนไขก่อนทุกรอบ และวนซ้ำตราบที่เงื่อนไขเป็น True ต้องระวัง **ลูปไม่สิ้นสุด** ที่ไม่ตั้งใจ
- **ลูปซ้อน** ทำให้จำนวนรอบเพิ่มเป็นผลคูณ จึงควรใช้เท่าที่จำเป็น
- **`break`** ออกจากลูปที่ทำงานอยู่ปัจจุบันหนึ่งลำดับชั้น และ **`continue`** ข้ามโค้ดที่เหลือในรอบปัจจุบัน แล้ววนรอบถัดไป

    return list(range(f, t + 1, i))       # +1 เพื่อให้รวม t

    result = 1
    for k in range(2, n + 1):

    return result

    primes = []
    for num in range(2, n + 1):
        is_prime = True
        for d in range(2, int(num ** 0.5) + 1):   # ตรวจตัวหารถึงรากที่สองก็พอ
            if num % d == 0:
                is_prime = False
                break
        if is_prime:
            primes.append(num)

### 🏆 แบบทดสอบท้ายบท (ตรวจคะแนนอัตโนมัติ)
รันเซลล์ด้านล่าง แล้วพิมพ์ตัวเลือก (ก/ข/ค/ง) ของแต่ละข้อ

````python
quiz = [
    ("คำว่า elif ย่อมาจากอะไร",
     ["ก. else if", "ข. end if", "ค. elseif loop", "ง. eliminate if"], "ก"),
    ("ใน if…elif…else จะมีบล็อกที่ทำงานได้มากที่สุดกี่บล็อก",
     ["ก. ทุกบล็อกที่เงื่อนไขเป็นจริง", "ข. 1 บล็อก", "ค. 2 บล็อก", "ง. ไม่มีบล็อกใดทำงาน"], "ข"),
    ("list(range(2, 10, 3)) ได้ผลลัพธ์ใด",
     ["ก. [2, 5, 8]", "ข. [2, 5, 8, 11]", "ค. [3, 6, 9]", "ง. [2, 3, 4]"], "ก"),
    ("ถ้าต้องการวนรับข้อมูลจนกว่าผู้ใช้จะพิมพ์ 'q' ควรใช้คำสั่งใด",
     ["ก. for", "ข. while", "ค. if", "ง. range"], "ข"),
    ("คำสั่งใดใช้ออกจากลูปทันที",
     ["ก. continue", "ข. pass", "ค. break", "ง. exit"], "ค"),
    ("คำสั่งใดข้ามโค้ดที่เหลือในรอบนี้แล้วไปรอบถัดไป",
     ["ก. continue", "ข. break", "ค. pass", "ง. return"], "ก"),
    ("for i in range(4): for j in range(5): print('*') พิมพ์ * กี่ตัว",
     ["ก. 9", "ข. 20", "ค. 45", "ง. 4"], "ข"),
    ("ข้อใดถูกตีความเป็น False ในเงื่อนไขของ if",
     ["ก. 'False'", "ข. [0]", "ค. -1", "ง. ''"], "ง"),
    ("x = 10; while x > 0: x -= 3 เมื่อจบลูป x มีค่าเท่าใด",
     ["ก. 0", "ข. 1", "ค. -2", "ง. -3"], "ค"),
    ("คำสั่งใดใช้เมื่อต้องการบล็อกที่ไม่ทำอะไรเลย",
     ["ก. break", "ข. pass", "ค. continue", "ง. None"], "ข"),
]

def run_quiz():
    score = 0
    for i, (q, choices, ans) in enumerate(quiz, 1):
        print(f"\nข้อ {i}. {q}")
        for c in choices:
            print("   ", c)
        reply = input("คำตอบ (ก/ข/ค/ง): ").strip()
        if reply == ans:
            print("   ✅ ถูกต้อง"); score += 1
        else:
            print(f"   ❌ ยังไม่ถูก คำตอบที่ถูกคือ {ans}")
    print(f"\n🎯 คะแนนของคุณ: {score}/{len(quiz)}")
    print("🏆 ยอดเยี่ยม!" if score == len(quiz) else "👍 ดีมาก" if score >= 7 else "📖 ทบทวนเนื้อหาอีกครั้งนะ")

run_quiz()
````
