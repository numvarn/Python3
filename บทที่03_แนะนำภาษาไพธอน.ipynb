## 3.1 ประวัติความเป็นมาและจุดเด่นของภาษาไพธอน

ไพธอนเป็นภาษาโปรแกรมที่ **ง่ายต่อการเรียนรู้** ซึ่งเป็นประโยชน์อย่างมากสำหรับผู้เริ่มต้น โค้ด **อ่านและเข้าใจง่าย** เมื่อเทียบกับภาษาอื่น และมี **โหมดโต้ตอบ (Interactive Mode)** ผ่าน **ไพธอนเชลล์ (Python Shell)** ที่ทำให้ผู้เรียนทดลองคำสั่งและเห็นการทำงานได้ทีละขั้น นอกจากนี้ไพธอนยังมีคุณลักษณะอื่นอีกมากที่ช่วยการเรียนรู้ และถูกนำไปใช้ศึกษาในหลายสาขาวิชา

- ตัวแปลภาษาไพธอนมาตรฐาน (CPython) **เขียนขึ้นด้วยภาษา C** และทำงานแบบ **อินเตอร์พรีเตอร์ (Interpreter)** คือแปลคำสั่งทีละบรรทัดแล้วทำงานทันที (ทบทวนบทที่ 1)
- **รองรับการเขียนโปรแกรมเชิงวัตถุ (OOP: Object Oriented Programming)** ซึ่งจะได้เรียนในบทที่ 12
- เป็น **โอเพนซอร์ส (Open Source)** คือเปิดเผยซอร์สโค้ดให้สาธารณะใช้หรือดัดแปลงได้ นักพัฒนาทั่วโลกจึงร่วมกันพัฒนาและแบ่งปันการเปลี่ยนแปลงกับชุมชนได้

### 3.1.1 ประวัติของภาษาไพธอน
- ไพธอนถูกสร้างขึ้นในต้นทศวรรษ 1990 โดย **Guido van Rossum** ชาวดัตช์ ที่ Stichting Mathematisch Centrum (CWI) ประเทศเนเธอร์แลนด์
- ต้นแบบสืบทอดมาจาก **ภาษา ABC** ซึ่งออกแบบมาให้เรียนรู้ง่าย Van Rossum เคยร่วมพัฒนาภาษา ABC และเห็นปัญหาของภาษานี้ แต่ยังชอบจุดเด่นหลายอย่าง เช่น การจัดการสิ่งผิดปกติ (Exception Handling) และการเชื่อมต่อกับระบบปฏิบัติการ Amoeba จึงตัดสินใจสร้าง **ภาษาสคริปต์ (Scripting Language)** ใหม่ที่ใช้ไวยากรณ์และจุดเด่นของ ABC
- เขาเริ่มพัฒนาภาษาใหม่ในช่วง **วันหยุดคริสต์มาส ธันวาคม 1989** และตั้งชื่อว่า **"Python"** ตามซีรีส์ตลก **"Monty Python's Flying Circus"** ของ BBC จากยุค 1970 (ไม่ได้ตั้งตามชื่องูเหลือม 🐍 อย่างที่หลายคนเข้าใจ)

### 3.1.2 ลักษณะเด่นของภาษาไพธอน

| จุดเด่น | คำอธิบาย |
|---|---|
| **เริ่มต้นเข้าใจได้ง่าย** | ออกแบบมาให้เข้าใจง่ายและสนุกกับการใช้งาน มีบทเรียนแนะนำสำหรับผู้เริ่มต้น สร้างโปรแกรมและเครื่องมือได้อย่างรวดเร็ว จึงเป็นภาษาที่ผู้เริ่มต้นนิยมเลือก และเป็นภาษาที่มหาวิทยาลัยในสหรัฐอเมริกานิยมใช้สอนมากที่สุด |
| **อ่านเข้าใจง่าย** | เป็นภาษาระดับสูงมาก ไวยากรณ์อ่านได้เหมือนภาษาอังกฤษ ไม่ต้องจดจำกฎยิบย่อยมากเหมือนภาษาอื่น ผู้เริ่มต้นจึงมุ่งเรียนรู้ **แนวคิดการเขียนโปรแกรม** ได้ทันทีโดยไม่ต้องกังวลกับไวยากรณ์มากเกินไป |
| **ยืดหยุ่นมาก** | เป็นภาษาแบบ **ไดนามิก (Dynamic)** แก้ปัญหาได้หลายวิธี (แม้ปรัชญาของไพธอนจะสนับสนุนวิธีที่ชัดเจนที่สุด) และให้อภัยข้อผิดพลาดมากกว่า โดยโปรแกรมจะทำงานไปจนกว่าจะพบปัญหา |
| **ย้ายระบบได้ง่าย (Cross-platform)** | โปรแกรมไพธอนทำงานได้โดยไม่ยึดติดกับระบบปฏิบัติการ เช่น Windows, macOS, Linux เพราะตัวแปลภาษาเขียนด้วยภาษา C ซึ่งมีคอมไพเลอร์ ANSI C อยู่บนแทบทุกแพลตฟอร์ม |

### 🔬 ทดลอง: ไพธอนอ่านง่ายจริงหรือ?
เปรียบเทียบโปรแกรมแสดงข้อความ "Hello World" ในภาษา C, Java และ Python

**ภาษา C**
```c
#include <stdio.h>
int main() {
    printf("Hello World\n");
    return 0;
}
```
**ภาษา Java**
```java
public class Hello {
    public static void main(String[] args) {
        System.out.println("Hello World");
    }
}
```
**ภาษา Python** ใช้เพียงบรรทัดเดียว ลองรันดู

````python
print("Hello World")
````

อีกตัวอย่างหนึ่งคือการหาค่าเฉลี่ยคะแนนเฉพาะคนที่สอบผ่าน ซึ่งโค้ดอ่านออกมาได้เกือบเหมือนประโยคภาษาอังกฤษ

````python
scores = [45, 78, 90, 62, 38, 85]
passed = [s for s in scores if s >= 50]          # เลือกคะแนนที่ >= 50
print("คะแนนที่ผ่าน:", passed)
print("ค่าเฉลี่ยของคนที่ผ่าน:", sum(passed) / len(passed))
````

### 🔬 ทดลอง: โหมดสคริปต์ (Script Mode) แบบเดียวกับ PyCharm
PyCharm ทำงานแบบ **โหมดสคริปต์** คือเขียนโค้ดเก็บไว้ใน **ไฟล์ `.py`** แล้วสั่งรัน **ทั้งไฟล์ในครั้งเดียว** ซึ่งต่างจากไพธอนเชลล์ที่รันทีละคำสั่ง เซลล์ด้านล่างจำลองขั้นตอนนี้ คือสร้างไฟล์ `hello.py` แล้วสั่งตัวแปลภาษาไพธอนรันไฟล์นั้น (เทียบเท่าการพิมพ์ `python hello.py` ใน Anaconda Prompt หรือกด ▶ ใน PyCharm)

````python
# ขั้นที่ 1: เขียนซอร์สโค้ดลงไฟล์ hello.py (เหมือนพิมพ์ใน Editor ของ PyCharm แล้วกดบันทึก)
source = '''
name = "นักศึกษา"
print("Hello World")
print(f"สวัสดี {name} นี่คือโปรแกรมแรกในโหมดสคริปต์")
for i in range(1, 4):
    print("บรรทัดที่", i)
'''
with open("hello.py", "w", encoding="utf-8") as f:
    f.write(source)
print("✅ สร้างไฟล์ hello.py แล้ว")
````

````python
# ขั้นที่ 2: สั่งรันไฟล์ทั้งไฟล์ (เหมือนกด Run ▶ ใน PyCharm)
import subprocess, sys, os
env = dict(os.environ, PYTHONIOENCODING="utf-8")   # ให้แสดงภาษาไทยได้ถูกต้องทุกระบบปฏิบัติการ
result = subprocess.run([sys.executable, "hello.py"], capture_output=True, text=True, encoding="utf-8", env=env)
print("----- ผลลัพธ์ในคอนโซล -----")
print(result.stdout, result.stderr)
````

> 💡 ใน Jupyter มีคำสั่งลัด (magic command) สำหรับงานนี้ด้วย คือ `%%writefile hello.py` (เขียนไฟล์) และ `%run hello.py` หรือ `!python hello.py` (รันไฟล์)

## 3.4 การใช้งาน Jupyter Notebook

ผู้เริ่มต้นมักอยากทดลองเขียนโค้ดสั้น ๆ ทดลองใช้ไลบรารี (Library) หรือทดสอบโค้ดตัวอย่างจากอินเทอร์เน็ต การเปิดโปรแกรมใหญ่อย่าง PyCharm เพื่อทดลองเล็ก ๆ น้อย ๆ อาจไม่คล่องตัว

**Jupyter Notebook** จึงเป็นเครื่องมือที่มีบทบาทมากในปัจจุบัน โดยเฉพาะงาน **วิเคราะห์ข้อมูล (Data Analysis)** และ **วิทยาการข้อมูล (Data Science)** เพราะ
- **แบ่งโค้ดเป็นเซลล์ (Cell)** และเลือกรันทีละส่วนได้ ต่างจาก PyCharm ที่ต้องรันทั้งไฟล์ในครั้งเดียว
- **เห็นผลลัพธ์ทันทีใต้เซลล์** ในหน้าเดียว โดยไม่ต้องสลับหน้าจอระหว่างซอร์สโค้ดกับผลลัพธ์
- **บันทึกผลลัพธ์ไว้ได้** และเขียนคำอธิบาย/รายงานไว้ในหน้าเดียวกับโค้ด (**Notebook ที่ผู้เรียนกำลังอ่านอยู่นี้ก็สร้างด้วย Jupyter**)
- **ทำงานบนเว็บเบราว์เซอร์** ซึ่งมีอยู่ในคอมพิวเตอร์ทุกเครื่อง
- ไฟล์ Notebook มีนามสกุล **`.ipynb`** (ย่อมาจาก IPython Notebook)

### 🔬 ทัวร์ Jupyter ภาคปฏิบัติ
ผู้เรียนกำลังใช้งาน Jupyter อยู่แล้ว มาลองใช้ความสามารถสำคัญกัน

#### ① เซลล์มี 2 ชนิดหลัก
| ชนิดเซลล์ | ใช้ทำอะไร | สังเกตอย่างไร |
|---|---|---|
| **Code** | เขียนโค้ดไพธอน แล้วรันได้ | มี `[ ]:` อยู่ด้านหน้า |
| **Markdown** | เขียนข้อความ หัวข้อ ตาราง รูปภาพ สูตรคณิตศาสตร์ | กลายเป็นข้อความสวยงามเมื่อรัน (เหมือนเซลล์ที่กำลังอ่านอยู่นี้) |

#### ② ตัวเลขในวงเล็บ `[n]` บอกลำดับการรัน
ตัวเลขหน้าเซลล์โค้ดบอกว่าเซลล์นั้นถูกรันเป็นลำดับที่เท่าใด ถ้าเห็น `[*]` แสดงว่าเซลล์กำลังทำงานอยู่ ลองรันเซลล์สองเซลล์ด้านล่างนี้ **สลับลำดับกัน** แล้วสังเกตผลลัพธ์

````python
x = 10
print("กำหนด x =", x)
````

````python
x = x + 5
print("ตอนนี้ x =", x)    # ลองรันเซลล์นี้ซ้ำหลายครั้ง ค่า x จะเพิ่มขึ้นเรื่อย ๆ
````

## 📊 เปรียบเทียบวิธีรันโปรแกรมไพธอน
การรัน (Execute) โปรแกรมไพธอนแบ่งได้เป็น 3 วิธีหลัก

| วิธี | เครื่องมือ | ลักษณะการทำงาน | เหมาะกับ |
|---|---|---|---|
| **1. โหมดโต้ตอบ (Interactive Mode)** | ไพธอนเชลล์ (`python` ใน Anaconda Prompt / Terminal), IDLE | พิมพ์และรันทีละคำสั่ง เห็นผลทันที แต่ไม่ได้บันทึกโค้ดไว้ | ทดลองคำสั่งสั้น ๆ, ใช้เป็นเครื่องคิดเลข |
| **2. โหมดสคริปต์ (Script Mode)** | PyCharm, VS Code, Text Editor แล้วรันด้วย `python ชื่อไฟล์.py` | เขียนโค้ดเก็บในไฟล์ `.py` แล้วรันทั้งไฟล์ | โปรแกรมจริงที่มีหลายบรรทัดหรือหลายไฟล์ และใช้ซ้ำได้ |
| **3. Notebook** | Jupyter Notebook, JupyterLab, Google Colab, VS Code | แบ่งโค้ดเป็นเซลล์ รันทีละเซลล์ ผลลัพธ์อยู่ใต้เซลล์ และผสมข้อความอธิบายได้ | การเรียนรู้ ทดลอง วิเคราะห์ข้อมูล เขียนรายงาน |

```
                 ┌── โหมดโต้ตอบ ──►  >>> print("Hi")     (ทีละคำสั่ง)
ตัวแปลภาษาไพธอน ──┼── โหมดสคริปต์ ──►  python hello.py    (ทั้งไฟล์)
                 └── Notebook   ──►  [1]: print("Hi")    (ทีละเซลล์)
```

> 💡 ทั้ง 3 วิธีใช้ **ตัวแปลภาษาไพธอนตัวเดียวกัน** แตกต่างกันเพียงรูปแบบการป้อนโค้ดและการแสดงผล

**เครื่องมือยอดนิยมอื่น ๆ ในปัจจุบัน:** Visual Studio Code (VS Code), Spyder (มากับ Anaconda), IDLE (มากับไพธอนจาก python.org), Thonny (ออกแบบสำหรับผู้เริ่มต้น), Google Colab, Sublime Text และ Vim/Neovim

````python
quiz = [
    ("ผู้สร้างภาษาไพธอนคือใคร",
     ["ก. Dennis Ritchie", "ข. Guido van Rossum", "ค. James Gosling", "ง. Bill Gates"], "ข"),
    ("ชื่อ \"Python\" ตั้งตามสิ่งใด",
     ["ก. งูเหลือม", "ข. ภาษา ABC", "ค. ซีรีส์ตลก Monty Python's Flying Circus", "ง. ชื่อบริษัท"], "ค"),
    ("ภาษาไพธอนสืบทอดแนวคิดมาจากภาษาใด",
     ["ก. ABC", "ข. Java", "ค. COBOL", "ง. Pascal"], "ก"),
    ("Python 3.0 ออกเมื่อปี ค.ศ. ใด",
     ["ก. 1991", "ข. 2000", "ค. 2008", "ง. 2020"], "ค"),
    ("เครื่องหมาย >>> ปรากฏในโปรแกรมใด",
     ["ก. ไพธอนเชลล์", "ข. Microsoft Word", "ค. เว็บเบราว์เซอร์", "ง. File Explorer"], "ก"),
    ("ไฟล์ซอร์สโค้ดภาษาไพธอนมีนามสกุลใด",
     ["ก. .java", "ข. .py", "ค. .ipynb", "ง. .exe"], "ข"),
    ("ไฟล์ Jupyter Notebook มีนามสกุลใด",
     ["ก. .py", "ข. .txt", "ค. .ipynb", "ง. .nb"], "ค"),
    ("ข้อใดคือจุดเด่นของ Jupyter Notebook",
     ["ก. ต้องรันทั้งไฟล์พร้อมกัน", "ข. แบ่งโค้ดเป็นเซลล์และเห็นผลใต้เซลล์ทันที",
      "ค. ใช้ได้เฉพาะ Windows", "ง. ไม่ต้องใช้ตัวแปลภาษา"], "ข"),
    ("คีย์ลัดใดใช้รันเซลล์ใน Jupyter แล้วไปเซลล์ถัดไป",
     ["ก. Ctrl + C", "ข. Alt + F4", "ค. Shift + Enter", "ง. Ctrl + Z"], "ค"),
    ("(5 + 30) * 20 มีค่าเท่าใด",
     ["ก. 605", "ข. 700", "ค. 650", "ง. 125"], "ข"),
]

def run_quiz():
    score = 0
    for i, (q, choices, ans) in enumerate(quiz, 1):
        print(f"\nข้อ {i}. {q}")
        for c in choices:
            print("   ", c)
        reply = input("คำตอบ (ก/ข/ค/ง): ").strip()
        if reply == ans:
            print("   ✅ ถูกต้อง"); score += 1
        else:
            print(f"   ❌ ยังไม่ถูก คำตอบที่ถูกคือ {ans}")
    print(f"\n🎯 คะแนนของคุณ: {score}/{len(quiz)}")
    print("🏆 ยอดเยี่ยม!" if score == len(quiz) else "👍 ดีมาก" if score >= 7 else "📖 ทบทวนเนื้อหาอีกครั้งนะ")

run_quiz()
````
