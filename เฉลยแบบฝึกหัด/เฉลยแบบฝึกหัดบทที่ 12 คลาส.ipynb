````python
# ▶️ รันเซลล์นี้ก่อน (Shift + Enter) เพื่อโหลดระบบตรวจคำตอบและไฟล์โมดูลทดสอบ
import hashlib, io, contextlib, copy, os, sys, importlib

def _setup_ch12_fixtures():
    """สร้างไฟล์โมดูลจำลองสำหรับทำแบบฝึกหัดบทที่ 12"""
    import os, sys, importlib
    if "" not in sys.path and "." not in sys.path:
        sys.path.insert(0, os.getcwd())
    with open("shop_module.py", "w", encoding="utf-8") as f:
        f.write(
            '"""โมดูลจำลองสินค้าและตะกร้าสินค้าสำหรับแบบฝึกหัดบทที่ 12"""\n\n'
            'class Item:\n'
            '    """คลาสแทนสินค้า 1 รายการ"""\n'
            '    def __init__(self, name, price, qty=1):\n'
            '        self.name = name\n'
            '        self.price = price\n'
            '        self.qty = qty\n\n'
            '    def subtotal(self):\n'
            '        return self.price * self.qty\n\n'
            '    def get_info(self):\n'
            '        return f"{self.name} x{self.qty} = {self.subtotal()}"\n\n'
            'class DiscountItem(Item):\n'
            '    """คลาสสินค้าลดราคา สืบทอดจาก Item"""\n'
            '    def __init__(self, name, price, qty=1, discount_pct=10):\n'
            '        super().__init__(name, price, qty)\n'
            '        self.discount_pct = discount_pct\n\n'
            '    def subtotal(self):\n'
            '        full = super().subtotal()\n'
            '        return full * (100 - self.discount_pct) / 100\n'
        )
    importlib.invalidate_caches()
    if "shop_module" in sys.modules:
        importlib.reload(sys.modules["shop_module"])

_setup_ch12_fixtures()

_SALT = 'py3-ch12-v1'
_KEY = {'A1': '3f20a9b3f6dcd0ab', 'A2': 'ce23df56d3513e35', 'A3': 'ff5dae244c2cb3b8', 'A4': '8ea48fa01eeeb2ff', 'A5': 'f9a94bc51fdee6a8', 'A6': '5a990c12e2bd8e14', 'A7': 'd6b98a550e0793db', 'A8': '43db783c730ecf32', 'A9': '3710cfdfef2c29fa', 'A10': '8435ece7e456f687', 'A11': 'c3d2ade073714f44', 'A12': 'ad4ef32bec1b0ad6', 'A13': '54f5f2f974e50d3b', 'A14': '720d3114544ac5be', 'A15': '3ea661200fa333b7', 'A16': '9f33a572b4c5db1c', 'A17': '622b03b2188bc45b', 'A18': '0db9f0a39534df74', 'A19': 'aa548fcb36ba0c87', 'A20': '968b3b1a83eae748', 'B1': '46ea7ab341354398', 'B2': '244347239d3f8538', 'B3': 'ef060a97f1bfba64', 'B4': '7949d19cbc9bab87', 'B5': 'e8ec1be8fa8027d6', 'B6': '71862e167ca3a071', 'B7': 'a725a1c9729dcf70', 'B8': '9fb22d0792135969', 'B9': 'aa7383ad33a1ede4', 'B10': 'b1679a7b969666ff', 'C1': '000eb827d46ba95f', 'C2': 'cb5b2ca593f341a6', 'C3': '12b667fc93bfdba4', 'C4': '7f9a47c990779e20', 'C5': '9f425a77125136f3', 'C6': '6bd9b363793142cb', 'C7': '98bd0ac42e60ce6f', 'C8': '586d23c7cabaff58', 'C9': '27765a7622fa7bc6', 'C10': '40ac04ffd11917ee', 'E1': ['3ff722a8a92bcea8', 'a3bae93be7fbae27', 'f88b744ae385037c'], 'E2': ['dc5c2cd85dbe488c', 'd87f2ca5c7b88b37', 'b6734dc9df5820be'], 'E3': ['7f6b35b1e68e34de', '582bffc98f3af864', '2355fcf779e35b6d'], 'E4': ['1be7c376bae93f61', '2a2f813feb6f89b2', '166b5903910beb6d'], 'E5': ['2411226a29b70755', '02514611bb25a1b3', '54bc8c03a9f4ae7a'], 'E6': ['7789d32ea2c5c2d2', '9415b3cf470d53ca', '486b205e1c114351'], 'E7': ['3310d82d7e595d91', '3d2f639d5cd1251e', 'c758dd98b83bdf8b'], 'E8': ['eaf4a3b8cea2fa5a', 'c3ff744fbebcb9c0'], 'E9': ['ce25382b05137125', 'c8ca76e1d56e257e', 'e9f3797a63c5939a'], 'E10': ['198caf26f38f144c', '3f1baa855c9c10fb'], 'E11': ['0e7036d20bad9e12', '4e2c3fe4793e1263', '022c8fbf76445de9'], 'E12': ['6b60844551f87a75', 'ad8b8d60a1bd4b34', 'a27c7f6fcb6f0b35']}
_TOPIC = {'A1': '12.1.1', 'A2': '12.1.1', 'A3': '12.1.1', 'A4': '12.1.1, 12.1.2', 'A5': '12.1.3, 12.1.4', 'A6': '12.1.5', 'A7': '12.1.5', 'A8': '12.2.2', 'A9': '12.2.3, 12.2.4', 'A10': '12.2.4', 'A11': '12.2.4', 'A12': '12.2.4', 'A13': '12.3.1, 12.3.2', 'A14': '12.3.2', 'A15': '12.3.2', 'A16': '12.3.4', 'A17': '12.3.5', 'A18': '12.4.2', 'A19': '12.4.1, 12.4.3, 12.4.4', 'A20': '12.4.5', 'B1': '12.1.1', 'B2': '12.1.1, 12.1.4', 'B3': '12.1.5', 'B4': '12.2.2', 'B5': '12.2.4', 'B6': '12.3.1, 12.3.2', 'B7': '12.3.3', 'B8': '12.3.5', 'B9': '12.4.2', 'B10': '12.4.4', 'C1': '12.1.1, 12.1.3, 12.1.4', 'C2': '12.1.5', 'C3': '12.2.2, 12.2.3, 12.2.4', 'C4': '12.2.4', 'C5': '12.2.4', 'C6': '12.2.4', 'C7': '12.3.2, 12.3.3', 'C8': '12.3.5', 'C9': '12.3.4', 'C10': '12.4.1, 12.4.2', 'E1': '12.1.1, 12.1.2, 12.1.3, 12.1.4', 'E2': '12.1.4, 12.1.5', 'E3': '12.2.1, 12.2.2, 12.2.4', 'E4': '12.2.4', 'E5': '12.2.4', 'E6': '12.2.4', 'E7': '12.3.1, 12.3.2, 12.3.3', 'E8': '12.3.2, 12.3.5', 'E9': '12.3.5', 'E10': '12.3.4', 'E11': '12.4.2', 'E12': '12.4.1, 12.4.2, 12.4.3'}
_PARTS = [('ตอนที่ 1 ปรนัย', ['A1', 'A2', 'A3', 'A4', 'A5', 'A6', 'A7', 'A8', 'A9', 'A10', 'A11', 'A12', 'A13', 'A14', 'A15', 'A16', 'A17', 'A18', 'A19', 'A20']), ('ตอนที่ 2 ถูก/ผิด', ['B1', 'B2', 'B3', 'B4', 'B5', 'B6', 'B7', 'B8', 'B9', 'B10']), ('ตอนที่ 3 ทำนายผลลัพธ์', ['C1', 'C2', 'C3', 'C4', 'C5', 'C6', 'C7', 'C8', 'C9', 'C10']), ('ตอนที่ 5 เขียนโปรแกรม', ['E1', 'E2', 'E3', 'E4', 'E5', 'E6', 'E7', 'E8', 'E9', 'E10', 'E11', 'E12'])]
_SNIP = {'C1': 'class Dog:\n    def __init__(self, name, age):\n        self.name = name\n        self.age = age\n\n    def sit(self):\n        print(f"{self.name.title()} is sitting.")\n\nd1 = Dog("kanoon", 3)\nd2 = Dog("kati", 2)\nprint(d1.name, d1.age)\nd2.sit()', 'C2': 'class Book:\n    def __init__(self, title, pages):\n        self.title = title\n        self.pages = pages\n\n    def __str__(self):\n        return f"{self.title} ({self.pages}p)"\n\nb = Book("Python 3", 320)\nprint(b)', 'C3': 'class Counter:\n    def __init__(self, start=0):\n        self.value = start\n\n    def add(self, n=1):\n        self.value += n\n\nc = Counter(10)\nc.add(5)\nc.add()\nc.value = 4\nc.add(3)\nprint(c.value)', 'C4': 'class Car:\n    def __init__(self):\n        self.odometer = 100\n\n    def update_odometer(self, mileage):\n        if mileage >= self.odometer:\n            self.odometer = mileage\n        else:\n            print("Reject!")\n\nc = Car()\nc.update_odometer(150)\nprint(c.odometer)\nc.update_odometer(120)\nprint(c.odometer)', 'C5': 'class Student:\n    uni = "SSKRU"\n    count = 0\n\n    def __init__(self, name):\n        self.name = name\n        Student.count += 1\n\ns1 = Student("Ann")\ns2 = Student("Bob")\ns1.uni = "MIT"\nprint(s1.uni, s2.uni, Student.count)', 'C6': 'class Box:\n    def __init__(self):\n        self._tag = "normal"\n        self.__code = 42\n\n    def reveal(self):\n        return self.__code\n\nb = Box()\nprint(b._tag)\nprint(b.reveal())\nprint(hasattr(b, "__code"), hasattr(b, "_Box__code"))', 'C7': 'class Vehicle:\n    def __init__(self, brand):\n        self.brand = brand\n\n    def info(self):\n        return f"Brand: {self.brand}"\n\nclass Bike(Vehicle):\n    def __init__(self, brand, gear):\n        super().__init__(brand)\n        self.gear = gear\n\n    def info(self):\n        return f"{super().info()}, Gear: {self.gear}"\n\nb = Bike("Trek", 21)\nprint(b.info())\nprint(isinstance(b, Vehicle), issubclass(Bike, Vehicle))', 'C8': 'class A:\n    def who(self):\n        return "A"\n\n    def greet(self):\n        return "Hello from " + self.who()\n\nclass B(A):\n    def who(self):\n        return "B"\n\nprint(A().greet())\nprint(B().greet())', 'C9': 'class Flyable:\n    def move(self):\n        return "Fly"\n\nclass Swimmable:\n    def move(self):\n        return "Swim"\n\nclass Duck(Flyable, Swimmable):\n    pass\n\nclass Penguin(Swimmable, Flyable):\n    pass\n\nprint(Duck().move(), Penguin().move())\nprint([cls.__name__ for cls in Duck.__mro__[:3]])', 'C10': 'from shop_module import Item, DiscountItem\n\nclass Cart:\n    def __init__(self):\n        self.items = []\n\n    def add(self, item):\n        self.items.append(item)\n\n    def total(self):\n        return sum(x.subtotal() for x in self.items)\n\ncart = Cart()\ncart.add(Item("Book", 100, 2))\ncart.add(DiscountItem("Pen", 50, 2, 20))\nprint(cart.items[0].get_info())\nprint(cart.total())'}
_TESTS = {'E1': [(('Milo', 'dog', 2), {}), (('Luna', 'cat', 1), {}), (('Coco', 'parrot', 5), {})], 'E2': [((4, 5), {}), ((6, 6), {}), ((10, 2), {})], 'E3': [(('audi', 'a4', 2020, 100, 80, 50), {}), (('honda', 'civic', 2022, 200, 250, -10), {}), (('toyota', 'yaris', 2019, 50, 50, 25), {})], 'E4': [(('Ann', 100, [('D', 200), ('W', 50), ('W', 500), ('D', -10)]), {}), (('Bob', 0, [('W', 10), ('D', 500), ('W', 120)]), {}), (('Cat', 300, [('W', 300), ('D', 50)]), {})], 'E5': [((['Somchai', 'Somsri'],), {}), ((['Alice', 'Bob', 'Charlie'],), {}), (([],), {})], 'E6': [(('1234', 500, '1234', 150), {}), (('1234', 500, '0000', 150), {}), (('9999', 200, '9999', 300), {})], 'E7': [(('Tesla', 'Model 3', 2023, 60), {}), (('BYD', 'Atto 3', 2024, 50), {}), (('Nissan', 'Leaf', 2021, 40), {})], 'E8': [(('Koi', 20000, 'Ek', 50000, 100000), {}), (('Ann', 30000, 'Bob', 40000, 50000), {})], 'E9': [(([('C', 10), ('S', 5)],), {}), (([('S', 4), ('S', 3), ('C', 2)],), {}), (([],), {})], 'E10': [(('0812345678',), {}), (('191',), {})], 'E11': [(('Model S', 75), {}), (('Model X', 100), {}), (('Model 3', 60), {})], 'E12': [(([('Mouse', 300, 1, 0), ('Keyboard', 1000, 2, 15)],), {}), (([('Book', 250, 2, 10), ('Pen', 20, 5, 0)],), {}), (([],), {})]}
_results = {}

def _canon(v):
    if isinstance(v, (set, frozenset)):
        return "set{" + ",".join(sorted(_canon(x) for x in v)) + "}"
    if isinstance(v, dict):
        return "dict{" + ",".join(sorted(_canon(k) + ":" + _canon(x) for k, x in v.items())) + "}"
    if isinstance(v, (list, tuple)):
        return type(v).__name__ + "(" + ",".join(_canon(x) for x in v) + ")"
    if isinstance(v, bool) or v is None:
        return repr(v)
    if isinstance(v, (int, float)):
        return "num:" + format(round(float(v), 6) + 0.0, ".6f")
    return type(v).__name__ + ":" + repr(v)

def _h(qid, value):
    return hashlib.sha256(f"{_SALT}|{qid}|{value}".encode("utf-8")).hexdigest()[:16]

def _norm_text(s):
    lines = [ln.rstrip() for ln in str(s).replace("\r\n", "\n").split("\n")]
    while lines and lines[0] == "":
        lines.pop(0)
    while lines and lines[-1] == "":
        lines.pop()
    return "\n".join(lines)

def _norm_choice(prefix, ans):
    a = str(ans).strip().upper()
    if prefix == "B":
        if a in ("T", "TRUE", "ถูก", "ท", "1"): return "T"
        if a in ("F", "FALSE", "ผิด", "ผ", "0"): return "F"
    return a

def grade(part, answers):
    """ตรวจคำตอบตอนที่ 1 (part='A') และตอนที่ 2 (part='B')"""
    n_ok, n_all = 0, 0
    for n in sorted(answers):
        qid = f"{part}{n}"
        if qid not in _KEY:
            continue
        n_all += 1
        a = _norm_choice(part, answers[n])
        if a == "":
            print(f"ข้อ {n:>2}: ⚪ ยังไม่ได้ตอบ")
            _results[qid] = False
            continue
        ok = _h(qid, a) == _KEY[qid]
        _results[qid] = ok
        n_ok += ok
        print(f"ข้อ {n:>2}: {'✅ ถูกต้อง' if ok else '❌ ยังไม่ถูก  (ทบทวนหัวข้อ ' + _TOPIC[qid] + ')'}")
    print("-" * 40)
    print(f"ได้ {n_ok} / {n_all} คะแนน")

def predict(qid, prediction):
    """ตอนที่ 3: เปรียบเทียบผลลัพธ์ที่ทำนายกับผลลัพธ์จริง (บันทึกคะแนนจากการตรวจครั้งแรก)"""
    _setup_ch12_fixtures()
    pred = _norm_text(prediction)
    if pred == "":
        print("⚪ ยังไม่ได้เขียนคำตอบที่ทำนายไว้ในตัวแปร")
        return
    buf = io.StringIO()
    with contextlib.redirect_stdout(buf):
        exec(_SNIP[qid], {})
    actual = _norm_text(buf.getvalue())
    ok = _h(qid, pred) == _KEY[qid]
    first = qid not in _results
    if first:
        _results[qid] = ok
    print("✅ ทำนายถูกต้อง" if ok else "❌ ทำนายไม่ตรงกับผลลัพธ์จริง")
    print("\nผลลัพธ์ที่ทำนาย:")
    print(pred)
    print("\nผลลัพธ์จริงจากการรันโค้ด:")
    print(actual)
    if not ok:
        print("\n(สังเกตเครื่องหมาย ช่องว่าง วงเล็บ และเครื่องหมายคำพูดให้ดี — ทบทวนหัวข้อ " + _TOPIC[qid] + ")")
    if not first:
        print("\nℹ️ คะแนนข้อนี้บันทึกจากการตรวจครั้งแรกไปแล้ว")

def check(qid, value):
    """ตอนที่ 5: ตรวจค่าที่ได้จากโปรแกรม"""
    ok = _h(qid, _canon(value)) == _KEY[qid]
    _results[qid] = ok
    if ok:
        print("✅ ถูกต้อง เยี่ยมมาก!")
    else:
        print("❌ ยังไม่ถูกต้อง ค่าที่ได้คือ:")
        print(repr(value))
        print("ตรวจสอบค่าและชนิดข้อมูลของแต่ละตัวแปรอีกครั้ง (ทบทวนหัวข้อ " + _TOPIC[qid] + ")")

def check_func(qid, func):
    """ตอนที่ 5: ทดสอบฟังก์ชันและคลาสด้วยข้อมูลทดสอบหลายชุด"""
    _setup_ch12_fixtures()
    if not callable(func):
        print("❌ ยังไม่ได้สร้างฟังก์ชัน")
        _results[qid] = False
        return
    tests = _TESTS[qid]
    n_ok = 0
    for k, (args, kwargs) in enumerate(tests):
        call = func.__name__ + "(" + ", ".join([repr(a) for a in args] +
                                              [f"{kk}={vv!r}" for kk, vv in kwargs.items()]) + ")"
        try:
            got = func(*copy.deepcopy(args), **copy.deepcopy(kwargs))
        except Exception as e:
            print(f"❌ {call}  →  เกิดข้อผิดพลาด {type(e).__name__}: {e}")
            continue
        ok = _h(f"{qid}#{k}", _canon(got)) == _KEY[qid][k]
        n_ok += ok
        print(f"{'✅' if ok else '❌'} {call}  →  {got!r}")
    _results[qid] = n_ok == len(tests)
    print("-" * 40)
    if n_ok == len(tests):
        print(f"✅ ผ่านการทดสอบครบทั้ง {len(tests)} กรณี เยี่ยมมาก!")
    else:
        print(f"❌ ผ่าน {n_ok} จาก {len(tests)} กรณี — ตรวจสอบกรณีที่ไม่ผ่าน (ทบทวนหัวข้อ " + _TOPIC[qid] + ")")
        print("   อย่าลืมว่าฟังก์ชันต้อง return ผลลัพธ์ ไม่ใช่ print")

def summary():
    """สรุปผลการประเมินตนเอง"""
    total_ok, total_all, review = 0, 0, []
    print(f"{'ส่วน':<28}{'คะแนน':>10}")
    print("=" * 38)
    for name, qids in _PARTS:
        ok = sum(1 for q in qids if _results.get(q))
        total_ok += ok
        total_all += len(qids)
        print(f"{name:<28}{ok:>5} / {len(qids):<3}")
        review += [q for q in qids if not _results.get(q)]
    print("=" * 38)
    pct = 100 * total_ok / total_all
    print(f"{'รวม (ส่วนที่ตรวจอัตโนมัติ)':<28}{total_ok:>5} / {total_all:<3}  ({pct:.1f}%)")
    print()
    if pct >= 80:
        print("🌟 ดีมาก — เข้าใจเนื้อหาบทที่ 12 เป็นอย่างดี ยินดีด้วยที่เรียนครบทุกบท!")
    elif pct >= 60:
        print("👍 ดี — เข้าใจเนื้อหาส่วนใหญ่ ควรทบทวนหัวข้อที่ยังทำไม่ได้")
    else:
        print("📖 ควรทบทวนบทเรียนอีกครั้ง แล้วกลับมาทำแบบฝึกหัดใหม่")
    topics = sorted({t.strip() for q in review for t in _TOPIC[q].split(",")},
                    key=lambda s: [int(x) for x in s.split(".")])
    if topics:
        print("\nหัวข้อที่ควรทบทวน:", ", ".join(topics))
    print("\n* ตอนที่ 4 และตอนที่ 6 ให้ตรวจสอบกับผลลัพธ์ที่โจทย์กำหนด/ผู้สอน")

print("✅ ระบบตรวจคำตอบและโมดูลทดสอบ (shop_module.py) พร้อมใช้งาน — เริ่มทำแบบฝึกหัดได้เลย")
````

    def __init__(self, name, age):

    def sit(self):
        print(f"{self.name.title()} is sitting.")

d2.sit()
```

class Book:

```

**ข้อ E1.** (การสร้างคลาส อินสแตนซ์ และเมธอดพื้นฐาน)
จงสร้างคลาส `Pet` ที่มีรายละเอียดดังนี้
- `__init__(self, name, species, age)` เก็บแอตทริบิวต์ `name`, `species` และ `age`
- `describe(self)` คืนค่าสตริงรูปแบบ `f"{self.name} is a {self.age}-year-old {self.species}."`
- `birthday(self)` เพิ่มค่า `self.age` ขึ้น `1` ปี แล้วคืนค่าอายุใหม่ (`self.age`)

และเขียนฟังก์ชัน `test_pet(name, species, age)` ที่สร้างอินสแตนซ์ `p = Pet(name, species, age)` บันทึกข้อความจาก `p.describe()` ก่อนวันเกิด ต่อด้วยเรียก `p.birthday()` แล้วคืนค่าเป็นทูเพิล `(ข้อความก่อนวันเกิด, อายุหลังวันเกิด, p.describe())`
```
test_pet("Milo", "dog", 2)
→ ('Milo is a 2-year-old dog.', 3, 'Milo is a 3-year-old dog.')
```

````python
class Pet:
    def __init__(self, name, species, age):
        self.name = name
        self.species = species
        self.age = age

    def describe(self):
        return f"{self.name} is a {self.age}-year-old {self.species}."

    def birthday(self):
        self.age += 1
        return self.age

def test_pet(name, species, age):
    p = Pet(name, species, age)
    before = p.describe()
    new_age = p.birthday()
    return (before, new_age, p.describe())

print(test_pet("Milo", "dog", 2))
check_func("E1", test_pet)
````

**ข้อ E2.** (เมธอดคำนวณและเมธอดพิเศษ `__str__`)
จงสร้างคลาส `Rectangle` แทนรูปสี่เหลี่ยมผืนผ้า
- `__init__(self, width, height)` เก็บแอตทริบิวต์ `width` และ `height`
- `area(self)` คืนค่าพื้นที่ (`width * height`)
- `perimeter(self)` คืนค่าความยาวรอบรูป (`2 * (width + height)`)
- `is_square(self)` คืนค่า `True` เมื่อ `width == height` มิฉะนั้นคืนค่า `False`
- `__str__(self)` คืนค่าสตริงรูปแบบ `f"Rectangle({self.width}x{self.height})"`

และเขียนฟังก์ชัน `inspect_rect(w, h)` สร้างอินสแตนซ์ `r = Rectangle(w, h)` แล้วคืนค่าทูเพิล `(str(r), r.area(), r.perimeter(), r.is_square())`
```
inspect_rect(4, 5)  →  ('Rectangle(4x5)', 20, 18, False)
inspect_rect(6, 6)  →  ('Rectangle(6x6)', 36, 24, True)
```

````python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

    def is_square(self):
        return self.width == self.height

    def __str__(self):
        return f"Rectangle({self.width}x{self.height})"

def inspect_rect(w, h):
    r = Rectangle(w, h)
    return (str(r), r.area(), r.perimeter(), r.is_square())

print(inspect_rect(4, 5))
print(inspect_rect(6, 6))
check_func("E2", inspect_rect)
````

**ข้อ E3.** (การตั้งค่าเริ่มต้นและการแก้ไขค่าผ่านเมธอดพร้อมตรวจสอบเงื่อนไข)
จงสร้างคลาส `CarOdometer`
- `__init__(self, make, model, year)` เก็บแอตทริบิวต์ `make`, `model`, `year` และตั้งค่าเริ่มต้น `self.odometer = 0`
- `get_name(self)` คืนค่าสตริง `f"{self.year} {self.make} {self.model}".title()`
- `update_odometer(self, mileage)` หาก `mileage >= self.odometer` ให้อัปเดต `self.odometer = mileage` แล้วคืนค่า `True` แต่ถ้าน้อยกว่าเดิม **ห้ามเปลี่ยนค่า** และคืนค่า `False`
- `increment_odometer(self, miles)` หาก `miles > 0` ให้บวกเพิ่ม `self.odometer += miles` แล้วคืนค่า `True` แต่ถ้า `miles <= 0` ห้ามเปลี่ยนค่าและคืนค่า `False`

และเขียนฟังก์ชัน `simulate_car(make, model, year, m1, m2, inc)` ที่สร้าง `c = CarOdometer(make, model, year)` แล้วเรียก `r1 = c.update_odometer(m1)`, `r2 = c.update_odometer(m2)`, `r3 = c.increment_odometer(inc)` ตามลำดับ คืนค่าเป็นทูเพิล `(c.get_name(), c.odometer, [r1, r2, r3])`
```
simulate_car("audi", "a4", 2020, 100, 80, 50)
→ ('2020 Audi A4', 150, [True, False, True])
```

````python
class CarOdometer:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
        self.odometer = 0

    def get_name(self):
        return f"{self.year} {self.make} {self.model}".title()

    def update_odometer(self, mileage):
        if mileage >= self.odometer:
            self.odometer = mileage
            return True
        return False

    def increment_odometer(self, miles):
        if miles > 0:
            self.odometer += miles
            return True
        return False

def simulate_car(make, model, year, m1, m2, inc):
    c = CarOdometer(make, model, year)
    r1 = c.update_odometer(m1)
    r2 = c.update_odometer(m2)
    r3 = c.increment_odometer(inc)
    return (c.get_name(), c.odometer, [r1, r2, r3])

print(simulate_car("audi", "a4", 2020, 100, 80, 50))
check_func("E3", simulate_car)
````

**ข้อ E4.** (การจัดการสถานะและประวัติธุรกรรมในอินสแตนซ์ — คลาส `BankAccount`)
จงสร้างคลาส `BankAccount`
- `__init__(self, owner, balance=0)` เก็บ `owner`, `balance` และสร้างลิสต์ว่าง `self.history = []` แยกเฉพาะของแต่ละบัญชี
- `deposit(self, amount)` ถ้า `amount > 0` ให้เพิ่ม `balance`, ต่อท้าย `("D", amount)` ลงใน `self.history` และคืนค่า `True` (ถ้า `amount <= 0` ไม่เปลี่ยนค่าและคืน `False`)
- `withdraw(self, amount)` ถ้า `0 < amount <= self.balance` ให้หัก `balance`, ต่อท้าย `("W", amount)` ลงใน `self.history` และคืนค่า `True` (มิฉะนั้นไม่เปลี่ยนค่าและคืน `False`)

และเขียนฟังก์ชัน `run_account(owner, initial, actions)` รับ `actions` เป็นลิสต์ของทูเพิล `(ชนิดธุรกรรม, จำนวนเงิน)` เช่น `[("D", 200), ("W", 50)]` ให้เรียก `deposit` เมื่อชนิดเป็น `"D"` และ `withdraw` เมื่อเป็น `"W"` ตามลำดับ แล้วคืนค่าทูเพิล `(acc.owner, acc.balance, acc.history)`
```
run_account("Ann", 100, [("D", 200), ("W", 50), ("W", 500), ("D", -10)])
→ ('Ann', 250, [('D', 200), ('W', 50)])
```

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
        self.history = []

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            self.history.append(("D", amount))
            return True
        return False

    def withdraw(self, amount):
        if 0 < amount <= self.balance:
            self.balance -= amount
            self.history.append(("W", amount))
            return True
        return False

def run_account(owner, initial, actions):
    acc = BankAccount(owner, initial)
    for act, amt in actions:
        if act == "D":
            acc.deposit(amt)
        elif act == "W":
            acc.withdraw(amt)
    return (acc.owner, acc.balance, acc.history)

print(run_account("Ann", 100, [("D", 200), ("W", 50), ("W", 500), ("D", -10)]))
check_func("E4", run_account)
````

**ข้อ E5.** (แอตทริบิวต์ของคลาส Class Attribute vs แอตทริบิวต์ของอินสแตนซ์)
จงสร้างคลาส `Member`
- มีแอตทริบิวต์ของคลาส `club = "Python Club"` และ `total_members = 0`
- `__init__(self, name)` เพิ่มค่า `Member.total_members` ขึ้น `1` ทุกครั้งที่สร้างอินสแตนซ์ แล้วกำหนด `self.name = name` และรหัสสมาชิก `self.member_id = Member.total_members`
- `badge(self)` คืนค่าสตริง `f"#{self.member_id} {self.name} ({self.club})"`

และเขียนฟังก์ชัน `register_members(names)` ที่เริ่มต้นด้วยการรีเซ็ต `Member.total_members = 0` จากนั้นสร้างอินสแตนซ์ `Member` ตามรายชื่อในลิสต์ `names` ตามลำดับ แล้วคืนค่าทูเพิล `(Member.total_members, [m.badge() for m in members])`
```
register_members(["Somchai", "Somsri"])
→ (2, ['#1 Somchai (Python Club)', '#2 Somsri (Python Club)'])
```

````python
class Member:
    club = "Python Club"
    total_members = 0

    def __init__(self, name):
        Member.total_members += 1
        self.name = name
        self.member_id = Member.total_members

    def badge(self):
        return f"#{self.member_id} {self.name} ({self.club})"

def register_members(names):
    Member.total_members = 0
    members = [Member(n) for n in names]
    return (Member.total_members, [m.badge() for m in members])

print(register_members(["Somchai", "Somsri"]))
check_func("E5", register_members)
````

**ข้อ E6.** (การห่อหุ้มข้อมูล Encapsulation ด้วย `__` Name Mangling)
จงสร้างคลาส `SafeWallet`
- `__init__(self, pin, balance=0)` เก็บแอตทริบิวต์ส่วนตัว `self.__pin = pin` และ `self.__balance = balance`
- `get_balance(self, pin)` ถ้า `pin == self.__pin` ให้คืนค่ายอดเงิน `self.__balance` แต่ถ้ารหัสผิดให้คืนค่า `"Access Denied"`
- `spend(self, pin, amount)` ถ้ารหัสไม่ตรงกับ `self.__pin` ให้คืนค่า `"Access Denied"` ถ้ารหัสถูกแต่ `amount <= 0` หรือ `amount > self.__balance` ให้คืนค่า `"Invalid Amount"` มิฉะนั้นให้หักยอด `self.__balance -= amount` แล้วคืนค่ายอดคงเหลือใหม่

และเขียนฟังก์ชัน `test_wallet(pin, init_bal, try_pin, spend_amt)` สร้าง `w = SafeWallet(pin, init_bal)` แล้วคืนค่าทูเพิล `(hasattr(w, "__balance"), w.spend(try_pin, spend_amt), w.get_balance(pin))`
```
test_wallet("1234", 500, "1234", 150)  →  (False, 350, 350)
test_wallet("1234", 500, "0000", 150)  →  (False, 'Access Denied', 500)
```

````python
class SafeWallet:
    def __init__(self, pin, balance=0):
        self.__pin = pin
        self.__balance = balance

    def get_balance(self, pin):
        if pin == self.__pin:
            return self.__balance
        return "Access Denied"

    def spend(self, pin, amount):
        if pin != self.__pin:
            return "Access Denied"
        if amount <= 0 or amount > self.__balance:
            return "Invalid Amount"
        self.__balance -= amount
        return self.__balance

def test_wallet(pin, init_bal, try_pin, spend_amt):
    w = SafeWallet(pin, init_bal)
    return (hasattr(w, "__balance"), w.spend(try_pin, spend_amt), w.get_balance(pin))

print(test_wallet("1234", 500, "1234", 150))
print(test_wallet("1234", 500, "0000", 150))
check_func("E6", test_wallet)
````

**ข้อ E7.** (การสืบทอดคลาสและการใช้ `super()` เพิ่มแอตทริบิวต์/เมธอดในคลาสลูก)
จงสร้างคลาสแม่ `Vehicle` และคลาสลูก `ElectricVehicle(Vehicle)`
- `Vehicle.__init__(self, make, model, year)` เก็บแอตทริบิวต์ `make`, `model`, `year`
- `Vehicle.description(self)` คืนค่าสตริง `f"{self.year} {self.make} {self.model}"`
- `ElectricVehicle.__init__(self, make, model, year, battery_kwh=75)` เรียก `super().__init__(make, model, year)` และเก็บแอตทริบิวต์ `self.battery_kwh = battery_kwh`
- `ElectricVehicle.estimated_range(self)` คืนค่าระยะทางวิ่งได้เท่ากับ `self.battery_kwh * 5`

และเขียนฟังก์ชัน `check_ev(make, model, year, kwh)` สร้าง `ev = ElectricVehicle(make, model, year, kwh)` แล้วคืนค่าทูเพิล `(ev.description(), ev.estimated_range(), isinstance(ev, Vehicle), issubclass(ElectricVehicle, Vehicle))`
```
check_ev("Tesla", "Model 3", 2023, 60)
→ ('2023 Tesla Model 3', 300, True, True)
```

````python
class Vehicle:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def description(self):
        return f"{self.year} {self.make} {self.model}"

class ElectricVehicle(Vehicle):
    def __init__(self, make, model, year, battery_kwh=75):
        super().__init__(make, model, year)
        self.battery_kwh = battery_kwh

    def estimated_range(self):
        return self.battery_kwh * 5

def check_ev(make, model, year, kwh):
    ev = ElectricVehicle(make, model, year, kwh)
    return (ev.description(), ev.estimated_range(), isinstance(ev, Vehicle), issubclass(ElectricVehicle, Vehicle))

print(check_ev("Tesla", "Model 3", 2023, 60))
check_func("E7", check_ev)
````

**ข้อ E8.** (การลบล้างเมธอด Method Overriding และการเรียกต่อยอดด้วย `super()`)
จงสร้างคลาส `Employee` และคลาสลูก `Manager(Employee)`
- `Employee.__init__(self, name, salary)` เก็บชื่อ `name` และเงินเดือน `salary`
- `Employee.annual_pay(self)` คืนค่ารายได้ต่อปี `self.salary * 12`
- `Employee.summary(self)` คืนค่าสตริง `f"{self.name}: {self.annual_pay()}"`
- `Manager.__init__(self, name, salary, bonus)` เรียก `super().__init__(name, salary)` และเก็บ `self.bonus = bonus`
- `Manager.annual_pay(self)` ลบล้างเมธอดโดยคืนค่า `super().annual_pay() + self.bonus` (สังเกตว่าไม่ต้องเขียน `summary()` ซ้ำใน `Manager` เพราะ `summary()` เรียก `self.annual_pay()`)

และเขียนฟังก์ชัน `compare_pay(emp_name, emp_sal, mgr_name, mgr_sal, bonus)` สร้างอินสแตนซ์ของทั้งสองคลาสแล้วคืนค่าทูเพิล `(emp.summary(), mgr.summary(), mgr.annual_pay() - emp.annual_pay())`
```
compare_pay("Koi", 20000, "Ek", 50000, 100000)
→ ('Koi: 240000', 'Ek: 700000', 460000)
```

````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def annual_pay(self):
        return self.salary * 12

    def summary(self):
        return f"{self.name}: {self.annual_pay()}"

class Manager(Employee):
    def __init__(self, name, salary, bonus):
        super().__init__(name, salary)
        self.bonus = bonus

    def annual_pay(self):
        return super().annual_pay() + self.bonus

def compare_pay(emp_name, emp_sal, mgr_name, mgr_sal, bonus):
    emp = Employee(emp_name, emp_sal)
    mgr = Manager(mgr_name, mgr_sal, bonus)
    return (emp.summary(), mgr.summary(), mgr.annual_pay() - emp.annual_pay())

print(compare_pay("Koi", 20000, "Ek", 50000, 100000))
check_func("E8", compare_pay)
````

**ข้อ E9.** (พอลิมอร์ฟิซึม Polymorphism กับรูปทรงเรขาคณิต)
จงสร้างคลาสแม่ `Shape(name)` และคลาสลูก 2 คลาส ได้แก่
- `Circle(Shape)` รับ `radius` เรียก `super().__init__("Circle")` และมีเมธอด `area(self)` คืนค่า `round(3.14159 * self.radius ** 2, 2)`
- `Square(Shape)` รับ `side` เรียก `super().__init__("Square")` และมีเมธอด `area(self)` คืนค่า `round(self.side ** 2, 2)`

และเขียนฟังก์ชัน `summarize_shapes(specs)` รับลิสต์ของทูเพิล เช่น `[("C", 10), ("S", 5)]` (ถ้าตัวแรกเป็น `"C"` ให้สร้าง `Circle` ถ้าเป็น `"S"` ให้สร้าง `Square`) แล้วคืนค่าทูเพิล `(ลิสต์ของ f"{s.name}:{s.area()}", ผลรวมพื้นที่ทั้งหมดปัดทศนิยม 2 ตำแหน่งด้วย round(..., 2))`
```
summarize_shapes([("C", 10), ("S", 5)])
→ (['Circle:314.16', 'Square:25'], 339.16)
```

````python
class Shape:
    def __init__(self, name):
        self.name = name

    def area(self):
        return 0

class Circle(Shape):
    def __init__(self, radius):
        super().__init__("Circle")
        self.radius = radius

    def area(self):
        return round(3.14159 * self.radius ** 2, 2)

class Square(Shape):
    def __init__(self, side):
        super().__init__("Square")
        self.side = side

    def area(self):
        return round(self.side ** 2, 2)

def summarize_shapes(specs):
    shapes = [Circle(val) if kind == "C" else Square(val) for kind, val in specs]
    labels = [f"{s.name}:{s.area()}" for s in shapes]
    total = round(sum(s.area() for s in shapes), 2)
    return (labels, total)

print(summarize_shapes([("C", 10), ("S", 5)]))
check_func("E9", summarize_shapes)
````

**ข้อ E11.** (ความสัมพันธ์แบบมีส่วนประกอบ Composition `has-a`)
จงสร้างคลาส `Battery` และคลาส `ElectricCar`
- `Battery.__init__(self, size=75)` เก็บแอตทริบิวต์ `self.size = size`
- `Battery.upgrade(self)` ถ้า `self.size < 100` ให้เปลี่ยน `self.size = 100`
- `Battery.get_range(self)` คืนค่า `260` เมื่อ `self.size == 75` และคืนค่า `315` เมื่อ `self.size == 100` (กรณีอื่นคืน `self.size * 3`)
- `ElectricCar.__init__(self, model, battery_size=75)` เก็บ `self.model = model` และสร้างอินสแตนซ์ `self.battery = Battery(battery_size)` เป็นแอตทริบิวต์

และเขียนฟังก์ชัน `test_battery_upgrade(model, initial_size)` สร้าง `ecar = ElectricCar(model, initial_size)` บันทึกระยะทางก่อนอัปเกรด จากนั้นเรียก `ecar.battery.upgrade()` แล้วคืนค่าทูเพิล `(ecar.model, ระยะทางก่อนอัปเกรด, ecar.battery.size, ecar.battery.get_range())`
```
test_battery_upgrade("Model S", 75)  →  ('Model S', 260, 100, 315)
test_battery_upgrade("Model X", 100) →  ('Model X', 315, 100, 315)
```

````python
class Battery:
    def __init__(self, size=75):
        self.size = size

    def upgrade(self):
        if self.size < 100:
            self.size = 100

    def get_range(self):
        if self.size == 75:
            return 260
        elif self.size == 100:
            return 315
        return self.size * 3

class ElectricCar:
    def __init__(self, model, battery_size=75):
        self.model = model
        self.battery = Battery(battery_size)

def test_battery_upgrade(model, initial_size):
    ecar = ElectricCar(model, initial_size)
    before = ecar.battery.get_range()
    ecar.battery.upgrade()
    return (ecar.model, before, ecar.battery.size, ecar.battery.get_range())

print(test_battery_upgrade("Model S", 75))
check_func("E11", test_battery_upgrade)
````

**ข้อ E12.** (การนำเข้าคลาสจากโมดูล `shop_module` และประยุกต์ใช้งานร่วมกัน)
ระบบได้สร้างไฟล์โมดูล `shop_module.py` ซึ่งภายในมีคลาส `Item(name, price, qty=1)` และคลาสลูก `DiscountItem(name, price, qty=1, discount_pct=10)` ไว้ให้แล้ว
จงนำเข้าทั้งสองคลาสจาก `shop_module` แล้วสร้างคลาส `ShoppingCart`
- `__init__(self)` สร้างลิสต์ว่าง `self.items = []`
- `add_item(self, item)` เพิ่มวัตถุ `item` เข้าไปใน `self.items`
- `total_cost(self)` คืนค่าผลรวมของ `item.subtotal()` ของสินค้าทุกชิ้นในตะกร้า
- `receipt_lines(self)` คืนค่าลิสต์ของสตริง `item.get_info()` ของสินค้าทุกชิ้นตามลำดับ

และเขียนฟังก์ชัน `checkout(orders)` รับลิสต์ของทูเพิล `(name, price, qty, discount_pct)`
- ถ้า `discount_pct == 0` ให้สร้างเป็น `Item(name, price, qty)`
- ถ้า `discount_pct > 0` ให้สร้างเป็น `DiscountItem(name, price, qty, discount_pct)`
นำเข้าตะกร้า `ShoppingCart` แล้วคืนค่าทูเพิล `(cart.receipt_lines(), cart.total_cost())`
```
checkout([("Mouse", 300, 1, 0), ("Keyboard", 1000, 2, 15)])
→ (['Mouse x1 = 300', 'Keyboard x2 = 1700.0'], 2000.0)
```

````python
from shop_module import Item, DiscountItem

class ShoppingCart:
    def __init__(self):
        self.items = []

    def add_item(self, item):
        self.items.append(item)

    def total_cost(self):
        return sum(it.subtotal() for it in self.items)

    def receipt_lines(self):
        return [it.get_info() for it in self.items]

def checkout(orders):
    cart = ShoppingCart()
    for name, price, qty, disc in orders:
        if disc == 0:
            cart.add_item(Item(name, price, qty))
        else:
            cart.add_item(DiscountItem(name, price, qty, disc))
    return (cart.receipt_lines(), cart.total_cost())

print(checkout([("Mouse", 300, 1, 0), ("Keyboard", 1000, 2, 15)]))
check_func("E12", checkout)
````

**ข้อ 1.** จงอธิบายความหมายและความสัมพันธ์ระหว่าง **คลาส (Class)** และ **อินสแตนซ์ (Instance / Object)** พร้อมระบุองค์ประกอบหลักของคลาส และอธิบายหน้าที่ของเมธอด `__init__()` กับพารามิเตอร์ `self`

✏️ **แนวคำตอบ:** **คลาส (Class)** คือแม่แบบ (Template/Blueprint) ที่นิยามว่าวัตถุชนิดหนึ่งมีข้อมูลอะไร (**แอตทริบิวต์**) และทำพฤติกรรมอะไรได้บ้าง (**เมธอด**) ส่วน **อินสแตนซ์ (Instance)** คือวัตถุที่ถูกสร้างขึ้นจริงจากคลาสนั้น โดยคลาสเดียวสามารถสร้างอินสแตนซ์ได้ไม่จำกัดและแต่ละอินสแตนซ์จะมีค่าแอตทริบิวต์แยกเป็นอิสระของตัวเอง เมธอดพิเศษ `__init__()` จะถูกไพธอนเรียกใช้อัตโนมัติทันทีที่สร้างอินสแตนซ์เพื่อกำหนดค่าเริ่มต้นให้แอตทริบิวต์ โดยมีพารามิเตอร์ตัวแรกคือ `self` ซึ่งอ้างอิงถึงตัวอินสแตนซ์ที่กำลังถูกสร้างหรือเรียกใช้เมธอดนั้น

**ข้อ 2.** การแก้ไขค่าแอตทริบิวต์ของอินสแตนซ์สามารถทำได้กี่วิธี อะไรบ้าง และเหตุใดการแก้ไขค่าผ่านเมธอดร่วมกับการห่อหุ้มข้อมูล (Encapsulation) จึงปลอดภัยกว่าการแก้ไขค่าโดยตรง

✏️ **แนวคำตอบ:** ทำได้ **3 วิธี** ได้แก่
1. **แก้ไขค่าโดยตรงผ่านอินสแตนซ์** เช่น `my_car.odometer = 23`
2. **แก้ไขผ่านเมธอดที่กำหนดค่าใหม่** เช่น `my_car.update_odometer(50)`
3. **แก้ไขผ่านเมธอดที่เพิ่ม/ลดค่าทีละส่วน** เช่น `my_car.increment_odometer(100)`

การแก้ไขผ่านเมธอดควบคู่กับการตั้งชื่อแอตทริบิวต์แบบห่อหุ้ม (`_attr` หรือ `__attr`) ปลอดภัยกว่า เพราะเราสามารถเขียนเงื่อนไขตรวจสอบความถูกต้องของข้อมูล (Validation) ภายในเมธอดก่อนอนุญาตให้เปลี่ยนค่าได้ เช่น ป้องกันการกรอกระยะทางติดลบหรือการปรับลดเลขไมล์ย้อนหลัง

**ข้อ 3.** จงเปรียบเทียบความแตกต่างระหว่าง **แอตทริบิวต์ของอินสแตนซ์ (Instance Attribute)** และ **แอตทริบิวต์ของคลาส (Class Attribute)** พร้อมยกตัวอย่างกรณีที่ควรเลือกใช้แอตทริบิวต์แต่ละแบบ และข้อควรระวังเมื่อใช้ชนิดข้อมูลเปลี่ยนแปลงค่าได้ (เช่น `list`) เป็นแอตทริบิวต์ของคลาส

✏️ **แนวคำตอบ:** - **Instance Attribute** กำหนดผ่าน `self.ชื่อ = ค่า` (มักอยู่ใน `__init__`) โดยแต่ละอินสแตนซ์จะมีสำเนาข้อมูลเป็นของตัวเอง เหมาะกับข้อมูลเฉพาะตัว เช่น ชื่อนักศึกษา ยอดเงินในบัญชี หรือเลขไมล์ของรถแต่ละคัน
- **Class Attribute** ประกาศไว้ในบล็อกของคลาสนอกเมธอด ใช้ค่าร่วมกันทุกอินสแตนซ์ของคลาสนั้น เหมาะกับค่าคงที่ที่ใช้ร่วมกัน (เช่น ชื่อมหาวิทยาลัย อัตราดอกเบี้ยมาตรฐาน) หรือตัวนับจำนวนอินสแตนซ์ทั้งหมด (`count`)
- **ข้อควรระวัง:** ห้ามประกาศโครงสร้างข้อมูลแบบเปลี่ยนแปลงค่าได้ (Mutable) เช่น `history = []` เป็น Class Attribute หากต้องการให้แต่ละอินสแตนซ์เก็บข้อมูลแยกกัน เพราะทุกอินสแตนซ์จะแก้ไขลิสต์ใบเดียวกัน

**ข้อ 4.** การสืบทอดคลาส (**Inheritance**) คืออะไร มีประโยชน์อย่างไร และฟังก์ชัน `super()` มีบทบาทอย่างไรทั้งในเมธอด `__init__()` และในเมธอดทั่วไปที่ถูกลบล้าง (Method Overriding)

✏️ **แนวคำตอบ:** **การสืบทอดคลาส** คือการสร้างคลาสใหม่ (**คลาสลูก**) จากคลาสที่มีอยู่เดิม (**คลาสแม่**) โดยคลาสลูกจะได้รับแอตทริบิวต์และเมธอดทั้งหมดของคลาสแม่มาทันที ช่วยลดการเขียนโค้ดซ้ำซ้อน (Code Reusability) ขยายความสามารถเพิ่มได้โดยไม่กระทบคลาสเดิม และจัดโครงสร้างคลาสเป็นลำดับชั้นตามความสัมพันธ์แบบ `is-a`

ฟังก์ชัน **`super()`** ทำหน้าที่อ้างอิงถึงคลาสแม่:
- ใน `__init__()` ใช้เรียก `super().__init__(...)` เพื่อให้คลาสแม่กำหนดแอตทริบิวต์พื้นฐานให้ครบถ้วนก่อนที่คลาสลูกจะกำหนดแอตทริบิวต์เฉพาะของตนเพิ่ม
- ในเมธอดที่ทำ **Method Overriding** สามารถเรียก `super().ชื่อเมธอด()` เพื่อนำผลลัพธ์เดิมของคลาสแม่มาต่อยอดแทนการเขียนโค้ดซ้ำทั้งหมด

**ข้อ 5.** จงอธิบายแนวคิดของ **การลบล้างเมธอด (Method Overriding)** และ **พอลิมอร์ฟิซึม (Polymorphism)** พร้อมยกตัวอย่างว่าการสืบทอดจากหลายคลาส (Multiple Inheritance) มีหลักการค้นหาเมธอด (MRO) อย่างไรเมื่อคลาสแม่มีเมธอดชื่อซ้ำกัน

✏️ **แนวคำตอบ:** - **Method Overriding** คือการที่คลาสลูกนิยามเมธอดชื่อเดียวกับคลาสแม่เพื่อเปลี่ยนการทำงานให้สอดคล้องกับคลาสลูก เช่น `ElectricCar` ลบล้างเมธอด `fill_gas_tank()` ของ `Car` หรือ `Circle` ลบล้างเมธอด `area()` ของ `Shape`
- **Polymorphism (พอลิมอร์ฟิซึม)** คือการที่โค้ดชุดเดียวกันสามารถเรียกเมธอดชื่อเดียวกัน (เช่น `s.area()`) บนวัตถุหลายชนิดได้ โดยแต่ละวัตถุจะตอบสนองตามเมธอดของคลาสตนเองโดยไม่ต้องใช้ `if-elif` แยกชนิดคลาส
- ใน **การสืบทอดหลายคลาส** เช่น `class C(A, B):` ไพธอนใช้ลำดับการค้นหาเมธอด (**Method Resolution Order: MRO**) โดยค้นหาที่คลาสลูก `C` ก่อน หากไม่พบจะค้นหาจากคลาสแม่ตามลำดับที่ระบุในวงเล็บจากซ้ายไปขวา คือ `A` ก่อน แล้วจึงไป `B` (ตรวจสอบได้ด้วย `C.__mro__`)

**ข้อ 6.** จงเปรียบเทียบความสัมพันธ์แบบ **การสืบทอด (`is-a` / Inheritance)** กับ **การประกอบสร้าง (`has-a` / Composition)** ว่าควรเลือกใช้อย่างไร และอธิบายแนวทางปฏิบัติที่ดีในการจัดเก็บคลาสไว้ในโมดูล (`.py`) และการนำเข้า (`import`) มาใช้งาน

✏️ **แนวคำตอบ:** - **Inheritance (`is-a`)** ใช้เมื่อคลาสลูก 'เป็นชนิดหนึ่งของ' คลาสแม่ เช่น `ElectricCar` เป็น `Car` ส่วน **Composition (`has-a`)** ใช้เมื่อวัตถุหนึ่ง 'มี' อีกวัตถุหนึ่งเป็นองค์ประกอบย่อย เช่น `ElectricCar` มี `Battery` (`self.battery = Battery()`) การแยกส่วนประกอบที่ซับซ้อนออกเป็นคลาสย่อยด้วย Composition ช่วยให้แต่ละคลาสไม่ใหญ่เกินไปและดูแลรักษาง่าย
- **การจัดเก็บและนำเข้าคลาสจากโมดูล:** ควรแยกคลาสที่เกี่ยวข้องกันเก็บไว้ในไฟล์โมดูล `.py` และนำเข้าอย่างชัดเจนด้วย `from module import ClassA, ClassB` หรือ `import module` แล้วเรียก `module.ClassA` โดย **หลีกเลี่ยง** การใช้ `from module import *` เพื่อป้องกันปัญหาชื่อคลาสชนกันและทำให้โค้ดอ่านเข้าใจง่ายว่าแต่ละคลาสมาจากโมดูลใด
