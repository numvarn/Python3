````python
#@title ▶ รันเซลล์นี้ก่อนเริ่มทำแบบฝึกหัด (ระบบตรวจคำตอบอัตโนมัติ — ไม่ต้องแก้ไข)
import hashlib, io, contextlib

_SALT = "py3-ch5"
_KEY = {"A1": "08d8f0ac400cddb1", "A2": "e07a670a069a1af0", "A3": "969362c312c287c0", "A4": "6c66cf87490fc1f8", "A5": "611858f7b48b11b9", "A6": "2b787c90c0c7544b", "A7": "8b2356eb42f9239b", "A8": "b0174ea3cf76f57f", "A9": "e547b01f1241a216", "A10": "0b5793593488689a", "A11": "c7b18d209e0137ba", "A12": "bec0e8f783f5b767", "A13": "1f973a68dc5eb613", "A14": "44a19da9cb5d25c5", "A15": "2c50fc7b2883a443", "A16": "750d4067b12f3845", "A17": "a3a64dc6831e0f59", "A18": "4dfdaa9efea50620", "A19": "07793540bb14557b", "A20": "e2f58fdaf59a14e3", "B1": "0589a1006fdf0a46", "B2": "b6ab2410ccb73720", "B3": "54667aaec22b49ff", "B4": "bb0ee2a48a7d9aca", "B5": "fd3d0127a4ca21eb", "B6": "e3ec126f71517246", "B7": "aa4deb563f1df4eb", "B8": "8681f8c2c29f5ffd", "B9": "06a7ae4d4e56ee4e", "B10": "77ab851c8fe8aeb6", "C1": "088857d38d985d96", "C2": "993f8251545a1311", "C3": "3daaeeaf1f25b266", "C4": "23f500933e079730", "C5": "2183b42bfb32d11e", "C6": "90fbee034302e059", "C7": "d0ed90506399ddeb", "C8": "4acbffeeeb4180f2", "C9": "363e7b30cf50c3ec", "C10": "4e1e8f26fbce2d4d", "E1": "738beeb3383d0a66", "E2": "8211d033f6cca7e1", "E3": "21d8d37f76ae0ac3", "E4": "3c29dd7ebdd42aa5", "E5": "6a08753466666311", "E6": "f7ad2949d0d10e01", "E7": "af4fc3992f98f6a3", "E8": "b88b53622f26fce0", "E9": "798bbf2bed4dad41", "E10": "d40abf4acedec06d", "E11": "82b121f86eb46156"}
_TOPIC = {"A1": "5.1", "A2": "5.2.1", "A3": "5.2.1", "A4": "5.3.1", "A5": "5.3.2", "A6": "5.4.1", "A7": "5.4.1.1", "A8": "5.4.1.1", "A9": "5.4.1.2", "A10": "5.4.2.2", "A11": "5.4.2.2", "A12": "5.4.2.4", "A13": "5.4.2.6", "A14": "5.5.2.1", "A15": "5.5.2.5", "A16": "5.5.1.6", "A17": "5.5.3.1", "A18": "5.5.4", "A19": "5.5.4", "A20": "5.5.4.3", "B1": "5.3.1", "B2": "5.4.2.2", "B3": "5.4.2.3", "B4": "5.5.1.1", "B5": "5.5.1.2", "B6": "5.5.2.3", "B7": "5.5.4.2", "B8": "5.5.3", "B9": "5.5.4", "B10": "5.4.1", "C1": "5.3.2", "C2": "5.4.2.2", "C3": "5.4.2.5", "C4": "5.4.2.5", "C5": "5.4.2.4", "C6": "5.5.1.5", "C7": "5.5.1.6", "C8": "5.5.2.2, 5.5.2.5", "C9": "5.5.3.1, 5.5.3.2", "C10": "5.5.4.3", "E1": "5.4.1", "E2": "5.4.2.4", "E3": "5.4.2.5", "E4": "5.4.1.1, 5.4.1.2", "E5": "5.4.2.2, 5.4.2.6", "E6": "5.4.2.4", "E7": "5.4.2.6", "E8": "5.5.1.5, 5.5.1.6", "E9": "5.5.2", "E10": "5.5.3", "E11": "5.5.4, 5.5.4.3"}
_PARTS = [('ตอนที่ 1 ปรนัย', ['A1', 'A2', 'A3', 'A4', 'A5', 'A6', 'A7', 'A8', 'A9', 'A10', 'A11', 'A12', 'A13', 'A14', 'A15', 'A16', 'A17', 'A18', 'A19', 'A20']), ('ตอนที่ 2 ถูก/ผิด', ['B1', 'B2', 'B3', 'B4', 'B5', 'B6', 'B7', 'B8', 'B9', 'B10']), ('ตอนที่ 3 ทำนายผลลัพธ์', ['C1', 'C2', 'C3', 'C4', 'C5', 'C6', 'C7', 'C8', 'C9', 'C10']), ('ตอนที่ 5 เขียนโปรแกรม', ['E1', 'E2', 'E3', 'E4', 'E5', 'E6', 'E7', 'E8', 'E9', 'E10', 'E11'])]
_SNIP = {"C1": "a, b = 1, 2\na, b = b, a + b\nprint(a, b)", "C2": "s = \"Python Programming\"\nprint(s[7:])\nprint(s[:6])\nprint(s[-3:])", "C3": "print(\"{2} {0} {1}\".format(\"apple\",\n                         \"pineapple\",\n                         \"banana\"))", "C4": "print(\"Binary of {0} is {0:b}\".format(10))", "C5": "print('at' not in 'battle', 'a' in 'program')", "C6": "lt = [1, 2, 3, 4]\nprint(lt * 2)\nprint(lt + [5])", "C7": "my_list = ['p', 'r', 'o', 'b', 'l', 'e', 'm']\nmy_list.remove('p')\nprint(my_list.pop(1))\nprint(my_list)", "C8": "t = ('a', 'p', 'p', 'l', 'e')\nprint(t.count('p'), t.index('l'))\nprint(t[1:3])", "C9": "d = {'name': 'Jack', 'age': 26}\nd['age'] = 27\nd['city'] = 'Sisaket'\nprint(d)\nprint(d.get('email'))", "C10": "A = set([1, 3, 5, 7, 9])\nB = set([3, 4, 5, 6, 7, 8])\nprint(A - B)\nprint(A & B)"}
_results = {}

def _canon(v):
    if isinstance(v, (set, frozenset)):
        return "set{" + ",".join(sorted(_canon(x) for x in v)) + "}"
    if isinstance(v, dict):
        return "dict{" + ",".join(sorted(_canon(k) + ":" + _canon(x) for k, x in v.items())) + "}"
    if isinstance(v, (list, tuple)):
        return type(v).__name__ + "(" + ",".join(_canon(x) for x in v) + ")"
    return type(v).__name__ + ":" + repr(v)

def _h(qid, value):
    return hashlib.sha256(f"{_SALT}|{qid}|{value}".encode("utf-8")).hexdigest()[:16]

def _norm_text(s):
    lines = [ln.rstrip() for ln in str(s).replace("\r\n", "\n").split("\n")]
    while lines and lines[0] == "":
        lines.pop(0)
    while lines and lines[-1] == "":
        lines.pop()
    return "\n".join(lines)

def _norm_choice(prefix, ans):
    a = str(ans).strip().upper()
    if prefix == "B":
        if a in ("T", "TRUE", "ถูก", "ท", "1"): return "T"
        if a in ("F", "FALSE", "ผิด", "ผ", "0"): return "F"
    return a

def grade(part, answers):
    """ตรวจคำตอบตอนที่ 1 (part='A') และตอนที่ 2 (part='B')"""
    n_ok, n_all = 0, 0
    for n in sorted(answers):
        qid = f"{part}{n}"
        if qid not in _KEY:
            continue
        n_all += 1
        a = _norm_choice(part, answers[n])
        if a == "":
            print(f"ข้อ {n:>2}: ⚪ ยังไม่ได้ตอบ")
            _results[qid] = False
            continue
        ok = _h(qid, a) == _KEY[qid]
        _results[qid] = ok
        n_ok += ok
        print(f"ข้อ {n:>2}: {'✅ ถูกต้อง' if ok else '❌ ยังไม่ถูก  (ทบทวนหัวข้อ ' + _TOPIC[qid] + ')'}")
    print("-" * 40)
    print(f"ได้ {n_ok} / {n_all} คะแนน")

def predict(qid, prediction):
    """ตอนที่ 3: เปรียบเทียบผลลัพธ์ที่ทำนายกับผลลัพธ์จริง (บันทึกคะแนนจากการตรวจครั้งแรก)"""
    pred = _norm_text(prediction)
    if pred == "":
        print("⚪ ยังไม่ได้เขียนคำตอบที่ทำนายไว้ในตัวแปร")
        return
    buf = io.StringIO()
    with contextlib.redirect_stdout(buf):
        exec(_SNIP[qid], {})
    actual = _norm_text(buf.getvalue())
    ok = _h(qid, pred) == _KEY[qid]
    first = qid not in _results
    if first:
        _results[qid] = ok
    print("✅ ทำนายถูกต้อง" if ok else "❌ ทำนายไม่ตรงกับผลลัพธ์จริง")
    print("\nผลลัพธ์ที่ทำนาย:")
    print(pred)
    print("\nผลลัพธ์จริงจากการรันโค้ด:")
    print(actual)
    if not ok:
        print("\n(สังเกตเครื่องหมาย ช่องว่าง วงเล็บ และเครื่องหมายคำพูดให้ดี — ทบทวนหัวข้อ " + _TOPIC[qid] + ")")
    if not first:
        print("\nℹ️ คะแนนข้อนี้บันทึกจากการตรวจครั้งแรกไปแล้ว")

def check(qid, value):
    """ตอนที่ 5: ตรวจค่าที่ได้จากโปรแกรม"""
    ok = _h(qid, _canon(value)) == _KEY[qid]
    _results[qid] = ok
    if ok:
        print("✅ ถูกต้อง เยี่ยมมาก!")
    else:
        print("❌ ยังไม่ถูกต้อง ค่าที่ได้คือ:")
        print(repr(value))
        print("ตรวจสอบค่าและชนิดข้อมูลของแต่ละตัวแปรอีกครั้ง (ทบทวนหัวข้อ " + _TOPIC[qid] + ")")

def summary():
    """สรุปผลการประเมินตนเอง"""
    total_ok, total_all, review = 0, 0, []
    print(f"{'ส่วน':<28}{'คะแนน':>10}")
    print("=" * 38)
    for name, qids in _PARTS:
        ok = sum(1 for q in qids if _results.get(q))
        total_ok += ok
        total_all += len(qids)
        print(f"{name:<28}{ok:>5} / {len(qids):<3}")
        review += [q for q in qids if not _results.get(q)]
    print("=" * 38)
    pct = 100 * total_ok / total_all
    print(f"{'รวม (ส่วนที่ตรวจอัตโนมัติ)':<28}{total_ok:>5} / {total_all:<3}  ({pct:.1f}%)")
    print()
    if pct >= 80:
        print("🌟 ดีมาก — เข้าใจเนื้อหาบทที่ 5 เป็นอย่างดี พร้อมเรียนบทถัดไป")
    elif pct >= 60:
        print("👍 ดี — เข้าใจเนื้อหาส่วนใหญ่ ควรทบทวนหัวข้อที่ยังทำไม่ได้")
    else:
        print("📖 ควรทบทวนบทเรียนอีกครั้ง แล้วกลับมาทำแบบฝึกหัดใหม่")
    topics = sorted({t.strip() for q in review for t in _TOPIC[q].split(",")},
                    key=lambda s: [int(x) for x in s.split(".")])
    if topics:
        print("\nหัวข้อที่ควรทบทวน:", ", ".join(topics))
    print("\n* ตอนที่ 4 และตอนที่ 6 ให้ตรวจสอบกับผลลัพธ์ที่โจทย์กำหนด/ผู้สอน")

print("✅ ระบบตรวจคำตอบพร้อมใช้งาน — เริ่มทำแบบฝึกหัดได้เลย")
````

**ข้อ E3.** (คำถามท้ายบทข้อ 5) กำหนด `pi = 3.141592653589` และ `n = 12` ให้ใช้เมธอด `format()` สร้างสตริงต่อไปนี้
- `s_pi` แสดงค่า pi ทศนิยม 3 ตำแหน่ง → `3.142`
- `s_bin` แสดงค่า n ในรูปเลขฐานสอง → `1100`
- `s_exp` แสดงค่า `1566.345` ในรูปเลขชี้กำลัง → `1.566345e+03`

````python
pi = 3.141592653589
n = 12

s_pi = "{0:.3f}".format(pi)
s_bin = "{0:b}".format(n)
s_exp = "{0:e}".format(1566.345)

print(s_pi, s_bin, s_exp)
check("E3", (s_pi, s_bin, s_exp))
````

**ข้อ E4.** (การแปลงชนิดข้อมูลและฟังก์ชันทางคณิตศาสตร์) ข้อมูลที่ได้จาก `input()` เป็นสตริงเสมอ กำหนด `x = "25"` และ `y = "3.5"`
1. `total` = ผลบวกของ x (แปลงเป็นจำนวนเต็ม) กับ y (แปลงเป็นจำนวนจริง) → `28.5`
2. `r1` = ค่าสัมบูรณ์ของ `-7.5` ด้วย `abs()`
3. `r2` = รากที่สองของ `144` ด้วย `math.sqrt()`
4. `r3` = ลอการิทึมฐาน 10 ของ `1000` ด้วย `math.log10()`
5. `r4` = `2` ยกกำลัง `10` ด้วยตัวดำเนินการ `**`

````python
import math

x = "25"
y = "3.5"

total = int(x) + float(y)
r1 = abs(-7.5)
r2 = math.sqrt(144)
r3 = math.log10(1000)
r4 = 2 ** 10

print(total, r1, r2, r3, r4)
check("E4", (total, r1, r2, r3, r4))
````

|---|---|---|
| `first` | อักขระตัวแรก | `'P'` |

**ข้อ E6.** (คำถามท้ายบทข้อ 10 — การวนซ้ำและการทดสอบสมาชิกในสตริง) กำหนด `text = "Love me love my dog"`
1. ใช้ลูป `for` นับจำนวนตัวอักษร `'o'` ใน `text` เก็บไว้ใน `count`
2. ใช้ตัวดำเนินการ `in` ตรวจสอบว่ามี `"dog"` อยู่ใน `text` หรือไม่ เก็บผล (True/False) ไว้ใน `has_dog`

````python
text = "Love me love my dog"

count = 0
for letter in text:
    if letter == 'o':
        count += 1

has_dog = "dog" in text

print(count, "letters found")
print(has_dog)
check("E6", (count, has_dog))
````

**ข้อ E7.** (split และ join) กำหนด `uni = "Sisaket Rajabhat University"`
1. แยกคำด้วย `split()` เก็บไว้ใน `words` → `['Sisaket', 'Rajabhat', 'University']`
2. รวมคำใน `words` ด้วยเครื่องหมาย `-` โดยใช้ `join()` เก็บไว้ใน `joined` → `'Sisaket-Rajabhat-University'`
3. แปลง `joined` เป็นตัวพิมพ์เล็กทั้งหมด เก็บไว้ใน `slug` → `'sisaket-rajabhat-university'`

````python
uni = "Sisaket Rajabhat University"

words = uni.split()
joined = "-".join(words)
slug = joined.lower()

print(words, joined, slug, sep="\n")
check("E7", (words, joined, slug))
````

**ข้อ E8.** (ลิสต์) กำหนด `scores = [70, 85, 60]` ให้ทำตามขั้นตอนต่อไปนี้ **ตามลำดับ** ด้วยคำสั่ง/เมธอดของลิสต์
1. เพิ่ม `90` ต่อท้ายด้วย `append()`
2. เปลี่ยนสมาชิกตัวแรกเป็น `75`
3. เพิ่ม `55` และ `80` ต่อท้ายพร้อมกันด้วย `extend()`
4. ลบค่า `60` ออกด้วย `remove()`
5. ลบสมาชิกตัวสุดท้ายด้วย `pop()` แล้วเก็บค่าที่ได้ไว้ใน `last`

ผลลัพธ์สุดท้าย: `scores` = `[75, 85, 90, 55]` และ `last` = `80`

````python
scores = [70, 85, 60]

scores.append(90)
scores[0] = 75
scores.extend([55, 80])
scores.remove(60)
last = scores.pop()

print(scores, last)
check("E8", (scores, last))
````

**ข้อ E10.** (ดิกชันนารี — สมุดโทรศัพท์)
1. สร้างดิกชันนารี `phonebook` ที่มีข้อมูล `'Ann': '081'` และ `'Bob': '082'`
2. เพิ่มข้อมูล `'Cat': '083'`
3. แก้ไขเบอร์ของ `'Bob'` เป็น `'089'`
4. ลบข้อมูลของ `'Ann'` ด้วย `pop()` เก็บค่าที่ได้ไว้ใน `removed`
5. ค้นหาเบอร์ของ `'Dan'` ด้วย `get()` เก็บไว้ใน `missing` (ต้องไม่เกิดข้อผิดพลาด)

````python
phonebook = {'Ann': '081', 'Bob': '082'}
phonebook['Cat'] = '083'
phonebook['Bob'] = '089'
removed = phonebook.pop('Ann')
missing = phonebook.get('Dan')

print(phonebook, removed, missing)
check("E10", (phonebook, removed, missing))
````

**ข้อ E11.** (เซต) กำหนด `A = {1, 2, 3, 4, 5}` และ `B = {4, 5, 6, 7, 8}` ให้หาผลลัพธ์ต่อไปนี้ด้วย **ตัวดำเนินการของเซต**
- `u` = union, `i` = intersection, `d` = A − B, `sd` = symmetric difference
- `unique` = เซตที่สร้างจากลิสต์ `[1, 2, 2, 3, 3, 3]` ด้วยฟังก์ชัน `set()`

````python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

u = A | B
i = A & B
d = A - B
sd = A ^ B
unique = set([1, 2, 2, 3, 3, 3])

print(u, i, d, sd, unique, sep="\n")
check("E11", (u, i, d, sd, unique))
````
