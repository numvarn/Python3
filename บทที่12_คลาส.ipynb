## 12.1 การสร้างและใช้งานคลาส

เราสร้างแบบจำลองของทุกสิ่ง รวมทั้งเหตุการณ์ต่าง ๆ ด้วยคลาสได้ทั้งสิ้น เริ่มจากสิ่งมีชีวิตใกล้ตัว คือ **สุนัข (Dog)** ลองคิดว่าสุนัขทั่วไปมีอะไรบ้าง

| สิ่งที่สุนัข **มี** (ข้อมูล) → **แอตทริบิวต์ (Attribute)** | สิ่งที่สุนัข **ทำได้** (พฤติกรรม) → **เมธอด (Method)** |
|---|---|
| ชื่อ (name), อายุ (age) | นั่ง (sit), กลิ้งไปกลิ้งมา (roll over) |

นำสิ่งเหล่านี้ไปสร้างเป็น **คลาส `Dog`** ซึ่งบอกไพธอนว่าเราจะแทน "สุนัขทั่วไป" ด้วยโปรแกรมอย่างไร เมื่อมีคลาสแล้ว จึงนำคลาสไปสร้าง **วัตถุ** หรือ **อินสแตนซ์ (Instance)** ซึ่งหมายถึง **สุนัขตัวใดตัวหนึ่งอย่างเจาะจง**

> 🍪 เปรียบ **คลาส** เหมือน **แม่พิมพ์ขนม** และ **อินสแตนซ์** คือ **ขนมแต่ละชิ้น** ที่ได้จากแม่พิมพ์ — แม่พิมพ์เดียวทำขนมได้ไม่จำกัด แต่ละชิ้นมีรูปทรงเหมือนกันแต่ใส่ไส้ต่างกันได้

<p align="center"><br><b>แผนภาพ</b> คลาส Dog และอินสแตนซ์ที่สร้างจากคลาส</p>

### 12.1.1 การสร้างคลาสในภาษาไพธอน

````python
class Dog():
    """การจำลองสุนัขทั่วไปด้วยคลาส"""
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def sit(self):
        """การจำลองพฤติกรรมการนั่ง"""
        print("{} is now sitting.".format(self.name.title()))

    def roll_over(self):
        """การจำลองพฤติกรรมการกลิ้ง"""
        print("{} is rolled over!".format(self.name.title()))
````

**อธิบายคลาส `Dog`**

| บรรทัด | คำอธิบาย |
|:---:|---|
| 1 | ประกาศคลาสชื่อ `Dog` (นิยมตั้งชื่อคลาสแบบ **CapWords** คือขึ้นต้นแต่ละคำด้วยตัวพิมพ์ใหญ่ เช่น `ElectricCar`) |
| 2 | **docstring** อธิบายว่าคลาสนี้ทำอะไร |
| 3 | ฟังก์ชันที่อยู่ในคลาสเรียกว่า **เมธอด (Method)** — `__init__()` เป็น **เมธอดพิเศษ** ที่ไพธอนเรียกให้ **อัตโนมัติทุกครั้งที่สร้างอินสแตนซ์** ชื่อขึ้นต้นและลงท้ายด้วย `_` สองตัว (double underscores หรือ "dunder") เพื่อไม่ให้ชนกับชื่อเมธอดที่ผู้พัฒนาตั้งเอง |
| 4–5 | ตัวแปรที่นำหน้าด้วย `self.` ใช้ได้ **ในทุกเมธอดของคลาส** และเข้าถึงผ่านอินสแตนซ์ได้ เรียกว่า **แอตทริบิวต์ (Attribute)** เช่น `self.name = name` นำค่าจากพารามิเตอร์ `name` ไปเก็บในแอตทริบิวต์ `self.name` ของอินสแตนซ์นั้น |
| 7, 11 | เมธอด `sit()` และ `roll_over()` ไม่ต้องรับข้อมูลเพิ่ม แต่ **ยังต้องมีพารามิเตอร์ `self`** |

**เรื่องสำคัญของ `self`** 🔑
- `__init__()` ของ `Dog` มีพารามิเตอร์ 3 ตัว คือ `self`, `name` และ `age` แต่ตอนสร้างอินสแตนซ์เราส่งแค่ 2 ค่า เพราะไพธอน **ส่งค่าให้ `self` อัตโนมัติ**
- `self` คือ **ตัวอินสแตนซ์เอง** ที่กำลังถูกใช้งาน ทำให้เมธอดรู้ว่ากำลังทำงานกับสุนัขตัวไหน
- **ทุกเมธอดของคลาสต้องมี `self` เป็นพารามิเตอร์ตัวแรกเสมอ** แม้จะไม่ต้องรับข้อมูลจากภายนอกเลยก็ตาม

### 12.1.2 การสร้างอินสแตนซ์จากคลาส
คลาส `Dog` คือแบบจำลองของสุนัขทั่วไป ไม่ได้ชี้เจาะจงตัวใด การ **สร้างอินสแตนซ์** คือการสร้างวัตถุที่เจาะจงสุนัขเพียงตัวเดียว โดยเรียกชื่อคลาสเหมือนเรียกฟังก์ชัน

````python
my_dog = Dog("Kanoon", 3)

print("My dog's name is {}.".format(my_dog.name.title()))

print("My dog is {} year old.".format(my_dog.age))
````

บรรทัดที่ 1 สร้างอินสแตนซ์ของ `Dog` เก็บในตัวแปร `my_dog` คือสุนัขชื่อ "Kanoon" (ขนุน) อายุ 3 ปี ไพธอนจะ
1. สร้างวัตถุใหม่ว่าง ๆ
2. เรียก `__init__(วัตถุใหม่, "Kanoon", 3)` ให้อัตโนมัติ → กำหนด `self.name` และ `self.age`
3. คืนวัตถุนั้นมาเก็บใน `my_dog`

````python
print(type(my_dog))
print(isinstance(my_dog, Dog))
print(vars(my_dog))          # ดูแอตทริบิวต์ทั้งหมดของอินสแตนซ์ในรูปดิกชันนารี
````

### 12.1.3 การเข้าถึงแอตทริบิวต์ของอินสแตนซ์
เข้าถึงแอตทริบิวต์ด้วย **เครื่องหมายจุด** `อินสแตนซ์.แอตทริบิวต์` เช่น `my_dog.name` (บรรทัดที่ 3 ของตัวอย่างข้างบน) และ `my_dog.age` (บรรทัดที่ 5)

````python
my_dog.name
````

### 12.1.4 การเรียกเมธอด
เรียกเมธอดด้วยเครื่องหมายจุดเช่นกัน `อินสแตนซ์.เมธอด()` เมื่อไพธอนอ่าน `my_dog.sit()` จะค้นหาเมธอด `sit()` ในคลาส `Dog` แล้วรันโค้ดนั้น โดยส่ง `my_dog` ไปเป็น `self`

````python
my_dog.sit()

my_dog.roll_over()
````

🆕 **เบื้องหลัง:** `my_dog.sit()` ที่จริงคือ `Dog.sit(my_dog)` — นี่คือที่มาของ `self`

### 12.1.5 การสร้างหลายอินสแตนซ์จากคลาสเดียว
คลาสหนึ่งคลาส **สร้างอินสแตนซ์ได้มากเท่าที่ต้องการ** แต่ละอินสแตนซ์เป็นอิสระต่อกัน มีค่าแอตทริบิวต์ของตัวเอง

````python
thungtong = Dog("Thungtong", 8)
thunggnern = Dog("Thunggnern", 9)

print("My dog's name is {}.".format(thungtong.name.title()))
print("My dog's name is {}.".format(thunggnern.name.title()))

thungtong.age += 1                       # เปลี่ยนอายุทองต้อง ไม่กระทบตัวอื่น
print(thungtong.age, thunggnern.age, my_dog.age)
````

💡 เก็บอินสแตนซ์จำนวนมากไว้ใน **ลิสต์** แล้ววนลูปเรียกเมธอดได้

````python
dogs = [Dog("ขนุน", 3), Dog("ทองต้อง", 8), Dog("ทองเงิน", 9)]
for d in dogs:
    d.sit()
print("อายุเฉลี่ย:", sum(d.age for d in dogs) / len(dogs), "ปี")
````

### 🆕 เมธอดพิเศษ `__str__` — กำหนดวิธีแสดงผลของวัตถุ
ถ้า `print(my_dog)` จะได้ข้อความที่อ่านไม่รู้เรื่อง เช่น `<__main__.Dog object at 0x7f...>` เราเขียนเมธอด `__str__()` ให้ **คืนสตริง** ที่ต้องการแสดง แล้ว `print()` และ `str()` จะเรียกใช้อัตโนมัติ

````python
print(my_dog)

class Dog():
    """การจำลองสุนัขทั่วไปด้วยคลาส (เพิ่ม __str__)"""
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def sit(self):
        print("{} is now sitting.".format(self.name.title()))

    def roll_over(self):
        print("{} is rolled over!".format(self.name.title()))

    def __str__(self):
        return f"สุนัขชื่อ {self.name} อายุ {self.age} ปี"

my_dog = Dog("Kanoon", 3)
print(my_dog)
````

## 12.3 การสืบทอด

### 12.3.1 ความหมายของการสืบทอด
**การสืบทอด (Inheritance)** คือการสร้างคลาสใหม่จากคลาสเดิมที่มีอยู่แล้ว **โดยไม่ต้องแก้ไขคลาสเดิม**

- คลาสต้นแบบ เรียกว่า **คลาสแม่** (Parent class, Base class, Superclass)
- คลาสใหม่ เรียกว่า **คลาสลูก** (Child class, Derived class, Subclass)
- คลาสลูกได้ **แอตทริบิวต์และเมธอดทั้งหมด** ของคลาสแม่ และ **เพิ่มของตัวเอง** หรือ **เปลี่ยนการทำงาน** ของเมธอดที่ได้รับมาได้

ความสัมพันธ์แบบนี้เรียกว่า **"is-a"** เช่น รถยนต์ไฟฟ้า **เป็น** รถยนต์ชนิดหนึ่ง


คลาสลูกเพิ่มแอตทริบิวต์และเมธอดใหม่ได้ไม่จำกัด เพื่อแยกความแตกต่างจากคลาสแม่ ตัวอย่างนี้เพิ่ม **ขนาดแบตเตอรี่ (battery_size)** ค่าเริ่มต้น 100 และเมธอดรายงานขนาดแบตเตอรี่ แอตทริบิวต์นี้มีเฉพาะในอินสแตนซ์ของ `ElectricCar` เท่านั้น ไม่มีในอินสแตนซ์ของ `Car`

### 12.3.4 การสืบทอดจากหลายคลาส
ไพธอนรองรับ **การสืบทอดจากหลายคลาส (Multiple inheritance)** คือคลาสลูกสืบทอดจากคลาสแม่มากกว่าหนึ่งคลาสพร้อมกัน เช่น การจำลอง **นาฬิกาดิจิทัลที่แสดงปฏิทินได้ (Digital Calendar Clock)** โดยนำคลาสนาฬิกาดิจิทัลและคลาสปฏิทินดิจิทัลที่มีอยู่แล้ว มาสร้างคลาส `DigitalCalendarClock`

<p align="center"><br><b>ภาพที่ 12.1</b> แนวคิดในการจำลองแบบวัตถุด้วยการสืบทอดจากหลายคลาส<br><small>ที่มา: https://www.python-course.eu/python3_multiple_inheritance.php</small></p>

ไวยากรณ์ (🆕 เอกสารพิมพ์ `Class` ตัวพิมพ์ใหญ่ ซึ่งผิด ต้องเป็น `class` ตัวพิมพ์เล็ก)
```python
class Base1:
    Body of the class

class Base2:
    Body of the class

class Derived(Base1, Base2):
    Body of the class
```

**ตัวอย่าง:** คลาส `Resident` สืบทอดจากทั้ง `Person` และ `Student`

<p align="center"></p>

````python
class Person:
    #defining constructor
    def __init__(self, personName, personAge):
        self.name = personName
        self.age = personAge

    #defining class methods
    def showName(self):
        print(self.name)

    def showAge(self):
        print(self.age)

    #end of class definition

class Student:
    def __init__(self, studentId):
        self.studentId = studentId

    def getId(self):
        return self.studentId


class Resident(Person, Student):
    def __init__(self, name, age, id):
        Person.__init__(self, name, age)
        Student.__init__(self, id)


# สร้างอินสแตนซ์จากคลาสลูก
resident1 = Resident('John', 30, '102')
resident1.showName()
print(resident1.getId())
````

- เพราะมีคลาสแม่สองคลาส จึงเรียก `__init__` ของแต่ละคลาสแม่ **ระบุชื่อคลาสตรง ๆ** และส่ง `self` ไปเอง
- 🆕 ถ้าคลาสแม่หลายคลาสมีเมธอดชื่อซ้ำกัน ไพธอนค้นหาตามลำดับที่เรียกว่า **MRO (Method Resolution Order)** — จากคลาสลูก → คลาสแม่ตามลำดับที่เขียนในวงเล็บ (ซ้ายไปขวา) → `object`

````python
print([c.__name__ for c in Resident.__mro__])
````

### 12.3.5 การลบล้างหรือแก้ไขเมธอดที่ได้รับสืบทอดจากคลาสแม่ (Overriding)
ถ้าเมธอดของคลาสแม่ทำงานไม่ตรงกับที่ต้องการ ให้ **นิยามเมธอดชื่อเดียวกัน** ในคลาสลูก ไพธอนจะใช้เมธอดของคลาสลูกแทน เรียกว่า **Method Overriding**

ตัวอย่าง: คลาส `Car` มีเมธอด `fill_gas_tank()` (เติมน้ำมัน) แต่รถยนต์ไฟฟ้าไม่มีถังน้ำมัน จึง override ใน `ElectricCar`

<div style="border-left:5px solid #7048e8;padding:8px 14px;background:rgba(112,72,232,.08)">

### 🆕 หมายเหตุ
คลาส `Car` ในเอกสารไม่ได้นิยามเมธอด `fill_gas_tank()` ไว้ ตัวอย่าง override จึงไม่เห็นความแตกต่าง Notebook นี้เพิ่มเมธอดนี้ใน `Car` ไว้แล้ว (หัวข้อ 12.2.4) เพื่อเปรียบเทียบผล
</div>

````python
class ElectricCar(Car):
    """คลาสรถยนต์ไฟฟ้าสืบทอดคุณสมบัติจากคลาสรถยนต์"""
    def __init__(self, make, model, year):
        super().__init__(make, model, year)
        self.battery_size = 100

    def describe_battery(self):
        print("This car has a {}-kWh battery.".format(self.battery_size))

    def fill_gas_tank(self):
        """ลบล้างเมธอดจากคลาสแม่
        เนื่องจากในรถยนต์ไฟฟ้าไม่มีถังสำหรับเติมน้ำมัน"""
        print("This car doesn't need a gas tank!")

my_new_car = Car('Toyota', 'Altis', 2016)
my_ecar = ElectricCar('Tesla', 'model s', 2020)

my_new_car.fill_gas_tank()      # ใช้เมธอดของ Car
my_ecar.fill_gas_tank()         # ใช้เมธอดของ ElectricCar (override)
````

🆕 **Override แบบต่อยอด:** บางครั้งไม่ได้ต้องการทิ้งเมธอดของคลาสแม่ทั้งหมด แต่อยาก **เพิ่ม** การทำงาน ให้เรียก `super().ชื่อเมธอด()` ภายในเมธอดที่ override

````python
class ElectricCar(Car):
    def __init__(self, make, model, year):
        super().__init__(make, model, year)
        self.battery_size = 100

    def describe_battery(self):
        print("This car has a {}-kWh battery.".format(self.battery_size))

    def fill_gas_tank(self):
        print("This car doesn't need a gas tank!")

    def get_descriptive_name(self):
        """ต่อยอดจากเมธอดของคลาสแม่"""
        return super().get_descriptive_name() + " ⚡ (EV)"

my_ecar = ElectricCar('Tesla', 'model s', 2020)
print(my_ecar.get_descriptive_name())
````

### ✍️ แบบฝึก 12.3 — รูปทรงเรขาคณิต (สืบทอด + Polymorphism)
คลาสแม่ `Shape` ให้มาแล้ว จงสร้างคลาสลูก
- `Circle(Shape)` รับ `radius` — `__init__` ต้องเรียก `super().__init__("วงกลม")` และ override `area()` คืน `math.pi * r ** 2`
- `Square(Shape)` รับ `side` — เรียก `super().__init__("สี่เหลี่ยมจัตุรัส")` และ override `area()` คืน `side ** 2`

แล้วเขียนฟังก์ชัน `total_area(shapes)` คืนผลรวมพื้นที่ของทุกรูปในลิสต์ (ใช้ polymorphism — ไม่ต้องตรวจว่าเป็นคลาสใด)
</div>

````python
import math

class Shape:
    def __init__(self, name):
        self.name = name

    def area(self):
        return 0

    def describe(self):
        return f"{self.name} พื้นที่ {self.area():.2f}"

# เขียนคลาส Circle, Square และฟังก์ชัน total_area ที่นี่
````

````python
# ✅ เซลล์ตรวจคำตอบ (รันได้ซ้ำ ไม่ต้องแก้ไข)
def _safe(f, *a):
    try:
        return f(*a)
    except Exception as e:
        return f"Error → {type(e).__name__}: {e}"
def _raises(f, *a):
    try:
        r = f(*a)
    except Exception as e:
        return type(e).__name__
    return f"ไม่เกิด Exception (คืนค่า {r!r})"
def _eq(a, b):
    try:
        return a == b
    except Exception:
        return False
_tests = [
    ("Circle(1).area()", _safe(lambda: round(Circle(1).area(), 4)), 3.1416),
    ("Square(3).area()", _safe(lambda: Square(3).area()), 9),
    ("Circle(2).name", _safe(lambda: Circle(2).name), "วงกลม"),
    ("Square(2).describe()", _safe(lambda: Square(2).describe()), "สี่เหลี่ยมจัตุรัส พื้นที่ 4.00"),
    ("isinstance(Circle(1), Shape)", _safe(lambda: isinstance(Circle(1), Shape)), True),
    ("total_area([Square(2), Square(3), Circle(1)])", _safe(lambda: round(total_area([Square(2), Square(3), Circle(1)]), 4)), 16.1416),
    ("total_area([])", _safe(lambda: total_area([])), 0),
]
_ok = 0
for _label, _got, _exp in _tests:
    _pass = bool(_eq(_got, _exp))
    _ok += _pass
    _show = repr(_got)
    _show = _show if len(_show) < 80 else _show[:77] + "..."
    print(("✅ " if _pass else "❌ ") + _label + " = " + _show + ("" if _pass else ("\n      ควรได้ " + repr(_exp))[:240]))
print(f"\nผ่าน {_ok}/{len(_tests)} ข้อ" + ("  🎉" if _ok == len(_tests) else ""))
````

### 12.4.1 การนำเข้าเพียงคลาสเดียว
บันทึกคลาส `Car` ไว้ในไฟล์ **`car.py`** — ไฟล์นี้คือ **โมดูล `car`**

````python
%%writefile car.py
"""โมดูลสำหรับจำลองรถยนต์"""

class Car():
    """คลาสสำหรับจำลองรถยนต์"""
    def __init__(self, make, model, year):
        """กำหนดค่าเริ่มต้นให้กับแอตทริบิวต์ของคลาสรถยนต์"""
        self.make = make
        self.model = model
        self.year = year
        self.odometer = 0

    def get_descriptive_name(self):
        """ส่งคืนค่ารายละเอียดต่าง ๆ ของรถยนต์"""
        long_name = str(self.year) + " " + self.make + " " + self.model
        return long_name.title()

    def read_odometer(self):
        """แสดงรายละเอียดระยะสะสมของรถยนต์"""
        print("This car has " + str(self.odometer) + " miles")

    def update_odometer(self, mileage):
        """กำหนดค่าระยะทางสะสมใหม่จากพารามิเตอร์"""
        if mileage >= self.odometer:
            self.odometer = mileage
        else:
            print("คุณไม่สามารถลดค่าระยะสะสมได้")
````

จากนั้นสร้างโปรแกรมใหม่ **`my_car.py`** ที่นำเข้าคลาส `Car` จากโมดูล `car` แล้วรันด้วยคำสั่ง `python my_car.py` (ใน Notebook ใช้ `%run`)

````python
%%writefile my_car.py
from car import Car

my_new_car = Car('Toyota', 'Altis', 2016)

print(my_new_car.get_descriptive_name())
my_new_car.read_odometer()
````

### 12.4.2 การจัดเก็บหลายคลาสในโมดูล
โมดูลหนึ่งเก็บคลาสได้มากเท่าที่ต้องการ เพิ่มคลาส `Battery` และ `ElectricCar` ลงใน `car.py`

### 🆕 Composition — ความสัมพันธ์แบบ "has-a"
สังเกต `ElectricCar` เวอร์ชันนี้: แทนที่จะเก็บ `battery_size` ไว้ตรง ๆ กลับสร้าง **อินสแตนซ์ของคลาส `Battery`** เก็บในแอตทริบิวต์ `self.battery` เรียกเทคนิคนี้ว่า **Composition** (การประกอบวัตถุ) — รถยนต์ไฟฟ้า **มี** แบตเตอรี่ (has-a) ไม่ใช่ **เป็น** แบตเตอรี่ (is-a)

ข้อดี: เมื่อเรื่องแบตเตอรี่ซับซ้อนขึ้น (ขนาด ระยะวิ่ง การชาร์จ) แยกไปอยู่ในคลาส `Battery` ทำให้ `ElectricCar` ไม่รก และนำ `Battery` ไปใช้กับคลาสอื่นได้ (เช่น มอเตอร์ไซค์ไฟฟ้า) — ดูแผนภาพในหัวข้อ 12.3.1

<div style="border-left:5px solid #7048e8;padding:8px 14px;background:rgba(112,72,232,.08)">

### 🆕 หมายเหตุ: ตัวแปรชื่อ range ในเมธอด get_range
เอกสารตั้งชื่อตัวแปร `range = 0` ซึ่ง **ทับชื่อฟังก์ชัน `range()`** ที่มากับไพธอนภายในเมธอดนั้น (ทำงานได้ แต่เป็นนิสัยที่ไม่ดี อาจเกิดบั๊กถ้าเรียก `range()` ในเมธอดเดียวกัน) Notebook นี้เปลี่ยนเป็น `driving_range` และใช้ `else` แทน `elif self.battery_size > 70` ที่ซ้ำซ้อน
</div>

````python
%%writefile car.py
"""โมดูลสำหรับจำลองรถยนต์ทั่วไปและรถยนต์ไฟฟ้า"""

class Car():
    """คลาสสำหรับจำลองรถยนต์"""
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
        self.odometer = 0

    def get_descriptive_name(self):
        long_name = str(self.year) + " " + self.make + " " + self.model
        return long_name.title()

    def read_odometer(self):
        print("This car has " + str(self.odometer) + " miles")

    def update_odometer(self, mileage):
        if mileage >= self.odometer:
            self.odometer = mileage
        else:
            print("คุณไม่สามารถลดค่าระยะสะสมได้")


class Battery():
    """คลาสจำลองแบตเตอรี่ของรถยนต์ไฟฟ้า"""
    def __init__(self, battery_size=60):
        self.battery_size = battery_size

    def describe_battery(self):
        print("This car has a " + str(self.battery_size) + "-kWh battery.")

    def get_range(self):
        if self.battery_size <= 70:
            driving_range = 240
        else:
            driving_range = 270

        message = "This car can go approximately " + str(driving_range)
        message += " miles on a full charge."
        print(message)


class ElectricCar(Car):
    """คลาสรถยนต์ไฟฟ้าสืบทอดคุณสมบัติจากคลาสรถยนต์"""
    def __init__(self, make, model, year):
        super().__init__(make, model, year)
        self.battery = Battery()          # composition: รถยนต์ไฟฟ้า "มี" แบตเตอรี่
````

สร้างไฟล์ **`my_electric_car.py`** นำเข้าคลาส `ElectricCar` สังเกตการเรียกเมธอดผ่าน **จุดสองชั้น** `my_ecar.battery.describe_battery()` (อินสแตนซ์ → แอตทริบิวต์ที่เป็นวัตถุ → เมธอดของวัตถุนั้น)

````python
%%writefile my_electric_car.py
from car import ElectricCar

my_ecar = ElectricCar('Tesla', 'model s', 2020)

print(my_ecar.get_descriptive_name())

my_ecar.battery.describe_battery()

my_ecar.battery.get_range()
````

## 12.5 สรุป

- **คลาส (Class)** คือแม่แบบสำหรับจำลองสิ่งต่าง ๆ หรือเหตุการณ์ให้อยู่ในรูปโปรแกรม ประกอบด้วย **แอตทริบิวต์** (ข้อมูล) และ **เมธอด** (พฤติกรรม = ฟังก์ชันภายในคลาส)
- **อินสแตนซ์ (Instance)** หรือวัตถุ คือสิ่งที่สร้างจากคลาส แต่ละอินสแตนซ์มีค่าแอตทริบิวต์ของตัวเอง คลาสเดียวสร้างอินสแตนซ์ได้ไม่จำกัด
- เมธอด **`__init__()`** ถูกเรียกอัตโนมัติเมื่อสร้างอินสแตนซ์ ใช้กำหนดค่าเริ่มต้นของแอตทริบิวต์
- **`self`** คือตัวอินสแตนซ์เอง ต้องเป็นพารามิเตอร์ตัวแรกของทุกเมธอด และไพธอนส่งค่าให้อัตโนมัติ
- แก้ไขค่าแอตทริบิวต์ได้ **โดยตรง** (`obj.attr = ค่า`) หรือ **ผ่านเมธอด** ซึ่งตรวจสอบความถูกต้องได้ (เช่น ไม่ให้ระยะทางสะสมลดลง)
- **การสืบทอด** สร้างคลาสลูกจากคลาสแม่ `class Child(Parent):` ใช้ **`super()`** เรียกเมธอดของคลาสแม่ คลาสลูกเพิ่มแอตทริบิวต์/เมธอดได้โดยไม่กระทบคลาสแม่ และสืบทอดจาก **หลายคลาส** ได้
- **Overriding** คือการนิยามเมธอดชื่อเดิมในคลาสลูกเพื่อเปลี่ยนการทำงาน
- เก็บคลาสใน **โมดูล** (ไฟล์ `.py`) แล้วนำเข้าด้วย `from module import Class`, `from module import A, B` หรือ `import module` (หลีกเลี่ยง `from module import *`)
- 🆕 เพิ่มเติม: `__str__()` กำหนดการแสดงผล, แอตทริบิวต์ของคลาสกับของอินสแตนซ์, การห่อหุ้มด้วย `_`/`__`, Polymorphism และ Composition (has-a)

| หลักการ OOP | ความหมาย | ในบทนี้ |
|---|---|---|
| **Encapsulation** (การห่อหุ้ม) | รวมข้อมูลกับเมธอดไว้ด้วยกัน และควบคุมการเข้าถึง | `update_odometer()` ตรวจค่าก่อนแก้, `__miles` |
| **Inheritance** (การสืบทอด) | สร้างคลาสใหม่ต่อยอดจากคลาสเดิม | `ElectricCar(Car)`, `Resident(Person, Student)` |
| **Polymorphism** (หลายรูปแบบ) | เมธอดชื่อเดียวกัน ทำงานต่างกันตามคลาส | `fill_gas_tank()`, `area()` |
| **Abstraction** (การนามธรรม) | ซ่อนรายละเอียดซับซ้อน ให้ใช้งานผ่านเมธอดที่เข้าใจง่าย | `my_ecar.battery.get_range()` |

✏️ *คำตอบของฉัน:* ...

<details><summary>แนวคำตอบข้อ 4</summary>

แอตทริบิวต์คือ **ตัวแปรที่ผูกกับวัตถุ** ใช้เก็บข้อมูลหรือสถานะของวัตถุ เช่น `name`, `age`, `odometer` สร้างด้วย `self.ชื่อ = ค่า` (มักทำใน `__init__`) และเข้าถึงด้วย `อินสแตนซ์.ชื่อ` นอกจากนี้ยังมีแอตทริบิวต์ของคลาสซึ่งใช้ร่วมกันทุกอินสแตนซ์
</details>

**5. จงอธิบายถึงความสัมพันธ์ระหว่าง คลาส และ วัตถุ**

✏️ *คำตอบของฉัน:* ...

<details><summary>แนวคำตอบข้อ 5</summary>

คลาสเป็น **แม่แบบ** ส่วนวัตถุ (อินสแตนซ์) เป็น **ของจริงที่สร้างจากแม่แบบ** — เหมือนแม่พิมพ์กับขนม หรือแบบแปลนบ้านกับบ้านแต่ละหลัง คลาสกำหนดว่าวัตถุ *มีอะไร* และ *ทำอะไรได้* ส่วนวัตถุแต่ละชิ้นมีค่าข้อมูลเฉพาะตัว คลาสหนึ่งสร้างวัตถุได้หลายชิ้น และวัตถุทุกชิ้นของคลาสเดียวกันมีโครงสร้างและเมธอดเหมือนกัน
</details>

**6. จงอธิบายถึงการแก้ไขค่าแอตทริบิวต์ของอินสแตนซ์ ว่าสามารถทำได้ด้วยวิธีใดบ้าง**

✏️ *คำตอบของฉัน:* ...

<details><summary>แนวคำตอบข้อ 6</summary>

1. **แก้ไขโดยตรง** ผ่านอินสแตนซ์ เช่น `my_new_car.odometer = 23` — ง่าย แต่ไม่มีการตรวจสอบ
2. **แก้ไขผ่านเมธอดที่กำหนดค่าใหม่** เช่น `update_odometer(50)` — ตรวจสอบความถูกต้องก่อนได้ (เช่นห้ามลดค่า)
3. **แก้ไขผ่านเมธอดที่เพิ่มค่าทีละส่วน** เช่น `increment_odometer(120)`

วิธีที่ 2–3 ปลอดภัยกว่า และควรใช้ร่วมกับธรรมเนียมตั้งชื่อ `_` หรือ `__` เพื่อบอกว่าไม่ควรแก้ไขโดยตรง
</details>

**7. จงอธิบายถึงวิธีการลบล้างหรือแก้ไขเมธอดที่ได้รับสืบทอดจากคลาสแม่ว่าสามารถทำได้อย่างไร**

✏️ *คำตอบของฉัน:* ...

<details><summary>แนวคำตอบข้อ 7</summary>

ทำได้โดย **นิยามเมธอดชื่อเดียวกัน** (และพารามิเตอร์สอดคล้องกัน) ในคลาสลูก เรียกว่า **Method Overriding** เมื่อเรียกเมธอดนั้นจากอินสแตนซ์ของคลาสลูก ไพธอนจะค้นหาในคลาสลูกก่อนจึงใช้ของคลาสลูกแทน เช่น `ElectricCar.fill_gas_tank()` ถ้าต้องการต่อยอดแทนการแทนที่ทั้งหมด ให้เรียก `super().ชื่อเมธอด()` ภายในเมธอดใหม่
</details>

**8. จงอธิบายว่า การสืบทอด คืออะไร และมีความสำคัญอย่างไรในการเขียนโปรแกรมเชิงวัตถุ**

✏️ *คำตอบของฉัน:* ...

<details><summary>แนวคำตอบข้อ 8</summary>

การสืบทอดคือการสร้าง **คลาสลูก** จาก **คลาสแม่** โดยคลาสลูกได้แอตทริบิวต์และเมธอดทั้งหมดของคลาสแม่ และเพิ่มหรือเปลี่ยนการทำงานของตนเองได้ (`class ElectricCar(Car):`) ความสำคัญ: (1) **นำโค้ดกลับมาใช้ใหม่** ไม่ต้องเขียนซ้ำ (2) **ขยายความสามารถโดยไม่แก้คลาสเดิม** ลดความเสี่ยงต่อโค้ดที่ทำงานได้อยู่แล้ว (3) แก้ไขที่คลาสแม่จุดเดียว คลาสลูกทั้งหมดได้รับผล (4) จัดโครงสร้างโปรแกรมเป็นลำดับชั้นที่สะท้อนโลกจริง และเป็นพื้นฐานของ Polymorphism
</details>

### 🏆 แบบทดสอบท้ายบท
รันเซลล์ด้านล่าง แล้วพิมพ์คำตอบ (ก/ข/ค/ง) ในช่องที่ปรากฏ

````python
quiz = [
    ("เมธอดใดถูกเรียกอัตโนมัติเมื่อสร้างอินสแตนซ์",
     ["ก. __str__()", "ข. __init__()", "ค. __new_instance__()", "ง. create()"], "ข"),
    ("พารามิเตอร์ self ในเมธอดหมายถึงอะไร",
     ["ก. ชื่อคลาส", "ข. อินสแตนซ์ที่กำลังเรียกใช้เมธอด", "ค. คลาสแม่", "ง. โมดูลปัจจุบัน"], "ข"),
    ("d = Dog('Kanoon', 3) โดย __init__(self, name, age) ผู้เรียกส่งอาร์กิวเมนต์เองกี่ตัว",
     ["ก. 1", "ข. 2", "ค. 3", "ง. 0"], "ข"),
    ("ข้อใดสร้างคลาส ElectricCar ที่สืบทอดจาก Car ได้ถูกต้อง",
     ["ก. class ElectricCar extends Car:", "ข. class ElectricCar(Car):", "ค. class Car(ElectricCar):", "ง. ElectricCar = class(Car)"], "ข"),
    ("super().__init__(make, model, year) ในคลาสลูกทำหน้าที่อะไร",
     ["ก. สร้างคลาสแม่ใหม่", "ข. เรียก __init__ ของคลาสแม่เพื่อกำหนดแอตทริบิวต์", "ค. ลบคลาสแม่", "ง. คัดลอกคลาสลูก"], "ข"),
    ("การนิยามเมธอดชื่อเดียวกับคลาสแม่ในคลาสลูกเรียกว่าอะไร",
     ["ก. Overloading", "ข. Overriding", "ค. Overflow", "ง. Overwriting class"], "ข"),
    ("ข้อใดเป็นข้อดีของการแก้ไขแอตทริบิวต์ผ่านเมธอดแทนการแก้ไขโดยตรง",
     ["ก. รันเร็วกว่า", "ข. ตรวจสอบความถูกต้องของค่าใหม่ได้", "ค. ไม่ต้องสร้างอินสแตนซ์", "ง. ไม่ต้องใช้ self"], "ข"),
    ("ElectricCar มีแอตทริบิวต์ self.battery = Battery() เป็นความสัมพันธ์แบบใด",
     ["ก. is-a (สืบทอด)", "ข. has-a (composition)", "ค. multiple inheritance", "ง. overriding"], "ข"),
    ("วิธีนำเข้าใดที่เอกสารไม่แนะนำ",
     ["ก. from car import Car", "ข. from car import Car, ElectricCar", "ค. import car", "ง. from car import *"], "ง"),
    ("class C(A, B) ถ้า A และ B มีเมธอด hello() ทั้งคู่ C().hello() จะใช้ของคลาสใด",
     ["ก. A", "ข. B", "ค. เกิด Error", "ง. สุ่ม"], "ก"),
]

def run_quiz():
    score = 0
    for i, (q, choices, ans) in enumerate(quiz, 1):
        print(f"\nข้อ {i}. {q}")
        for c in choices:
            print("   ", c)
        a = input("ตอบ (ก/ข/ค/ง): ").strip()
        if a == ans:
            score += 1
            print("✅ ถูกต้อง")
        else:
            print(f"❌ ยังไม่ถูก คำตอบคือ {ans}")
    print(f"\nคะแนนของคุณ: {score}/{len(quiz)}" + ("  🎉 ยอดเยี่ยม!" if score == len(quiz) else ""))

run_quiz()
````
