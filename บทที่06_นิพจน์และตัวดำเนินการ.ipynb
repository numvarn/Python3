## 6.1 นิพจน์

### 6.1.1 ความหมายของนิพจน์
> 📌 **นิพจน์ (Expression)** คือ การดำเนินการที่ประกอบด้วย **ตัวดำเนินการ (Operator)** และ **ตัวถูกดำเนินการ (Operand)** ซึ่ง **ให้ผลลัพธ์ออกมาเป็นค่าค่าหนึ่ง**

```
        A   +   B
        ▲   ▲   ▲
        │   │   └── ตัวถูกดำเนินการ (Operand)
        │   └────── ตัวดำเนินการ (Operator)
        └────────── ตัวถูกดำเนินการ (Operand)
```
ตัวอย่างนิพจน์ เช่น `2 + 1` หรือ `x**2 + y` ซึ่งนิพจน์ที่มีตัวดำเนินการแบ่งได้ 2 ชนิด คือ **นิพจน์เลขคณิต** และ **นิพจน์ทางตรรกะ**

**คำสั่ง (Statement) ต่างจากนิพจน์ (Expression) อย่างไร**

| | นิพจน์ (Expression) | คำสั่ง (Statement) |
|---|---|---|
| ความหมาย | ส่วนของโค้ดที่ **คำนวณแล้วได้ค่า** | **หน่วยของการสั่งงาน** ให้โปรแกรมทำบางอย่าง |
| ตัวอย่าง | `3 + 4`, `x > 5`, `len("abc")` | `x = 3 + 4`, `if x > 5:`, `print(x)`, `import math` |
| ความสัมพันธ์ | มักเป็น **ส่วนประกอบ** ของคำสั่ง | คำสั่งหนึ่งอาจมีนิพจน์หลายนิพจน์อยู่ข้างใน |

### 6.1.2 นิพจน์เลขคณิต
**นิพจน์เลขคณิต (Arithmetic Expression)** เป็นนิพจน์ที่ให้ผลลัพธ์เป็น **ตัวเลข** (จำนวนเต็มหรือจำนวนจริง) ซึ่งเป็นพื้นฐานสำคัญของการทำงานของคอมพิวเตอร์ ตัวถูกดำเนินการเป็น **ค่าคงที่** หรือ **ตัวแปร** ก็ได้

````python
print(6 + 2)
print(6 * 2)
print(6 - 2)
print(6 / 2)      # การหารด้วย / ได้ผลเป็น float เสมอ
print(6 % 2)
````

````python
a = 9
b = 3
print(a + b, a - b, a * b, a / b, a % b)
````

### 6.1.3 นิพจน์ทางตรรกะ
**นิพจน์ทางตรรกะ (Boolean Expression)** เป็นนิพจน์ที่ให้ผลลัพธ์ได้เพียง **2 ค่า** คือ **จริง (`True`)** หรือ **เท็จ (`False`)** โดยใช้ตัวดำเนินการ 2 ประเภท คือ **ตัวดำเนินการสัมพันธ์ (Relational Operator)** และ **ตัวดำเนินการตรรกะ (Logical Operator)** นิพจน์ทางตรรกะเป็นหัวใจของ **การตัดสินใจ** ในโปรแกรม (บทที่ 7)

````python
print(10 > 9)
print(10 >= 10)
print(10 < 9)
print(10 < 10)
````

````python
a = 5
b = 6
print("a == b :", a == b)
print("a != b :", a != b)
print("a > b  :", a > b)
print("a < b  :", a < b)
print("a >= b :", a >= b)
print("a <= b :", a <= b)
````

## 6.2 ตัวดำเนินการ
ไพธอนมีตัวดำเนินการ 7 กลุ่ม

| กลุ่ม | ตัวดำเนินการ |
|---|---|
| กำหนดค่า (Assignment) | `=  +=  -=  *=  /=  %=  //=  **=  &=  \|=  ^=  >>=  <<=` |
| คณิตศาสตร์ (Arithmetic) | `+  -  *  /  %  **  //` |
| สัมพันธ์ (Relational / Comparison) | `>  >=  <  <=  ==  !=` |
| ตรรกะ (Logical) | `and  or  not` |
| สมาชิก (Membership) | `in  not in` |
| เอกลักษณ์ (Identity) | `is  is not` |
| ระดับบิต (Bitwise) | `&  \|  ^  ~  <<  >>` |

### 6.2.1 ตัวดำเนินการกำหนดค่า (Assignment Operators)
ใช้กำหนดค่าให้ตัวแปร เครื่องหมาย `=` กำหนดค่า **ทางขวา** ให้ตัวแปร **ทางซ้าย** และไพธอนมี **ตัวดำเนินการผสม (Compound Operators)** ซึ่งรวมการคำนวณกับการกำหนดค่าไว้ด้วยกัน เช่น `a += 5` มีความหมายเดียวกับ `a = a + 5`

**ตารางที่ 6.1** ตัวดำเนินการกำหนดค่าในภาษาไพธอน

| ตัวดำเนินการ | ตัวอย่าง | ความหมาย |
|---|---|---|
| `=` | `x = 5` | `x = 5` |
| `+=` | `x += 5` | `x = x + 5` |
| `-=` | `x -= 5` | `x = x - 5` |
| `*=` | `x *= 5` | `x = x * 5` |
| `/=` | `x /= 5` | `x = x / 5` |
| `%=` | `x %= 5` | `x = x % 5` |
| `//=` | `x //= 5` | `x = x // 5` |
| `**=` | `x **= 5` | `x = x ** 5` |
| `&=` | `x &= 5` | `x = x & 5` |
| `\|=` | `x \|= 5` | `x = x \| 5` |
| `^=` | `x ^= 5` | `x = x ^ 5` |
| `>>=` | `x >>= 5` | `x = x >> 5` |
| `<<=` | `x <<= 5` | `x = x << 5` |

````python
x = 10
print("เริ่มต้น       x =", x)
x += 5;  print("x += 5   ->   x =", x)
x -= 3;  print("x -= 3   ->   x =", x)
x *= 2;  print("x *= 2   ->   x =", x)
x /= 4;  print("x /= 4   ->   x =", x)
x //= 2; print("x //= 2  ->   x =", x)
x **= 3; print("x **= 3  ->   x =", x)
x %= 5;  print("x %= 5   ->   x =", x)
````

> 💡 ตัวดำเนินการผสมใช้กับสตริงและลิสต์ได้ด้วย เช่น `msg += "!"` และ `items += [4]` ส่วน `x++` แบบภาษา C **ไม่มีในไพธอน** ต้องใช้ `x += 1`

### 6.2.2 ตัวดำเนินการทางคณิตศาสตร์ (Arithmetic Operators)

**ตารางที่ 6.2** การทำงานของตัวดำเนินการทางคณิตศาสตร์ (กำหนดให้ `a = 11` และ `b = 20`)

| ตัวดำเนินการ | ชื่อเรียก | ความหมาย | ตัวอย่าง |
|---|---|---|---|
| `+` | Addition | บวกค่าทั้งสองด้าน | `a + b` = 31 |
| `-` | Subtraction | ลบค่าทางขวาออกจากค่าทางซ้าย | `a - b` = -9 |
| `*` | Multiplication | คูณค่าทั้งสองด้าน | `a * b` = 220 |
| `/` | Division | หารค่าทางซ้ายด้วยค่าทางขวา (ได้ **float เสมอ**) | `b / a` = 1.8181… |
| `%` | Modulus | หารแล้วเอา **เศษ** | `b % a` = 9 |
| `**` | Exponent | ค่าทางซ้าย **ยกกำลัง** ค่าทางขวา | `a ** 2` = 121 |
| `//` | Floor Division | หารแล้ว **ปัดลง** เป็นจำนวนเต็มที่ไม่มากกว่าผลหาร | `9 // 2` = 4, `9.0 // 2.0` = 4.0, `-11 // 3` = -4 |

````python
a, b = 11, 20
print("a + b  =", a + b)
print("a - b  =", a - b)
print("a * b  =", a * b)
print("b / a  =", b / a)
print("b % a  =", b % a)
print("a ** 2 =", a ** 2)
print("b // a =", b // a)
````

````python
a = 21
b = 10
c = a + b;   print("a + b  =", c)
c = a - b;   print("a - b  =", c)
c = a * b;   print("a * b  =", c)
c = a / b;   print("a / b  =", c)
c = a % b;   print("a % b  =", c)
c = a ** b;  print("a ** b =", c)
a, b = 10, 5
c = a // b;  print("10 // 5 =", c)
````

#### `//` และ `%` คู่หูของการหาร
สำหรับจำนวนเต็มบวก `a // b` คือ **ผลหารที่ตัดเศษ** และ `a % b` คือ **เศษ** ซึ่งเป็นจริงเสมอว่า `a == (a // b) * b + (a % b)`

```
   17 ÷ 5  =  3  เศษ  2
   17 // 5 =  3
   17 %  5 =  2         ตรวจสอบ: 3 × 5 + 2 = 17 ✓
```
**การประยุกต์ที่พบบ่อย:** ตรวจเลขคู่/คี่ (`n % 2`), แปลงหน่วยเวลา, ทอนเงิน, ดึงหลักของตัวเลข

````python
total_seconds = 7384
hours = total_seconds // 3600
minutes = (total_seconds % 3600) // 60
seconds = total_seconds % 60
print(f"{total_seconds} วินาที = {hours} ชั่วโมง {minutes} นาที {seconds} วินาที")

n = 2569
print(f"{n} เป็นเลข{'คู่' if n % 2 == 0 else 'คี่'}")
print(f"หลักหน่วยของ {n} คือ {n % 10} และหลักสิบคือ {n // 10 % 10}")
````

### ⚠️ `//` กับจำนวนลบ: ปัดลง ไม่ใช่ตัดทิ้ง
`//` ปัดผลลงไปหาจำนวนที่ **น้อยกว่า** (floor) เสมอ เช่น −11 / 3 = −3.67 จึงปัดลงเป็น **−4** (ไม่ใช่ −3) ซึ่งต่างจาก `int(-11/3)` ที่ตัดทศนิยมทิ้งได้ −3 และ `%` จะให้เศษที่มีเครื่องหมายเดียวกับตัวหาร
</div>

````python
print(-11 // 3, "| int(-11 / 3) =", int(-11 / 3))
print(-11.0 // 3)
print(-11 % 3)          # เศษเป็นบวก เพราะตัวหาร (3) เป็นบวก
print((-11 // 3) * 3 + (-11 % 3))   # ยังได้ -11 เสมอ
````

### 6.2.3 ตัวดำเนินการสัมพันธ์ (Relational Operators)
เมื่อโปรแกรมต้อง **ตัดสินใจ** เลือกทำงานทางใดทางหนึ่ง มักต้องเปรียบเทียบข้อมูล เช่น เท่ากันหรือไม่ น้อยกว่าหรือมากกว่า ซึ่งใช้ **ตัวดำเนินการสัมพันธ์** และให้ผลเป็น `True` หรือ `False` เท่านั้น

**ตารางที่ 6.3** ชื่อเรียกและความหมายของตัวดำเนินการสัมพันธ์

| ตัวดำเนินการ | ชื่อเรียก | ความหมาย | ตัวอย่าง |
|---|---|---|---|
| `>` | Greater than | มากกว่า | `x > y` |
| `>=` | Greater than or equal to | มากกว่าหรือเท่ากับ | `x >= y` |
| `<` | Less than | น้อยกว่า | `x < y` |
| `<=` | Less than or equal to | น้อยกว่าหรือเท่ากับ | `x <= y` |
| `==` | Equal to | เท่ากับ | `x == y` |
| `!=` | Not equal to | ไม่เท่ากับ | `x != y` |

> ⚠️ **ข้อผิดพลาดที่พบบ่อยที่สุด:** สับสนระหว่าง `=` (กำหนดค่า) กับ `==` (เปรียบเทียบ)

````python
x = 10
y = 12
print("x > y  :", x > y)
print("x >= y :", x >= y)
print("x < y  :", x < y)
print("x <= y :", x <= y)
print("x == y :", x == y)
print("x != y :", x != y)
````

#### 🔬 ความสามารถพิเศษของไพธอน
- **เปรียบเทียบต่อเนื่อง (Chained Comparison):** เขียน `0 <= score <= 100` ได้เหมือนในคณิตศาสตร์ (ภาษา C ต้องเขียน `score >= 0 && score <= 100`)
- **เปรียบเทียบสตริง:** เทียบตามลำดับรหัส Unicode ทีละอักขระ (ทบทวนบทที่ 1) โดยตัวพิมพ์ใหญ่มีรหัสน้อยกว่าตัวพิมพ์เล็ก
- **เปรียบเทียบ float:** ระวังความคลาดเคลื่อนของทศนิยม (บทที่ 1) ควรใช้ `math.isclose()`

````python
score = 75
print("0 <= score <= 100 :", 0 <= score <= 100)

print("'apple' < 'banana' :", 'apple' < 'banana')
print("'Zebra' < 'apple'  :", 'Zebra' < 'apple', " (เพราะ ord('Z') =", ord('Z'), "< ord('a') =", ord('a'), ")")

import math
print("0.1 + 0.2 == 0.3          :", 0.1 + 0.2 == 0.3)
print("math.isclose(0.1+0.2, 0.3):", math.isclose(0.1 + 0.2, 0.3))
````

### 6.2.4 ตัวดำเนินการตรรกะ (Logical Operators)
ใช้ **รวมนิพจน์ทางตรรกะหลายนิพจน์** เข้าด้วยกัน และให้ผลเป็น `True` หรือ `False`

**ตารางที่ 6.4** ตัวดำเนินการตรรกะและความหมาย

| ตัวดำเนินการ | ความหมาย | ตัวอย่าง |
|---|---|---|
| `and` | ให้ `True` เมื่อตัวถูกดำเนินการ **เป็น True ทั้งหมด** | `x and y` |
| `or` | ให้ `True` เมื่อตัวถูกดำเนินการ **ตัวใดตัวหนึ่งเป็น True** | `x or y` |
| `not` | **กลับค่า** True เป็น False และ False เป็น True | `not x` |

**ตารางที่ 6.5** ตารางค่าความจริง (Truth Table)

| X | Y | X and Y | X or Y | not X |
|---|---|---|---|---|
| True | True | **True** | True | False |
| True | False | False | True | False |
| False | True | False | **True** | True |
| False | False | False | **False** | True |

> 💡 **จำง่าย ๆ:** `and` เป็น True **กรณีเดียว** คือทุกตัวเป็น True ส่วน `or` เป็น False **กรณีเดียว** คือทุกตัวเป็น False

### 6.2.6 ตัวดำเนินการเอกลักษณ์ (Identity Operators)
`is` และ `is not` ใช้ตรวจว่า **ตัวแปรสองตัวอ้างถึงวัตถุ (object) เดียวกันในหน่วยความจำหรือไม่** ซึ่งต่างจาก `==` ที่ตรวจว่า **ค่าเท่ากันหรือไม่** ตัวแปรสองตัวที่ค่าเท่ากันจึง **ไม่จำเป็นต้องเป็นวัตถุเดียวกัน**

| | `==` | `is` |
|---|---|---|
| ตรวจอะไร | ค่าเท่ากันหรือไม่ | เป็นวัตถุเดียวกันหรือไม่ (ตำแหน่งเดียวกันในหน่วยความจำ) |
| เทียบเท่ากับ | | `id(a) == id(b)` |

````python
x1 = 5
y1 = 5
x2 = 'Hello'
y2 = 'Hello'
x3 = [1, 2, 3]
y3 = [1, 2, 3]

print("x1 is y1     :", x1 is y1)
print("x2 is y2     :", x2 is y2)
print("x1 is not y1 :", x1 is not y1)
print("x3 is y3     :", x3 is y3)
print("x3 == y3     :", x3 == y3)
print("id(x3), id(y3):", id(x3), id(y3))
````

`x3` และ `y3` เป็นลิสต์ที่มี **ค่าเท่ากัน (`==` ได้ True)** แต่ **ไม่ใช่ลิสต์เดียวกัน (`is` ได้ False)** เพราะลิสต์เปลี่ยนแปลงได้ ไพธอนจึงสร้างแยกไว้คนละที่ในหน่วยความจำ (ทบทวนกับดัก `b = a` ในบทที่ 5)

> ⚠️ ที่ `x1 is y1` และ `x2 is y2` ได้ True เป็นเพราะไพธอน **ใช้วัตถุเดิมซ้ำ** สำหรับจำนวนเต็มเล็ก ๆ และสตริงสั้น ๆ เพื่อประหยัดหน่วยความจำ ซึ่ง **ไม่รับประกันเสมอไป** ดังนั้น **เปรียบเทียบค่าให้ใช้ `==` เสมอ** และใช้ `is` เฉพาะกับ `None` เช่น `if result is None:`

````python
a = [1, 2, 3]
b = a            # ชี้วัตถุเดียวกัน
c = a.copy()     # วัตถุใหม่ ค่าเท่ากัน
print("b is a :", b is a, "| c is a :", c is a, "| c == a :", c == a)

result = None
print("result is None :", result is None)
````

### 6.2.7 ตัวดำเนินการระดับบิต (Bitwise Operators)
ทำงานกับ **เลขฐานสองทีละบิต** (ทบทวนบทที่ 1)

**ตารางที่ 6.6** การทำงานของตัวดำเนินการระดับบิต

| ตัวดำเนินการ | ชื่อเรียก | ความหมาย |
|---|---|---|
| `&` | Binary AND | ได้ 1 เมื่อบิตตำแหน่งเดียวกัน **เป็น 1 ทั้งคู่** นอกนั้นได้ 0 |
| `\|` | Binary OR | ได้ 1 เมื่อบิตตำแหน่งเดียวกัน **เป็น 1 อย่างน้อยหนึ่งตัว** และได้ 0 เมื่อเป็น 0 ทั้งคู่ |
| `^` | Binary XOR | ได้ 1 เมื่อบิตตำแหน่งเดียวกัน **ต่างกัน** และได้ 0 เมื่อเหมือนกัน |
| `~` | Binary Ones Complement | กลับบิต 0 เป็น 1 และ 1 เป็น 0 (ผลคือ `~a = -(a + 1)`) |
| `<<` | Binary Left Shift | เลื่อนบิตไปทางซ้ายตามจำนวนที่กำหนด (เท่ากับ **คูณ 2 ยกกำลัง n**) |
| `>>` | Binary Right Shift | เลื่อนบิตไปทางขวาตามจำนวนที่กำหนด (เท่ากับ **หาร 2 ยกกำลัง n** ปัดลง) |

ตัวอย่าง: `a = 60` และ `b = 13`
```
  a      = 0011 1100   (60)
  b      = 0000 1101   (13)
  a & b  = 0000 1100   (12)
  a | b  = 0011 1101   (61)
  a ^ b  = 0011 0001   (49)
  ~a     = 1100 0011   (-61 ในรูป two's complement)
  a << 2 = 1111 0000   (240)
  a >> 2 = 0000 1111   (15)
```

````python
a = 60
b = 13
print(f"a      = {a:08b} ({a})")
print(f"b      = {b:08b} ({b})")
print(f"a & b  = {a & b:08b} ({a & b})")
print(f"a | b  = {a | b:08b} ({a | b})")
print(f"a ^ b  = {a ^ b:08b} ({a ^ b})")
print(f"~a     = {~a & 0xFF:08b} ({~a})   <- 8 บิตล่าง (two's complement จากบทที่ 1)")
print(f"a << 2 = {a << 2:08b} ({a << 2})")
print(f"a >> 2 = {a >> 2:08b} ({a >> 2})")
````

> 💡 **ใช้ทำอะไรในงานจริง?** เช่น การกำหนดสิทธิ์ไฟล์ (read/write/execute), การคำนวณ **Subnet Mask** ในระบบเครือข่าย (นำ IP Address มา AND กับ Subnet Mask จะได้ Network Address) และการบีบอัดหรือเข้ารหัสข้อมูล

````python
# ตัวอย่าง: หา Network Address จาก IP 192.168.10.77 / Subnet mask 255.255.255.0
ip   = [192, 168, 10, 77]
mask = [255, 255, 255, 0]
network = [i & m for i, m in zip(ip, mask)]
print("IP Address      :", ".".join(map(str, ip)))
print("Subnet Mask     :", ".".join(map(str, mask)))
print("Network Address :", ".".join(map(str, network)))
````

## 6.3 ลำดับการดำเนินการ
**ลำดับการดำเนินการ (Operator Precedence)** คือกฎที่กำหนดว่าในนิพจน์ที่มีตัวดำเนินการหลายตัว **ต้องคำนวณส่วนใดก่อนหลัง** ซึ่งนักคณิตศาสตร์ทั่วโลกตกลงร่วมกัน เพื่อให้แก้ปัญหาได้ถูกต้องตรงกัน มิฉะนั้นคำตอบจะผิดเพี้ยน

### 6.3.1 กฎพื้นฐาน
1. **วงเล็บ** ทำก่อน
2. **เลขยกกำลัง**
3. **คูณ และ หาร** (รวมถึง `//` และ `%`) ทำจากซ้ายไปขวา
4. **บวก และ ลบ** ทำจากซ้ายไปขวา

> ⚠️ **ข้อยกเว้นสำคัญในไพธอน:** เลขยกกำลัง `**` ทำจาก **ขวาไปซ้าย** เช่น `2 ** 3 ** 2` = `2 ** 9` = 512 (ไม่ใช่ `8 ** 2` = 64) และ `-2 ** 2` = **−4** เพราะยกกำลังทำก่อนเครื่องหมายลบ

````python
print("2 ** 3 ** 2   =", 2 ** 3 ** 2)      # = 2 ** 9
print("(2 ** 3) ** 2 =", (2 ** 3) ** 2)    # = 8 ** 2
print("-2 ** 2       =", -2 ** 2)          # = -(2 ** 2)
print("(-2) ** 2     =", (-2) ** 2)
````

#### ตารางลำดับความสำคัญของตัวดำเนินการในไพธอน (จากสูงไปต่ำ)

| ลำดับ | ตัวดำเนินการ | คำอธิบาย |
|---|---|---|
| 1 | `( )` | วงเล็บ |
| 2 | `**` | ยกกำลัง (ขวาไปซ้าย) |
| 3 | `+x`, `-x`, `~x` | เครื่องหมายบวก ลบ หน้าตัวเลข และกลับบิต |
| 4 | `*`, `/`, `//`, `%` | คูณ หาร หารปัดลง หารเอาเศษ |
| 5 | `+`, `-` | บวก ลบ |
| 6 | `<<`, `>>` | เลื่อนบิต |
| 7 | `&` | AND ระดับบิต |
| 8 | `^` | XOR ระดับบิต |
| 9 | `\|` | OR ระดับบิต |
| 10 | `==`, `!=`, `>`, `>=`, `<`, `<=`, `is`, `is not`, `in`, `not in` | เปรียบเทียบ สมาชิก เอกลักษณ์ |
| 11 | `not` | ตรรกะ NOT |
| 12 | `and` | ตรรกะ AND |
| 13 | `or` | ตรรกะ OR |
| 14 | `=`, `+=`, `-=`, … | กำหนดค่า (ทำหลังสุด) |

> 💡 **เคล็ดลับ:** ถ้าไม่แน่ใจลำดับ **ให้ใส่วงเล็บ** เพราะทำให้โค้ดอ่านง่ายขึ้นด้วย (Explicit is better than implicit)

### 6.3.2 วงเล็บและเลขยกกำลัง
เมื่อพบวงเล็บหรือเลขยกกำลัง ให้คำนวณนิพจน์ย่อยในวงเล็บหรือเลขยกกำลังก่อน

**ตัวอย่างที่ 1:** `(2 + 3) * (4 - 1) + 2 ** 3`
```
(2 + 3) * (4 - 1) + 2 ** 3
=   5   * (4 - 1) + 2 ** 3      ← วงเล็บแรก
=   5   *    3    + 2 ** 3      ← วงเล็บที่สอง
=   5   *    3    +   8         ← ยกกำลัง
=      15         +   8         ← คูณ
=      23                       ← บวก
```
**ตัวอย่างที่ 2:** `(2 ** 3) ** 2`
```
(2 ** 3) ** 2  =  8 ** 2  =  64
```

### 6.3.3 การคูณและการหาร
การคูณ **ไม่จำเป็นต้องทำก่อนการหาร** เมื่อพบคูณและหารต่อเนื่องกัน ให้ **ทำจากซ้ายไปขวา** เสมอ
```
5 * 4 - 9 / 3
=  20 - 9 / 3
=  20 - 3.0
=  17.0
```
> ตัวอย่างความต่างของซ้ายไปขวา: `8 / 4 * 2` = `2.0 * 2` = **4.0** (ไม่ใช่ `8 / 8` = 1.0)

### 6.3.4 การบวกและการลบ
ทำลำดับสุดท้าย และทำจากซ้ายไปขวาเช่นกัน ตัวอย่างเมื่อใช้กฎทั้งหมด
```
(1 + 8) * (4 - 1) + 16 / 2 ** 3
=   9   *    3    + 16 / 2 ** 3     ← วงเล็บ
=   9   *    3    + 16 / 8          ← ยกกำลัง
=      27         + 2.0             ← คูณ หาร (ซ้ายไปขวา)
=      29.0                         ← บวก
```

#### 🔬 เครื่องมือช่วยดูลำดับการคำนวณ
ฟังก์ชันด้านล่างใช้โมดูล `ast` ของไพธอนแสดงว่า **ไพธอนจับกลุ่มนิพจน์อย่างไร** (ใส่วงเล็บให้ทุกขั้น) ลองเปลี่ยนนิพจน์ในลิสต์แล้วรันดู

````python
import ast

def show_precedence(expr):
    tree = ast.parse(expr, mode="eval").body
    def wrap(node):
        if isinstance(node, ast.BinOp):
            return f"({wrap(node.left)} {_op(node.op)} {wrap(node.right)})"
        if isinstance(node, ast.UnaryOp):
            return f"({_op(node.op)}{wrap(node.operand)})"
        if isinstance(node, ast.BoolOp):
            op = " and " if isinstance(node.op, ast.And) else " or "
            return "(" + op.join(wrap(v) for v in node.values) + ")"
        if isinstance(node, ast.Compare):
            parts = [wrap(node.left)]
            for o, c in zip(node.ops, node.comparators):
                parts += [_op(o), wrap(c)]
            return "(" + " ".join(parts) + ")"
        return ast.unparse(node)
    print(f"{expr:<28} ->  {wrap(tree):<40} = {eval(expr)}")

def _op(op):
    return {ast.Add: "+", ast.Sub: "-", ast.Mult: "*", ast.Div: "/", ast.FloorDiv: "//",
            ast.Mod: "%", ast.Pow: "**", ast.USub: "-", ast.UAdd: "+", ast.Not: "not ",
            ast.Invert: "~", ast.BitAnd: "&", ast.BitOr: "|", ast.BitXor: "^",
            ast.LShift: "<<", ast.RShift: ">>", ast.Gt: ">", ast.Lt: "<", ast.GtE: ">=",
            ast.LtE: "<=", ast.Eq: "==", ast.NotEq: "!="}[type(op)]

for e in ["2 + 3 * 4", "2 ** 3 ** 2", "-2 ** 2", "8 / 4 * 2",
          "17 / 2 * 3 + 2", "5 > 3 and 2 > 4 or 1 < 2", "not 1 + 1 == 3"]:
    show_precedence(e)
````

### ✍️ ลงมือทำ 1: คำนวณด้วยมือก่อน แล้วค่อยให้ไพธอนตรวจ
คำนวณค่าของนิพจน์ต่อไปนี้ **ด้วยมือ** (แสดงขั้นตอนบนกระดาษ) แล้วใส่คำตอบในตัวแปร `p1`–`p5` จากนั้นรันเซลล์ตรวจคำตอบ (ห้ามพิมพ์นิพจน์ลงไปตรง ๆ)
1. `p1` : `10 - 4 * 2 + 1`
2. `p2` : `(10 - 4) * (2 + 1)`
3. `p3` : `20 // 6 + 20 % 6`
4. `p4` : `3 ** 2 ** 2`
5. `p5` : `10 > 5 and not 3 == 3 or 2 < 1` (ตอบ True/False)

## 6.4 สรุป
- **นิพจน์ (Expression)** คือการดำเนินการที่ประกอบด้วย **ตัวดำเนินการ (Operator)** และ **ตัวถูกดำเนินการ (Operand)** แล้วให้ผลลัพธ์เป็นค่าหนึ่ง
- **นิพจน์เลขคณิต** ให้ผลเป็นตัวเลข (จำนวนเต็มหรือจำนวนจริง) และ **นิพจน์ทางตรรกะ** ให้ผลเพียง `True` หรือ `False`
- **ตัวดำเนินการกำหนดค่า** `=` และตัวดำเนินการผสม เช่น `+=`, `-=`, `*=`, `//=`, `**=`
- **ตัวดำเนินการทางคณิตศาสตร์** `+ - * / % ** //` โดย `/` ได้ float เสมอ `//` หารปัดลง และ `%` หารเอาเศษ
- **ตัวดำเนินการสัมพันธ์** `> >= < <= == !=` ใช้เปรียบเทียบค่า 2 ค่า ได้ผลเป็น True/False และเขียนต่อเนื่องได้ เช่น `0 <= x <= 100`
- **ตัวดำเนินการตรรกะ** `and` (True เมื่อทุกตัวเป็น True), `or` (False เมื่อทุกตัวเป็น False) และ `not` (กลับค่า) โดยประเมินแบบลัดวงจร
- **ตัวดำเนินการสมาชิก** `in`, `not in` ตรวจความเป็นสมาชิก (ดิกชันนารีตรวจที่คีย์)
- **ตัวดำเนินการเอกลักษณ์** `is`, `is not` ตรวจว่าเป็นวัตถุเดียวกันหรือไม่ (เปรียบเทียบค่าให้ใช้ `==`)
- **ตัวดำเนินการระดับบิต** `& | ^ ~ << >>` ทำงานกับเลขฐานสองทีละบิต
- **ลำดับการดำเนินการ (Precedence)** คือลำดับก่อนหลังเมื่อนิพจน์มีตัวดำเนินการหลายตัว: วงเล็บ → ยกกำลัง (ขวาไปซ้าย) → คูณ/หาร (ซ้ายไปขวา) → บวก/ลบ (ซ้ายไปขวา) → เปรียบเทียบ → not → and → or

### 🏆 แบบทดสอบท้ายบท (ตรวจคะแนนอัตโนมัติ)
รันเซลล์ด้านล่าง แล้วพิมพ์ตัวเลือก (ก/ข/ค/ง) ของแต่ละข้อ

````python
quiz = [
    ("ในนิพจน์ a * b เครื่องหมาย * เรียกว่าอะไร",
     ["ก. ตัวถูกดำเนินการ", "ข. ตัวดำเนินการ", "ค. ตัวแปร", "ง. คำสงวน"], "ข"),
    ("15 // 4 มีค่าเท่าใด",
     ["ก. 3.75", "ข. 3", "ค. 4", "ง. 3.0"], "ข"),
    ("15 % 4 มีค่าเท่าใด",
     ["ก. 3", "ข. 3.75", "ค. 0", "ง. 4"], "ก"),
    ("x = 6 แล้ว x //= 4 ค่า x เท่ากับเท่าใด",
     ["ก. 1.5", "ข. 2", "ค. 1", "ง. 24"], "ค"),
    ("False or True ได้ผลใด",
     ["ก. True", "ข. False", "ค. None", "ง. Error"], "ก"),
    ("not (3 < 5) and 2 == 2 ได้ผลใด",
     ["ก. True", "ข. False", "ค. 2", "ง. Error"], "ข"),
    ("ข้อใดตรวจว่าตัวแปรสองตัวเป็นวัตถุเดียวกันในหน่วยความจำ",
     ["ก. ==", "ข. in", "ค. is", "ง. ="], "ค"),
    ("6 & 3 มีค่าเท่าใด (6 = 110, 3 = 011)",
     ["ก. 2", "ข. 7", "ค. 5", "ง. 9"], "ก"),
    ("2 ** 3 ** 2 มีค่าเท่าใด",
     ["ก. 64", "ข. 512", "ค. 36", "ง. 12"], "ข"),
    ("10 - 2 * 3 + 4 / 2 มีค่าเท่าใด",
     ["ก. 6.0", "ข. 14.0", "ค. 26.0", "ง. 8.0"], "ก"),
]

def run_quiz():
    score = 0
    for i, (q, choices, ans) in enumerate(quiz, 1):
        print(f"\nข้อ {i}. {q}")
        for c in choices:
            print("   ", c)
        reply = input("คำตอบ (ก/ข/ค/ง): ").strip()
        if reply == ans:
            print("   ✅ ถูกต้อง"); score += 1
        else:
            print(f"   ❌ ยังไม่ถูก คำตอบที่ถูกคือ {ans}")
    print(f"\n🎯 คะแนนของคุณ: {score}/{len(quiz)}")
    print("🏆 ยอดเยี่ยม!" if score == len(quiz) else "👍 ดีมาก" if score >= 7 else "📖 ทบทวนเนื้อหาอีกครั้งนะ")

run_quiz()
````
