### 🧪 เตรียมโฟลเดอร์ทดลอง
รันเซลล์ด้านล่างก่อนเสมอ (รวมถึงทุกครั้งที่ Restart Kernel) เซลล์นี้จะสร้างโฟลเดอร์ `ch10_lab` และย้าย **ไดเรกทอรีทำงานปัจจุบัน (Current Working Directory)** เข้าไปในโฟลเดอร์นั้น ทำให้ชื่อไฟล์สั้น ๆ เช่น `'somefile.txt'` หมายถึงไฟล์ในโฟลเดอร์นี้

````python
import os

LAB = "ch10_lab"
if os.path.basename(os.getcwd()) != LAB:      # ป้องกันการซ้อนโฟลเดอร์เมื่อรันซ้ำ
    os.makedirs(LAB, exist_ok=True)
    os.chdir(LAB)

print("ไดเรกทอรีทำงานปัจจุบัน:", os.getcwd())
````

## 10.1 พื้นฐานเกี่ยวกับไฟล์

**ไฟล์ (File)** หรือ **แฟ้มข้อมูล** ในทางคอมพิวเตอร์ หมายถึงกลุ่มระเบียนสารสนเทศ หรือทรัพยากรสำหรับเก็บบันทึกสารสนเทศ ที่โปรแกรมคอมพิวเตอร์ใช้งานได้ โดยปกติอยู่บน **หน่วยเก็บบันทึกถาวร** (ฮาร์ดดิสก์ SSD แฟลชไดรฟ์) จึง **คงอยู่ต่อไป** แม้โปรแกรมที่สร้างไฟล์จะทำงานเสร็จแล้ว โปรแกรมอื่นยังนำไปใช้ได้

> คำว่า "แฟ้ม" มาจากการเปรียบเทียบกับเอกสารกระดาษที่แต่เดิมเก็บไว้ในตู้แฟ้มของสำนักงานและห้องสมุด ไฟล์คอมพิวเตอร์จึงถือเป็นเอกสารยุคใหม่

| การกระทำกับไฟล์ | ใครทำได้ |
|---|---|
| สร้าง แก้ไข ขยาย/ยุบขนาด | ส่วนใหญ่ทำโดยโปรแกรม เช่น ไฟล์ .docx สร้างและแก้ไขด้วย Microsoft Word |
| ย้าย เปลี่ยนชื่อ ลบ | ผู้ใช้ทำได้โดยตรงผ่านโปรแกรมจัดการไฟล์ เช่น Windows Explorer หรือ Finder และ **โปรแกรมไพธอนของเราก็ทำได้** (หัวข้อ 10.5) |

**ไฟล์ในระบบปฏิบัติการตระกูลยูนิกซ์ (Unix, Linux, macOS)** กระบวนการในพื้นที่ผู้ใช้ (User space) ไม่ได้จัดการกับไฟล์โดยตรง แต่ระบบปฏิบัติการจัดเตรียม **ระดับการกำหนดสาระสำคัญ (Abstraction)** ไว้ให้ การติดต่อกับไฟล์เกือบทั้งหมดทำผ่าน **ฮาร์ดลิงก์ (Hard link)** ซึ่งทำหน้าที่เชื่อมชื่อหนึ่งชื่อไปยังข้อมูลของไฟล์

- โปรแกรมในพื้นที่ผู้ใช้ **ไม่ได้ลบไฟล์โดยตรง** แต่ลบ **ลิงก์** ที่อ้างถึงไฟล์ (เช่นคำสั่งเชลล์ `rm`)
- เมื่อ **เคอร์เนล (Kernel)** พบว่าไม่มีลิงก์ใดชี้มาที่ไฟล์แล้ว จึงคืนพื้นที่ของไฟล์นั้น — ในความเป็นจริงมีเพียงเคอร์เนลเท่านั้นที่จัดการไฟล์บนดิสก์ได้โดยตรง

### 🆕 ไฟล์ข้อความ (Text file) กับไฟล์ไบนารี (Binary file)

| | ไฟล์ข้อความ | ไฟล์ไบนารี |
|---|---|---|
| เนื้อหา | ตัวอักษรที่คนอ่านได้ แบ่งเป็นบรรทัด | ไบต์ดิบตามรูปแบบเฉพาะของโปรแกรม |
| ตัวอย่าง | `.txt` `.py` `.csv` `.html` `.json` | `.jpg` `.png` `.mp3` `.docx` `.xlsx` `.pdf` |
| เปิดด้วย Notepad | อ่านรู้เรื่อง | เห็นเป็นตัวอักษรแปลก ๆ |
| ในไพธอน | ได้/ส่งข้อมูลเป็น `str` | ได้/ส่งข้อมูลเป็น `bytes` (โหมด `'b'`) |

บทนี้เน้น **ไฟล์ข้อความ** ซึ่งพบบ่อยที่สุดในงานประมวลผลข้อมูล

### 🆕 พาธ (Path) — ที่อยู่ของไฟล์
**พาธ** คือสตริงที่บอกตำแหน่งของไฟล์หรือไดเรกทอรี (โฟลเดอร์) มี 2 แบบ

| แบบ | ความหมาย | ตัวอย่าง |
|---|---|---|
| **พาธสัมบูรณ์ (Absolute path)** | เริ่มจากรากของระบบไฟล์ ชี้ตำแหน่งได้แน่นอน | `C:\text\somefile.txt` (Windows), `/home/student/text/somefile.txt` (Linux/macOS) |
| **พาธสัมพัทธ์ (Relative path)** | นับจาก **ไดเรกทอรีทำงานปัจจุบัน** | `somefile.txt`, `data/file.csv`, `../other.txt` |

ในพาธสัมพัทธ์ `.` หมายถึงไดเรกทอรีปัจจุบัน และ `..` หมายถึงไดเรกทอรีแม่ (ระดับเหนือขึ้นไปหนึ่งชั้น)

````python
import os
print("ไดเรกทอรีทำงานปัจจุบัน :", os.getcwd())
print("พาธสัมบูรณ์ของ 'somefile.txt' :", os.path.abspath("somefile.txt"))
print("พาธสัมบูรณ์ของ '..'          :", os.path.abspath(".."))
````

## 10.2 การเปิดไฟล์

สิ่งแรกที่ต้องเรียนรู้คือการ **เปิดไฟล์** ด้วยฟังก์ชัน `open()` และการกำหนด **โหมด (mode)** ของการเปิดไฟล์

### 10.2.1 คำสั่ง `open`
ฟังก์ชัน `open()` มีอาร์กิวเมนต์ที่จำเป็นคือ **พาธของไฟล์** ที่ต้องการเปิด และอาร์กิวเมนต์ทางเลือกคือ **โหมด** เมื่อเปิดสำเร็จจะคืนค่าเป็น **ไฟล์ออบเจกต์ (File Object)**

ไวยากรณ์ในเอกสาร (อ้างอิงจากหนังสือยุคไพธอน 2)
```
open(name [, mode [, buffering]])
```

<div style="border-left:5px solid #7048e8;padding:8px 14px;background:rgba(112,72,232,.08)">

### 🆕 ไวยากรณ์ของ open() ในไพธอน 3
ในไพธอน 3 ฟังก์ชัน `open()` มีพารามิเตอร์มากกว่านั้น ที่ใช้บ่อยคือ

```python
open(file, mode='r', buffering=-1, encoding=None, errors=None, newline=None)
```

| พารามิเตอร์ | ความหมาย |
|---|---|
| `file` | พาธของไฟล์ |
| `mode` | โหมดการเปิด ค่าเริ่มต้น `'r'` (อ่านไฟล์ข้อความ) |
| `encoding` | การเข้ารหัสตัวอักษร **ควรระบุ `encoding="utf-8"` เสมอเมื่อมีภาษาไทย** |
| `newline` | การจัดการอักขระขึ้นบรรทัดใหม่ ใช้ `newline=""` เมื่อเขียนไฟล์ CSV (หัวข้อ 10.6) |

แนะนำให้ส่งอาร์กิวเมนต์ตัวที่สามเป็นต้นไปแบบระบุชื่อ เช่น `open("a.txt", "w", encoding="utf-8")`
</div>

สมมติว่ามีไฟล์ข้อความชื่อ `somefile.txt` เก็บไว้ในไดเรกทอรี `C:\text` (หรือ `~/text` ใน Linux) เอกสารเขียนคำสั่งเปิดไฟล์ไว้ดังนี้
```python
>>> f = open('C:\text\somefile.txt')
```
คำสั่งนี้ **มีกับดักซ่อนอยู่** ลองรันเซลล์ด้านล่าง

````python
p1 = 'C:\text\somefile.txt'      # \t ในสตริงคือ escape character "แท็บ"!
print(p1)
print(repr(p1))
print("ความยาว:", len(p1))
````

### 🆕 ข้อแก้ไข: แบ็กสแลชในพาธของ Windows
ในสตริงของไพธอน `\t` คือ **แท็บ** และ `\n` คือ **ขึ้นบรรทัดใหม่** พาธ `'C:\text\somefile.txt'` จึงกลายเป็น `C:<แท็บ>ext\somefile.txt` ซึ่งไม่ใช่ไฟล์ที่ต้องการ (และ `\s` ยังเป็น escape ที่ไม่ถูกต้อง ไพธอนรุ่นใหม่จะแสดง `SyntaxWarning`) วิธีเขียนพาธ Windows ที่ถูกต้องมี 3 วิธี

```python
f = open(r'C:\text\somefile.txt')     # 1) raw string ใส่ r นำหน้า (แนะนำ)
f = open('C:\\text\\somefile.txt')    # 2) เขียนแบ็กสแลชซ้อน 2 ตัว
f = open('C:/text/somefile.txt')      # 3) ใช้ / ซึ่ง Windows ก็รองรับ
```
</div>

````python
p2 = r'C:\text\somefile.txt'
p3 = 'C:\\text\\somefile.txt'
print(p2)
print(p3)
print("เหมือนกันไหม:", p2 == p3)
````

หากไฟล์ที่ต้องการเปิด **ไม่มีอยู่จริง** โปรแกรมจะแจ้งข้อผิดพลาด `FileNotFoundError`
```
Traceback (most recent call last):
  File "<stdin>", line 1, in <module>
FileNotFoundError: [Errno 2] No such file or directory: 'C:\\text\\somefile.txt'
```
เซลล์ด้านล่างใช้ `try ... except` (จะเรียนละเอียดในบทที่ 11) เพื่อดักข้อผิดพลาดไว้ ไม่ให้ Notebook หยุดทำงาน

````python
try:
    f = open("ไฟล์ที่ไม่มีอยู่จริง.txt")
except FileNotFoundError as e:
    print("เกิดข้อผิดพลาด:", type(e).__name__)
    print(e)
````

### 10.2.2 โหมดอาร์กิวเมนต์
ถ้าเรียก `open()` โดยส่งพาธเพียงอย่างเดียว จะได้ไฟล์ออบเจกต์ที่ **อ่านได้อย่างเดียว** ถ้าต้องการเขียนต้องระบุโหมด

**ตารางโหมดในเอกสาร**

| โหมด | ความหมาย |
|:---:|---|
| `'r'` | อ่านเพียงอย่างเดียว |
| `'w'` | เขียนเพียงอย่างเดียว ถ้ายังไม่มีไฟล์จะสร้างไฟล์ใหม่ |
| `'a'` | เปิดไฟล์แล้วเขียนข้อมูลต่อท้ายข้อมูลเดิม |
| `'w+'` | เขียน สร้างไฟล์ใหม่ หรือแทนที่ข้อมูลเดิม |
| `'r+'` | เปิดไฟล์เพื่ออ่านหรือเขียนข้อมูลทับไฟล์ |

### 🆕 ตารางโหมดฉบับสมบูรณ์ (ไพธอน 3)
| โหมด | อ่าน | เขียน | ไฟล์ไม่มีอยู่ | ไฟล์มีอยู่แล้ว | ตำแหน่งเริ่มต้น |
|:---:|:---:|:---:|---|---|---|
| `'r'` (ค่าเริ่มต้น) | ✅ | ❌ | ❌ `FileNotFoundError` | อ่านได้ | ต้นไฟล์ |
| `'w'` | ❌ | ✅ | สร้างใหม่ | ⚠️ **ลบข้อมูลเดิมทิ้งทันที** | ต้นไฟล์ |
| `'a'` | ❌ | ✅ | สร้างใหม่ | เก็บข้อมูลเดิมไว้ | ท้ายไฟล์ |
| `'x'` | ❌ | ✅ | สร้างใหม่ | ❌ `FileExistsError` (กันเขียนทับ) | ต้นไฟล์ |
| `'r+'` | ✅ | ✅ | ❌ `FileNotFoundError` | เก็บข้อมูลเดิม เขียนทับตามตำแหน่ง | ต้นไฟล์ |
| `'w+'` | ✅ | ✅ | สร้างใหม่ | ⚠️ ลบข้อมูลเดิม | ต้นไฟล์ |
| `'a+'` | ✅ | ✅ | สร้างใหม่ | เก็บข้อมูลเดิม | ท้ายไฟล์ |

**ตัวขยายโหมด** ต่อท้ายได้: `'t'` = ไฟล์ข้อความ (ค่าเริ่มต้น) และ `'b'` = ไฟล์ไบนารี เช่น `'rb'` อ่านไฟล์รูปภาพ, `'wb'` เขียนไฟล์ไบนารี

> ข้อควรจำ: ในเอกสารบอกว่า `'w'` "ถ้ายังไม่มีไฟล์จะสร้างไฟล์ใหม่" แต่ **ถ้ามีไฟล์อยู่แล้ว ข้อมูลเดิมจะถูกลบทั้งหมดทันทีที่เปิด** แม้ยังไม่ได้เขียนอะไรเลย

## 10.3 ไฟล์ออบเจกต์

สิ่งที่ได้จาก `open()` คือ **ไฟล์ออบเจกต์ (File Object)** ขั้นต่อไปคือใช้เมธอดของไฟล์ออบเจกต์ในการอ่านและเขียน

### 10.3.1 การอ่านและเขียนไฟล์
ความสามารถที่สำคัญที่สุดของไฟล์คือการรองรับและบันทึกข้อมูล ถ้า `f` เป็นไฟล์ออบเจกต์

| เมธอด | หน้าที่ |
|---|---|
| `f.write(s)` | เขียนสตริง `s` ต่อจากที่เขียนไว้ก่อนหน้า **คืนค่าจำนวนอักขระที่เขียน** |
| `f.read()` | อ่าน **ทั้งหมด** ที่เหลือในไฟล์ ได้สตริงหนึ่งก้อน |
| `f.read(n)` | อ่าน `n` อักขระ |
| `f.close()` | ปิดไฟล์ — **ข้อปฏิบัติที่ดี** คือปิดไฟล์ทุกครั้งที่ทำงานเสร็จ เพื่อให้ข้อมูลถูกเขียนลงดิสก์ครบและคืนทรัพยากรให้ระบบ |

**ตัวอย่างการเขียนไฟล์** (ในโหมดโต้ตอบของไพธอน 3 จะเห็นตัวเลข 7 และ 6 ปรากฏหลัง `write` ซึ่งคือจำนวนอักขระที่เขียน)

````python
f = open('somefile.txt', 'w')
n1 = f.write('Hello, ')
n2 = f.write('World!')
f.close()
print("เขียนไป", n1, "และ", n2, "อักขระ")
````

**การอ่านไฟล์** ทำได้ 2 รูปแบบ

รูปแบบแรก — ระบุจำนวนอักขระที่ต้องการอ่าน

````python
f = open('somefile.txt', 'r')
print(repr(f.read(4)))
f.close()
````

รูปแบบที่สอง — อ่านทั้งไฟล์ โดยไม่ต้องระบุอาร์กิวเมนต์ให้ `read()`

````python
f = open('somefile.txt', 'r')
content = f.read()
f.close()
content
````

### 🆕 ตัวชี้ตำแหน่งในไฟล์ (File pointer)
สังเกตว่าเอกสาร **เปิดไฟล์ใหม่** ก่อนอ่านทั้งไฟล์ เพราะไฟล์ออบเจกต์จำ **ตำแหน่งปัจจุบัน** ไว้ การอ่านแต่ละครั้งเริ่มจากจุดที่อ่านค้างไว้ ถ้าไม่เปิดใหม่ `read()` จะได้เฉพาะส่วนที่เหลือ

- `f.tell()` บอกตำแหน่งปัจจุบัน
- `f.seek(0)` ย้ายตำแหน่งกลับไปต้นไฟล์

````python
f = open('somefile.txt')
print("อ่าน 4 ตัวแรก :", repr(f.read(4)))
print("ตำแหน่งตอนนี้ :", f.tell())
print("อ่านที่เหลือ  :", repr(f.read()))
print("อ่านอีกครั้ง  :", repr(f.read()), "← ถึงท้ายไฟล์แล้ว ได้สตริงว่าง")
f.seek(0)
print("หลัง seek(0)  :", repr(f.read()))
f.close()
````

**การเขียนแบบผนวก (Append)** — โดยปกติการเปิดไฟล์เพื่อเขียน (`'w'`) จะ **เขียนทับข้อมูลเดิม** ถ้าต้องการเพิ่มข้อมูลใหม่ต่อท้ายข้อมูลเดิม ต้องเปิดด้วยโหมด `'a'`

````python
# สร้างไฟล์เริ่มต้น
f = open('helloworld.txt', 'w')
f.write('first line')
f.close()

# ตัวอย่างจากเอกสาร: ผนวกข้อมูลต่อท้าย
f = open('helloworld.txt', 'a')
f.write('\n' + 'hello world')
f.close()

print(open('helloworld.txt').read())
````

````python
# เปรียบเทียบ: ถ้าใช้ 'w' ข้อมูลเดิมหายหมด
f = open('helloworld.txt', 'w')
f.write('only this line')
f.close()
print(open('helloworld.txt').read())
````

## 10.4 การอ่านและเขียนไฟล์ทีละบรรทัด

`read()` และ `write()` เหมาะกับไฟล์ขนาดเล็ก แต่ไฟล์ในงานจริงทุกวันนี้มีขนาดใหญ่ขึ้นเรื่อย ๆ บางไฟล์ใหญ่ระดับ **GB** การอ่านทั้งไฟล์ในครั้งเดียวต้องใช้หน่วยความจำเท่าขนาดไฟล์และใช้เวลานาน ทางเลือกที่นิยมที่สุดคือ **อ่านและเขียนทีละบรรทัด**

สร้างไฟล์ตัวอย่าง `pi_digit.txt` ก่อน (เนื้อหาตามเอกสาร)

````python
with open('pi_digit.txt', 'w', encoding='utf-8') as fp:
    fp.write("3.1415926535\n")
    fp.write("8979323846\n")
    fp.write("2643383279\n")
print(open('pi_digit.txt', encoding='utf-8').read())
````

**`read_line1.py`** — ใช้เมธอด `readline()` อ่านทีละบรรทัด ร่วมกับ `while` เมื่อถึงท้ายไฟล์ `readline()` จะคืน **สตริงว่าง** `''` ซึ่งมีค่าความจริงเป็น `False` ลูปจึงหยุด

````python
# read_line1.py
with open('pi_digit.txt', 'r') as fp:
    line = fp.readline()
    cnt = 1
    while line:
        print("Line {}: {}".format(cnt, line.strip()))
        line = fp.readline()
        cnt += 1
````

`line.strip()` ตัดช่องว่างและอักขระขึ้นบรรทัดใหม่ `\n` ที่หัวและท้ายบรรทัดออก ลองดูว่าแต่ละบรรทัดที่อ่านมามีอะไรติดมาบ้างด้วย `repr()`

````python
with open('pi_digit.txt') as fp:
    for _ in range(4):
        print(repr(fp.readline()))
````

**`read_line2.py`** — วิธีที่นิยมที่สุดในปัจจุบัน: **วนลูป `for` บนไฟล์ออบเจกต์โดยตรง** ไฟล์ออบเจกต์เป็น iterable ที่ให้ข้อมูลทีละบรรทัด อ่านง่ายและประหยัดหน่วยความจำ

````python
# read_line2.py
with open('pi_digit.txt') as fp:
    cnt = 1
    for line in fp:
        print("Line {}: {}".format(cnt, line.strip()))
        cnt += 1
````

### 🆕 เทคนิคเพิ่มเติมในการอ่านทีละบรรทัด
| วิธี | ผลลัพธ์ | หมายเหตุ |
|---|---|---|
| `for line in f:` | ทีละบรรทัด (มี `\n`) | ประหยัดหน่วยความจำที่สุด ✅ |
| `f.readlines()` | ลิสต์ของทุกบรรทัด (มี `\n`) | โหลดทั้งไฟล์เข้าหน่วยความจำ |
| `f.read().splitlines()` | ลิสต์ของทุกบรรทัด (ไม่มี `\n`) | สะดวก แต่โหลดทั้งไฟล์ |
| `enumerate(f, start=1)` | คู่ (เลขบรรทัด, บรรทัด) | ไม่ต้องใช้ตัวนับ `cnt` เอง |

> ข้อสรุปในเอกสารว่า "อ่านทีละบรรทัดจะเร็วกว่า" ที่จริงข้อดีหลักคือ **ประหยัดหน่วยความจำ** และเริ่มประมวลผลได้ทันทีโดยไม่ต้องรอโหลดทั้งไฟล์ ทำให้จัดการไฟล์ขนาดใหญ่กว่าหน่วยความจำได้
</div>

````python
with open('pi_digit.txt') as fp:
    print(fp.readlines())

with open('pi_digit.txt') as fp:
    print(fp.read().splitlines())

with open('pi_digit.txt') as fp:
    for cnt, line in enumerate(fp, start=1):
        print(f"Line {cnt}: {line.strip()}")
````

**`write_line.py`** — เขียนข้อมูลลงไฟล์ทีละบรรทัดด้วย `write()` ในโหมด `'a'` (เพิ่มต่อท้าย) หรือ `'w'` (เขียนทับทั้งหมด) การขึ้นบรรทัดใหม่ต้องเติม escape character `'\n'` เอง เพราะ `write()` ไม่เติมให้

**`os.listdir`** — คืนลิสต์ของ **ชื่อ** ไฟล์และไดเรกทอรีย่อยที่อยู่ในไดเรกทอรีที่ระบุ

ข้อสังเกตจากผลลัพธ์ของ `os.listdir`

2. ไฟล์และไดเรกทอรีปนกันมา **ไม่ได้บอก** ว่ารายการใดเป็นไฟล์หรือไดเรกทอรี
3. **ไม่รับประกันลำดับ** ของผลลัพธ์

ฟังก์ชัน `print_dir` แสดงเนื้อหาของไดเรกทอรีเป็นพาธเต็มบรรทัดละรายการ

````python
import os

def print_dir(dir_path):
    for name in os.listdir(dir_path):
        print(os.path.join(dir_path, name))

print_dir("zoo")
````

ผลลัพธ์ของ `os.listdir` เป็นลิสต์ของสตริง จึงเรียงลำดับเองได้ด้วย `sorted()` ค่าเริ่มต้นเรียงตามรหัสตัวอักษร ซึ่ง **ตัวพิมพ์ใหญ่ (A–Z) มาก่อนตัวพิมพ์เล็ก (a–z)** ถ้าต้องการเรียงโดยไม่สนตัวพิมพ์ ใช้ `key=str.lower`

````python
print(sorted(os.listdir("zoo")))
print(sorted(os.listdir("zoo"), key=str.lower))
````

**`os.path.isfile`** และ **`os.path.isdir`** — ตรวจว่าพาธเป็นไฟล์หรือไดเรกทอรี

````python
print(os.path.isfile("zoo"), os.path.isdir("zoo"))
print(os.path.isfile("zoo/README.txt"), os.path.isdir("zoo/README.txt"))
````

นำ `isdir` มาใช้กับ **ฟังก์ชันเวียนเกิด (Recursive Function)** (บทที่ 8) เพื่อแสดงเนื้อหาของไดเรกทอรีย่อยทุกชั้น: เมื่อพบว่ารายการใดเป็นไดเรกทอรี ฟังก์ชันจะ **เรียกตัวเองซ้ำ** เพื่อแสดงรายการภายในไดเรกทอรีนั้นก่อนทำงานต่อ (บรรทัดที่ 5–6)

````python
def print_tree(dir_path):
    for name in os.listdir(dir_path):
        full_path = os.path.join(dir_path, name)
        print(full_path)
        if os.path.isdir(full_path):
            print_tree(full_path)

print_tree("zoo")
````

🆕 **ปรับให้สวยขึ้น:** เรียงชื่อ และย่อหน้าตามระดับความลึก โดยเพิ่มพารามิเตอร์ `level`

````python
def print_tree2(dir_path, level=0):
    for name in sorted(os.listdir(dir_path)):
        full_path = os.path.join(dir_path, name)
        icon = "📁" if os.path.isdir(full_path) else "📄"
        print("    " * level + icon, name)
        if os.path.isdir(full_path):
            print_tree2(full_path, level + 1)

print_tree2("zoo")
````

### 10.5.3 การเปลี่ยนชื่อ เคลื่อนย้าย สำเนา และการลบไฟล์
โมดูล **`shutil`** (shell utilities) มีฟังก์ชันจัดการไฟล์ระดับสูง

| งาน | ฟังก์ชัน |
|---|---|
| เปลี่ยนชื่อ / ย้าย | `shutil.move(src, dst)` (หรือ `os.rename(src, dst)`) |
| สำเนา | `shutil.copy(src, dst)` — `dst` เป็นชื่อไฟล์ใหม่หรือไดเรกทอรีก็ได้ |
| สำเนา (ระบุชื่อไฟล์ปลายทาง) | `shutil.copyfile(src, dst)` |
| ลบไฟล์ | `os.remove(path)` |

````python
import shutil

make_file("server.log", "log data\n")
make_file("old mail.txt", "จดหมายเก่า\n")
make_file("important.dat", "ข้อมูลสำคัญ\n")
make_file("junk.dat", "ขยะ\n")
os.makedirs("archive", exist_ok=True)
os.makedirs("backups", exist_ok=True)

shutil.move("server.log", "server.log.backup")      # เปลี่ยนชื่อ
shutil.move("old mail.txt", "archive")              # ย้ายเข้าไดเรกทอรี
shutil.copy("important.dat", "backups")             # สำเนาเข้าไดเรกทอรี
os.remove("junk.dat")                               # ลบไฟล์

for p in ["server.log", "server.log.backup", "archive/old mail.txt",
          "important.dat", "backups/important.dat", "junk.dat"]:
    print(f"{p:25} มีอยู่?", os.path.exists(p))
````

### 10.5.4 การสร้าง เคลื่อนย้าย และการลบไดเรกทอรี

| ฟังก์ชัน | หน้าที่ | ข้อควรระวัง |
|---|---|---|
| `os.mkdir(p)` | สร้างไดเรกทอรี **ชั้นเดียว** | ถ้าไม่มีไดเรกทอรีแม่ → `FileNotFoundError`; ถ้ามีอยู่แล้ว → `FileExistsError` |
| `os.makedirs(p, exist_ok=True)` | สร้างไดเรกทอรีพร้อม **ไดเรกทอรีแม่ทุกชั้น** ที่ยังไม่มี | `exist_ok=True` ทำให้ไม่ error ถ้ามีอยู่แล้ว |
| `os.rmdir(p)` | ลบไดเรกทอรี **ว่าง** เท่านั้น | ถ้าไม่ว่าง → `OSError` |
| `shutil.rmtree(p)` | ลบไดเรกทอรีพร้อม **ทุกสิ่งข้างใน** | ⚠️ อันตราย ข้อมูลหายถาวร ไม่ผ่านถังขยะ |
| `shutil.move(src, dst)` | ย้ายไดเรกทอรีทั้งก้อน | |

````python
def total_size(d):
````

### 🆕 ข้อแก้ไข: find_word.py
โค้ดในเอกสารมีจุดที่ควรปรับ 3 จุด
1. **คำสุดท้ายของบรรทัดมี `\n` ติดมา** เช่น `'เบาหวาน\n'` ทำให้ `keyword in words` เป็น `False` ทั้งที่มีคำนั้นอยู่ → ต้อง `line.strip()` ก่อน `split`
2. `dst_path + '/' + filename` เมื่อ `dst_path` ลงท้ายด้วย `/` อยู่แล้วจะได้ `//` → ใช้ `os.path.join` แทน
3. ถ้ายังไม่มีไดเรกทอรีปลายทาง `copyfile` จะ error → สร้างก่อนด้วย `os.makedirs(..., exist_ok=True)`

````python
# ทดสอบปัญหาข้อ 1
line = "เหมาะ|กับ|ผู้|ป่วย|เบาหวาน\n"
print(line.split("|"))
print("พบ 'เบาหวาน' (ไม่ strip)?", "เบาหวาน" in line.split("|"))
print("พบ 'เบาหวาน' (strip แล้ว)?", "เบาหวาน" in line.strip().split("|"))
````

````python
# find_word.py (ฉบับแก้ไข)
from shutil import copyfile
import os.path

def find_word(keyword, base_path, dst_path):
    with open(base_path, 'r', encoding='utf-8') as f:
        for line in f:
            words = line.strip().split("|")
            if keyword in words:
                print("Found: ", keyword, "in", base_path)
                base, filename = os.path.split(base_path)
                os.makedirs(dst_path, exist_ok=True)
                copyfile(base_path, os.path.join(dst_path, filename))
                return True
    return False

if __name__ == '__main__':
    keyword = 'เบาหวาน'
    dst_path = 'filtered'
    # ประยุกต์ร่วมกับ list_file.py: วนทุกไฟล์ทุกเว็บไซต์
    for site in sorted(os.listdir('webdata')):
        site_dir = os.path.join('webdata', site)
        for name in sorted(os.listdir(site_dir)):
            path = os.path.join(site_dir, name)
            if os.path.isfile(path) and name.endswith('.txt'):
                find_word(keyword, path, dst_path)

    print("ไฟล์ที่คัดกรองได้:", sorted(os.listdir('filtered')))
````

💡 เงื่อนไข `if __name__ == '__main__':` ทำให้ส่วนนั้นทำงานเฉพาะเมื่อ **รันไฟล์นี้โดยตรง** ไม่ทำงานเมื่อไฟล์ถูก `import` เป็นโมดูลจากโปรแกรมอื่น (ใน Notebook ค่า `__name__` คือ `'__main__'` จึงทำงานเสมอ)

นำความรู้ทั้งบทมารวมกัน: อ่านไฟล์ทุกไฟล์ → นับความถี่ของคำที่สนใจ → เขียนผลเป็นตารางในไฟล์ CSV (แถว = เอกสาร, คอลัมน์ = คำ)

````python
import csv

vocab = ["ขิง", "ช่วย", "บรรเทา", "เบาหวาน", "อาการ"]       # คำที่พิจารณา (เช่น จากพจนานุกรม)

rows = []
for dirpath, dirnames, filenames in os.walk("webdata"):
    for name in sorted(filenames):
        if not name.endswith(".txt"):
            continue
        with open(os.path.join(dirpath, name), encoding="utf-8") as f:
            words = []
            for line in f:
                words.extend(line.strip().split("|"))
        rows.append([name] + [words.count(w) for w in vocab])

rows.sort()
with open("word_vector.csv", "w", newline="", encoding="utf-8") as f:
    w = csv.writer(f)
    w.writerow(["document"] + vocab)
    w.writerows(rows)

print(open("word_vector.csv", encoding="utf-8").read())
````

## 10.8 สรุป

- เปิดไฟล์ด้วย `open()` โดยส่ง **พาธ** ของไฟล์เป็นอาร์กิวเมนต์ ผลลัพธ์คือ **ไฟล์ออบเจกต์** ที่ใช้จัดการไฟล์ต่อ
- ถ้าส่งเฉพาะพาธ จะเปิดเพื่อ **อ่านอย่างเดียว** (`'r'`) ถ้าต้องการทำอย่างอื่นต้องระบุ **โหมด**
- ไฟล์ออบเจกต์ `f` อ่านด้วย `f.read()` และเขียนด้วย `f.write()`
- โหมด `'w'` **ลบข้อมูลเดิมทั้งหมด** แล้วเขียนใหม่ ส่วน `'a'` **เขียนต่อท้าย** โดยข้อมูลเดิมยังอยู่ (🆕 `'x'` สร้างไฟล์ใหม่โดยไม่ยอมเขียนทับ)
- เมื่อทำงานกับไฟล์เสร็จต้องปิดด้วย `close()` เสมอ
- คำสั่ง `with` ควบคุมการจัดการไฟล์ **ปิดไฟล์ให้อัตโนมัติ** แม้เกิดข้อผิดพลาด จึงนิยมใช้กับงานไฟล์ทุกงาน
- 🆕 ระบุ `encoding="utf-8"` ทุกครั้งเมื่อทำงานกับข้อความภาษาไทย และใช้ raw string (`r'...'`) กับพาธ Windows
- ไฟล์ขนาดใหญ่นิยมอ่าน/เขียน **ทีละบรรทัด** (`for line in f:`) ซึ่ง **ประหยัดหน่วยความจำ** และเริ่มประมวลผลได้ทันที
- จัดการพาธด้วย `os.path` (`join`, `split`, `exists`, ...) แทนการต่อสตริงเอง เพื่อให้โปรแกรมทำงานข้ามแพลตฟอร์มได้
- `os.listdir` แสดงเนื้อหาในไดเรกทอรี (ได้เฉพาะชื่อ ไม่เรียงลำดับ) และตรวจชนิดด้วย `os.path.isfile` / `os.path.isdir`
- เปลี่ยนชื่อ/ย้าย สำเนา และลบไฟล์ ใช้ `shutil.move`, `shutil.copy` และ **`os.remove`** ตามลำดับ (🆕 เอกสารเขียน `shutil.remove` ซึ่งไม่มีอยู่จริง)
- `os.mkdir` สร้างไดเรกทอรีได้ชั้นเดียว ต้องมีไดเรกทอรีแม่อยู่ก่อน ส่วน **`os.makedirs`** (🆕 ไม่ใช่ `os.mkdirs`) สร้างทุกชั้นที่ยังไม่มีให้
- `os.rmdir` ลบได้เฉพาะไดเรกทอรีว่าง ส่วน `shutil.rmtree` ลบทั้งต้นไม้ (ระวัง!)
- ไฟล์ CSV อ่านด้วย `csv.reader` (แต่ละแถวเป็นลิสต์ของสตริง) และเขียนด้วย `csv.writer` + `writerow`/`writerows` (🆕 เปิดด้วย `newline=''`)

````python
# ✅ เซลล์ตรวจคำตอบ (รันได้ซ้ำ ไม่ต้องแก้ไข)
def _safe(f, *a):
    try:
        return f(*a)
    except Exception as e:
        return f"Error → {type(e).__name__}: {e}"
def _eq(a, b):
    try:
        return a == b
    except Exception:
        return False
_tests = [
    ("1. first_line", _safe(first_line, "story.txt"), "Python is a programming language"),
    ("2. last_line", _safe(last_line, "story.txt"), "Learning Python is fun"),
    ("3. read_all", _safe(read_all, "story.txt"), open("story.txt", encoding="utf-8").read()),
    ("4. read_lines", _safe(read_lines, "story.txt"), ["Python is a programming language", "Python is easy to learn", "Files store data permanently", "Learning Python is fun"]),
    ("5. longest_word", _safe(longest_word, "story.txt"), "programming"),
    ("6. count_lines", _safe(count_lines, "story.txt"), 4),
    ("7. word_freq 'Python'", _safe(word_freq, "story.txt", "Python"), 3),
    ("7. word_freq 'Java'", _safe(word_freq, "story.txt", "Java"), 0),
    ("8. word_count จำนวนคำ", _safe(lambda: word_count("story.txt")[0]), 18),
    ("8. word_count ความถี่ 'is'", _safe(lambda: word_count("story.txt")[1]["is"]), 3),
    ("8. word_count จำนวนคำไม่ซ้ำ", _safe(lambda: len(word_count("story.txt")[1])), 14),
    ("9. find_file 'c.csv'", _safe(find_file, "qdir", "c.csv"), True),
    ("9. find_file 'z.txt'", _safe(find_file, "qdir", "z.txt"), False),
    ("10. min_max_file", _safe(min_max_file, "qdir"), ("a.txt", "b.txt")),
]
_ok = 0
for _label, _got, _exp in _tests:
    _pass = bool(_eq(_got, _exp))
    _ok += _pass
    _show = repr(_got)
    _show = _show if len(_show) < 80 else _show[:77] + "..."
    print(("✅ " if _pass else "❌ ") + _label + " = " + _show + ("" if _pass else ("\n      ควรได้ " + repr(_exp))[:240]))
print(f"\nผ่าน {_ok}/{len(_tests)} ข้อ" + ("  🎉" if _ok == len(_tests) else ""))
````

```python
def first_line(path):
    with open(path, encoding="utf-8") as f:
        return f.readline().rstrip("\n")

def last_line(path):
    last = ""
    with open(path, encoding="utf-8") as f:
        for line in f:              # อ่านทีละบรรทัดจนถึงบรรทัดสุดท้าย
            if line.strip():
                last = line.rstrip("\n")
    return last

def read_all(path):
    with open(path, encoding="utf-8") as f:
        return f.read()

def read_lines(path):
    with open(path, encoding="utf-8") as f:
        return [line.rstrip("\n") for line in f]

def longest_word(path):
    best = ""
    with open(path, encoding="utf-8") as f:
        for line in f:
            for w in line.split():
                if len(w) > len(best):      # ใช้ > ไม่ใช่ >= จึงได้คำที่พบก่อน
                    best = w
    return best

def count_lines(path):
    with open(path, encoding="utf-8") as f:
        return sum(1 for _ in f)

def word_freq(path, word):
    with open(path, encoding="utf-8") as f:
        return f.read().split().count(word)

def word_count(path):
    freq = {}
    total = 0
    with open(path, encoding="utf-8") as f:
        for line in f:
            for w in line.split():
                total += 1
                freq[w] = freq.get(w, 0) + 1
    return total, freq

def find_file(d, name):
    return name in os.listdir(d)
    # ถ้าต้องการค้นทุกไดเรกทอรีย่อย: any(name in files for _, _, files in os.walk(d))

def min_max_file(d):
    files = [n for n in os.listdir(d) if os.path.isfile(os.path.join(d, n))]
    size = lambda n: os.path.getsize(os.path.join(d, n))
    return min(files, key=size), max(files, key=size)
```
