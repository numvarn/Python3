## 1.1 ความสำคัญของคอมพิวเตอร์

ลองนึกถึงรูปแบบการใช้งานคอมพิวเตอร์ของผู้คนในปัจจุบัน

| สถานที่ | ตัวอย่างการใช้งาน |
|---|---|
| 🏫 **โรงเรียน/มหาวิทยาลัย** | เขียนบทความ ค้นหาบทความ ส่งอีเมล เข้าร่วมชั้นเรียนออนไลน์ |
| 🏢 **ที่ทำงาน** | วิเคราะห์ข้อมูล ทำเอกสารนำเสนอ ทำธุรกรรมทางธุรกิจ สื่อสารกับลูกค้าและเพื่อนร่วมงาน ควบคุมเครื่องจักรในโรงงาน |
| 🏠 **ที่บ้าน** | จ่ายค่าตั๋ว ซื้อของออนไลน์ สื่อสารกับเพื่อนและครอบครัว เล่นเกมคอมพิวเตอร์ |

นอกจากนี้ **โทรศัพท์มือถือ ระบบนำทางรถยนต์** และอุปกรณ์อื่น ๆ อีกมากมาย **ก็ถือเป็นคอมพิวเตอร์ด้วยเช่นกัน** การใช้คอมพิวเตอร์ในยุคปัจจุบันจึงแทบไม่มีขีดจำกัดในชีวิตประจำวันของเรา

### คอมพิวเตอร์ทำงานได้เพราะ "โปรแกรม"
คอมพิวเตอร์ทำสิ่งต่าง ๆ ได้มากมายเพราะ **เราตั้งโปรแกรมให้มันทำงานได้** คอมพิวเตอร์ไม่ได้ถูกออกแบบมาให้ทำงานด้วยตัวเอง แต่ทำงานตามที่โปรแกรมกำหนดเท่านั้น

> 📌 **โปรแกรม (Program)** คือ ชุดของคำสั่งที่คอมพิวเตอร์ต้องดำเนินการตามที่คำสั่งกำหนด

ตัวอย่างในภาพที่ 1.1 คือโปรแกรม **Microsoft Word** ซึ่งเป็นโปรแกรมประมวลผลคำสำหรับสร้าง แก้ไข และพิมพ์เอกสาร และ **Adobe Photoshop** ซึ่งเป็นโปรแกรมแก้ไขภาพสำหรับทำงานกับภาพกราฟิก เช่น ภาพถ่ายจากกล้องดิจิทัล

### ซอฟต์แวร์และนักพัฒนาซอฟต์แวร์
> 📌 **ซอฟต์แวร์ (Software)** คือ ชุดของข้อมูล คำสั่ง และโปรแกรม ที่ใช้สั่งงานคอมพิวเตอร์ให้ทำงานใด ๆ อย่างเฉพาะเจาะจง

ซอฟต์แวร์เป็นสิ่งจำเป็นต่อคอมพิวเตอร์ เพราะทำหน้าที่ **ควบคุมทุกอย่าง** ที่คอมพิวเตอร์ทำ ซอฟต์แวร์ทั้งหมดที่เราใช้ถูกสร้างขึ้นโดย **โปรแกรมเมอร์ (Programmer)** หรือ **นักพัฒนาซอฟต์แวร์ (Software Developer)** ซึ่งเป็นผู้ที่ได้รับการฝึกฝนทักษะในการ **ออกแบบ สร้าง และทดสอบ** โปรแกรมคอมพิวเตอร์

การเขียนโปรแกรมเป็นอาชีพที่น่าตื่นเต้นและคุ้มค่า นักพัฒนาโปรแกรมทำงานได้หลากหลายสายงาน เช่น การปกครอง การบังคับใช้กฎหมาย การเกษตร งานวิชาการ และงานบันเทิง เพราะทุกงานล้วนต้องใช้คอมพิวเตอร์และซอฟต์แวร์

**ลำดับเนื้อหาของบทนี้**
1. ส่วนประกอบทางกายภาพของคอมพิวเตอร์ (ฮาร์ดแวร์)
2. วิธีที่คอมพิวเตอร์เก็บข้อมูลและรันโปรแกรม
3. ภาษาโปรแกรมและซอฟต์แวร์ที่ใช้ในการเขียนโปรแกรมไพธอน

### 🔬 ทดลอง: โปรแกรมคือชุดคำสั่งที่ทำงานตามลำดับ
เซลล์ด้านล่างคือ "โปรแกรม" ขนาดเล็กที่มี 4 คำสั่ง คอมพิวเตอร์จะทำงาน **ทีละคำสั่งจากบนลงล่าง** ตามที่เราเขียนไว้เท่านั้น ลองรัน แล้วสลับลำดับบรรทัดดู ผลลัพธ์จะเปลี่ยนไปอย่างไร

````python
ราคาสินค้า = 250          # คำสั่งที่ 1: เก็บราคาสินค้า
จำนวน = 3                 # คำสั่งที่ 2: เก็บจำนวนที่ซื้อ
รวม = ราคาสินค้า * จำนวน   # คำสั่งที่ 3: ให้คอมพิวเตอร์คำนวณ
print("ยอดที่ต้องชำระ =", รวม, "บาท")   # คำสั่งที่ 4: แสดงผล
````

> 💡 ไพธอน 3 ตั้งชื่อตัวแปรเป็นภาษาไทยได้ (เพราะรองรับ Unicode ซึ่งจะได้เรียนในหัวข้อ 1.3.2) แต่ในทางปฏิบัติแนะนำให้ใช้ภาษาอังกฤษ เช่น `price`, `qty`

## 1.2 ฮาร์ดแวร์และซอฟต์แวร์

### 1.2.1 ฮาร์ดแวร์ (Hardware)
> 📌 **ฮาร์ดแวร์** หมายถึง อุปกรณ์หรือส่วนประกอบทางกายภาพทั้งหมดที่ประกอบขึ้นเป็นคอมพิวเตอร์

คอมพิวเตอร์ **ไม่ใช่อุปกรณ์ชิ้นเดียว** แต่เป็น **ระบบของอุปกรณ์ที่ทำงานร่วมกัน** เปรียบได้กับเครื่องดนตรีในวงดุริยางค์ซิมโฟนี ที่เครื่องดนตรีแต่ละชิ้นมีบทบาทของตัวเอง

ระบบคอมพิวเตอร์โดยทั่วไปประกอบด้วยองค์ประกอบหลัก 5 ส่วน (ภาพที่ 1.2)
1. หน่วยประมวลผลกลาง (CPU)
2. หน่วยความจำหลัก (Main Memory)
3. หน่วยความจำสำรอง (Secondary Storage Devices)
4. อุปกรณ์อินพุต (Input Devices)
5. อุปกรณ์เอาต์พุต (Output Devices)

<p align="center"><img src="images/บทที่01/fig1.2.jpg" width="620"><br><b>ภาพที่ 1.2</b> องค์ประกอบทั่วไปของระบบคอมพิวเตอร์<br><sub>ที่มา: “Introduction to Computers and Programming” (2016)</sub></p>

#### ① หน่วยประมวลผลกลาง — ซีพียู (CPU: Central Processing Unit)
เมื่อคอมพิวเตอร์ทำงานตามที่โปรแกรมสั่ง เราเรียกว่าคอมพิวเตอร์กำลัง **รัน (run)** หรือ **เอ็กซีคิวท์ (execute)** โปรแกรม

- **ซีพียู** คือส่วนที่ใช้ **ประมวลผลคำสั่งของโปรแกรม**
- ซีพียูเป็น **ส่วนประกอบที่สำคัญที่สุด** เพราะถ้าไม่มีซีพียู คอมพิวเตอร์จะประมวลผลซอฟต์แวร์ไม่ได้เลย

**วิวัฒนาการของซีพียู**
- คอมพิวเตอร์ยุคแรกใช้ซีพียูขนาดใหญ่ที่ทำจากชิ้นส่วนไฟฟ้าและเครื่องกล เช่น **หลอดสุญญากาศ** และ **สวิตช์**
- **ENIAC** สร้างขึ้นในปี พ.ศ. 2488 (ค.ศ. 1945) เพื่อคำนวณวิถีการยิงปืนใหญ่ให้กองทัพสหรัฐฯ และถือเป็นคอมพิวเตอร์อิเล็กทรอนิกส์ที่ตั้งโปรแกรมได้เครื่องแรกของโลก มีความ **สูง 8 ฟุต ยาว 100 ฟุต หนัก 30 ตัน**

<p align="center"><img src="images/บทที่01/fig1.3.jpg" width="620"><br><b>ภาพที่ 1.3</b> ENIAC Computer<br><sub>ที่มา: Wink (2011)</sub></p>

- ปัจจุบันซีพียูเป็นชิปขนาดเล็กที่เรียกว่า **ไมโครโพรเซสเซอร์ (Microprocessor)** ซึ่งนอกจากเล็กกว่ามากแล้วยังมี **ประสิทธิภาพสูงกว่า** ซีพียูยุคแรกอีกด้วย

<p align="center"><img src="images/บทที่01/fig1.4.jpg" width="620"><br><b>ภาพที่ 1.4</b> ช่างเทคนิคห้องปฏิบัติการกับไมโครโพรเซสเซอร์ที่ทันสมัย<br><sub>ที่มา: Derhacobian (2016)</sub></p>

#### ② หน่วยความจำหลัก (Main Memory)
- เป็นที่ที่คอมพิวเตอร์ใช้ **เก็บโปรแกรมขณะที่โปรแกรมกำลังทำงาน** รวมถึง **ข้อมูลที่โปรแกรมใช้ในขณะนั้น**
- ตัวอย่าง: ขณะใช้โปรแกรมประมวลผลคำเขียนเรียงความ ทั้งตัวโปรแกรมและข้อมูลเรียงความจะอยู่ในหน่วยความจำหลัก
- นิยมเรียกว่า **แรม (RAM: Random Access Memory)** หรือ "หน่วยความจำเข้าถึงโดยสุ่ม" เพราะซีพียูเข้าถึงข้อมูลตำแหน่งใดก็ได้อย่างรวดเร็ว
- แรมเป็นหน่วยความจำแบบ **ลบเลือนได้ (Volatile)** ใช้เก็บข้อมูลชั่วคราว **เมื่อปิดเครื่อง ข้อมูลในแรมจะหายไปทั้งหมด**

<p align="center"><img src="images/บทที่01/fig1.5.jpg" width="620"><br><b>ภาพที่ 1.5</b> Memory Chip<br><sub>ที่มา: “Memory Chip” (2016)</sub></p>

#### ③ หน่วยความจำสำรอง (Secondary Storage Devices)
- เก็บข้อมูลได้ **เป็นเวลานาน แม้ไม่มีไฟฟ้าเข้าเครื่อง (Non-volatile)**
- โปรแกรมถูกเก็บไว้ในหน่วยความจำสำรอง และ **โหลดเข้าหน่วยความจำหลักเมื่อต้องการใช้งาน**
- ใช้เก็บข้อมูลสำคัญ เช่น เอกสาร ข้อมูลเงินเดือน และบันทึกสินค้าคงคลัง

| ประเภท | หลักการ / จุดเด่น |
|---|---|
| **ดิสก์ไดร์ฟ (Disk Drive)** | พบมากที่สุด เก็บข้อมูลโดย **เข้ารหัสแม่เหล็ก** บนแผ่นดิสก์วงกลม มีทั้งแบบติดตั้งภายในและแบบภายนอก ซึ่งใช้สำรองหรือย้ายข้อมูลได้ |
| **ฟล็อปปี้ดิสก์ (Floppy Disk)** | เคยนิยมใช้หลายปี แต่มีข้อเสียมาก ปัจจุบันแทบไม่มีผู้ใช้แล้ว |
| **ไดร์ฟ USB / แฟลชไดร์ฟ** | เสียบผ่านพอร์ต USB (Universal Serial Bus) **ไม่มีดิสก์อยู่จริง** แต่เก็บข้อมูลใน **หน่วยความจำแฟลช** ราคาไม่แพง เชื่อถือได้ และพกพาสะดวก |
| **ออปติคัลดิสก์ (CD / DVD)** | ข้อมูลถูกเข้ารหัสเป็น **ชุดของหลุม** บนผิวแผ่น ไดร์ฟใช้ **เลเซอร์** ตรวจจับหลุมเพื่ออ่านข้อมูล และบันทึกข้อมูลได้จำนวนมาก |

#### ④ อุปกรณ์อินพุต (Input Devices)
- **อินพุต (Input)** คือข้อมูลใด ๆ ที่คอมพิวเตอร์รวบรวมจากผู้คนหรืออุปกรณ์อื่น
- **อุปกรณ์อินพุต** คืออุปกรณ์ที่รวบรวมข้อมูลแล้วส่งให้คอมพิวเตอร์ เช่น **แป้นพิมพ์ เมาส์ สแกนเนอร์ ไมโครโฟน กล้องดิจิทัล**
- **ดิสก์ไดร์ฟและออปติคัลไดร์ฟ** ก็นับเป็นอุปกรณ์อินพุตได้ เพราะส่งโปรแกรมและข้อมูลเข้าหน่วยความจำของคอมพิวเตอร์

#### ⑤ อุปกรณ์เอาต์พุต (Output Devices)
- **เอาต์พุต (Output)** คือข้อมูลที่คอมพิวเตอร์สร้างขึ้นเพื่อแสดงแก่บุคคลหรือส่งต่อไปยังอุปกรณ์อื่น เช่น รายงานการขาย รายชื่อ ภาพกราฟิก
- **อุปกรณ์เอาต์พุต** เช่น **จอแสดงผล เครื่องพิมพ์ ลำโพง**
- **ดิสก์ไดร์ฟและเครื่องบันทึกซีดี** ก็นับเป็นอุปกรณ์เอาต์พุตได้ เพราะระบบส่งข้อมูลไปบันทึกที่อุปกรณ์เหล่านั้น

> 💡 สังเกตว่า **ดิสก์ไดร์ฟเป็นได้ทั้งอินพุตและเอาต์พุต** ขึ้นอยู่กับทิศทางการไหลของข้อมูล

### 🔬 ทดลอง: สำรวจฮาร์ดแวร์ของเครื่องที่ใช้อยู่
ไพธอนมีโมดูลมาตรฐานที่ช่วยให้เราสอบถามข้อมูลฮาร์ดแวร์จากระบบปฏิบัติการได้ ลองรันดูว่าเครื่องของผู้เรียน (หรือเซิร์ฟเวอร์ที่รัน Notebook นี้) มีสเปกอย่างไร

````python
import os, platform, shutil

print("=== ① CPU ===")
print("สถาปัตยกรรมเครื่อง  :", platform.machine())
print("ข้อมูลโพรเซสเซอร์    :", platform.processor() or "(ระบบไม่รายงาน)")
print("จำนวนแกนประมวลผล   :", os.cpu_count(), "cores")

print("\n=== ② หน่วยความจำหลัก (RAM) ===")
try:
    total_ram = os.sysconf("SC_PAGE_SIZE") * os.sysconf("SC_PHYS_PAGES")
    print(f"ขนาดแรมทั้งหมด      : {total_ram / 1024**3:.2f} GB")
except (ValueError, AttributeError, OSError):
    print("ระบบปฏิบัติการนี้ไม่รองรับการอ่านค่าแรมด้วยวิธีนี้ (เช่น Windows)")

print("\n=== ③ หน่วยความจำสำรอง (ดิสก์) ===")
disk = shutil.disk_usage(os.path.abspath(os.sep))
print(f"ความจุทั้งหมด        : {disk.total / 1024**3:.2f} GB")
print(f"ใช้ไปแล้ว            : {disk.used  / 1024**3:.2f} GB")
print(f"พื้นที่ว่าง           : {disk.free  / 1024**3:.2f} GB")
````

รันเซลล์แล้วพิมพ์ชื่อกับปีเกิด (พ.ศ.) ในช่องที่ปรากฏ จากนั้นกด Enter

## 1.3 วิธีเก็บข้อมูลของคอมพิวเตอร์

- หน่วยความจำถูกแบ่งเป็นพื้นที่เล็ก ๆ เรียกว่า **ไบต์ (Byte)** ซึ่ง 1 ไบต์มีขนาดพอเก็บตัวอักษร 1 ตัว หรือตัวเลขขนาดเล็ก 1 จำนวน

### 1.3.1 การจัดเก็บข้อมูลประเภทตัวเลข
บิตแต่ละตัวมีได้ 2 สถานะ คือ ปิด = **0** และ เปิด = **1** ซึ่งตรงกับ **ระบบเลขฐานสอง (Binary)** ที่เขียนตัวเลขทุกจำนวนด้วย 0 และ 1 เท่านั้น เช่น

```
10011101
```

แต่ละหลักมี **ค่าประจำตำแหน่ง** โดยเริ่มจากหลักขวาสุดไปทางซ้ายเป็น 2⁰, 2¹, 2², 2³, … (ภาพที่ 1.9) หรือเท่ากับ 1, 2, 4, 8, 16, 32, 64, 128 (ภาพที่ 1.10)

#### วิธีแปลงเลขฐานสองเป็นเลขฐานสิบ
นำค่าประจำตำแหน่ง **เฉพาะบิตที่เป็น 1** มารวมกัน เช่น `10011101` มีบิตที่เป็น 1 อยู่ในตำแหน่งที่มีค่า 1, 4, 8, 16 และ 128

$$1 + 4 + 8 + 16 + 128 = 157$$

ดังนั้น $10011101_2 = 157_{10}$

| ตำแหน่ง | 2⁷ | 2⁶ | 2⁵ | 2⁴ | 2³ | 2² | 2¹ | 2⁰ |
|---|---|---|---|---|---|---|---|---|
| ค่าประจำตำแหน่ง | 128 | 64 | 32 | 16 | 8 | 4 | 2 | 1 |
| บิต | **1** | 0 | 0 | **1** | **1** | **1** | 0 | **1** |
| นำมารวม | 128 | – | – | 16 | 8 | 4 | – | 1 |

### 🔬 ทดลอง: แปลงเลขฐานสองด้วยไพธอน
ไพธอนมีเครื่องมือในตัวสำหรับแปลงเลขฐาน
- `bin(n)` แปลงเลขฐานสิบเป็นข้อความเลขฐานสอง (ขึ้นต้นด้วย `0b`)
- `int("...", 2)` แปลงข้อความเลขฐานสองเป็นเลขฐานสิบ
- `0b10011101` เขียนเลขฐานสองลงในโค้ดได้โดยตรง

````python
print(int("10011101", 2))     # ฐานสอง -> ฐานสิบ
print(bin(157))               # ฐานสิบ -> ฐานสอง
print(0b10011101)             # เขียนเลขฐานสองในโค้ดโดยตรง
print(format(77, "08b"))      # แสดงเลข 77 เป็นฐานสองครบ 8 บิต (เหมือนภาพที่ 1.8)
````

ฟังก์ชันด้านล่างแสดง **ขั้นตอนการคำนวณ** แบบเดียวกับที่เราทำด้วยมือ ลองเปลี่ยนค่าเลขฐานสองเป็นค่าอื่นดู

````python
def binary_to_decimal(bits):
    """แปลงเลขฐานสองเป็นฐานสิบ พร้อมแสดงวิธีคิดทีละขั้น"""
    n = len(bits)
    parts = []
    total = 0
    print(f"{'ตำแหน่ง':>8} {'ค่าประจำตำแหน่ง':>16} {'บิต':>5}")
    for i, b in enumerate(bits):
        power = n - 1 - i                 # หลักซ้ายสุดมีกำลังสูงสุด
        value = 2 ** power
        mark = "✔" if b == "1" else ""
        print(f"{'2^' + str(power):>8} {value:>16} {b:>5} {mark}")
        if b == "1":
            parts.append(str(value))
            total += value
    print("\nผลรวม =", " + ".join(reversed(parts)), "=", total)
    return total

binary_to_decimal("10011101")
````

#### มองเห็นไบต์เป็นชุดสวิตช์
ฟังก์ชัน `show_byte()` วาดไบต์เป็นสวิตช์ 8 ตัวเหมือนภาพที่ 1.12 (🟩 = เปิด/1, ⬜ = ปิด/0)

````python
def show_byte(number):
    if not 0 <= number <= 255:
        print("ไบต์เดียวเก็บได้เฉพาะ 0 – 255 เท่านั้น!")
        return
    bits = format(number, "08b")
    values = [128, 64, 32, 16, 8, 4, 2, 1]
    print("สวิตช์ :", "  ".join("🟩" if b == "1" else "⬜" for b in bits))
    print("บิต    :", "   ".join(bits))
    print("ค่า    :", " ".join(f"{v:<3}" for v in values))
    used = [str(v) for v, b in zip(values, bits) if b == "1"]
    print(f"ค่าที่เก็บ = {' + '.join(used) if used else '0'} = {number}\n")

show_byte(157)
show_byte(77)
show_byte(255)
````

#### ขีดจำกัดของ 1 ไบต์ และการใช้หลายไบต์

### ✍️ ลงมือทำ 1: แปลงเลขฐานสิบเป็นฐานสอง (วิธีหารด้วย 2)
วิธีแปลงด้วยมือคือ **หารด้วย 2 ไปเรื่อย ๆ แล้วเก็บเศษ** จนผลหารเป็น 0 จากนั้นอ่านเศษจากล่างขึ้นบน

```
157 ÷ 2 = 78  เศษ 1
 78 ÷ 2 = 39  เศษ 0
 39 ÷ 2 = 19  เศษ 1
 19 ÷ 2 =  9  เศษ 1
  9 ÷ 2 =  4  เศษ 1
  4 ÷ 2 =  2  เศษ 0
  2 ÷ 2 =  1  เศษ 0
  1 ÷ 2 =  0  เศษ 1     อ่านจากล่างขึ้นบน → 10011101
```
จงเติมโค้ดในฟังก์ชัน `decimal_to_binary(n)` ให้คืนค่าเป็นข้อความเลขฐานสอง **โดยห้ามใช้ `bin()` หรือ `format()`**
คำแนะนำ: ตัวดำเนินการ `n % 2` ให้เศษ และ `n // 2` ให้ผลหาร
</div>

````python
def decimal_to_binary(n):
    if n == 0:
        return "0"
    result = ""
    # ---- เขียนโค้ดของคุณตรงนี้ ----
    # ใช้ while n > 0: แล้วในลูกต้อง (1) นำ n % 2 ไปต่อด้านหน้า result
    #                              (2) อัปเดต n = n // 2  (ถ้าลืม ลูปจะไม่มีวันจบ!)

    # -------------------------------
    return result

print(decimal_to_binary(157))
````

````python
# ✅ เซลล์ตรวจคำตอบ (รันได้เลยโดยไม่ต้องแก้ไข)
def _check_dec2bin():
    tests = [0, 1, 2, 5, 77, 157, 255, 1000]
    ok = True
    for t in tests:
        try:
            got = decimal_to_binary(t)
        except Exception as e:
            got = f"เกิดข้อผิดพลาด: {e}"
        exp = bin(t)[2:]
        status = "✅" if got == exp else "❌"
        ok &= got == exp
        print(f"{status} decimal_to_binary({t}) = {got!r}  (ควรได้ {exp!r})")
    print("\n🎉 ยอดเยี่ยม! ผ่านทุกกรณี" if ok else "\nยังไม่ถูกต้องทุกกรณี ลองแก้ไขอีกครั้ง")
_check_dec2bin()
````

### 1.3.2 การจัดเก็บข้อมูลประเภทอักขระ
ข้อมูลทุกอย่างในหน่วยความจำต้องเก็บเป็นเลขฐานสอง รวมถึง **อักขระ** เช่น ตัวอักษรและเครื่องหมายวรรคตอน เมื่อจะเก็บอักขระ คอมพิวเตอร์จะ **แปลงอักขระเป็นรหัสตัวเลขก่อน** แล้วจึงเก็บรหัสนั้นเป็นเลขฐานสอง

#### รหัส ASCII
- **ASCII** ย่อมาจาก *American Standard Code for Information Interchange*
- เป็นชุดรหัสตัวเลข **128 รหัส** (0–127 ใช้ 7 บิต) แทนตัวอักษรภาษาอังกฤษ ตัวเลข เครื่องหมายวรรคตอน และอักขระควบคุม
- ตัวอย่าง: รหัส ASCII ของ **A (ตัวพิมพ์ใหญ่) คือ 65** เมื่อกดแป้น A ค่า 65 จะถูกเก็บในหน่วยความจำเป็นเลขฐานสอง `01000001`
- พัฒนาขึ้นในต้นทศวรรษ 1960 และผู้ผลิตคอมพิวเตอร์เกือบทั้งหมดนำไปใช้

<p align="center"><img src="images/บทที่01/fig1.14.jpg" width="620"><br><b>ภาพที่ 1.14</b> ตัวอักษร A ถูกเก็บในหน่วยความจำโดยแทนด้วยตัวเลข 65<br><sub>ที่มา: “Introduction to Computers and Programming” (2016)</sub></p>

#### รหัส Unicode
- ASCII มีข้อจำกัดเพราะกำหนดรหัสได้เพียง 128 อักขระ จึงไม่พอสำหรับภาษาไทย จีน ญี่ปุ่น ฯลฯ
- **Unicode** พัฒนาขึ้นในต้นทศวรรษ 1990 และ **ใช้ร่วมกับ ASCII ได้** (128 รหัสแรกตรงกัน)
- แสดงอักขระของ **ทุกภาษาในโลก** และปัจจุบันเป็นมาตรฐานของอุตสาหกรรมคอมพิวเตอร์ ภาษาไพธอน 3 ใช้ Unicode เป็นค่าเริ่มต้นของข้อความ (`str`)

### 🔬 ทดลอง: ดูรหัสของอักขระ
- `ord(ตัวอักษร)` คืนค่ารหัส (code point) ของอักขระ
- `chr(รหัส)` แปลงรหัสกลับเป็นอักขระ

````python
for ch in ["A", "B", "a", "0", " ", "!"]:
    code_point = ord(ch)
    print(f"{ch!r:5} -> รหัส {code_point:>3} -> ฐานสอง {code_point:08b}")
````

### 1.3.3 การจัดเก็บข้อมูลตัวเลขขั้นสูง
เทคนิคเลขฐานสองแบบง่ายข้างต้นใช้ได้กับ **จำนวนเต็มบวก** เท่านั้น ส่วน **จำนวนลบ** และ **จำนวนจริง** (เช่น 3.14159) ต้องใช้รูปแบบการเข้ารหัสร่วมกับเลขฐานสอง ดังนี้
- **จำนวนลบ** เข้ารหัสด้วยเทคนิค **Two's complement (ส่วนเติมเต็มสอง)**
- **จำนวนจริง** เข้ารหัสด้วย **Floating-point notation (จำนวนจุดลอยตัว)**

โดยทั่วไปการเขียนโปรแกรมไม่จำเป็นต้องรู้รายละเอียดของการเข้ารหัสเหล่านี้ แต่การลองดูสักครั้งจะช่วยให้เข้าใจพฤติกรรมบางอย่างของคอมพิวเตอร์ได้ดีขึ้น

#### 🔬 (เสริม) Two's complement แบบ 8 บิต
วิธีหารูปแบบบิตของ **−n**: เขียน n เป็นฐานสอง **กลับบิตทุกตัว** (0↔1) แล้ว **บวก 1**

````python
def twos_complement_8bit(n):
    """คืนรูปแบบบิต 8 บิตของจำนวนเต็ม n (ช่วง -128 ถึง 127)"""
    return format(n & 0xFF, "08b")

n = 5
print(f"+{n}  = {twos_complement_8bit(n)}")
inverted = format(~n & 0xFF, "08b")
print(f"กลับบิต = {inverted}")
print(f"บวก 1  = {twos_complement_8bit(-n)}   <- นี่คือ -{n}")

print()
for x in [0, 1, 127, -1, -5, -128]:
    print(f"{x:>5} -> {twos_complement_8bit(x)}")
````

#### 🔬 (เสริม) Floating-point ไม่ได้แม่นยำเสมอไป
จำนวนจริงหลายค่า เช่น 0.1 เขียนเป็นเลขฐานสองให้ลงตัวพอดีไม่ได้ คอมพิวเตอร์จึงเก็บเป็น **ค่าประมาณ** และนี่คือเหตุผลที่ `0.1 + 0.2` ไม่เท่ากับ `0.3` พอดี

````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
print(f"ค่าจริงที่เก็บของ 0.1 = {0.1:.25f}")
print("ใช้ round() ช่วยเปรียบเทียบได้:", round(0.1 + 0.2, 10) == 0.3)
````

### 🔬 ทดลอง: เสียงคือชุดของตัวเลข (samples)
สร้างคลื่นเสียงโน้ต A (440 Hz) ที่อัตรา 44,100 samples/วินาที แล้วดูว่าเสียงเพียง 1 มิลลิวินาทีประกอบด้วยตัวเลขกี่ตัว

````python
import numpy as np
import matplotlib.pyplot as plt

sample_rate = 44_100          # samples ต่อวินาที (คุณภาพ CD)
freq = 440                    # ความถี่โน้ต A (Hz)
duration = 0.005              # 5 มิลลิวินาที

t = np.arange(0, duration, 1 / sample_rate)
samples = np.round(np.sin(2 * np.pi * freq * t) * 32767).astype(int)  # 16-bit

print("จำนวน samples ใน 5 มิลลิวินาที =", len(samples))
print("5 samples แรก =", samples[:5].tolist())
print("เพลงยาว 3 นาที (สเตอริโอ 16 บิต) ใช้พื้นที่ประมาณ",
      round(sample_rate * 180 * 2 * 2 / 1024**2, 1), "MB")

plt.figure(figsize=(9, 3))
plt.plot(t * 1000, samples, color="lightgray")
plt.stem(t[::4] * 1000, samples[::4], basefmt=" ")
plt.xlabel("time (ms)"); plt.ylabel("sample value")
plt.title("A 440 Hz sound wave stored as numbers (every 4th sample shown)")
plt.show()
````

## 1.4 วิธีการทำงานของโปรแกรม

ซีพียูมักถูกเรียกว่า "สมองของคอมพิวเตอร์" และถูกบรรยายว่า "ฉลาด" แต่นั่นเป็นเพียงคำอุปมา **ความจริงซีพียูไม่ใช่สมองและไม่ได้ฉลาด** ซีพียูเป็นเพียงอุปกรณ์อิเล็กทรอนิกส์ที่ถูกออกแบบมาให้ทำงานพื้นฐานต่อไปนี้
- อ่านข้อมูลจากหน่วยความจำหลัก
- บวกเลข
- ลบเลข
- คูณเลข
- หารเลข
- ย้ายข้อมูลจากตำแหน่งหน่วยความจำหนึ่งไปยังอีกตำแหน่งหนึ่ง
- เปรียบเทียบค่า

ซีพียูจะทำงานได้ก็ต่อเมื่อเราบอกว่าต้องทำอะไรผ่าน **คำสั่ง (instruction)** ในโปรแกรม ตัวอย่างคำสั่งที่อาจอยู่ในโปรแกรม

```
10110000
```

สำหรับคนทั่วไปนี่เป็นเพียงชุดของ 0 และ 1 แต่สำหรับซีพียูคือคำสั่งให้ดำเนินการ เพราะ **ซีพียูเข้าใจเฉพาะคำสั่งที่เขียนด้วยภาษาเครื่อง (Machine Language)** ซึ่งมีรูปแบบเป็นเลขฐานสองเท่านั้น

> 📌 ชุดคำสั่งทั้งหมดที่ซีพียูตัวหนึ่งทำได้ เรียกว่า **ชุดคำสั่งของซีพียู (CPU's instruction set)**

งานจริงต้องใช้คำสั่งภาษาเครื่องจำนวนมากเรียงตามลำดับที่ถูกต้อง เช่น การคำนวณดอกเบี้ยเงินฝากของปีนี้ และไม่ใช่เรื่องแปลกที่โปรแกรมหนึ่งจะมีคำสั่งภาษาเครื่อง **นับพันหรือหลายล้านคำสั่ง**

### โปรแกรมต้องถูกคัดลอกลง RAM ก่อนรันเสมอ
- โปรแกรมถูกเก็บในหน่วยความจำสำรอง เช่น ดิสก์ไดร์ฟ (ติดตั้งจากซีดีรอมหรือดาวน์โหลดจากเว็บไซต์)
- **ก่อนรันทุกครั้ง** โปรแกรมต้องถูกคัดลอกลงหน่วยความจำหลัก (RAM)
- ตัวอย่าง: เมื่อดับเบิลคลิกไอคอนโปรแกรมประมวลผลคำ โปรแกรมจะถูกคัดลอกจากดิสก์ไปยัง RAM แล้วซีพียูจึงรัน **สำเนาของโปรแกรมที่อยู่ใน RAM**

### วงจร Fetch–Decode–Execute
เมื่อซีพียูรันคำสั่งในโปรแกรม จะทำงานเป็นวงจรวนซ้ำที่เรียกว่า **fetch-decode-execute cycle** ซึ่งแต่ละรอบมี 3 ขั้นตอน

| ขั้นตอน | ความหมาย |
|---|---|
| **1. Fetch (ดึงคำสั่ง)** | อ่านคำสั่งถัดไปจากหน่วยความจำหลักมายังซีพียู |
| **2. Decode (ถอดรหัส)** | ถอดรหัสคำสั่งที่เป็นเลขฐานสอง เพื่อดูว่าต้องทำการดำเนินการใด |
| **3. Execute (ดำเนินการ)** | ทำตามที่คำสั่งกำหนด |

ซีพียูทำวงจรนี้ซ้ำไปเรื่อย ๆ กับทุกคำสั่งในโปรแกรม โดยซีพียูสมัยใหม่ทำได้ **หลายพันล้านรอบต่อวินาที**

<p align="center"><img src="images/บทที่01/fig1.17.jpg" width="620"><br><b>ภาพที่ 1.17</b> วงจร fetch-decode-execute cycle<br><sub>ที่มา: “Introduction to Computers and Programming” (2016)</sub></p>

### 🔬 ทดลอง: จำลองซีพียูอย่างง่าย
เราจะสร้าง "ซีพียูของเล่น" ด้วยไพธอน โดยมีหน่วยความจำ (RAM) เก็บคำสั่งภาษาเครื่อง 8 บิต และมี **รีจิสเตอร์ (register)** ชื่อ `ACC` ไว้เก็บผลการคำนวณ

รูปแบบคำสั่ง: **4 บิตแรก = รหัสคำสั่ง (opcode)** และ **4 บิตหลัง = ตัวเลขที่ใช้ (operand)**

| opcode | ความหมาย (mnemonic) |
|---|---|
| `0001` | LOAD n — ใส่ค่า n ลงใน ACC |
| `0010` | ADD n — บวก n เข้ากับ ACC |
| `0011` | SUB n — ลบ n ออกจาก ACC |
| `0100` | MUL n — คูณ ACC ด้วย n |
| `0101` | PRINT — แสดงค่า ACC |
| `1111` | HALT — หยุดทำงาน |

````python
def run_cpu(memory, verbose=True):
    ACC = 0          # รีจิสเตอร์เก็บผลลัพธ์
    PC = 0           # Program Counter: ตำแหน่งคำสั่งถัดไปในหน่วยความจำ
    names = {"0001": "LOAD", "0010": "ADD", "0011": "SUB",
             "0100": "MUL", "0101": "PRINT", "1111": "HALT"}
    while True:
        # 1) FETCH : ดึงคำสั่งจากหน่วยความจำ
        instruction = memory[PC]
        PC += 1
        # 2) DECODE : แยก opcode และ operand
        opcode, operand = instruction[:4], int(instruction[4:], 2)
        op = names.get(opcode, "???")
        if verbose:
            print(f"FETCH {instruction} -> DECODE {op:<5} {operand:<2} -> EXECUTE", end=" ")
        # 3) EXECUTE : ทำตามคำสั่ง
        if op == "LOAD":
            ACC = operand
        elif op == "ADD":
            ACC += operand
        elif op == "SUB":
            ACC -= operand
        elif op == "MUL":
            ACC *= operand
        elif op == "PRINT":
            print(f"\n   >>> ผลลัพธ์บนจอ: {ACC}", end="")
        elif op == "HALT":
            print("(หยุดทำงาน)")
            break
        else:
            print("ไม่รู้จักคำสั่งนี้!")
            break
        if verbose:
            print(f"  [ACC = {ACC}]")

# โปรแกรมภาษาเครื่อง: คำนวณ (7 + 5) x 3
program = [
    "00010111",   # LOAD 7
    "00100101",   # ADD 5
    "01000011",   # MUL 3
    "01010000",   # PRINT
    "11110000",   # HALT
]
run_cpu(program)
````

> 🤔 ลองคิด: ถ้าต้องเขียนโปรแกรมยาวหลายพันบรรทัดด้วยเลข 0 และ 1 แบบนี้ จะยากและผิดพลาดง่ายแค่ไหน แค่วาง 0 หรือ 1 ผิดตำแหน่งเดียว โปรแกรมก็ทำงานผิดแล้ว ลองเปลี่ยน `"00100101"` เป็น `"00110101"` (ต่างกันเพียงบิตเดียว) แล้วรันใหม่ดู

### จากภาษาเครื่องไปสู่ภาษาแอสเซมบลี
- คอมพิวเตอร์รันได้เฉพาะโปรแกรมภาษาเครื่อง แต่การเขียนภาษาเครื่องที่มีคำสั่งเลขฐานสองนับพันนับล้านคำสั่งนั้น **น่าเบื่อ เสียเวลา และผิดพลาดง่ายมาก**
- **ภาษาแอสเซมบลี (Assembly Language)** จึงถูกสร้างขึ้นในยุคแรกของคอมพิวเตอร์ โดยใช้ **คำสั้น ๆ ที่เรียกว่า mnemonics** แทนเลขฐานสอง เช่น
  - `add` = บวกเลข
  - `mul` = คูณเลข
  - `mov` = ย้ายค่าในหน่วยความจำ
- ซีพียูรันภาษาแอสเซมบลีโดยตรงไม่ได้ จึงต้องใช้โปรแกรมพิเศษชื่อ **แอสเซมเบลอร์ (Assembler)** แปลงเป็นภาษาเครื่องก่อน

<p align="center"><img src="images/บทที่01/fig1.18.jpg" width="620"><br><b>ภาพที่ 1.18</b> แอสเซมเบลอร์แปลโปรแกรมภาษาแอสเซมบลีเป็นภาษาเครื่อง<br><sub>ที่มา: “Introduction to Computers and Programming” (2016)</sub></p>

### 🔬 ทดลอง: เขียนแอสเซมเบลอร์ของเล่น
ฟังก์ชัน `assemble()` แปลงโปรแกรมที่เขียนด้วย mnemonic ให้เป็นภาษาเครื่องของซีพียูของเล่นข้างต้น จากนั้นนำไปรันได้ทันที ลองแก้โปรแกรมเป็นสูตรอื่นดู (ตัวเลขต้องอยู่ระหว่าง 0–15)

````python
OPCODES = {"LOAD": "0001", "ADD": "0010", "SUB": "0011",
           "MUL": "0100", "PRINT": "0101", "HALT": "1111"}

def assemble(source):
    machine_code = []
    for line in source.strip().splitlines():
        parts = line.split()
        mnemonic = parts[0].upper()
        operand = int(parts[1]) if len(parts) > 1 else 0
        binary = OPCODES[mnemonic] + format(operand, "04b")
        print(f"{line:<10} -> {binary}")
        machine_code.append(binary)
    return machine_code

assembly_program = """
LOAD 9
SUB 4
MUL 6
PRINT
HALT
"""

print("=== แอสเซมเบลอร์กำลังแปล ===")
code_out = assemble(assembly_program)
print("\n=== ซีพียูกำลังรัน ===")
run_cpu(code_out, verbose=False)
````

### คอมไพเลอร์และอินเตอร์พรีเตอร์ (Compilers and Interpreters)
ซีพียูเข้าใจเฉพาะภาษาเครื่อง ดังนั้นโปรแกรมภาษาระดับสูงจึง **ต้องถูกแปลเป็นภาษาเครื่องก่อน** โดยใช้ **คอมไพเลอร์** หรือ **อินเตอร์พรีเตอร์** ขึ้นอยู่กับภาษาที่ใช้

#### คอมไพเลอร์ (Compiler)
แปล **ทั้งโปรแกรม** ภาษาระดับสูงให้เป็น **โปรแกรมภาษาเครื่องแยกต่างหาก** แล้วนำไปรันได้ทุกเมื่อโดยไม่ต้องแปลใหม่

#### อินเตอร์พรีเตอร์ (Interpreter)
**ภาษาไพธอนใช้อินเตอร์พรีเตอร์** ซึ่งอ่านคำสั่ง **ทีละคำสั่ง** แปลเป็นภาษาเครื่อง แล้ว **รันทันที** และทำซ้ำเช่นนี้กับทุกคำสั่ง โดยปกติจะไม่สร้างไฟล์ภาษาเครื่องแยกออกมา

| การรันซ้ำ | รันไฟล์ที่แปลแล้วได้เลย | ต้องแปลใหม่ทุกครั้งที่รัน |
| ตัวอย่างภาษา | C, C++ | Python |

#### ซอร์สโค้ดและข้อผิดพลาดทางไวยากรณ์
- คำสั่งที่นักพัฒนาเขียนด้วยภาษาระดับสูงเรียกว่า **ซอร์สโค้ด (Source Code)** หรือเรียกสั้น ๆ ว่า **โค้ด (Code)**
- นักพัฒนาพิมพ์โค้ดใน **โปรแกรมแก้ไขข้อความ (Text Editor)** บันทึกเป็นไฟล์ แล้วใช้คอมไพเลอร์หรืออินเตอร์พรีเตอร์แปล
- ทุกภาษามีกฎที่ต้องปฏิบัติอย่างเคร่งครัด เรียกว่า **ไวยากรณ์ (Syntax)** หากโค้ดผิดไวยากรณ์ จะแปลไม่ได้
- **ข้อผิดพลาดทางไวยากรณ์ (Syntax Error)** เช่น สะกดคำสั่งผิด ขาดอักขระหรือเครื่องหมายวรรคตอน หรือใช้ตัวดำเนินการไม่ถูกต้อง เมื่อเกิดขึ้น ตัวแปลภาษาจะแสดงข้อความแจ้งข้อผิดพลาด และนักพัฒนาต้องแก้ไขก่อนจึงจะแปลได้อีกครั้ง

### 🔬 ทดลอง: อินเตอร์พรีเตอร์ทำงานทีละคำสั่ง
สังเกตว่าคำสั่งที่ 1 และ 2 ทำงานและแสดงผลไปแล้ว **ก่อน** ที่อินเตอร์พรีเตอร์จะพบข้อผิดพลาดในคำสั่งที่ 3 (หารด้วยศูนย์) ซึ่งเป็นพฤติกรรมของการแปลและรันทีละคำสั่ง

````python
print("คำสั่งที่ 1 ทำงานแล้ว")
print("คำสั่งที่ 2 ทำงานแล้ว")
print("คำสั่งที่ 3:", 10 / 0)      # ข้อผิดพลาดขณะรัน
print("คำสั่งที่ 4 จะไม่ถูกรัน")
````

### 🔬 ทดลอง: ข้อผิดพลาดทางไวยากรณ์ (Syntax Error)
โค้ดด้านล่างขาดวงเล็บปิด `)` ไพธอนจะตรวจไวยากรณ์ของเซลล์ **ก่อนรัน** จึงไม่มีบรรทัดใดในเซลล์นี้ถูกรันเลย แม้แต่บรรทัดแรก ลองรัน อ่านข้อความแจ้งข้อผิดพลาด แล้วแก้ไขให้ถูกต้อง

````python
print("บรรทัดนี้ถูกต้อง")
print("Hello world"
````

ข้อผิดพลาดทางไวยากรณ์ที่พบบ่อยในภาษาไพธอน (เราใช้ `compile()` ตรวจโค้ดโดยไม่รัน)

````python
examples = {
    "สะกดคำสั่งผิด (ไม่ใช่ syntax error แต่เป็น NameError ตอนรัน)": 'pritn("hi")',
    "ขาดเครื่องหมายคำพูดปิด": 'print("hi)',
    "ขาดเครื่องหมาย : หลัง if": 'if 5 > 3\n    print("yes")',
    "ใช้ตัวดำเนินการผิด": 'x = 5 +* 3',
}
for desc, src in examples.items():
    try:
        compile(src, "<code>", "exec")
        print(f"✔ {desc}\n   ไวยากรณ์ถูกต้อง: {src!r}\n")
    except SyntaxError as e:
        print(f"✘ {desc}\n   SyntaxError: {e.msg}  ->  {src!r}\n")
````

> 💡 **ความรู้เพิ่มเติม:** ตัวแปลภาษาไพธอนมาตรฐาน (CPython) ทำงานเป็นสองช่วง โดยแปลซอร์สโค้ดเป็น **ไบต์โค้ด (bytecode)** ซึ่งเป็นคำสั่งระดับต่ำก่อน แล้วจึงให้ **เครื่องเสมือนไพธอน (Python Virtual Machine)** อินเตอร์พรีตไบต์โค้ดนั้นทีละคำสั่ง เราดูไบต์โค้ดได้ด้วยโมดูล `dis` ซึ่งหน้าตาคล้ายภาษาแอสเซมบลีที่เรียนมา

````python
import dis

def total_price(price, qty):
    return price * qty + 10

dis.dis(total_price)
````

## 1.5 สรุป

บทเรียนนี้ปูพื้นฐานความรู้เกี่ยวกับคอมพิวเตอร์และการเขียนโปรแกรมเบื้องต้น ดังนี้

- **ความสำคัญของคอมพิวเตอร์:** มีอิทธิพลต่อการใช้ชีวิตและการทำงาน และคอมพิวเตอร์ทำงานตาม **โปรแกรม** ซึ่งเป็นชุดคำสั่ง
- **ฮาร์ดแวร์** มี 5 ส่วน ได้แก่ CPU, หน่วยความจำหลัก (RAM ซึ่งลบเลือนได้), หน่วยความจำสำรอง (เก็บถาวร), อุปกรณ์อินพุต และอุปกรณ์เอาต์พุต
- **ซอฟต์แวร์** แบ่งเป็นซอฟต์แวร์ระบบ (ระบบปฏิบัติการ, โปรแกรมอรรถประโยชน์, เครื่องมือพัฒนาซอฟต์แวร์) และซอฟต์แวร์ประยุกต์
- **การจัดเก็บข้อมูล:** ทุกอย่างเก็บเป็น **เลขฐานสอง** โดย 1 ไบต์ = 8 บิต
  - ตัวเลข: ใช้ค่าประจำตำแหน่ง 2ⁿ โดย 1 ไบต์เก็บได้ 0–255
  - อักขระ: ASCII (128 รหัส) และ Unicode (ทุกภาษา)
  - จำนวนลบใช้ Two's complement ส่วนจำนวนจริงใช้ Floating-point
  - ภาพเก็บเป็นพิกเซล ส่วนเสียงเก็บเป็น samples
- **การทำงานของโปรแกรม:** โปรแกรมถูกคัดลอกจากดิสก์ลง RAM แล้ว CPU ทำงานด้วยวงจร **Fetch → Decode → Execute**
- **ภาษาโปรแกรม:** วิวัฒนาการจากภาษาเครื่อง ไปสู่ภาษาแอสเซมบลี (ใช้ Assembler แปล ซึ่งเป็นภาษาระดับต่ำ) และภาษาระดับสูง
- **ตัวแปลภาษา:** คอมไพเลอร์แปลทั้งโปรแกรมเป็นไฟล์ภาษาเครื่อง ส่วนอินเตอร์พรีเตอร์แปลและรันทีละคำสั่ง ซึ่ง **ไพธอนใช้อินเตอร์พรีเตอร์**
- โค้ดที่ผิดกฎไวยากรณ์ (Syntax) จะเกิด **Syntax Error** และแปลไม่ได้

````python
quiz = [
    ("ส่วนใดสำคัญที่สุดในคอมพิวเตอร์ เพราะใช้ประมวลผลคำสั่งของโปรแกรม",
     ["ก. RAM", "ข. CPU", "ค. ดิสก์ไดร์ฟ", "ง. จอภาพ"], "ข"),
    ("ข้อใดเป็นหน่วยความจำแบบลบเลือนได้ (ข้อมูลหายเมื่อปิดเครื่อง)",
     ["ก. แฟลชไดร์ฟ", "ข. DVD", "ค. RAM", "ง. ฮาร์ดดิสก์"], "ค"),
    ("10000001 ในฐานสองมีค่าเท่าใดในฐานสิบ",
     ["ก. 129", "ข. 128", "ค. 65", "ง. 255"], "ก"),
    ("1 ไบต์ประกอบด้วยกี่บิต",
     ["ก. 2", "ข. 4", "ค. 8", "ง. 16"], "ค"),
    ("รหัส ASCII ของตัวอักษร A คือข้อใด",
     ["ก. 1", "ข. 65", "ค. 97", "ง. 128"], "ข"),
    ("ชุดอักขระใดรองรับภาษาไทย",
     ["ก. ASCII", "ข. Binary", "ค. Unicode", "ง. Two's complement"], "ค"),
    ("ขั้นตอนแรกของวงจรการทำงานของ CPU คือข้อใด",
     ["ก. Execute", "ข. Decode", "ค. Compile", "ง. Fetch"], "ง"),
    ("โปรแกรมที่แปลภาษาแอสเซมบลีเป็นภาษาเครื่องคือข้อใด",
     ["ก. Assembler", "ข. Compiler", "ค. Interpreter", "ง. Editor"], "ก"),
    ("ข้อใดเป็นซอฟต์แวร์ระบบ",
     ["ก. Microsoft Word", "ข. โปรแกรมสแกนไวรัส", "ค. เกม", "ง. Photoshop"], "ข"),
    ("ภาษาไพธอนใช้ตัวแปลภาษาแบบใด",
     ["ก. Assembler", "ข. Compiler เท่านั้น", "ค. Interpreter", "ง. ไม่ต้องแปล"], "ค"),
]

def run_quiz():
    score = 0
    for i, (q, choices, ans) in enumerate(quiz, 1):
        print(f"\nข้อ {i}. {q}")
        for c in choices:
            print("   ", c)
        reply = input("คำตอบ (ก/ข/ค/ง): ").strip()
        if reply == ans:
            print("   ✅ ถูกต้อง")
            score += 1
        else:
            print(f"   ❌ ยังไม่ถูก คำตอบที่ถูกคือ {ans}")
    print(f"\n🎯 คะแนนของคุณ: {score}/{len(quiz)}")
    if score == len(quiz):
        print("🏆 สุดยอด! เข้าใจบทที่ 1 อย่างครบถ้วน")
    elif score >= 7:
        print("👍 ดีมาก ทบทวนข้อที่ผิดอีกเล็กน้อย")
    else:
        print("📖 แนะนำให้ย้อนกลับไปอ่านเนื้อหาอีกครั้ง แล้วลองทำใหม่")

run_quiz()
````
