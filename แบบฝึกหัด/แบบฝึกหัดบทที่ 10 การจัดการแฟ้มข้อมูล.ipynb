````python
#@title ▶ รันเซลล์นี้ก่อนเริ่มทำแบบฝึกหัด (ระบบตรวจคำตอบอัตโนมัติ — ไม่ต้องแก้ไข)
import hashlib, io, contextlib, copy, os, shutil, csv

def _setup_ch10_fixtures():
    with open("story.txt", "w", encoding="utf-8") as f:
        f.write("Python is a programming language\n"
                "Python is easy to learn\n"
                "Files store data permanently\n"
                "Learning Python is fun\n")
    with open("poem.txt", "w", encoding="utf-8") as f:
        f.write("Roses are red\n"
                "Violets are blue\n"
                "Python is awesome\n")
    os.makedirs("qdir/sub", exist_ok=True)
    for name, size in [("a.txt", 10), ("b.txt", 250), ("c.csv", 42), ("sub/big.bin", 999)]:
        with open(os.path.join("qdir", name), "wb") as f:
            f.write(b"x" * size)
    os.makedirs("qdir2/nested", exist_ok=True)
    for name, size in [("x.py", 120), ("y.md", 35), ("z.json", 500), ("nested/huge.dat", 2000)]:
        with open(os.path.join("qdir2", name), "wb") as f:
            f.write(b"y" * size)
    with open("scores.csv", "w", encoding="utf-8", newline="") as f:
        w = csv.writer(f)
        w.writerows([
            ["name", "midterm", "final"],
            ["Ann", 35, 48],
            ["Bob", 28, 39],
            ["Cat", 42, 45],
            ["Dan", 20, 30],
        ])
    with open("items.csv", "w", encoding="utf-8", newline="") as f:
        w = csv.writer(f)
        w.writerows([
            ["item", "price", "qty"],
            ["Pen", 15, 10],
            ["Book", 120, 3],
            ["Ruler", 20, 5],
        ])

_setup_ch10_fixtures()

_SALT = 'py3-ch10'
_KEY = {"A1": "8d7dcfc10b1cc8d0", "A2": "c3d9d07432dbd924", "A3": "f2b6f2069439747a", "A4": "04880bdcf80b6e9c", "A5": "468f95405f30f5b8", "A6": "bdf700c1983f3bf1", "A7": "0df31ca8af9d8639", "A8": "6e05aa0ee10dab51", "A9": "bd4ef88c3319a540", "A10": "f10c19dd13dff590", "A11": "a09442e82c35a326", "A12": "fc76f117c2fbc16e", "A13": "5f34a6ec57f1e932", "A14": "7479d23471cf83c3", "A15": "a0dfeff0847be92e", "A16": "393b2eab66b503a3", "A17": "7ba34bedffae1cb6", "A18": "a0e1acd50b5b8c68", "A19": "33270f3f2156686f", "A20": "962347c2c5d6077f", "B1": "4655fa6c3529f77c", "B2": "349b7061848b3c6f", "B3": "0f2749328e814772", "B4": "538bfed257754866", "B5": "e73e6bf3f719edc0", "B6": "1ea13285da6e65f8", "B7": "9d8e1b5f5c5bee27", "B8": "809e6a8d35a425b3", "B9": "ca71409b9219a1e3", "B10": "fd9ef4a502f41698", "C1": "b8f22aad773a9dc8", "C2": "77e617d660b6a812", "C3": "ae12411f363fb19b", "C4": "7a55c578baae1d62", "C5": "5890ddaebd234e11", "C6": "9fb7ab6099c3a311", "C7": "74f91b5f63d7aa21", "C8": "fb18a4ed2690a75a", "C9": "7f76747c7ae1face", "C10": "5622c3c6ae62e7b5", "E1": ["e8818e407d1fa623", "323a7e6372117bc8"], "E2": ["a91a26b264aef144", "27e279cbda1db6e9"], "E3": ["0061845d58d25e50", "2c3129f98665ff92"], "E4": ["60a3beba465f5406", "9c46bf2aac945710"], "E5": ["12c6e99b1c636eeb", "754f2af0fd6cbea7"], "E6": ["502dd6fdc58b6934", "40eaf582d07ed9e9"], "E7": ["1aa4c5b4c66296fd", "415f5b69eee0a408", "7026ed644c4b50fb"], "E8": ["0a95c6f7ac379e41", "d49fae7686130e6d"], "E9": ["16cef93ae1e3295d", "0609059357a5771e", "c49c49a93946c3a1"], "E10": ["97964647b1ac9b61", "e154933a88a34f81"], "E11": ["0f9b0abc5e6b7455", "a55ca2552e56ab83", "eedc60945430ea7e"], "E12": ["eff59744b11896f1", "f91af19ead3cda6a", "a4a26459716afa5d", "cb57f804ba78924d"]}
_TOPIC = {"A1": "10.2.1", "A2": "10.2.2", "A3": "10.2.1", "A4": "10.2.2", "A5": "10.2.2", "A6": "10.2.2", "A7": "10.2.1", "A8": "10.3.1", "A9": "10.3.1", "A10": "10.3.2", "A11": "10.4", "A12": "10.4", "A13": "10.5.1", "A14": "10.5.1", "A15": "10.5.2", "A16": "10.5.2", "A17": "10.5.4", "A18": "10.5.4", "A19": "10.6.2", "A20": "10.6.3", "B1": "10.2.2", "B2": "10.3.2", "B3": "10.4", "B4": "10.4", "B5": "10.5.2", "B6": "10.5.2", "B7": "10.5.3", "B8": "10.5.4", "B9": "10.6.2", "B10": "10.6.3", "C1": "10.3.1", "C2": "10.2.2, 10.3.1", "C3": "10.3.1", "C4": "10.4", "C5": "10.4", "C6": "10.5.1", "C7": "10.5.2", "C8": "10.5.2", "C9": "10.6.2", "C10": "10.6.2, 10.6.3", "E1": "10.3.1, 10.4", "E2": "10.4", "E3": "10.3.1", "E4": "10.4", "E5": "10.4", "E6": "10.4", "E7": "10.3.1, 10.4", "E8": "10.4", "E9": "10.5.2", "E10": "10.5.2", "E11": "10.3.1, 10.4", "E12": "10.6.2"}
_PARTS = [('ตอนที่ 1 ปรนัย', ['A1', 'A2', 'A3', 'A4', 'A5', 'A6', 'A7', 'A8', 'A9', 'A10', 'A11', 'A12', 'A13', 'A14', 'A15', 'A16', 'A17', 'A18', 'A19', 'A20']), ('ตอนที่ 2 ถูก/ผิด', ['B1', 'B2', 'B3', 'B4', 'B5', 'B6', 'B7', 'B8', 'B9', 'B10']), ('ตอนที่ 3 ทำนายผลลัพธ์', ['C1', 'C2', 'C3', 'C4', 'C5', 'C6', 'C7', 'C8', 'C9', 'C10']), ('ตอนที่ 5 เขียนโปรแกรม', ['E1', 'E2', 'E3', 'E4', 'E5', 'E6', 'E7', 'E8', 'E9', 'E10', 'E11', 'E12'])]
_SNIP = {"C1": "with open(\"_c1.txt\", \"w\", encoding=\"utf-8\") as f:\n    n1 = f.write(\"Hello, \")\n    n2 = f.write(\"World!\")\n\nwith open(\"_c1.txt\", \"r\", encoding=\"utf-8\") as f:\n    print(n1, n2)\n    print(f.read(5))\n    print(f.read())", "C2": "with open(\"_c2.txt\", \"w\", encoding=\"utf-8\") as f:\n    f.write(\"First\\n\")\nwith open(\"_c2.txt\", \"w\", encoding=\"utf-8\") as f:\n    f.write(\"Second\\n\")\nwith open(\"_c2.txt\", \"a\", encoding=\"utf-8\") as f:\n    f.write(\"Third\\n\")\n\nwith open(\"_c2.txt\", \"r\", encoding=\"utf-8\") as f:\n    print(f.read(), end=\"\")", "C3": "with open(\"_c3.txt\", \"w\", encoding=\"utf-8\") as f:\n    f.write(\"ABCDEFG\")\n\nwith open(\"_c3.txt\", \"r\", encoding=\"utf-8\") as f:\n    print(f.read(3))\n    print(f.tell())\n    f.seek(0)\n    print(f.read(2))", "C4": "with open(\"_c4.txt\", \"w\", encoding=\"utf-8\") as f:\n    f.write(\"Ant\\nBat\\nCat\\n\")\n\nwith open(\"_c4.txt\", \"r\", encoding=\"utf-8\") as f:\n    lines = [line.strip() for line in f]\nprint(len(lines), lines)", "C5": "with open(\"_c5.txt\", \"w\", encoding=\"utf-8\") as f:\n    f.writelines([\"One\", \"Two\", \"Three\"])\n\nwith open(\"_c5.txt\", \"r\", encoding=\"utf-8\") as f:\n    print(f.read())", "C6": "import posixpath\n\np = posixpath.join(\"project\", \"data\", \"report.csv\")\nparent, name = posixpath.split(p)\nstem, ext = posixpath.splitext(name)\nprint(p)\nprint(parent, \"|\", name)\nprint(stem, \"|\", ext)", "C7": "import os\n\nitems = sorted(os.listdir(\"qdir\"))\nfiles_only = [n for n in items if os.path.isfile(os.path.join(\"qdir\", n))]\ndirs_only = [n for n in items if os.path.isdir(os.path.join(\"qdir\", n))]\nprint(items)\nprint(files_only)\nprint(dirs_only)", "C8": "import os\n\nfor name in [\"a.txt\", \"b.txt\", \"c.csv\"]:\n    p = os.path.join(\"qdir\", name)\n    print(name, os.path.getsize(p))", "C9": "import csv\n\nwith open(\"_c9.csv\", \"w\", encoding=\"utf-8\", newline=\"\") as f:\n    f.write(\"01/06/2017, 4\\n02/06/2017, 8\\n03/06/2017, 3\\n\")\n\ndates, pts = [], []\nwith open(\"_c9.csv\", \"r\", encoding=\"utf-8\") as f:\n    for row in csv.reader(f, skipinitialspace=True):\n        dates.append(row[0])\n        pts.append(int(row[1]))\n\nprint(dates)\nprint(pts, sum(pts))", "C10": "import csv\n\nrows = [[\"id\", \"score\"], [\"S1\", 80], [\"S2\", 95]]\nwith open(\"_c10.csv\", \"w\", encoding=\"utf-8\", newline=\"\") as f:\n    csv.writer(f, delimiter=\";\").writerows(rows)\n\nwith open(\"_c10.csv\", \"r\", encoding=\"utf-8\") as f:\n    for r in csv.DictReader(f, delimiter=\";\"):\n        print(r[\"id\"], int(r[\"score\"]) + 5)"}
_TESTS = {'E1': [(('story.txt',), {}), (('poem.txt',), {})], 'E2': [(('story.txt',), {}), (('poem.txt',), {})], 'E3': [(('story.txt',), {}), (('poem.txt',), {})], 'E4': [(('story.txt',), {}), (('poem.txt',), {})], 'E5': [(('story.txt',), {}), (('poem.txt',), {})], 'E6': [(('story.txt',), {}), (('poem.txt',), {})], 'E7': [(('story.txt', 'Python'), {}), (('story.txt', 'Java'), {}), (('poem.txt', 'are'), {})], 'E8': [(('story.txt',), {}), (('poem.txt',), {})], 'E9': [(('qdir', 'c.csv'), {}), (('qdir', 'z.txt'), {}), (('qdir2', 'y.md'), {})], 'E10': [(('qdir',), {}), (('qdir2',), {})], 'E11': [(('_out1.txt', ['Alpha', 'Beta', 'Gamma']), {}), (('_out2.txt', ['Python']), {}), (('_out3.txt', []), {})], 'E12': [(('scores.csv', 1), {}), (('scores.csv', 2), {}), (('items.csv', 1), {}), (('items.csv', 2), {})]}
_results = {}

def _canon(v):
    if isinstance(v, (set, frozenset)):
        return "set{" + ",".join(sorted(_canon(x) for x in v)) + "}"
    if isinstance(v, dict):
        return "dict{" + ",".join(sorted(_canon(k) + ":" + _canon(x) for k, x in v.items())) + "}"
    if isinstance(v, (list, tuple)):
        return type(v).__name__ + "(" + ",".join(_canon(x) for x in v) + ")"
    if isinstance(v, bool) or v is None:
        return repr(v)
    if isinstance(v, (int, float)):
        return "num:" + format(round(float(v), 6) + 0.0, ".6f")
    return type(v).__name__ + ":" + repr(v)

def _h(qid, value):
    return hashlib.sha256(f"{_SALT}|{qid}|{value}".encode("utf-8")).hexdigest()[:16]

def _norm_text(s):
    lines = [ln.rstrip() for ln in str(s).replace("\r\n", "\n").split("\n")]
    while lines and lines[0] == "":
        lines.pop(0)
    while lines and lines[-1] == "":
        lines.pop()
    return "\n".join(lines)

def _norm_choice(prefix, ans):
    a = str(ans).strip().upper()
    if prefix == "B":
        if a in ("T", "TRUE", "ถูก", "ท", "1"): return "T"
        if a in ("F", "FALSE", "ผิด", "ผ", "0"): return "F"
    return a

def grade(part, answers):
    """ตรวจคำตอบตอนที่ 1 (part='A') และตอนที่ 2 (part='B')"""
    n_ok, n_all = 0, 0
    for n in sorted(answers):
        qid = f"{part}{n}"
        if qid not in _KEY:
            continue
        n_all += 1
        a = _norm_choice(part, answers[n])
        if a == "":
            print(f"ข้อ {n:>2}: ⚪ ยังไม่ได้ตอบ")
            _results[qid] = False
            continue
        ok = _h(qid, a) == _KEY[qid]
        _results[qid] = ok
        n_ok += ok
        print(f"ข้อ {n:>2}: {'✅ ถูกต้อง' if ok else '❌ ยังไม่ถูก  (ทบทวนหัวข้อ ' + _TOPIC[qid] + ')'}")
    print("-" * 40)
    print(f"ได้ {n_ok} / {n_all} คะแนน")

def predict(qid, prediction):
    """ตอนที่ 3: เปรียบเทียบผลลัพธ์ที่ทำนายกับผลลัพธ์จริง (บันทึกคะแนนจากการตรวจครั้งแรก)"""
    _setup_ch10_fixtures()
    pred = _norm_text(prediction)
    if pred == "":
        print("⚪ ยังไม่ได้เขียนคำตอบที่ทำนายไว้ในตัวแปร")
        return
    buf = io.StringIO()
    with contextlib.redirect_stdout(buf):
        exec(_SNIP[qid], {})
    actual = _norm_text(buf.getvalue())
    ok = _h(qid, pred) == _KEY[qid]
    first = qid not in _results
    if first:
        _results[qid] = ok
    print("✅ ทำนายถูกต้อง" if ok else "❌ ทำนายไม่ตรงกับผลลัพธ์จริง")
    print("\nผลลัพธ์ที่ทำนาย:")
    print(pred)
    print("\nผลลัพธ์จริงจากการรันโค้ด:")
    print(actual)
    if not ok:
        print("\n(สังเกตเครื่องหมาย ช่องว่าง วงเล็บ และเครื่องหมายคำพูดให้ดี — ทบทวนหัวข้อ " + _TOPIC[qid] + ")")
    if not first:
        print("\nℹ️ คะแนนข้อนี้บันทึกจากการตรวจครั้งแรกไปแล้ว")

def check_func(qid, func):
    """ตอนที่ 5: ทดสอบฟังก์ชันด้วยข้อมูลทดสอบหลายชุด"""
    _setup_ch10_fixtures()
    if not callable(func):
        print("❌ ยังไม่ได้สร้างฟังก์ชัน")
        _results[qid] = False
        return
    tests = _TESTS[qid]
    n_ok = 0
    for k, (args, kwargs) in enumerate(tests):
        call = func.__name__ + "(" + ", ".join([repr(a) for a in args] +
                                              [f"{kk}={vv!r}" for kk, vv in kwargs.items()]) + ")"
        try:
            got = func(*copy.deepcopy(args), **copy.deepcopy(kwargs))
        except Exception as e:
            print(f"❌ {call}  →  เกิดข้อผิดพลาด {type(e).__name__}: {e}")
            continue
        ok = _h(f"{qid}#{k}", _canon(got)) == _KEY[qid][k]
        n_ok += ok
        print(f"{'✅' if ok else '❌'} {call}  →  {got!r}")
    _results[qid] = n_ok == len(tests)
    print("-" * 40)
    if n_ok == len(tests):
        print(f"✅ ผ่านการทดสอบครบทั้ง {len(tests)} กรณี เยี่ยมมาก!")
    else:
        print(f"❌ ผ่าน {n_ok} จาก {len(tests)} กรณี — ตรวจสอบกรณีที่ไม่ผ่าน (ทบทวนหัวข้อ " + _TOPIC[qid] + ")")
        print("   อย่าลืมว่าฟังก์ชันต้อง return ผลลัพธ์ ไม่ใช่ print")

def summary():
    """สรุปผลการประเมินตนเอง"""
    total_ok, total_all, review = 0, 0, []
    print(f"{'ส่วน':<28}{'คะแนน':>10}")
    print("=" * 38)
    for name, qids in _PARTS:
        ok = sum(1 for q in qids if _results.get(q))
        total_ok += ok
        total_all += len(qids)
        print(f"{name:<28}{ok:>5} / {len(qids):<3}")
        review += [q for q in qids if not _results.get(q)]
    print("=" * 38)
    pct = 100 * total_ok / total_all
    print(f"{'รวม (ส่วนที่ตรวจอัตโนมัติ)':<28}{total_ok:>5} / {total_all:<3}  ({pct:.1f}%)")
    print()
    if pct >= 80:
        print("🌟 ดีมาก — เข้าใจเนื้อหาบทที่ 10 เป็นอย่างดี พร้อมเรียนบทถัดไป")
    elif pct >= 60:
        print("👍 ดี — เข้าใจเนื้อหาส่วนใหญ่ ควรทบทวนหัวข้อที่ยังทำไม่ได้")
    else:
        print("📖 ควรทบทวนบทเรียนอีกครั้ง แล้วกลับมาทำแบบฝึกหัดใหม่")
    topics = sorted({t.strip() for q in review for t in _TOPIC[q].split(",")},
                    key=lambda s: [int(x) for x in s.split(".")])
    if topics:
        print("\nหัวข้อที่ควรทบทวน:", ", ".join(topics))
    print("\n* ตอนที่ 4 และตอนที่ 6 ให้ตรวจสอบกับผลลัพธ์ที่โจทย์กำหนด/ผู้สอน")

print("✅ ระบบตรวจคำตอบและไฟล์ข้อมูลทดสอบพร้อมใช้งาน — เริ่มทำแบบฝึกหัดได้เลย")
````
