````python
#@title ▶ รันเซลล์นี้ก่อนเริ่มทำแบบฝึกหัด (ระบบตรวจคำตอบอัตโนมัติ — ไม่ต้องแก้ไข)
import hashlib, io, contextlib, copy

_SALT = "py3-ch8"
_KEY = {"A1": "5040c23d53b2c86f", "A2": "5dc67f151c112d0d", "A3": "e4012d32b342c876", "A4": "bc6741f1e6f6bace", "A5": "fd2a934ba0d62222", "A6": "661d837793263c21", "A7": "56f2d50c2fff0b4c", "A8": "3d20ccba540c61fc", "A9": "40fa9c99b6503cb4", "A10": "e5016722693b3b7c", "A11": "cba0165375aa9428", "A12": "83d93ebcadd2c424", "A13": "134bb88bb10be73b", "A14": "e181b7b3aac4fec9", "A15": "5d73723189563ebd", "A16": "dcc18c14198af4a4", "A17": "3536e1ee91d964cc", "A18": "6c08256c017cc99f", "A19": "02b39b20078469cb", "A20": "830ac2b1accef96d", "B1": "ce182380d5072cf3", "B2": "b89b5e4df8e790e2", "B3": "8244041f38e6159d", "B4": "a81fae093d3b1130", "B5": "903f95787776e514", "B6": "347cd01002ff3e5c", "B7": "e1e2db9601314455", "B8": "7bb2b65d9cac5018", "B9": "99af6b8b164fda8a", "B10": "395bba13605e0ce4", "C1": "87b704710074eb8c", "C2": "fbdffeda38b6baf0", "C3": "88c19a83eac74588", "C4": "937c652b7e494498", "C5": "4c0b63a9b059e9d6", "C6": "3e46d3326f98b5f9", "C7": "79e67646b5f64c46", "C8": "20730a225ccb3a44", "C9": "9a8fae7e3f239b8c", "C10": "8728bf66104c53b6", "E1": ["87b0af3873f9b57c", "ebd9342c7bcec8b9", "76de58228cf21309", "20c27991cea22673", "a6cc898940c653e3"], "E2": ["80e93e65bec4aca5", "f83b9a228c78038e", "69dd802980f7ed94", "f13b137d81206d90", "48d63ceae1139125"], "E3": ["573fc820f0bb1d2d", "23b5c8fa253b1917", "c5431d8e4ce39ec2", "00d92c5d2bc8a2aa", "86a79d459eb86c88"], "E4": ["6d801edcba5e94d6", "d3fb6dcb1df74c8b", "cb3940aca3acee5e", "c7bc0491ff4ff183", "4175a1e0f9a10928", "5f3f3b56cda6b4c7", "dc317a87082c41b4", "ee190b5252b327ca", "39904987e756e8d8"], "E5": ["7d386ad666447136", "0668e85398094153", "de91055c1d0d1e82", "f1974322c0c83e4f"], "E6": ["aeeacff36d49a9aa", "48b97451edbc351d", "8263928223bb77cc", "fc4d26202f839af5"], "E7": ["513826c7acdbc4ce", "2c1175465d764149", "f6f184de2d54dfef", "e0368f0f482c7969", "ef1f934246bf5993"], "E8": ["b66839b13720b612", "c433b84d3aab7242", "faada782c498b519", "296712f0dc6fb8c1"], "E9": ["b05f2aa629955076", "ae9f003b424c2745", "e5ef40101569b646", "7fec0ade7a049f0d", "832cc3f477536c03"], "E10": ["023fdc187df538f0", "17143ccc14d737c3", "8d89a304c6541841", "8a8e3367af361785"], "E11": ["1ac82b19116d316b", "0a51e2031056dbf6", "6c76ffeb656b3ffc", "9c143805b839f119"], "E12": ["c26e363a105a32f8", "986677df7ec7be1e", "ff270b90ef6d599e", "9b29836ccdd7ea2d", "901186c253af3bd4", "6002116adbcc87da"]}
_TOPIC = {"A1": "8.1", "A2": "8", "A3": "8", "A4": "8.2.2", "A5": "8.2.3", "A6": "8.2.3", "A7": "8.4", "A8": "8.3", "A9": "8.3", "A10": "8.3", "A11": "8.5.1", "A12": "8.5.2", "A13": "8.5.3", "A14": "8.5.4", "A15": "8.6", "A16": "8.6.1", "A17": "8.6.2", "A18": "8.4.2", "A19": "8.4.2", "A20": "8.6.2", "B1": "8.2.1", "B2": "8.1", "B3": "8.2.3", "B4": "8.3", "B5": "8.3", "B6": "8.5.3", "B7": "8.5.2", "B8": "8.4.2", "B9": "8.4.2", "B10": "8.6.1", "C1": "8.2.3", "C2": "8.2.3", "C3": "8.3", "C4": "8.3", "C5": "8.5.2, 8.5.3", "C6": "8.5.4", "C7": "8.4.2", "C8": "8.4.2", "C9": "8.2.3", "C10": "8.6.1", "E1": "8.1, 8.2.3", "E2": "8.2.3", "E3": "8.2.3", "E4": "8.2.3", "E5": "8.2.3", "E6": "8.2.3", "E7": "8.2.3", "E8": "8.6.1", "E9": "8.6.2", "E10": "8.5.2, 8.5.3", "E11": "8.5.4", "E12": "8.7"}
_PARTS = [('ตอนที่ 1 ปรนัย', ['A1', 'A2', 'A3', 'A4', 'A5', 'A6', 'A7', 'A8', 'A9', 'A10', 'A11', 'A12', 'A13', 'A14', 'A15', 'A16', 'A17', 'A18', 'A19', 'A20']), ('ตอนที่ 2 ถูก/ผิด', ['B1', 'B2', 'B3', 'B4', 'B5', 'B6', 'B7', 'B8', 'B9', 'B10']), ('ตอนที่ 3 ทำนายผลลัพธ์', ['C1', 'C2', 'C3', 'C4', 'C5', 'C6', 'C7', 'C8', 'C9', 'C10']), ('ตอนที่ 5 เขียนโปรแกรม', ['E1', 'E2', 'E3', 'E4', 'E5', 'E6', 'E7', 'E8', 'E9', 'E10', 'E11', 'E12'])]
_SNIP = {"C1": "def greet(name):\n    print(\"Hello, \" + name + \"!\")\n\nr = greet(\"May\")\nprint(r)", "C2": "def absolute_value(num):\n    if num >= 0:\n        return num\n    else:\n        return -num\n\nprint(absolute_value(-4), absolute_value(2))", "C3": "def my_func():\n    x = 10\n    print(\"inside:\", x)\n\nx = 20\nmy_func()\nprint(\"outside:\", x)", "C4": "def my_func():\n    global x\n    x = 10\n    print(\"inside:\", x)\n\nx = 20\nmy_func()\nprint(\"outside:\", x)", "C5": "def printinfo(name, age=35):\n    print(name, age)\n\nprintinfo(\"miki\")\nprintinfo(age=50, name=\"Ann\")", "C6": "def printinfo(arg1, *vartuple):\n    print(arg1, vartuple)\n\nprintinfo(10)\nprintinfo(70, 60, 50)", "C7": "def changeme(mylist):\n    mylist.append(4)\n\nnums = [1, 2, 3]\nchangeme(nums)\nprint(nums)", "C8": "def reset(mylist):\n    mylist = [0]\n    print(\"inside:\", mylist)\n\nnums = [1, 2, 3]\nreset(nums)\nprint(\"outside:\", nums)", "C9": "def div_mod(a, b):\n    return a // b, a % b\n\nq, r = div_mod(17, 5)\nprint(q, r)\nprint(div_mod(9, 2))", "C10": "def fact(n):\n    if n <= 1:\n        return 1\n    return n * fact(n - 1)\n\nprint(fact(5), fact(0))"}
_TESTS = {'E1': [(([8, 2, 3, 0, 7],), {}), (([],), {}), (([-1, 1, 10],), {}), (((5, 5, 5),), {}), (([2.5, 2.5],), {})], 'E2': [(('1234abcd',), {}), (('Python',), {}), (('',), {}), (('a',), {}), (('สวัสดี',), {})], 'E3': [(([1, 2, 3, 3, 3, 3, 4, 5],), {}), ((['a', 'b', 'c', 'c', 'c', 'a', 'd', 'e', 'e'],), {}), (([3, 1, 3, 2, 1],), {}), (([],), {}), (([7, 7, 7],), {})], 'E4': [((2,), {}), ((1,), {}), ((0,), {}), ((17,), {}), ((18,), {}), ((97,), {}), ((100,), {}), ((7919,), {}), ((-7,), {})], 'E5': [(('green-red-yellow-black-white',), {}), (('b-a',), {}), (('python',), {}), (('cat-ant-bee-ant',), {})], 'E6': [(('The quick Brow Fox',), {}), (('ABC def',), {}), (('123 !?',), {}), (('PyThOn',), {})], 'E7': [((130,), {}), ((45,), {}), ((600,), {}), ((61,), {}), ((0,), {})], 'E8': [((0,), {}), ((1,), {}), ((5,), {}), ((10,), {})], 'E9': [((0,), {}), ((1,), {}), ((2,), {}), ((10,), {}), ((15,), {})], 'E10': [(('Ann',), {}), (('Bob', 'Hi'), {}), (('Cat',), {'punct': '?'}), (('Dan',), {'greeting': 'สวัสดี', 'punct': ''})], 'E11': [((1, 2, 3), {}), ((10,), {}), ((), {}), ((4, 5), {})], 'E12': [((5800,), {}), ((3800,), {}), ((100,), {}), ((250,), {}), ((0,), {}), ((1500,), {})]}
_results = {}

def _canon(v):
    if isinstance(v, (set, frozenset)):
        return "set{" + ",".join(sorted(_canon(x) for x in v)) + "}"
    if isinstance(v, dict):
        return "dict{" + ",".join(sorted(_canon(k) + ":" + _canon(x) for k, x in v.items())) + "}"
    if isinstance(v, (list, tuple)):
        return type(v).__name__ + "(" + ",".join(_canon(x) for x in v) + ")"
    if isinstance(v, bool) or v is None:
        return repr(v)
    if isinstance(v, (int, float)):
        return "num:" + format(round(float(v), 6) + 0.0, ".6f")
    return type(v).__name__ + ":" + repr(v)

def _h(qid, value):
    return hashlib.sha256(f"{_SALT}|{qid}|{value}".encode("utf-8")).hexdigest()[:16]

def _norm_text(s):
    lines = [ln.rstrip() for ln in str(s).replace("\r\n", "\n").split("\n")]
    while lines and lines[0] == "":
        lines.pop(0)
    while lines and lines[-1] == "":
        lines.pop()
    return "\n".join(lines)

def _norm_choice(prefix, ans):
    a = str(ans).strip().upper()
    if prefix == "B":
        if a in ("T", "TRUE", "ถูก", "ท", "1"): return "T"
        if a in ("F", "FALSE", "ผิด", "ผ", "0"): return "F"
    return a

def grade(part, answers):
    """ตรวจคำตอบตอนที่ 1 (part='A') และตอนที่ 2 (part='B')"""
    n_ok, n_all = 0, 0
    for n in sorted(answers):
        qid = f"{part}{n}"
        if qid not in _KEY:
            continue
        n_all += 1
        a = _norm_choice(part, answers[n])
        if a == "":
            print(f"ข้อ {n:>2}: ⚪ ยังไม่ได้ตอบ")
            _results[qid] = False
            continue
        ok = _h(qid, a) == _KEY[qid]
        _results[qid] = ok
        n_ok += ok
        print(f"ข้อ {n:>2}: {'✅ ถูกต้อง' if ok else '❌ ยังไม่ถูก  (ทบทวนหัวข้อ ' + _TOPIC[qid] + ')'}")
    print("-" * 40)
    print(f"ได้ {n_ok} / {n_all} คะแนน")

def predict(qid, prediction):
    """ตอนที่ 3: เปรียบเทียบผลลัพธ์ที่ทำนายกับผลลัพธ์จริง (บันทึกคะแนนจากการตรวจครั้งแรก)"""
    pred = _norm_text(prediction)
    if pred == "":
        print("⚪ ยังไม่ได้เขียนคำตอบที่ทำนายไว้ในตัวแปร")
        return
    buf = io.StringIO()
    with contextlib.redirect_stdout(buf):
        exec(_SNIP[qid], {})
    actual = _norm_text(buf.getvalue())
    ok = _h(qid, pred) == _KEY[qid]
    first = qid not in _results
    if first:
        _results[qid] = ok
    print("✅ ทำนายถูกต้อง" if ok else "❌ ทำนายไม่ตรงกับผลลัพธ์จริง")
    print("\nผลลัพธ์ที่ทำนาย:")
    print(pred)
    print("\nผลลัพธ์จริงจากการรันโค้ด:")
    print(actual)
    if not ok:
        print("\n(สังเกตเครื่องหมาย ช่องว่าง วงเล็บ และเครื่องหมายคำพูดให้ดี — ทบทวนหัวข้อ " + _TOPIC[qid] + ")")
    if not first:
        print("\nℹ️ คะแนนข้อนี้บันทึกจากการตรวจครั้งแรกไปแล้ว")

def check(qid, value):
    """ตอนที่ 5: ตรวจค่าที่ได้จากโปรแกรม"""
    ok = _h(qid, _canon(value)) == _KEY[qid]
    _results[qid] = ok
    if ok:
        print("✅ ถูกต้อง เยี่ยมมาก!")
    else:
        print("❌ ยังไม่ถูกต้อง ค่าที่ได้คือ:")
        print(repr(value))
        print("ตรวจสอบค่าและชนิดข้อมูลของแต่ละตัวแปรอีกครั้ง (ทบทวนหัวข้อ " + _TOPIC[qid] + ")")

try:
    from IPython.core.magic import register_cell_magic
    from IPython import get_ipython

    @register_cell_magic
    def check_output(line, cell):
        """ตอนที่ 5: รันโค้ดในเซลล์ แล้วตรวจข้อความที่แสดงผลด้วย print()"""
        qid = line.strip()
        buf = io.StringIO()
        try:
            with contextlib.redirect_stdout(buf):
                exec(cell, get_ipython().user_ns)
        except Exception as e:
            print(buf.getvalue())
            print("❌ เกิดข้อผิดพลาด:", type(e).__name__, e)
            _results[qid] = False
            return
        out = buf.getvalue()
        print(out if out.strip() else "(ไม่มีผลลัพธ์แสดงออกมา)", end="" if out.endswith("\n") else "\n")
        ok = _h(qid, _norm_text(out)) == _KEY[qid]
        _results[qid] = ok
        print("-" * 40)
        print("✅ ผลลัพธ์ถูกต้อง เยี่ยมมาก!" if ok else
              "❌ ผลลัพธ์ยังไม่ตรงตามที่โจทย์กำหนด — ตรวจสอบจำนวนบรรทัด ช่องว่าง และข้อความ (ทบทวนหัวข้อ " + _TOPIC[qid] + ")")
except Exception:
    pass

def check_func(qid, func):
    """ตอนที่ 5: ทดสอบฟังก์ชันด้วยข้อมูลทดสอบหลายชุด"""
    if not callable(func):
        print("❌ ยังไม่ได้สร้างฟังก์ชัน")
        _results[qid] = False
        return
    tests = _TESTS[qid]
    n_ok = 0
    for k, (args, kwargs) in enumerate(tests):
        call = func.__name__ + "(" + ", ".join([repr(a) for a in args] +
                                              [f"{kk}={vv!r}" for kk, vv in kwargs.items()]) + ")"
        try:
            got = func(*copy.deepcopy(args), **copy.deepcopy(kwargs))
        except Exception as e:
            print(f"❌ {call}  →  เกิดข้อผิดพลาด {type(e).__name__}: {e}")
            continue
        ok = _h(f"{qid}#{k}", _canon(got)) == _KEY[qid][k]
        n_ok += ok
        print(f"{'✅' if ok else '❌'} {call}  →  {got!r}")
    _results[qid] = n_ok == len(tests)
    print("-" * 40)
    if n_ok == len(tests):
        print(f"✅ ผ่านการทดสอบครบทั้ง {len(tests)} กรณี เยี่ยมมาก!")
    else:
        print(f"❌ ผ่าน {n_ok} จาก {len(tests)} กรณี — ตรวจสอบกรณีที่ไม่ผ่าน (ทบทวนหัวข้อ " + _TOPIC[qid] + ")")
        print("   อย่าลืมว่าฟังก์ชันต้อง return ผลลัพธ์ ไม่ใช่ print")

def summary():
    """สรุปผลการประเมินตนเอง"""
    total_ok, total_all, review = 0, 0, []
    print(f"{'ส่วน':<28}{'คะแนน':>10}")
    print("=" * 38)
    for name, qids in _PARTS:
        ok = sum(1 for q in qids if _results.get(q))
        total_ok += ok
        total_all += len(qids)
        print(f"{name:<28}{ok:>5} / {len(qids):<3}")
        review += [q for q in qids if not _results.get(q)]
    print("=" * 38)
    pct = 100 * total_ok / total_all
    print(f"{'รวม (ส่วนที่ตรวจอัตโนมัติ)':<28}{total_ok:>5} / {total_all:<3}  ({pct:.1f}%)")
    print()
    if pct >= 80:
        print("🌟 ดีมาก — เข้าใจเนื้อหาบทที่ 8 เป็นอย่างดี พร้อมเรียนบทถัดไป")
    elif pct >= 60:
        print("👍 ดี — เข้าใจเนื้อหาส่วนใหญ่ ควรทบทวนหัวข้อที่ยังทำไม่ได้")
    else:
        print("📖 ควรทบทวนบทเรียนอีกครั้ง แล้วกลับมาทำแบบฝึกหัดใหม่")
    topics = sorted({t.strip() for q in review for t in _TOPIC[q].split(",")},
                    key=lambda s: [int(x) for x in s.split(".")])
    if topics:
        print("\nหัวข้อที่ควรทบทวน:", ", ".join(topics))
    print("\n* ตอนที่ 4 และตอนที่ 6 ให้ตรวจสอบกับผลลัพธ์ที่โจทย์กำหนด/ผู้สอน")

print("✅ ระบบตรวจคำตอบพร้อมใช้งาน — เริ่มทำแบบฝึกหัดได้เลย")
````

## ตอนที่ 3 อ่านโค้ดและทำนายผลลัพธ์ (ข้อละ 1 คะแนน)
