### วัตถุประสงค์
หลังทำแบบฝึกหัดนี้ นักศึกษาสามารถ
1. อธิบายความแตกต่างระหว่างข้อผิดพลาดทางไวยากรณ์ (Syntax Errors) และสิ่งผิดปรกติขณะทำงาน (Exceptions) ได้ (หัวข้อ 11.1)
2. อ่านข้อความ Traceback จำแนกชนิดของ Built-in Exceptions และเข้าใจลำดับชั้นของ Exception ได้ (หัวข้อ 11.2)

5. สร้างสิ่งผิดปรกติด้วยคำสั่ง `raise` และสร้างคลาส Exception เองเบื้องต้นได้ (หัวข้อ 11.3.6)

````python
#@title ▶ รันเซลล์นี้ก่อนเริ่มทำแบบฝึกหัด (ระบบตรวจคำตอบอัตโนมัติ — ไม่ต้องแก้ไข)
import hashlib, io, contextlib, copy, os

os.makedirs("data", exist_ok=True)
with open("data/ch11_sample.txt", "w", encoding="utf-8") as _fp:
    _fp.write("สวัสดีไพธอน 3\nบรรทัดที่สอง\n")

_SALT = 'py3-ch11'
_KEY = {"A1": "29b67db76de22464", "A2": "03b4aef328f683b4", "A3": "976c496f54df6121", "A4": "40da977f342daa16", "A5": "59174450f1a91296", "A6": "baa3090db91fe8f0", "A7": "5b1162e801469f93", "A8": "f6c7c6e623a134a1", "A9": "673db771d61a8ffc", "A10": "da9757e42933419b", "A11": "16d5ad34adedcee6", "A12": "f5fa537679a526af", "A13": "b3f4151d627cc2bc", "A14": "b033d82be0b53803", "A15": "e1f3b686d9935de1", "A16": "2207d1d54b6e4928", "A17": "64cdbfe35d61fbc0", "A18": "ef70893ae4507405", "A19": "26aa0c650ab14485", "A20": "599763e0c954ccaf", "B1": "33aee8126126af63", "B2": "2c0ee5945caabb9c", "B3": "413ef86a11b0ceeb", "B4": "67b01c7856595109", "B5": "80cb9fb3e5c8b785", "B6": "ccadb5412b04557a", "B7": "256f427a658b1dff", "B8": "1380d507fe8e728e", "B9": "49bf581d13591ac2", "B10": "91a33c350779b7c1", "C1": "1304fc823f6735d8", "C2": "1b411c8a47e798c4", "C3": "5672cedc95e96156", "C4": "706e26cdd2a05a05", "C5": "aedd3185cfe78d16", "C6": "a7c66b172929b636", "C7": "5828dec491d75d0f", "C8": "ac3a142e6da9573f", "C9": "84b2bc4607de1192", "C10": "f47057537e6bef1e", "E1": ["b9f48ff5de663cb6", "234e1d578eace3d3", "c00162f6c4313d32", "528dacf8542dc992", "2ab6085a26bfcaa0"], "E2": ["78c107b25dc93d2b", "d782c6d252e22e55", "0e739a4de8ca4cc5", "3a6eddfbdad1b762", "c2541fd3fc71d084"], "E3": ["4749884b95bcb959", "0cbe2fec8c63f223", "42ffc2b509e40f7b", "4d27d5d98ed201a4", "c639bd05b6d33b64", "5194659f7ae0c78d"], "E4": ["361ead80dce58dfb", "15dc03b4329b89b1", "82d2b90c6cc1be56", "9e33fb0b562532e4", "18241978f1944bda", "d0d8e3d63bbfd81e", "a838235f23ead3c2"], "E5": ["a98202c9804e7edb", "4a785ac541602d7f", "b7e74aea877a02eb", "9a9b908d28e6282e"], "E6": ["19402454b6847e37", "0a4a174c83d23bd4", "d031d8f53228549f", "48c0c9fefebbc040", "ae9e954b66d438e2", "d0850ec19f873d54"], "E7": ["af3ef37af535b94a", "83e8b9b063e8b3b5"], "E8": ["a9363582d1ca614b", "3100fa2cecf55284"], "E9": ["2dcbb0309303c86b", "4709d462a1474dae", "0b239092e9cef678", "a3d2d1797b31738d"], "E10": ["ceecf03296443329", "f83420cecf1f04a4", "e0f728ee0aeeae91", "fee019adf5983c14", "7461618ddf68163a", "533c40e7f01483d1", "1b64dcf6a13e3e32"], "E11": ["0af8efc440593bea", "b9469acdff391d7a", "62a06625370d9155", "49375dfd9c223a71", "0bd64f77a12312a6"], "E12": ["d5d78529802b3447", "d868488948afc428", "cf0180f57610e8eb", "ea9d4b6ee7d8c909", "8d3d4bcd7ae138a1"]}
_TOPIC = {"A1": "11.1.1", "A2": "11.1.2", "A3": "11.2", "A4": "11.1.2, 11.2", "A5": "11.2", "A6": "11.2", "A7": "11.2", "A8": "11.2", "A9": "11.2", "A10": "11.2", "A11": "11.2", "A12": "11.3.1", "A13": "11.3.1", "A14": "11.3.2", "A15": "11.3.2", "A16": "11.3.3", "A17": "11.3.4", "A18": "11.3.5", "A19": "11.3.6", "A20": "11.3.6", "B1": "11.1.1, 11.2", "B2": "11.2", "B3": "11.2", "B4": "11.2", "B5": "11.3.2", "B6": "11.2, 11.3.3", "B7": "11.3.5", "B8": "11.3, 11.3.5", "B9": "11.3.3", "B10": "11.3.6", "C1": "11.3.1", "C2": "11.3.2", "C3": "11.2, 11.3.2", "C4": "11.2, 11.3.2", "C5": "11.3.4", "C6": "11.3.4, 11.3.5", "C7": "11.3.5", "C8": "11.3.2", "C9": "11.3.6", "C10": "11.3.6", "E1": "11.3.1", "E2": "11.3.1, 11.3.2", "E3": "11.2, 11.3.1", "E4": "11.2, 11.3.3, 11.3.6", "E5": "11.3.2, 11.3.4", "E6": "11.3.3", "E7": "11.3.2", "E8": "11.3.4", "E9": "11.3.4, 11.3.5", "E10": "11.3.6", "E11": "11.3.4, 11.3.6", "E12": "11.3.6"}
_PARTS = [('ตอนที่ 1 ปรนัย', ['A1', 'A2', 'A3', 'A4', 'A5', 'A6', 'A7', 'A8', 'A9', 'A10', 'A11', 'A12', 'A13', 'A14', 'A15', 'A16', 'A17', 'A18', 'A19', 'A20']), ('ตอนที่ 2 ถูก/ผิด', ['B1', 'B2', 'B3', 'B4', 'B5', 'B6', 'B7', 'B8', 'B9', 'B10']), ('ตอนที่ 3 ทำนายผลลัพธ์', ['C1', 'C2', 'C3', 'C4', 'C5', 'C6', 'C7', 'C8', 'C9', 'C10']), ('ตอนที่ 5 เขียนโปรแกรม', ['E1', 'E2', 'E3', 'E4', 'E5', 'E6', 'E7', 'E8', 'E9', 'E10', 'E11', 'E12'])]
_SNIP = {"C1": "try:\n    print(\"A\")\n    x = int(\"12x\")\n    print(\"B\")\nexcept ValueError:\n    print(\"C\")\nprint(\"D\")", "C2": "data = [\"25\", \"0\", \"abc\"]\nfor s in data:\n    try:\n        res = 100 // int(s)\n        print(\"OK:\", res)\n    except ValueError:\n        print(\"ERR: Value\")\n    except ZeroDivisionError:\n        print(\"ERR: Zero\")", "C3": "items = [10, 20]\ninfo = {\"a\": 1}\nfor k in [1, 5, \"a\", \"b\"]:\n    try:\n        val = items[k] if isinstance(k, int) else info[k]\n        print(\"Found:\", val)\n    except LookupError as e:\n        print(\"Missing:\", type(e).__name__)", "C4": "try:\n    open(\"no_such_file_12345.txt\", \"r\")\nexcept OSError:\n    print(\"Caught OSError\")\nexcept FileNotFoundError:\n    print(\"Caught FileNotFoundError\")", "C5": "for s in [\"10\", \"x\"]:\n    try:\n        n = int(s)\n    except ValueError:\n        print(s, \"-> bad\")\n    else:\n        print(s, \"-> good\", n * 2)", "C6": "def trace(x):\n    log = []\n    try:\n        log.append(\"T1\")\n        y = 10 // int(x)\n        log.append(f\"T2:{y}\")\n    except ValueError:\n        log.append(\"V\")\n    except ZeroDivisionError:\n        log.append(\"Z\")\n    else:\n        log.append(\"E\")\n    finally:\n        log.append(\"F\")\n    return \"-\".join(log)\n\nprint(trace(\"5\"))\nprint(trace(\"0\"))\nprint(trace(\"hi\"))", "C7": "def calc(s):\n    try:\n        n = int(s)\n    except ValueError:\n        print(\"E\", end=\" \")\n        return -1\n    else:\n        print(\"L\", end=\" \")\n        return n * 10\n    finally:\n        print(\"F\", end=\" \")\n\nr1 = calc(\"4\")\nprint(r1)\nr2 = calc(\"x\")\nprint(r2)", "C8": "import math\n\nfor v in [4, 0, -2, \"5\"]:\n    try:\n        print(\"fact:\", math.factorial(v))\n    except TypeError:\n        print(v, \"-> TypeError\")\n    except ValueError:\n        print(v, \"-> ValueError\")", "C9": "def check_positive(n):\n    if not isinstance(n, int):\n        raise TypeError(\"not int\")\n    if n <= 0:\n        raise ValueError(\"not positive\")\n    return n * 2\n\nfor x in [5, -3, \"5\"]:\n    try:\n        print(check_positive(x))\n    except (TypeError, ValueError) as e:\n        print(f\"{type(e).__name__}: {e}\")", "C10": "class OutOfStockError(Exception):\n    pass\n\ndef buy(stock, qty):\n    if qty > stock:\n        raise OutOfStockError(f\"need {qty}, have {stock}\")\n    return stock - qty\n\nfor q in [3, 8]:\n    try:\n        print(\"left:\", buy(5, q))\n    except OutOfStockError as e:\n        print(\"error:\", e)"}
_TESTS = {'E1': [(('42',), {}), ((' -7 ',), {}), (('abc',), {}), (('3.5', -1), {}), ((None, 99), {})], 'E2': [((10, 4), {}), ((-15, 3), {}), ((10, 0), {}), (('10', 2), {}), ((10, None), {})], 'E3': [(([10, 20, 30], 1), {}), (([10, 20, 30], 10), {}), (({'a': 1, 'b': 2}, 'b'), {}), (({'a': 1, 'b': 2}, 'c'), {}), (('Python', -1), {}), (('Python', 20, 'N/A'), {})], 'E4': [(('10', '+', '5'), {}), ((10, '/', 4), {}), (('10', '/', '0'), {}), (('10', '%', 0), {}), (('abc', '+', 5), {}), ((10, '^', 2), {}), ((None, '*', 3), {})], 'E5': [((['80', '90', 'ขาด', '70.5', ''],), {}), (([100, '50', -10, 120, None, 'abc'],), {}), ((['a', 'b', None],), {}), ((['75.5', 84.5, '90'],), {})], 'E6': [(('Anan,85',), {}), (('  Bua , 92 ',), {}), (('Chai,eighty',), {}), (('Dan',), {}), (('Eak,10,20',), {}), ((None,), {})], 'E7': [(([5, 0, -3, 'apple', None],), {}), (([4, '4', [2], -1],), {})], 'E8': [(('data/ch11_sample.txt',), {}), (('data/no_such_file_999.txt',), {})], 'E9': [(('10', '2'), {}), (('10', '0'), {}), (('abc', '2'), {}), ((15, 4), {})], 'E10': [((20,), {}), ((0,), {}), ((150,), {}), ((-1,), {}), ((200,), {}), (('20',), {}), ((20.5,), {})], 'E11': [(('สมชาย', 20), {}), (('สมหญิง', -5), {}), (('สมศรี', 180), {}), (('สมปอง', 'ยี่สิบ'), {}), (('สมหมาย', 19.5), {})], 'E12': [((1000, 300), {}), ((1000, 1000), {}), ((1000, 1500), {}), ((1000, 0), {}), ((1000, -200), {})]}
_results = {}

def _canon(v):
    try:
        import numpy as _np
        if isinstance(v, _np.ndarray):
            kind = {"u": "i"}.get(v.dtype.kind, v.dtype.kind)
            return "ndarray" + str(v.shape) + kind + ":" + _canon(v.tolist())
        if isinstance(v, _np.generic):
            v = v.item()
    except ImportError:
        pass
    if isinstance(v, (set, frozenset)):
        return "set{" + ",".join(sorted(_canon(x) for x in v)) + "}"
    if isinstance(v, dict):
        return "dict{" + ",".join(sorted(_canon(k) + ":" + _canon(x) for k, x in v.items())) + "}"
    if isinstance(v, (list, tuple)):
        return type(v).__name__ + "(" + ",".join(_canon(x) for x in v) + ")"
    if isinstance(v, bool) or v is None:
        return repr(v)
    if isinstance(v, (int, float)):
        return "num:" + format(round(float(v), 6) + 0.0, ".6f")
    return type(v).__name__ + ":" + repr(v)

def _h(qid, value):
    return hashlib.sha256(f"{_SALT}|{qid}|{value}".encode("utf-8")).hexdigest()[:16]

def _norm_text(s):
    lines = [ln.rstrip() for ln in str(s).replace("\r\n", "\n").split("\n")]
    while lines and lines[0] == "":
        lines.pop(0)
    while lines and lines[-1] == "":
        lines.pop()
    return "\n".join(lines)

def _norm_choice(prefix, ans):
    a = str(ans).strip().upper()
    if prefix == "B":
        if a in ("T", "TRUE", "ถูก", "ท", "1"): return "T"
        if a in ("F", "FALSE", "ผิด", "ผ", "0"): return "F"
    return a

def grade(part, answers):
    """ตรวจคำตอบตอนที่ 1 (part='A') และตอนที่ 2 (part='B')"""
    n_ok, n_all = 0, 0
    for n in sorted(answers):
        qid = f"{part}{n}"
        if qid not in _KEY:
            continue
        n_all += 1
        a = _norm_choice(part, answers[n])
        if a == "":
            print(f"ข้อ {n:>2}: ⚪ ยังไม่ได้ตอบ")
            _results[qid] = False
            continue
        ok = _h(qid, a) == _KEY[qid]
        _results[qid] = ok
        n_ok += ok
        print(f"ข้อ {n:>2}: {'✅ ถูกต้อง' if ok else '❌ ยังไม่ถูก  (ทบทวนหัวข้อ ' + _TOPIC[qid] + ')'}")
    print("-" * 40)
    print(f"ได้ {n_ok} / {n_all} คะแนน")

def predict(qid, prediction):
    """ตอนที่ 3: เปรียบเทียบผลลัพธ์ที่ทำนายกับผลลัพธ์จริง (บันทึกคะแนนจากการตรวจครั้งแรก)"""
    pred = _norm_text(prediction)
    if pred == "":
        print("⚪ ยังไม่ได้เขียนคำตอบที่ทำนายไว้ในตัวแปร")
        return
    buf = io.StringIO()
    with contextlib.redirect_stdout(buf):
        exec(_SNIP[qid], {})
    actual = _norm_text(buf.getvalue())
    ok = _h(qid, pred) == _KEY[qid]
    first = qid not in _results
    if first:
        _results[qid] = ok
    print("✅ ทำนายถูกต้อง" if ok else "❌ ทำนายไม่ตรงกับผลลัพธ์จริง")
    print("\nผลลัพธ์ที่ทำนาย:")
    print(pred)
    print("\nผลลัพธ์จริงจากการรันโค้ด:")
    print(actual)
    if not ok:
        print("\n(สังเกตเครื่องหมาย ช่องว่าง วงเล็บ และเครื่องหมายคำพูดให้ดี — ทบทวนหัวข้อ " + _TOPIC[qid] + ")")
    if not first:
        print("\nℹ️ คะแนนข้อนี้บันทึกจากการตรวจครั้งแรกไปแล้ว")

def check(qid, value):
    """ตอนที่ 5: ตรวจค่าที่ได้จากโปรแกรม"""
    ok = _h(qid, _canon(value)) == _KEY[qid]
    _results[qid] = ok
    if ok:
        print("✅ ถูกต้อง เยี่ยมมาก!")
    else:
        print("❌ ยังไม่ถูกต้อง ค่าที่ได้คือ:")
        print(repr(value))
        print("ตรวจสอบค่าและชนิดข้อมูลของแต่ละตัวแปรอีกครั้ง (ทบทวนหัวข้อ " + _TOPIC[qid] + ")")

def check_func(qid, func):
    """ตอนที่ 5: ทดสอบฟังก์ชันด้วยข้อมูลทดสอบหลายชุด"""
    if not callable(func):
        print("❌ ยังไม่ได้สร้างฟังก์ชัน")
        _results[qid] = False
        return
    tests = _TESTS[qid]
    n_ok = 0
    for k, (args, kwargs) in enumerate(tests):
        call = func.__name__ + "(" + ", ".join([repr(a) for a in args] +
                                              [f"{kk}={vv!r}" for kk, vv in kwargs.items()]) + ")"
        try:
            got = func(*copy.deepcopy(args), **copy.deepcopy(kwargs))
            c_val = _canon(got)
            disp = repr(got)
            raised = None
        except Exception as e:
            c_val = f"raise:{type(e).__name__}:{e}"
            disp = f"raise {type(e).__name__}({str(e)!r})"
            raised = e
        ok = _h(f"{qid}#{k}", c_val) == _KEY[qid][k]
        if not ok and raised is not None:
            print(f"❌ {call}  →  เกิดข้อผิดพลาด {type(raised).__name__}: {raised}")
            continue
        n_ok += ok
        print(f"{'✅' if ok else '❌'} {call}  →  {disp}")
    _results[qid] = n_ok == len(tests)
    print("-" * 40)
    if n_ok == len(tests):
        print(f"✅ ผ่านการทดสอบครบทั้ง {len(tests)} กรณี เยี่ยมมาก!")
    else:
        print(f"❌ ผ่าน {n_ok} จาก {len(tests)} กรณี — ตรวจสอบกรณีที่ไม่ผ่าน (ทบทวนหัวข้อ " + _TOPIC[qid] + ")")
        print("   อย่าลืมว่าฟังก์ชันต้อง return ผลลัพธ์ (หรือ raise ตามที่โจทย์กำหนด) ไม่ใช่เพียงแค่ print")

def summary():
    """สรุปผลการประเมินตนเอง"""
    total_ok, total_all, review = 0, 0, []
    print(f"{'ส่วน':<28}{'คะแนน':>10}")
    print("=" * 38)
    for name, qids in _PARTS:
        ok = sum(1 for q in qids if _results.get(q))
        total_ok += ok
        total_all += len(qids)
        print(f"{name:<28}{ok:>5} / {len(qids):<3}")
        review += [q for q in qids if not _results.get(q)]
    print("=" * 38)
    pct = 100 * total_ok / total_all
    print(f"{'รวม (ส่วนที่ตรวจอัตโนมัติ)':<28}{total_ok:>5} / {total_all:<3}  ({pct:.1f}%)")
    print()
    if pct >= 80:
        print("🌟 ดีมาก — เข้าใจเนื้อหาบทที่ 11 เป็นอย่างดี พร้อมเรียนบทถัดไป")
    elif pct >= 60:
        print("👍 ดี — เข้าใจเนื้อหาส่วนใหญ่ ควรทบทวนหัวข้อที่ยังทำไม่ได้")
    else:
        print("📖 ควรทบทวนบทเรียนอีกครั้ง แล้วกลับมาทำแบบฝึกหัดใหม่")
    topics = sorted({t.strip() for q in review for t in _TOPIC[q].split(",")},
                    key=lambda s: [int(x) for x in s.split(".")])
    if topics:
        print("\nหัวข้อที่ควรทบทวน:", ", ".join(topics))
    print("\n* ตอนที่ 4 และตอนที่ 6 ให้ตรวจสอบกับผลลัพธ์ที่โจทย์กำหนด/ผู้สอน")

print("✅ ระบบตรวจคำตอบพร้อมใช้งาน — เริ่มทำแบบฝึกหัดได้เลย")
````
