````python
#@title ▶ รันเซลล์นี้ก่อนเริ่มทำแบบฝึกหัด (ระบบตรวจคำตอบอัตโนมัติ — ไม่ต้องแก้ไข)
import hashlib, io, contextlib

_SALT = "py3-ch6"
_KEY = {"A1": "9cfa0b4c6e581105", "A2": "4c8038da27d9219e", "A3": "6ab3b57e1a207bd4", "A4": "72edefdb5eeae5f0", "A5": "d6b6b2e16bc0e73b", "A6": "c8dabc7e9048c552", "A7": "0626b3f957d00c39", "A8": "766338a7bf7e2eff", "A9": "34d3e84e0c861e56", "A10": "1d8197e0bce8c9a6", "A11": "76d6a33626740230", "A12": "157aebe4171fcc62", "A13": "9f45d1d410dc9ebd", "A14": "1f25c4dbcfe23d1e", "A15": "bb66d235c66232b4", "A16": "2d46314f118d1409", "A17": "223e8f34c2b8160d", "A18": "26b2af83dc38715d", "A19": "15802e268d4fbb7a", "A20": "2791dffbe3ad00cc", "B1": "443b6a938a014e5e", "B2": "aaa9fd2c2f63b887", "B3": "00e8be99d67c1151", "B4": "6f2cb43f9dea7a65", "B5": "7824f06a975e1faf", "B6": "96ca3ce57f0c8b6e", "B7": "0a743c5ccc895690", "B8": "261a8987017e9b32", "B9": "306c9475276a483e", "B10": "50bbfc0befd58d4b", "C1": "7548b4c62c34d892", "C2": "dca6c0b1895ea454", "C3": "f73888c026405677", "C4": "714545bde36979de", "C5": "aa3283e5c658a75f", "C6": "87219ed2ceec5ca6", "C7": "f5c4682be577f799", "C8": "09d9180abaaf96c3", "C9": "3c7091fbbf21282d", "C10": "fdcd87c6f41ddb6a", "E1": "9fb6c032d0505034", "E3": "6ac14a1b38ecfb00", "E4": "908916d4fa1a5a42", "E5": "b34f8908166cfee3", "E6": "c853adcb79305408", "E7": "d578861531bc759a", "E8": "add17102a04221f3", "E9": "3ec2de959f8ab65b", "E10": "be7cc329bb09dbdd", "E11": "1180110c47444992"}
_TOPIC = {"A1": "6.1.1", "A2": "6.1.3", "A3": "6.1.2", "A4": "6.2.1", "A5": "6.2.2", "A6": "6.2.2", "A7": "6.2.2", "A8": "6.3.2", "A9": "6.2.3", "A10": "6.2.1, 6.2.3", "A11": "6.2.4", "A12": "6.2.4", "A13": "6.2.4", "A14": "6.2.5", "A15": "6.2.6", "A16": "6.2.7", "A17": "6.2.7", "A18": "6.3.1", "A19": "6.3.3", "A20": "6.3.4", "B1": "6.1.2", "B2": "6.2.2", "B3": "6.2.4", "B4": "6.2.4", "B5": "6.2.6", "B6": "6.2.5", "B7": "6.3.3", "B8": "6.2.7", "B9": "6.2.1", "B10": "6.3.1", "C1": "6.2.2", "C2": "6.3", "C3": "6.3", "C4": "6.3", "C5": "6.2.1", "C6": "6.2.3", "C7": "6.2.4", "C8": "6.2.5", "C9": "6.2.7", "C10": "6.3.2", "E1": "6.2.2", "E2": "4.2.1, 6.2.2", "E3": "6.2.2", "E4": "6.2.2", "E5": "5.4.1.2, 6.3", "E6": "6.2.2", "E7": "6.2.3, 6.2.4", "E8": "6.2.2, 6.2.3, 6.2.4", "E9": "6.2.5, 6.2.6", "E10": "6.2.7", "E11": "6.2.1"}
_PARTS = [('ตอนที่ 1 ปรนัย', ['A1', 'A2', 'A3', 'A4', 'A5', 'A6', 'A7', 'A8', 'A9', 'A10', 'A11', 'A12', 'A13', 'A14', 'A15', 'A16', 'A17', 'A18', 'A19', 'A20']), ('ตอนที่ 2 ถูก/ผิด', ['B1', 'B2', 'B3', 'B4', 'B5', 'B6', 'B7', 'B8', 'B9', 'B10']), ('ตอนที่ 3 ทำนายผลลัพธ์', ['C1', 'C2', 'C3', 'C4', 'C5', 'C6', 'C7', 'C8', 'C9', 'C10']), ('ตอนที่ 5 เขียนโปรแกรม', ['E1', 'E3', 'E4', 'E5', 'E6', 'E7', 'E8', 'E9', 'E10', 'E11'])]
_SNIP = {"C1": "a = 21\nb = 10\nprint(a / b, a // b, a % b)", "C2": "print(17 / 2 * 3 + 2)\nprint(2 + 17 / 2 * 3)", "C3": "print(19 % 4 + 15 / 2 * 3)\nprint((15 + 6) - 10 * 4)", "C4": "print(17 / 2 % 2 * 3 ** 3)", "C5": "x = 10\nx += 5\nx *= 2\nx //= 4\nx %= 4\nprint(x)", "C6": "x = 10\ny = 12\nprint(x > y, x <= y, x == y, x != y)", "C7": "print(True and not False, False or False, not (3 > 2 and 2 > 1))", "C8": "s = 'Hello world'\nprint('H' in s, 'h' in s, 'ell' in s, 'ele' not in s)", "C9": "a = 60\nb = 13\nprint(a & b, a | b, a ^ b, ~a, a << 2, a >> 2)", "C10": "print(2 ** 3 ** 2, (2 ** 3) ** 2, -2 ** 2)"}
_results = {}

def _canon(v):
    if isinstance(v, (set, frozenset)):
        return "set{" + ",".join(sorted(_canon(x) for x in v)) + "}"
    if isinstance(v, dict):
        return "dict{" + ",".join(sorted(_canon(k) + ":" + _canon(x) for k, x in v.items())) + "}"
    if isinstance(v, (list, tuple)):
        return type(v).__name__ + "(" + ",".join(_canon(x) for x in v) + ")"
    if isinstance(v, bool) or v is None:
        return repr(v)
    if isinstance(v, (int, float)):
        return "num:" + format(round(float(v), 6) + 0.0, ".6f")
    return type(v).__name__ + ":" + repr(v)

def _h(qid, value):
    return hashlib.sha256(f"{_SALT}|{qid}|{value}".encode("utf-8")).hexdigest()[:16]

def _norm_text(s):
    lines = [ln.rstrip() for ln in str(s).replace("\r\n", "\n").split("\n")]
    while lines and lines[0] == "":
        lines.pop(0)
    while lines and lines[-1] == "":
        lines.pop()
    return "\n".join(lines)

def _norm_choice(prefix, ans):
    a = str(ans).strip().upper()
    if prefix == "B":
        if a in ("T", "TRUE", "ถูก", "ท", "1"): return "T"
        if a in ("F", "FALSE", "ผิด", "ผ", "0"): return "F"
    return a

def grade(part, answers):
    """ตรวจคำตอบตอนที่ 1 (part='A') และตอนที่ 2 (part='B')"""
    n_ok, n_all = 0, 0
    for n in sorted(answers):
        qid = f"{part}{n}"
        if qid not in _KEY:
            continue
        n_all += 1
        a = _norm_choice(part, answers[n])
        if a == "":
            print(f"ข้อ {n:>2}: ⚪ ยังไม่ได้ตอบ")
            _results[qid] = False
            continue
        ok = _h(qid, a) == _KEY[qid]
        _results[qid] = ok
        n_ok += ok
        print(f"ข้อ {n:>2}: {'✅ ถูกต้อง' if ok else '❌ ยังไม่ถูก  (ทบทวนหัวข้อ ' + _TOPIC[qid] + ')'}")
    print("-" * 40)
    print(f"ได้ {n_ok} / {n_all} คะแนน")

def predict(qid, prediction):
    """ตอนที่ 3: เปรียบเทียบผลลัพธ์ที่ทำนายกับผลลัพธ์จริง (บันทึกคะแนนจากการตรวจครั้งแรก)"""
    pred = _norm_text(prediction)
    if pred == "":
        print("⚪ ยังไม่ได้เขียนคำตอบที่ทำนายไว้ในตัวแปร")
        return
    buf = io.StringIO()
    with contextlib.redirect_stdout(buf):
        exec(_SNIP[qid], {})
    actual = _norm_text(buf.getvalue())
    ok = _h(qid, pred) == _KEY[qid]
    first = qid not in _results
    if first:
        _results[qid] = ok
    print("✅ ทำนายถูกต้อง" if ok else "❌ ทำนายไม่ตรงกับผลลัพธ์จริง")
    print("\nผลลัพธ์ที่ทำนาย:")
    print(pred)
    print("\nผลลัพธ์จริงจากการรันโค้ด:")
    print(actual)
    if not ok:
        print("\n(สังเกตเครื่องหมาย ช่องว่าง วงเล็บ และเครื่องหมายคำพูดให้ดี — ทบทวนหัวข้อ " + _TOPIC[qid] + ")")
    if not first:
        print("\nℹ️ คะแนนข้อนี้บันทึกจากการตรวจครั้งแรกไปแล้ว")

def check(qid, value):
    """ตอนที่ 5: ตรวจค่าที่ได้จากโปรแกรม"""
    ok = _h(qid, _canon(value)) == _KEY[qid]
    _results[qid] = ok
    if ok:
        print("✅ ถูกต้อง เยี่ยมมาก!")
    else:
        print("❌ ยังไม่ถูกต้อง ค่าที่ได้คือ:")
        print(repr(value))
        print("ตรวจสอบค่าและชนิดข้อมูลของแต่ละตัวแปรอีกครั้ง (ทบทวนหัวข้อ " + _TOPIC[qid] + ")")

def summary():
    """สรุปผลการประเมินตนเอง"""
    total_ok, total_all, review = 0, 0, []
    print(f"{'ส่วน':<28}{'คะแนน':>10}")
    print("=" * 38)
    for name, qids in _PARTS:
        ok = sum(1 for q in qids if _results.get(q))
        total_ok += ok
        total_all += len(qids)
        print(f"{name:<28}{ok:>5} / {len(qids):<3}")
        review += [q for q in qids if not _results.get(q)]
    print("=" * 38)
    pct = 100 * total_ok / total_all
    print(f"{'รวม (ส่วนที่ตรวจอัตโนมัติ)':<28}{total_ok:>5} / {total_all:<3}  ({pct:.1f}%)")
    print()
    if pct >= 80:
        print("🌟 ดีมาก — เข้าใจเนื้อหาบทที่ 6 เป็นอย่างดี พร้อมเรียนบทถัดไป")
    elif pct >= 60:
        print("👍 ดี — เข้าใจเนื้อหาส่วนใหญ่ ควรทบทวนหัวข้อที่ยังทำไม่ได้")
    else:
        print("📖 ควรทบทวนบทเรียนอีกครั้ง แล้วกลับมาทำแบบฝึกหัดใหม่")
    topics = sorted({t.strip() for q in review for t in _TOPIC[q].split(",")},
                    key=lambda s: [int(x) for x in s.split(".")])
    if topics:
        print("\nหัวข้อที่ควรทบทวน:", ", ".join(topics))
    print("\n* ตอนที่ 4 และตอนที่ 6 ให้ตรวจสอบกับผลลัพธ์ที่โจทย์กำหนด/ผู้สอน")

print("✅ ระบบตรวจคำตอบพร้อมใช้งาน — เริ่มทำแบบฝึกหัดได้เลย")
````

## ตอนที่ 5 เขียนโปรแกรม
เขียนโปรแกรมตามโจทย์ โดยเก็บผลลัพธ์ไว้ในตัวแปรที่กำหนด แล้วรันเซลล์ — ระบบจะตรวจด้วย `check(...)` ให้อัตโนมัติ

**ข้อ E1.** (คำถามท้ายบทข้อ 2 แบบไม่รับค่า) กำหนด `x = 17` และ `y = 5` ให้คำนวณผลลัพธ์ของตัวดำเนินการทางคณิตศาสตร์ทั้ง 7 ตัว เก็บไว้ในตัวแปรตามตาราง แล้วแสดงผลในรูปแบบ `17 + 5 = 22`

| ตัวแปร | `add` | `sub` | `mul` | `div` | `fdiv` | `mod` | `power` |
|---|---|---|---|---|---|---|---|
| ตัวดำเนินการ | `+` | `-` | `*` | `/` | `//` | `%` | `**` |

````python
x = 17
y = 5

add = x + y
sub = x - y
mul = x * y
div = x / y
fdiv = x // y
mod = x % y
power = x ** y

print("{} + {} = {}".format(x, y, add))
print("{} - {} = {}".format(x, y, sub))
print("{} * {} = {}".format(x, y, mul))
print("{} / {} = {}".format(x, y, div))
print("{} // {} = {}".format(x, y, fdiv))
print("{} % {} = {}".format(x, y, mod))
print("{} ** {} = {}".format(x, y, power))
check("E1", (add, sub, mul, div, fdiv, mod, power))
````

**ข้อ E2.** (คำถามท้ายบทข้อ 2) เขียนโปรแกรมรับค่า `x` และ `y` จากแป้นพิมพ์ (แปลงเป็นจำนวนจริงด้วย `float()`) แล้วแสดงผลลัพธ์ของการบวก ลบ คูณ หาร
> 💡 การให้ผู้ใช้เลือกตัวดำเนินการเองต้องใช้คำสั่ง `if` ซึ่งจะได้เรียนในบทที่ 7

ตัวอย่างการทำงาน
```
x = 8
y = 2
8.0 + 2.0 = 10.0
8.0 - 2.0 = 6.0
8.0 * 2.0 = 16.0
8.0 / 2.0 = 4.0
```

````python
x = float(input("x = "))
y = float(input("y = "))

print("{} + {} = {}".format(x, y, x + y))
print("{} - {} = {}".format(x, y, x - y))
print("{} * {} = {}".format(x, y, x * y))
print("{} / {} = {}".format(x, y, x / y))
````

**ข้อ E3.** (คำถามท้ายบทข้อ 3 — ตู้ ATM) ตู้ ATM มีธนบัตร 1000, 500 และ 100 บาท ให้คำนวณจำนวนธนบัตรแต่ละชนิด โดยจ่ายธนบัตรมูลค่าสูงสุดก่อน สำหรับจำนวนเงิน `amount = 3800`
- ใช้ตัวดำเนินการ `//` และ `%` (ยังไม่ต้องใช้คำสั่ง if)
- เก็บผลไว้ใน `n1000`, `n500`, `n100` → คำตอบที่ถูกต้องคือ 3, 1, 3

> 💡 ลองเปลี่ยน `amount` เป็นค่าอื่น เช่น 4700 แล้วตรวจสอบว่าโปรแกรมยังให้ผลถูกต้อง (ระบบตรวจเฉพาะ 3800)

````python
amount = 3800

n1000 = amount // 1000
remain = amount % 1000
n500 = remain // 500
remain = remain % 500
n100 = remain // 100

print("1000 บาท: {} ใบ".format(n1000))
print(" 500 บาท: {} ใบ".format(n500))
print(" 100 บาท: {} ใบ".format(n100))
check("E3", (n1000, n500, n100))
````

1. `total_in` = ความยาวทั้งหมดเป็นนิ้ว
2. `feet` = จำนวนฟุตเต็ม (ใช้ `//`) → `5`
3. `inch` = จำนวนนิ้วที่เหลือ (ใช้ `%`)
4. `text` = ข้อความ `5 ft 10.87 in` (นิ้วทศนิยม 2 ตำแหน่ง ด้วย `format()`)

````python
cm = 180

total_in = cm / 2.54
feet = int(total_in // 12)
inch = total_in % 12
text = "{} ft {:.2f} in".format(feet, inch)

print(text)
check("E4", (total_in, feet, inch, text))
````

**ข้อ E5.** (คำถามท้ายบทข้อ 5) แปลงสมการต่อไปนี้เป็นนิพจน์ภาษาไพธอน แล้วเก็บผลไว้ในตัวแปรที่กำหนด (ใช้ `math.pi` และ `math.sqrt()`)

| ข้อ | สมการ | ค่าที่กำหนด | ตัวแปรผลลัพธ์ |
|---|---|---|---|
| 5.1 | $a = \pi r^2$ | r = 7 | `ans1` |
| 5.2 | $c^2 = a^2 + b^2$ (หาค่า c) | a = 3, b = 4 | `ans2` |
| 5.3 | $x = \dfrac{\sqrt{b^2 - 4ac}}{2a}$ | a = 1, b = 5, c = 6 | `ans3` |
| 5.4 | $a = \dfrac{(b \times 2) \times d}{c}$ | b = 3, c = 4, d = 8 | `ans4` |
| 5.5 | $a = b^2 \times (5 + 2)$ | b = 3 | `ans5` |

````python
import math

# 5.1
r = 7
ans1 = math.pi * r ** 2

# 5.2
a, b = 3, 4
ans2 = math.sqrt(a ** 2 + b ** 2)

# 5.3
a, b, c = 1, 5, 6
ans3 = math.sqrt(b ** 2 - 4 * a * c) / (2 * a)

# 5.4
b, c, d = 3, 4, 8
ans4 = (b * 2) * d / c

# 5.5
b = 3
ans5 = b ** 2 * (5 + 2)

print(ans1, ans2, ans3, ans4, ans5, sep="\n")
check("E5", (ans1, ans2, ans3, ans4, ans5))
````

**ข้อ E6.** (แปลงวินาทีเป็นชั่วโมง นาที วินาที) กำหนด `seconds = 7384`
1. คำนวณ `h`, `m`, `s` ด้วย `//` และ `%` → 2 ชั่วโมง 3 นาที 4 วินาที
2. สร้าง `time_str` ในรูปแบบ `2:03:04` (นาทีและวินาทีแสดง 2 หลัก เติม 0 ข้างหน้า)
> 💡 ใบ้: ทบทวนการกำหนดความกว้างฟิลด์และอักขระเติม เช่น `{:0>2}` จากบทที่ 4

````python
seconds = 7384

h = seconds // 3600
m = seconds % 3600 // 60
s = seconds % 60
time_str = "{}:{:0>2}:{:0>2}".format(h, m, s)

print(time_str)
check("E6", (h, m, s, time_str))
````

**ข้อ E7.** (นิพจน์ทางตรรกะ) กำหนด `age = 20`, `score = 75`, `member = False` ให้เขียนนิพจน์และเก็บผล (True/False) ไว้ในตัวแปร
- `is_adult` — อายุตั้งแต่ 18 ปีขึ้นไป
- `passed` — คะแนนอยู่ระหว่าง 50 ถึง 100 (รวม 50 และ 100) ใช้ `and`
- `discount` — เป็นสมาชิก **หรือ** อายุน้อยกว่า 12 ปี
- `not_member` — ไม่ได้เป็นสมาชิก ใช้ `not`

````python
age = 20
score = 75
member = False

is_adult = age >= 18
passed = score >= 50 and score <= 100
discount = member or age < 12
not_member = not member

print(is_adult, passed, discount, not_member)
check("E7", (is_adult, passed, discount, not_member))
````

**ข้อ E8.** (ปีอธิกสุรทิน) ปี ค.ศ. ใดเป็นปีอธิกสุรทินเมื่อ **หารด้วย 4 ลงตัวแต่หารด้วย 100 ไม่ลงตัว** หรือ **หารด้วย 400 ลงตัว**
ให้เขียนนิพจน์ทางตรรกะเดียวกันนี้ตรวจสอบปี 2024, 1900, 2000 และ 2023 เก็บผลไว้ใน `leap1` ถึง `leap4`

````python
year = 2024
leap1 = (year % 4 == 0 and year % 100 != 0) or year % 400 == 0

year = 1900
leap2 = (year % 4 == 0 and year % 100 != 0) or year % 400 == 0

year = 2000
leap3 = (year % 4 == 0 and year % 100 != 0) or year % 400 == 0

year = 2023
leap4 = (year % 4 == 0 and year % 100 != 0) or year % 400 == 0

print(leap1, leap2, leap3, leap4)
check("E8", (leap1, leap2, leap3, leap4))
````

**ข้อ E9.** (ตัวดำเนินการสมาชิกและเอกลักษณ์) จากข้อมูลที่กำหนด ให้เขียนนิพจน์เก็บผลไว้ในตัวแปร
- `has_banana` — มี `"banana"` อยู่ใน `fruits`
- `no_cherry` — **ไม่มี** `"cherry"` อยู่ใน `fruits` (ใช้ `not in`)
- `has_age` — มีคีย์ `"age"` อยู่ใน `info`
- `same_value` — `x` และ `y` มีค่าเท่ากัน
- `same_object` — `x` และ `y` เป็นออบเจกต์เดียวกัน
- `alias` — `x` และ `z` เป็นออบเจกต์เดียวกัน

````python
fruits = ["apple", "banana", "mango"]
info = {"name": "Ann", "age": 20}
x = [1, 2]
y = [1, 2]
z = x

has_banana = "banana" in fruits
no_cherry = "cherry" not in fruits
has_age = "age" in info
same_value = x == y
same_object = x is y
alias = x is z

print(has_banana, no_cherry, has_age, same_value, same_object, alias)
check("E9", (has_banana, no_cherry, has_age, same_value, same_object, alias))
````

**ข้อ E10.** (ตัวดำเนินการระดับบิต) กำหนด `a = 60` และ `b = 13` ให้คำนวณ
- `r_and` = a AND b, `r_or` = a OR b, `r_xor` = a XOR b, `r_not` = NOT a
- `r_left` = a เลื่อนบิตซ้าย 2 ตำแหน่ง, `r_right` = a เลื่อนบิตขวา 2 ตำแหน่ง
- `bits` = ค่าของ `r_and` ในรูปเลขฐานสอง 8 หลัก เติม 0 ข้างหน้า → `'00001100'` (ใช้ `format()` กับรหัส `b`)

````python
a = 60
b = 13

r_and = a & b
r_or = a | b
r_xor = a ^ b
r_not = ~a
r_left = a << 2
r_right = a >> 2
bits = "{:08b}".format(r_and)

print(r_and, r_or, r_xor, r_not, r_left, r_right, bits)
check("E10", (r_and, r_or, r_xor, r_not, r_left, r_right, bits))
````

**ข้อ E11.** (ตัวดำเนินการกำหนดค่าแบบผสม) บัญชีเงินฝากมียอดเริ่มต้น `balance = 1000` ให้ใช้ **ตัวดำเนินการกำหนดค่าแบบผสม** (`+=`, `-=`, `*=`) ทำรายการตามลำดับ
1. ฝากเงิน 500 บาท
2. ถอนเงิน 200 บาท
3. ได้รับดอกเบี้ย 5% (คูณด้วย 1.05)

ยอดเงินสุดท้ายต้องเป็น `1365.0`

````python
balance = 1000

balance += 500
balance -= 200
balance *= 1.05

print(balance)
check("E11", balance)
````

**ข้อ 1.** (คำถามท้ายบทข้อ 1) จงอธิบายความแตกต่างของคำว่า **คำสั่ง (Statement)** และ **นิพจน์ (Expression)** พร้อมยกตัวอย่าง

✏️ **แนวคำตอบ:** นิพจน์คือส่วนที่ประกอบด้วยตัวดำเนินการและตัวถูกดำเนินการ ซึ่ง **คำนวณแล้วได้ค่า** เช่น `a + b`, `x > 5` ส่วนคำสั่งคือหน่วยของโปรแกรมที่ **สั่งให้ทำงาน** อย่างใดอย่างหนึ่ง ซึ่งอาจมีนิพจน์อยู่ภายใน เช่น `c = a + b` (คำสั่งกำหนดค่า), `print(c)`, `if x > 5:`

✏️ **แนวคำตอบ:** นิพจน์เลขคณิตให้ผลเป็นตัวเลข (int/float) เช่น `6 * 2`, `a / b` ส่วนนิพจน์ทางตรรกะให้ผลเป็น True หรือ False ใช้ตัวดำเนินการสัมพันธ์และตรรกะ เช่น `10 > 9`, `a == b and b != 0`

**ข้อ 3.** จงอธิบายความแตกต่างของตัวดำเนินการ `/`, `//` และ `%` โดยยกตัวอย่างกับ 17 และ 5

✏️ **แนวคำตอบ:** `/` หารได้ผลเป็นทศนิยม `17 / 5 = 3.4`, `//` หารแล้วปัดลงเป็นจำนวนเต็ม `17 // 5 = 3`, `%` หาเศษจากการหาร `17 % 5 = 2`

**ข้อ 4.** ตัวดำเนินการ `==` และ `is` แตกต่างกันอย่างไร

✏️ **แนวคำตอบ:** `==` ตรวจสอบว่า **ค่า** เท่ากันหรือไม่ ส่วน `is` ตรวจสอบว่าเป็น **ออบเจกต์เดียวกันในหน่วยความจำ** หรือไม่ เช่น `[1, 2] == [1, 2]` เป็น True แต่ลิสต์สองลิสต์ที่สร้างแยกกันจะ `is` เป็น False

**ข้อ 5.** จงอธิบายลำดับการดำเนินการ (Operator Precedence) และคำนวณ `x = 19 % 4 + 15 / 2 * 3` ทีละขั้น

✏️ **แนวคำตอบ:** วงเล็บ → ยกกำลัง (ขวาไปซ้าย) → คูณ หาร หารปัดเศษ หาเศษ (ซ้ายไปขวา) → บวก ลบ (ซ้ายไปขวา) — `19 % 4` = 3, `15 / 2` = 7.5, `7.5 * 3` = 22.5, `3 + 22.5` = **25.5**

**ข้อ 6.** เมื่อใดที่ `and` ให้ผลเป็น True และเมื่อใดที่ `or` ให้ผลเป็น False จงเขียนตารางค่าความจริงของ `and` และ `or`

✏️ **แนวคำตอบ:** `and` เป็น True เฉพาะเมื่อทุกตัวเป็น True ส่วน `or` เป็น False เฉพาะเมื่อทุกตัวเป็น False — (T,T): and=T, or=T · (T,F): and=F, or=T · (F,T): and=F, or=T · (F,F): and=F, or=F
