````python
#@title ▶ รันเซลล์นี้ก่อนเริ่มทำแบบฝึกหัด (ระบบตรวจคำตอบอัตโนมัติ — ไม่ต้องแก้ไข)
import hashlib, io, contextlib, copy

_SALT = "py3-ch9"
_KEY = {"A1": "5697c5708390283f", "A2": "179551a91f53fb54", "A3": "e70966c489ee540f", "A4": "e259c4d7ec74c7c9", "A5": "eb6b43d036fd36b6", "A6": "0b0d907a282cf823", "A7": "48dea394276dac9c", "A8": "8d07639215161c98", "A9": "b15a9136cd1f8466", "A10": "b93e37a7f8673af4", "A11": "ffb6ef61b840d323", "A12": "794cf3b3711b8d13", "A13": "119b30e3c7bb384d", "A14": "a47587482369999f", "A15": "d5a1c78e5c9d6e54", "A16": "838552364ba13430", "A17": "b53ed931265acd1b", "A18": "25f2e2873c0879e4", "A19": "0ad593a479e829f0", "A20": "5185d6efcf90fa0b", "B1": "1ff3170215dc9dd1", "B2": "c8e8ccd2297b070c", "B3": "524e4b9c19d2c33e", "B4": "13e0bc2a6a48f2e2", "B5": "63e6a1ad18d6ae73", "B6": "bf0e27c076b9b743", "B7": "b1854bb96201c262", "B8": "0a6a04f5ae5db64b", "B9": "12559280400e54ea", "B10": "d7fe29d875e6323d", "C1": "79255c9d5d77a7cd", "C2": "afd7d41282b92ba2", "C3": "a22c78c5ece87ebf", "C4": "697db5b63e77f025", "C5": "799651384861612e", "C6": "2a2be4d140325a97", "C7": "f856edb5aafaf0ad", "C8": "8ec239247106e7f6", "C9": "e4913de94b837ac7", "C10": "956fe6ad825994f8", "E1": "c4680364d04bb950", "E2": "3a1eabbc6948d31c", "E3": "1b21d3124819fac4", "E4": "83a8ff72ccb04776", "E5": "f062eb99951db363", "E6": "fa92ffa8ba7b47ec", "E7": "3a89e3a5081dbf11", "E8": "33fad849e4611d26", "E9": "b524774829865152", "E10": "b54ecdda19a4c083", "E11": "6c7acae667857ab2", "E12": "b70a6ee18aa3e5cb", "E13": ["10ca01807a0b1320", "36524824a073698d", "4d6fe0d9da2978f2", "75906be07746f475"]}
_TOPIC = {"A1": "9.1", "A2": "9.1.3", "A3": "9.2, 9.3.1", "A4": "9.3.1", "A5": "9.3.2", "A6": "9.3.2", "A7": "9.3.2", "A8": "9.3.2", "A9": "9.4", "A10": "9.4", "A11": "9.5", "A12": "9.6", "A13": "9.6", "A14": "9.6", "A15": "9.7", "A16": "9.8.1", "A17": "9.8.2", "A18": "9.10.3", "A19": "9.10.3", "A20": "9.11.1", "B1": "9.1.1", "B2": "9.1.2", "B3": "9.3.2", "B4": "9.4", "B5": "9.4", "B6": "9.6", "B7": "9.8.1", "B8": "9.11.1", "B9": "9.10.2", "B10": "9.11.2", "C1": "9.2, 9.4", "C2": "9.3.2", "C3": "9.6", "C4": "9.5, 9.6", "C5": "9.6", "C6": "9.6, 9.7", "C7": "9.8.1", "C8": "9.9", "C9": "9.10", "C10": "9.11.1", "E1": "9.1.3, 9.3.2", "E2": "9.3.2", "E3": "9.7", "E4": "9.4, 9.9", "E5": "9.5", "E6": "9.9", "E7": "9.9", "E8": "9.10", "E9": "9.3.2", "E10": "9.10", "E11": "9.9", "E12": "9.8", "E13": "9.4, 9.10"}
_PARTS = [('ตอนที่ 1 ปรนัย', ['A1', 'A2', 'A3', 'A4', 'A5', 'A6', 'A7', 'A8', 'A9', 'A10', 'A11', 'A12', 'A13', 'A14', 'A15', 'A16', 'A17', 'A18', 'A19', 'A20']), ('ตอนที่ 2 ถูก/ผิด', ['B1', 'B2', 'B3', 'B4', 'B5', 'B6', 'B7', 'B8', 'B9', 'B10']), ('ตอนที่ 3 ทำนายผลลัพธ์', ['C1', 'C2', 'C3', 'C4', 'C5', 'C6', 'C7', 'C8', 'C9', 'C10']), ('ตอนที่ 5 เขียนโปรแกรม', ['E1', 'E2', 'E3', 'E4', 'E5', 'E6', 'E7', 'E8', 'E9', 'E10', 'E11', 'E12', 'E13'])]
_SNIP = {"C1": "import numpy as np\n\narr = np.array([1, 2, 3])\nlst = [1, 2, 3]\nprint(arr * 2)\nprint(lst * 2)", "C2": "import numpy as np\n\na = np.array([[1, 2, 3], [4, 5, 6]])\nprint(a.shape, a.ndim, a.size)", "C3": "import numpy as np\n\nx = np.arange(10)\nprint(x[:5])\nprint(x[4:7])\nprint(x[1::2])", "C4": "import numpy as np\n\nx2 = np.array([[3, 5, 2, 4], [7, 6, 8, 8], [1, 6, 7, 7]])\nprint(x2[2, -1], x2[0, 0])\nprint(x2[:, 0])\nprint(x2[1])", "C5": "import numpy as np\n\nx2 = np.array([[3, 5, 2, 4], [7, 6, 8, 8], [1, 6, 7, 7]])\nprint(x2[:2, :3])", "C6": "import numpy as np\n\ngrid = np.arange(1, 10).reshape((3, 3))\nprint(grid[1:, 1:])", "C7": "import numpy as np\n\nx = np.array([1, 2, 3])\ny = np.array([3, 2, 1])\nprint(np.concatenate([x, y]))\nprint(np.vstack([x, y]))", "C8": "import numpy as np\n\nx = np.arange(4)\nprint(x // 2, x % 2, -x)", "C9": "import numpy as np\n\nM = np.array([[1, 5, 2], [4, 3, 6]])\nprint(M.sum(), M.min(axis=0), M.max(axis=1))", "C10": "import numpy as np\n\nx = np.array([2, 1, 4, 3, 5])\ni = np.argsort(x)\nprint(i)\nprint(x[i])"}
_TESTS = {'E13': [(([1, 2, 3],), {}), (([10, 20, 30, 40, 50],), {}), (([5, 0, 10],), {}), (([-4, 0, 4, 8],), {})]}
_results = {}

def _canon(v):
    try:
        import numpy as _np
        if isinstance(v, _np.ndarray):
            kind = {"u": "i"}.get(v.dtype.kind, v.dtype.kind)
            return "ndarray" + str(v.shape) + kind + ":" + _canon(v.tolist())
        if isinstance(v, _np.generic):
            v = v.item()
    except ImportError:
        pass
    if isinstance(v, (set, frozenset)):
        return "set{" + ",".join(sorted(_canon(x) for x in v)) + "}"
    if isinstance(v, dict):
        return "dict{" + ",".join(sorted(_canon(k) + ":" + _canon(x) for k, x in v.items())) + "}"
    if isinstance(v, (list, tuple)):
        return type(v).__name__ + "(" + ",".join(_canon(x) for x in v) + ")"
    if isinstance(v, bool) or v is None:
        return repr(v)
    if isinstance(v, (int, float)):
        return "num:" + format(round(float(v), 6) + 0.0, ".6f")
    return type(v).__name__ + ":" + repr(v)

def _h(qid, value):
    return hashlib.sha256(f"{_SALT}|{qid}|{value}".encode("utf-8")).hexdigest()[:16]

def _norm_text(s):
    lines = [ln.rstrip() for ln in str(s).replace("\r\n", "\n").split("\n")]
    while lines and lines[0] == "":
        lines.pop(0)
    while lines and lines[-1] == "":
        lines.pop()
    return "\n".join(lines)

def _norm_choice(prefix, ans):
    a = str(ans).strip().upper()
    if prefix == "B":
        if a in ("T", "TRUE", "ถูก", "ท", "1"): return "T"
        if a in ("F", "FALSE", "ผิด", "ผ", "0"): return "F"
    return a

def grade(part, answers):
    """ตรวจคำตอบตอนที่ 1 (part='A') และตอนที่ 2 (part='B')"""
    n_ok, n_all = 0, 0
    for n in sorted(answers):
        qid = f"{part}{n}"
        if qid not in _KEY:
            continue
        n_all += 1
        a = _norm_choice(part, answers[n])
        if a == "":
            print(f"ข้อ {n:>2}: ⚪ ยังไม่ได้ตอบ")
            _results[qid] = False
            continue
        ok = _h(qid, a) == _KEY[qid]
        _results[qid] = ok
        n_ok += ok
        print(f"ข้อ {n:>2}: {'✅ ถูกต้อง' if ok else '❌ ยังไม่ถูก  (ทบทวนหัวข้อ ' + _TOPIC[qid] + ')'}")
    print("-" * 40)
    print(f"ได้ {n_ok} / {n_all} คะแนน")

def predict(qid, prediction):
    """ตอนที่ 3: เปรียบเทียบผลลัพธ์ที่ทำนายกับผลลัพธ์จริง (บันทึกคะแนนจากการตรวจครั้งแรก)"""
    pred = _norm_text(prediction)
    if pred == "":
        print("⚪ ยังไม่ได้เขียนคำตอบที่ทำนายไว้ในตัวแปร")
        return
    buf = io.StringIO()
    with contextlib.redirect_stdout(buf):
        exec(_SNIP[qid], {})
    actual = _norm_text(buf.getvalue())
    ok = _h(qid, pred) == _KEY[qid]
    first = qid not in _results
    if first:
        _results[qid] = ok
    print("✅ ทำนายถูกต้อง" if ok else "❌ ทำนายไม่ตรงกับผลลัพธ์จริง")
    print("\nผลลัพธ์ที่ทำนาย:")
    print(pred)
    print("\nผลลัพธ์จริงจากการรันโค้ด:")
    print(actual)
    if not ok:
        print("\n(สังเกตเครื่องหมาย ช่องว่าง วงเล็บ และเครื่องหมายคำพูดให้ดี — ทบทวนหัวข้อ " + _TOPIC[qid] + ")")
    if not first:
        print("\nℹ️ คะแนนข้อนี้บันทึกจากการตรวจครั้งแรกไปแล้ว")

def check(qid, value):
    """ตอนที่ 5: ตรวจค่าที่ได้จากโปรแกรม"""
    ok = _h(qid, _canon(value)) == _KEY[qid]
    _results[qid] = ok
    if ok:
        print("✅ ถูกต้อง เยี่ยมมาก!")
    else:
        print("❌ ยังไม่ถูกต้อง ค่าที่ได้คือ:")
        print(repr(value))
        print("ตรวจสอบค่าและชนิดข้อมูลของแต่ละตัวแปรอีกครั้ง (ทบทวนหัวข้อ " + _TOPIC[qid] + ")")

try:
    from IPython.core.magic import register_cell_magic
    from IPython import get_ipython

    @register_cell_magic
    def check_output(line, cell):
        """ตอนที่ 5: รันโค้ดในเซลล์ แล้วตรวจข้อความที่แสดงผลด้วย print()"""
        qid = line.strip()
        buf = io.StringIO()
        try:
            with contextlib.redirect_stdout(buf):
                exec(cell, get_ipython().user_ns)
        except Exception as e:
            print(buf.getvalue())
            print("❌ เกิดข้อผิดพลาด:", type(e).__name__, e)
            _results[qid] = False
            return
        out = buf.getvalue()
        print(out if out.strip() else "(ไม่มีผลลัพธ์แสดงออกมา)", end="" if out.endswith("\n") else "\n")
        ok = _h(qid, _norm_text(out)) == _KEY[qid]
        _results[qid] = ok
        print("-" * 40)
        print("✅ ผลลัพธ์ถูกต้อง เยี่ยมมาก!" if ok else
              "❌ ผลลัพธ์ยังไม่ตรงตามที่โจทย์กำหนด — ตรวจสอบจำนวนบรรทัด ช่องว่าง และข้อความ (ทบทวนหัวข้อ " + _TOPIC[qid] + ")")
except Exception:
    pass

def check_func(qid, func):
    """ตอนที่ 5: ทดสอบฟังก์ชันด้วยข้อมูลทดสอบหลายชุด"""
    if not callable(func):
        print("❌ ยังไม่ได้สร้างฟังก์ชัน")
        _results[qid] = False
        return
    tests = _TESTS[qid]
    n_ok = 0
    for k, (args, kwargs) in enumerate(tests):
        call = func.__name__ + "(" + ", ".join([repr(a) for a in args] +
                                              [f"{kk}={vv!r}" for kk, vv in kwargs.items()]) + ")"
        try:
            got = func(*copy.deepcopy(args), **copy.deepcopy(kwargs))
        except Exception as e:
            print(f"❌ {call}  →  เกิดข้อผิดพลาด {type(e).__name__}: {e}")
            continue
        ok = _h(f"{qid}#{k}", _canon(got)) == _KEY[qid][k]
        n_ok += ok
        print(f"{'✅' if ok else '❌'} {call}  →  {got!r}")
    _results[qid] = n_ok == len(tests)
    print("-" * 40)
    if n_ok == len(tests):
        print(f"✅ ผ่านการทดสอบครบทั้ง {len(tests)} กรณี เยี่ยมมาก!")
    else:
        print(f"❌ ผ่าน {n_ok} จาก {len(tests)} กรณี — ตรวจสอบกรณีที่ไม่ผ่าน (ทบทวนหัวข้อ " + _TOPIC[qid] + ")")
        print("   อย่าลืมว่าฟังก์ชันต้อง return ผลลัพธ์ ไม่ใช่ print")

def summary():
    """สรุปผลการประเมินตนเอง"""
    total_ok, total_all, review = 0, 0, []
    print(f"{'ส่วน':<28}{'คะแนน':>10}")
    print("=" * 38)
    for name, qids in _PARTS:
        ok = sum(1 for q in qids if _results.get(q))
        total_ok += ok
        total_all += len(qids)
        print(f"{name:<28}{ok:>5} / {len(qids):<3}")
        review += [q for q in qids if not _results.get(q)]
    print("=" * 38)
    pct = 100 * total_ok / total_all
    print(f"{'รวม (ส่วนที่ตรวจอัตโนมัติ)':<28}{total_ok:>5} / {total_all:<3}  ({pct:.1f}%)")
    print()
    if pct >= 80:
        print("🌟 ดีมาก — เข้าใจเนื้อหาบทที่ 9 เป็นอย่างดี พร้อมเรียนบทถัดไป")
    elif pct >= 60:
        print("👍 ดี — เข้าใจเนื้อหาส่วนใหญ่ ควรทบทวนหัวข้อที่ยังทำไม่ได้")
    else:
        print("📖 ควรทบทวนบทเรียนอีกครั้ง แล้วกลับมาทำแบบฝึกหัดใหม่")
    topics = sorted({t.strip() for q in review for t in _TOPIC[q].split(",")},
                    key=lambda s: [int(x) for x in s.split(".")])
    if topics:
        print("\nหัวข้อที่ควรทบทวน:", ", ".join(topics))
    print("\n* ตอนที่ 4 และตอนที่ 6 ให้ตรวจสอบกับผลลัพธ์ที่โจทย์กำหนด/ผู้สอน")

print("✅ ระบบตรวจคำตอบพร้อมใช้งาน — เริ่มทำแบบฝึกหัดได้เลย")
````
