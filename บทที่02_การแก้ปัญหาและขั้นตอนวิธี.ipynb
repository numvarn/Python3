## 2.1 การแก้ปัญหาด้วยคอมพิวเตอร์

ปัญหาที่แก้ได้ด้วยคอมพิวเตอร์ **ไม่จำเป็นต้องเป็นปัญหาทางคณิตศาสตร์เสมอไป** แต่เนื่องจากโปรแกรมต้องระบุขั้นตอนการทำงานและเงื่อนไขอย่างชัดเจน ผู้เรียนจึงต้อง **เข้าใจปัญหาและความต้องการให้ชัดเจนก่อน** แล้วจึงพัฒนาขั้นตอนวิธีที่ใช้งานได้ การแก้ปัญหาด้วยคอมพิวเตอร์มี 7 ขั้นตอน ดังนี้

<p align="center"><img src="images/บทที่02/fsteps.png" width="900"><br><b>แผนภาพ</b> ขั้นตอนการแก้ปัญหาด้วยคอมพิวเตอร์ 7 ขั้น</p>

### ขั้นที่ 1 การนิยามปัญหา (Problem Definition)
สิ่งสำคัญที่สุดคือต้องเข้าใจให้ได้ว่า **ปัญหาคืออะไร ประเด็นหลักอยู่ที่ใด และต้องการผลลัพธ์อะไร**

### ขั้นที่ 2 การวิเคราะห์ปัญหา (Problem Analysis)
วิเคราะห์ปัญหาแล้วพัฒนา **สูตรเชิงคณิตศาสตร์** หรือวิธีที่ใช้แก้ปัญหา โดย
1. พิจารณา **ผลลัพธ์ (Output)** ก่อนว่าคืออะไร
2. หา **ข้อมูลนำเข้า (Input)** ที่ทำให้ได้ผลลัพธ์นั้น
3. คิด **สูตร/สมการ/วิธีการ** ที่ใช้แก้ปัญหา
4. แยกให้ได้ว่าข้อมูลใด **นำเข้าจากผู้ใช้** ข้อมูลใด **เกิดจากการคำนวณ** และข้อมูลใดเป็น **ค่าคงที่**

#### 📌 ตัวอย่าง: โปรแกรมคำนวณค่าจ้างพนักงานรายวัน
> ค่าแรง **40 บาทต่อชั่วโมง** ถ้าทำงานเกิน **8 ชั่วโมง** ส่วนที่เกินคิดค่าล่วงเวลา **7 บาทต่อชั่วโมง**

| ประเภทข้อมูล | รายละเอียด |
|---|---|
| ผลลัพธ์ (Output) | จำนวนเงินที่ต้องจ่ายให้พนักงาน |
| ข้อมูลนำเข้า (Input) | จำนวนชั่วโมงที่ทำงาน |
| ค่าคงที่ | 40 บาท/ชม., 8 ชม., 7 บาท/ชม. |
| ข้อมูลจากการคำนวณ | ค่าแรงปกติ, ค่าล่วงเวลา, ยอดรวม |

**สูตรเชิงคณิตศาสตร์**
- ค่าแรงพนักงาน = ชั่วโมงทำงาน × 40 *(ถ้าชั่วโมง ≤ 8)* หรือ = 8 × 40 *(ถ้าชั่วโมง > 8)*
- ค่าล่วงเวลา = (ชั่วโมงทำงาน − 8) × 7 *(ถ้าชั่วโมง > 8)* หรือ = 0 *(ถ้าชั่วโมง ≤ 8)*
- ค่าแรงพนักงานแต่ละคน = ค่าแรงพนักงาน + ค่าล่วงเวลา

### ขั้นที่ 3 การออกแบบขั้นตอนวิธีการแก้ปัญหา (Algorithm Design)
คิดหาขั้นตอนการแก้ปัญหา **ทีละขั้น** ให้แปลงเป็นคำสั่งของภาษาโปรแกรมได้ง่าย ขั้นตอนการแก้ปัญหาทีละขั้นนี้เรียกว่า **อัลกอริทึม (Algorithm)**

**อัลกอริทึมคำนวณค่าจ้างพนักงาน**
1. รับข้อมูลชั่วโมงทำงานของพนักงานจากแป้นพิมพ์
2. เปรียบเทียบจำนวนชั่วโมงทำงานกับ 8
   - ถ้า **มากกว่า 8** ให้ ค่าแรง = 8 × 40 และ ค่าล่วงเวลา = (ชั่วโมงทำงาน − 8) × 7
   - ถ้า **น้อยกว่าหรือเท่ากับ 8** ให้ ค่าแรง = ชั่วโมงทำงาน × 40 และ ค่าล่วงเวลา = 0
3. คำนวณ เงินที่ต้องจ่าย = ค่าแรง + ค่าล่วงเวลา
4. แสดงผลลัพธ์ให้ผู้ใช้ทราบ
5. จบการทำงาน

> แต่ละขั้นตอนต้องชัดเจนว่า **จะทำอะไร** และ **ได้ผลลัพธ์อะไร**

เขียนเป็นผังงานได้ดังนี้ (สัญลักษณ์ผังงานจะได้เรียนในหัวข้อ 2.4)

<p align="center"><img src="images/บทที่02/fpay.png" width="380"><br><b>ผังงาน</b> การคำนวณค่าจ้างพนักงานรายวัน</p>

### ขั้นที่ 4 การพัฒนาโปรแกรม (Program Development)
นำอัลกอริทึมแต่ละขั้นมา **แปลงเป็นคำสั่งของภาษาโปรแกรม** เช่น ภาษาไพธอน ซึ่งจะทำได้ง่ายถ้าอัลกอริทึมถูกต้องและชัดเจน

สังเกตว่าโค้ดด้านล่างมีโครงสร้างตรงกับอัลกอริทึมทีละขั้น ลองรัน แล้วป้อนจำนวนชั่วโมง

````python
# ขั้นที่ 1: รับข้อมูลชั่วโมงทำงาน
hours = float(input("จำนวนชั่วโมงที่ทำงาน: "))

# ขั้นที่ 2: เปรียบเทียบกับ 8
if hours > 8:
    wage = 8 * 40
    ot = (hours - 8) * 7
else:
    wage = hours * 40
    ot = 0

# ขั้นที่ 3: คำนวณเงินที่ต้องจ่าย
total = wage + ot

# ขั้นที่ 4: แสดงผล
print(f"ค่าแรงปกติ   = {wage:,.2f} บาท")
print(f"ค่าล่วงเวลา  = {ot:,.2f} บาท")
print(f"รวมต้องจ่าย  = {total:,.2f} บาท")
# ขั้นที่ 5: จบการทำงาน
````

### ขั้นที่ 5 การทดสอบความถูกต้อง (Testing)
ตรวจสอบการทำงานของโปรแกรมด้วย **ข้อมูลทดสอบ (Test data)** เพื่อดูผลลัพธ์หรือข้อผิดพลาดจากการคำนวณ โดย **ควรทดสอบทุกกรณีที่คาดว่าจะเกิดขึ้น** เช่น โปรแกรมค่าจ้างต้องทดสอบชั่วโมงทำงานอย่างน้อย 3 กรณี คือ **น้อยกว่า 8, เท่ากับ 8 และมากกว่า 8**

เพื่อให้ทดสอบซ้ำได้ง่าย เราจะห่อสูตรไว้ใน **ฟังก์ชัน** แล้วเปรียบเทียบผลกับค่าที่คำนวณด้วยมือ (ค่าที่คาดหวัง)

````python
def calc_pay(hours):
    """คำนวณค่าจ้างพนักงานรายวัน
    - ค่าแรง 40 บาท/ชม. ไม่เกิน 8 ชม.
    - ส่วนที่เกิน 8 ชม. คิดค่าล่วงเวลา 7 บาท/ชม.
    """
    if hours > 8:
        return 8 * 40 + (hours - 8) * 7
    return hours * 40

# ตารางข้อมูลทดสอบ: (ชั่วโมง, ค่าที่คาดหวัง ซึ่งคำนวณด้วยมือ)
test_data = [
    (5, 200),     # กรณีน้อยกว่า 8 : 5 x 40
    (8, 320),     # กรณีเท่ากับ 8  : 8 x 40  (ค่าขอบเขต สำคัญมาก!)
    (10, 334),    # กรณีมากกว่า 8  : 320 + 2 x 7
    (0, 0),       # กรณีไม่ได้ทำงาน
]

print(f"{'ชั่วโมง':>8} {'คาดหวัง':>8} {'ได้จริง':>8}  ผล")
for h, expected in test_data:
    got = calc_pay(h)
    print(f"{h:>8} {expected:>8} {got:>8}  {'✅ ผ่าน' if got == expected else '❌ ไม่ผ่าน'}")
````

### ขั้นที่ 6 การจัดทำเอกสาร (Documentation)
เป็นงานสำคัญที่นักเขียนโปรแกรมส่วนใหญ่มักละเลย ประกอบด้วย 2 แบบ
- **เอกสารคู่มือผู้ใช้ (User Manual)** อธิบายวิธีใช้งานโปรแกรม

- `""" ... """` **docstring** สำหรับอธิบายฟังก์ชัน ซึ่งเรียกดูได้ด้วย `help()`

### ขั้นที่ 7 การบำรุงรักษาและการแก้ไข (Maintenance)
การพัฒนาโปรแกรมใหม่ถือว่า **ง่ายกว่า** การบำรุงรักษาและแก้ไข และ **ต้นทุนการบำรุงรักษาจะสูงขึ้นมากหากออกแบบไม่ดี** เพราะทำให้เกิดข้อผิดพลาดเมื่อนำโปรแกรมไปใช้

ตัวอย่าง: ถ้าบริษัทปรับค่าแรงเป็น 45 บาท/ชม. โปรแกรมที่เขียนตัวเลข 40 กระจายอยู่หลายที่ต้องตามแก้ทุกจุด แต่ถ้าออกแบบให้ค่าคงที่อยู่ที่เดียว ก็แก้เพียงบรรทัดเดียว

````python
# ออกแบบให้บำรุงรักษาง่าย: รวมค่าคงที่ไว้ที่เดียว
RATE = 45          # ค่าแรงต่อชั่วโมง (ปรับใหม่)
NORMAL_HOURS = 8   # ชั่วโมงทำงานปกติ
OT_RATE = 7        # ค่าล่วงเวลาต่อชั่วโมง

def calc_pay_v2(hours):
    if hours > NORMAL_HOURS:
        return NORMAL_HOURS * RATE + (hours - NORMAL_HOURS) * OT_RATE
    return hours * RATE

for h in [5, 8, 10]:
    print(f"ทำงาน {h} ชม. ได้ {calc_pay_v2(h)} บาท")
````

## 2.2 การระบุข้อมูลเข้า ข้อมูลออก และเงื่อนไขของปัญหา

ก่อนระบุขั้นตอนวิธีที่ชัดเจนได้ ต้องวิเคราะห์และทำความเข้าใจปัญหา เพื่อให้รู้ว่ามีข้อมูลอะไรใช้ประมวลผลได้ มีเงื่อนไขอย่างไร และผลลัพธ์ที่ต้องการคืออะไร โดยแบ่งข้อมูลเป็น 2 ส่วน
1. **ข้อมูลเข้า (Input)** คือข้อมูลที่ใช้ประมวลผล
2. **ข้อมูลออก (Output)** คือข้อมูลผลลัพธ์ที่ต้องการ

นอกจากระบุว่าข้อมูลคืออะไรแล้ว ยังอาจระบุ **เงื่อนไขเพิ่มเติม** ได้ เช่น ขอบเขตของข้อมูลเข้า หรือคุณสมบัติที่ต้องการของข้อมูลออก การวิเคราะห์นี้ทำให้ข้อกำหนดของปัญหาชัดเจน ซึ่งจำเป็นต่อการออกแบบขั้นตอนวิธีที่ถูกต้อง

```
  ข้อมูลเข้า (Input)  ──►  [ ประมวลผล (ขั้นตอนวิธี) ]  ──►  ข้อมูลออก (Output)
          ▲                        ▲
     ขอบเขต/เงื่อนไข           เงื่อนไขของปัญหา
```

### 2.2.1 ปัญหาการหาตัวหารร่วมมาก (ห.ร.ม.)
**ห.ร.ม.** ของจำนวนเต็มสองจำนวน คือ **จำนวนเต็มบวกที่มีค่ามากที่สุดที่หารทั้งสองจำนวนลงตัว**

**วิธีหนึ่งที่ได้จากนิยาม:** นำ 1, 2, 3, … ไปหารทั้งสองจำนวน ไล่ไปจนถึงจำนวนที่น้อยกว่า และระหว่างนั้นจดจำค่ามากที่สุดที่หารทั้งสองลงตัว ค่าที่จดจำไว้สุดท้ายคือ ห.ร.ม.
- วิธีนี้สะดวกเมื่อตัวเลขมีค่าน้อย เช่น 21 กับ 14
- ถ้าตัวเลขมีค่ามาก เช่น 221 กับ 187 ต้องคำนวณถึง **187 ครั้ง** จึงได้คำตอบว่า ห.ร.ม. คือ **17**

**การระบุข้อมูลเข้า ข้อมูลออก และเงื่อนไข**
1. **ข้อมูลเข้า:** จำนวนเต็มบวกสองจำนวน คือ `a` และ `b`
2. **ข้อมูลออก:** จำนวนเต็มบวกหนึ่งจำนวน คือ `c` ที่มีคุณสมบัติ
   - 2.1 `c` เป็นจำนวนเต็มบวก
   - 2.2 `c` หาร `a` และ `b` ลงตัว
   - 2.3 `c` มีค่ามากที่สุดเท่าที่เป็นไปได้

````python
def gcd_brute_force(a, b, show=False):
    """หา ห.ร.ม. ตามนิยาม: ลองหารด้วย 1, 2, 3, ... จนถึงจำนวนที่น้อยกว่า"""
    best = 1
    rounds = 0
    for d in range(1, min(a, b) + 1):
        rounds += 1
        if a % d == 0 and b % d == 0:   # d หารทั้ง a และ b ลงตัว
            best = d                     # จดจำค่าที่มากที่สุดไว้
            if show:
                print(f"  {d} หาร {a} และ {b} ลงตัว -> จำไว้ best = {best}")
    return best, rounds

c, n = gcd_brute_force(21, 14, show=True)
print(f"ห.ร.ม.(21, 14) = {c}  ใช้การคำนวณ {n} ครั้ง\n")

c, n = gcd_brute_force(221, 187)
print(f"ห.ร.ม.(221, 187) = {c}  ใช้การคำนวณ {n} ครั้ง")
````

#### 🔬 (เสริม) ขั้นตอนวิธีที่ดีกว่า: วิธีของยุคลิด (Euclidean Algorithm)
ปัญหาเดียวกันมีได้หลายขั้นตอนวิธี และบางวิธีเร็วกว่ามาก วิธีของยุคลิดใช้หลักว่า **ห.ร.ม.(a, b) = ห.ร.ม.(b, a mod b)** แล้วทำซ้ำจนเศษเป็น 0

````python
def gcd_euclid(a, b, show=False):
    rounds = 0
    while b != 0:
        rounds += 1
        if show:
            print(f"  ห.ร.ม.({a}, {b}) -> {a} mod {b} = {a % b}")
        a, b = b, a % b
    return a, rounds

c, n = gcd_euclid(221, 187, show=True)
print(f"ห.ร.ม. = {c}  ใช้การคำนวณเพียง {n} ครั้ง (วิธีตามนิยามใช้ 187 ครั้ง)")

import math
print("ตรวจสอบกับ math.gcd():", math.gcd(221, 187))
````

### 2.2.2 ปัญหาคะแนนสอบ
> ครูตรวจข้อสอบของนักเรียน 40 คน และต้องการหา **คะแนนสูงสุด คะแนนต่ำสุด และคะแนนเฉลี่ย**

1. **ข้อมูลเข้า:** รายการคะแนนสอบของนักเรียน 40 คน
2. **ข้อมูลออก:** คะแนนสูงสุด คะแนนต่ำสุด คะแนนเฉลี่ย

ตัวอย่างด้านล่างสุ่มคะแนน 40 คนมาเป็นข้อมูลเข้า (ใช้ `random.seed` เพื่อให้ทุกคนได้ชุดข้อมูลเดียวกัน)

````python
import random
random.seed(2)
scores = [random.randint(20, 100) for _ in range(40)]   # ข้อมูลเข้า
print("คะแนนทั้ง 40 คน:", scores)

# ข้อมูลออก
print("คะแนนสูงสุด :", max(scores))
print("คะแนนต่ำสุด :", min(scores))
print(f"คะแนนเฉลี่ย : {sum(scores) / len(scores):.2f}")
````

### 2.2.3 การแบ่งกลุ่มทำงาน
ในหลายกรณีอาจระบุข้อมูลเข้าและข้อมูลออกได้ไม่ชัดเจนนัก แต่ **ความพยายามระบุข้อมูลทั้งสองช่วยให้เข้าใจปัญหามากขึ้น**

> นักเรียนจัดกิจกรรมวันภาษาไทย มีงานที่ต้องทำ 5 อย่าง ได้แก่ (1) จัดบอร์ดหน้าห้อง (2) จัดเตรียมงานโต้วาที (3) เป็นผู้โต้วาที 2 กลุ่ม กลุ่มละ 3 คน (4) อ่านกลอนทำนองเสนาะ (5) ร้องเพลงไทยสมัยใหม่
> นักเรียนทุกคนกรอกว่าทำงานใดได้บ้าง และอยากทำงานใดเป็นพิเศษ โดยมีเงื่อนไขว่า **คนหนึ่งไม่ควรทำงานเกิน 2 อย่าง** และ **ผู้โต้วาทีไม่ควรเป็นผู้จัดเตรียมงานโต้วาที**

1. **ข้อมูลเข้า:** รายการงานทั้งหมด และข้อมูลของนักเรียนแต่ละคนว่าทำงานใดได้บ้าง และอยากทำงานใดเป็นพิเศษ
2. **ข้อมูลออก:** ข้อมูลที่ระบุว่านักเรียนคนใดทำงานอะไร โดยมีเงื่อนไข
   - 2.1 นักเรียนควรได้ทำงานที่ตนเลือกว่าทำได้
   - 2.2 ถ้าเป็นไปได้ ควรได้ทำงานที่ตนอยากทำเป็นพิเศษ
   - 2.3 แต่ละคนควรมีงานอย่างน้อย 1 อย่าง แต่ไม่เกิน 2 อย่าง
   - 2.4 ผู้โต้วาทีไม่ควรเป็นผู้จัดเตรียมงานโต้วาที

**การวิเคราะห์ช่วยให้เห็นปัญหาที่ซ่อนอยู่:** ถ้ามีนักเรียนบางคน **ไม่ระบุงานที่ทำได้เลย** ก็จะจัดกลุ่มไม่ได้ (ขัดกับเงื่อนไข 2.3) จึงอาจเพิ่มเงื่อนไขว่า **นักเรียนทุกคนต้องเลือกงานที่ทำได้อย่างน้อย 1 อย่าง**

ตัวอย่างโปรแกรมด้านล่างแปลงข้อมูลเข้าและเงื่อนไขเป็นโค้ด เพื่อ **ตรวจสอบว่าผลการจัดกลุ่มถูกต้องตามเงื่อนไขหรือไม่**

````python
# ข้อมูลเข้า: งานที่แต่ละคนทำได้ (can) และงานที่อยากทำเป็นพิเศษ (want)
students = {
    "แก้ว":  {"can": ["บอร์ด", "กลอน"],          "want": ["กลอน"]},
    "ขวัญ":  {"can": ["เตรียมโต้วาที", "เพลง"],   "want": ["เพลง"]},
    "เคน":   {"can": ["โต้วาที", "บอร์ด"],        "want": ["โต้วาที"]},
    "งาม":   {"can": ["โต้วาที", "เตรียมโต้วาที"], "want": ["โต้วาที"]},
    "จัน":   {"can": [],                           "want": []},      # ไม่ระบุงานเลย!
}

# ข้อมูลออกที่ลองจัดไว้
assignment = {
    "แก้ว": ["กลอน"],
    "ขวัญ": ["เพลง", "เตรียมโต้วาที"],
    "เคน":  ["โต้วาที"],
    "งาม":  ["โต้วาที", "เตรียมโต้วาที"],
    "จัน":  [],
}

def check_assignment(students, assignment):
    for name, jobs in assignment.items():
        problems = []
        if any(j not in students[name]["can"] for j in jobs):
            problems.append("ได้งานที่ไม่ได้เลือกว่าทำได้ (ผิดข้อ 2.1)")
        if not 1 <= len(jobs) <= 2:
            problems.append(f"มีงาน {len(jobs)} อย่าง (ผิดข้อ 2.3)")
        if "โต้วาที" in jobs and "เตรียมโต้วาที" in jobs:
            problems.append("เป็นทั้งผู้โต้วาทีและผู้เตรียมงาน (ผิดข้อ 2.4)")
        liked = [j for j in jobs if j in students[name]["want"]]
        note = f"ได้งานที่อยากทำ {liked}" if liked else "ไม่ได้งานที่อยากทำ (ข้อ 2.2 'ถ้าเป็นไปได้')"
        status = "✅" if not problems else "❌ " + "; ".join(problems)
        print(f"{name:<5} {str(jobs):<28} {status}  | {note}")

check_assignment(students, assignment)
````

> 🤔 ลองแก้ `assignment` ให้ผ่านทุกเงื่อนไข และจะพบว่า "จัน" ทำให้ผ่านเงื่อนไข 2.3 ไม่ได้เลย นี่คือเหตุผลที่ต้องเพิ่มเงื่อนไขในข้อมูลเข้าว่า **ทุกคนต้องเลือกงานที่ทำได้อย่างน้อย 1 อย่าง**

## 2.3 การออกแบบขั้นตอนวิธี

> 📌 **ขั้นตอนวิธี (Algorithm)** หมายถึง กระบวนการแก้ปัญหาที่เข้าใจได้ มีลำดับหรือวิธีการแก้ปัญหาใดปัญหาหนึ่งอย่างเป็นขั้นเป็นตอนและชัดเจน **เมื่อนำเข้าอะไรแล้วต้องได้ผลลัพธ์อย่างไร**

ขั้นตอนวิธีมีมาตั้งแต่สมัยโบราณ เช่น ขั้นตอนวิธีการบวก ลบ คูณ หาร ที่พัฒนาโดยนักคณิตศาสตร์ชาวเปอร์เซีย (คำว่า *algorithm* มาจากชื่อ **al-Khwārizmī** นักคณิตศาสตร์ชาวเปอร์เซีย)

ขั้นตอนวิธีไม่ได้มีเฉพาะในการคำนวณทางคอมพิวเตอร์ แต่ยังพบได้ในชีวิตประจำวัน เช่น **การแนะนำหนังสือ** เมื่อเข้าเว็บไซต์ร้านหนังสือ

วิธีง่าย ๆ ในการวัด "ความใกล้เคียง" คือ **นับจำนวนหนังสือ (A–F) ที่ซื้อแตกต่างกัน** ยิ่งแตกต่างน้อยยิ่งใกล้เคียง เช่น สมพลกับสมชายซื้อแตกต่างกัน 2 เล่ม คือ D และ F

**ตารางที่ 2.2** ค่าความแตกต่างของการซื้อหนังสือ

| ลูกค้า | จำนวนหนังสือที่ซื้อแตกต่างกันกับสมพล |
|---|---|
| สมชาย | 0+0+0+1+0+1 = 2 |
| สมหญิง | 2 |
| สมศักดิ์ | **1** |
| สมฤดี | 2 |
| สมหมาย | 5 |

<sub>ที่มา: สถาบันส่งเสริมการสอนวิทยาศาสตร์และเทคโนโลยี (2561)</sub>

สมพลมีพฤติกรรมใกล้เคียงกับ **สมศักดิ์** มากที่สุด และสมศักดิ์ซื้อ "เวทมนตร์พ่อมดวัยรุ่น" ร้านจึงแนะนำหนังสือเล่มนี้ให้สมพล

### 🔬 ทดลอง: เขียนขั้นตอนวิธีแนะนำหนังสือเป็นโปรแกรม

````python
old_books = ["A", "B", "C", "D", "E", "F"]
new_books = ["เวทมนตร์พ่อมดวัยรุ่น", "การปลูกมะม่วง", "กลอนภาษาไทย"]

# ข้อมูลเข้า (จากตารางที่ 2.1): [ซื้อหนังสือ A-F], [ซื้อหนังสือใหม่ 3 เล่ม]
customers = {
    "สมชาย":   ([1, 1, 0, 1, 0, 0], [1, 0, 1]),
    "สมหญิง":  ([0, 1, 1, 0, 0, 1], [0, 1, 0]),
    "สมศักดิ์": ([1, 1, 0, 0, 0, 0], [1, 0, 0]),
    "สมฤดี":   ([1, 1, 0, 1, 1, 1], [1, 1, 0]),
    "สมหมาย":  ([0, 0, 0, 1, 1, 0], [0, 0, 1]),
}
sompol = [1, 1, 0, 0, 0, 1]

# ขั้นที่ 1: คำนวณค่าความแตกต่างกับลูกค้าแต่ละคน (ตารางที่ 2.2)
diffs = {}
for name, (bought, _) in customers.items():
    d = [abs(x - y) for x, y in zip(bought, sompol)]
    diffs[name] = sum(d)
    print(f"{name:<9} {'+'.join(map(str, d))} = {diffs[name]}")

# ขั้นที่ 2: เลือกลูกค้าที่แตกต่างน้อยที่สุด
nearest = min(diffs, key=diffs.get)
print("\nลูกค้าที่ใกล้เคียงสมพลที่สุด:", nearest)

# ขั้นที่ 3: แนะนำหนังสือใหม่ที่ลูกค้าคนนั้นซื้อ
recommend = [book for book, b in zip(new_books, customers[nearest][1]) if b == 1]
print("หนังสือที่แนะนำให้สมพล:", recommend)
````

> 💡 แนวคิดนี้คือรากฐานของ **ระบบแนะนำ (Recommender System)** ที่ใช้ใน Shopee, Netflix และ YouTube ลองเปลี่ยนค่า `sompol` เป็นพฤติกรรมของคนอื่น แล้วดูว่าระบบแนะนำหนังสืออะไร

### 2.4.4 รูปแบบการเขียนผังงาน
การเขียนผังงานไม่มีรูปแบบตายตัว ขึ้นกับการออกแบบของผู้เขียน แต่มี **โครงสร้างพื้นฐานสำคัญ 3 แบบ** ได้แก่ **โครงสร้างแบบเรียงลำดับ**, **โครงสร้างแบบทางเลือก** และ **โครงสร้างแบบวนซ้ำ**

#### ① โครงสร้างแบบเรียงลำดับ (Sequence)
ทำงานตามลำดับทีละขั้นจนถึงคำสั่งสุดท้าย **โดยไม่มีการข้ามขั้นตอน** เช่น การรับตัวเลข 3 จำนวนมาหาค่าเฉลี่ย

<p align="center"><img src="images/บทที่02/f2.2.png" width="220"><br><b>ภาพที่ 2.2</b> ผังงานแสดงการรับจำนวน 3 จำนวนเพื่อนำมาหาค่าเฉลี่ย<br><sub>ดัดแปลงจาก: ศิริเพ็ญ วิกัยสุขสกุล (2554)</sub></p>

เริ่มจากรับค่า 3 ค่าเก็บในตัวแปร `a`, `b`, `c` จากนั้นหาผลบวกเก็บใน `sum` นำ `sum` หารด้วยจำนวนตัวเลข (3) เก็บใน `avg` แล้วแสดงผลด้วย `print(avg)`

````python
a = float(input("a = "))
b = float(input("b = "))
c = float(input("c = "))
total = a + b + c        # ในไพธอนไม่ควรตั้งชื่อตัวแปรว่า sum เพราะซ้ำกับฟังก์ชัน sum() ที่มีอยู่แล้ว
avg = total / 3
print("ค่าเฉลี่ย =", avg)
````

#### ② โครงสร้างแบบทางเลือก (Selection)
ใช้เมื่อต้องการให้ทำงานตาม **เงื่อนไข** ที่กำหนด โดยเริ่มจากกำหนดเงื่อนไข แล้วกำหนดงานที่ทำภายใต้เงื่อนไขนั้น มี 2 ลักษณะ

**ก) ทางเลือกเดียว (`if`)** ทำงานเฉพาะเมื่อเงื่อนไขเป็นจริง ถ้าเป็นเท็จให้ข้ามไป

<p align="center"><img src="images/บทที่02/f2.3.png" width="300"><br><b>ภาพที่ 2.3</b> ผังงานแสดงการเลือกแบบทางเลือกเดียว</p>

รับค่าเก็บใน `x`, `y` แล้วเปรียบเทียบเงื่อนไข `x > y` ถ้า **จริง** นำ `x − y` เก็บใน `A` แล้วแสดงค่า `A` ถ้า **เท็จ** จะจบการทำงานทันที

````python
x = float(input("x = "))
y = float(input("y = "))
if x > y:                # Decision
    A = x - y            # ทำเฉพาะเมื่อเงื่อนไขเป็นจริง
    print("A =", A)
print("จบการทำงาน")
````

**ข) สองทางเลือก (`if … else`)** ถ้าเงื่อนไขเป็นจริงทำงานแบบหนึ่ง ถ้าเป็นเท็จทำอีกแบบหนึ่ง แล้วจึงกลับมาทำงานต่อร่วมกัน

<p align="center"><img src="images/บทที่02/f2.4.png" width="330"><br><b>ภาพที่ 2.4</b> ผังงานแสดงการเลือกแบบสองทางเลือก</p>

````python
a = float(input("a = "))
b = float(input("b = "))
if a < b:
    result = a - b       # เส้นทาง True
else:
    result = a + b       # เส้นทาง False
print("result =", result)
````

#### ③ โครงสร้างแบบวนซ้ำ (Repetition / Loop)

**ก) for loop** ใช้เมื่อ **รู้จำนวนรอบที่แน่นอน** ในภาษาไพธอนนิยมสร้างข้อมูลแบบลำดับ (ตัวเลขหรือสตริง) ตามจำนวนรอบที่ต้องการก่อน เช่น `range(10)` แล้ววนซ้ำไปจนกว่าจะเข้าถึงสมาชิกตัวสุดท้ายของลำดับ

<p align="center"><img src="images/บทที่02/f2.5.png" width="440"><br><b>ภาพที่ 2.5</b> การแสดงประโยค “Happy New Year” 10 ครั้งโดยใช้ for loop</p>

````python
seq = range(10)            # สร้างลำดับเลข 0 ถึง 9
print(list(seq))
for i in seq:              # วนจนกว่าจะเข้าถึงสมาชิกตัวสุดท้าย
    print(i + 1, "Happy new year")
````

**ข) while loop** ใช้เมื่อ **ไม่ทราบจำนวนรอบที่แน่นอน** วนซ้ำไปเรื่อย ๆ **ตราบที่เงื่อนไขเป็นจริง (True)** และ **ตรวจสอบเงื่อนไขก่อนทุกรอบ** หากเงื่อนไขเป็นเท็จ (False) จะหยุดทันที

<p align="center"><img src="images/บทที่02/f2.6.png" width="400"><br><b>ภาพที่ 2.6</b> การแสดงประโยค “Happy New Year” 10 ครั้งโดยใช้ while loop<br><sub>ดัดแปลงจาก: ศิริเพ็ญ วิกัยสุขสกุล (2554)</sub></p>

````python
count = 1                  # กำหนดค่าเริ่มต้น
while count <= 10:         # ตรวจเงื่อนไขก่อนทุกรอบ
    print(count, "Happy new year")
    count = count + 1      # ปรับค่า (ถ้าลืมบรรทัดนี้ จะวนไม่รู้จบ!)
````

#### ตัวอย่าง: รวมเลขคู่ทั้งหมดตั้งแต่ 1 ถึง 100 ด้วย while loop
โจทย์นี้ต้องวนซ้ำเพื่อพิจารณาตัวเลข 1 ถึง 100 โดยเลือกใช้ for หรือ while ก็ได้ ในที่นี้ออกแบบด้วย while

<p align="center"><img src="images/บทที่02/f2.7.png" width="440"><br><b>ภาพที่ 2.7</b> รวมเลขคู่ทั้งหมดตั้งแต่ 1 ถึง 100 ด้วยผังงานแบบ while loop</p>

**ลำดับการทำงาน**
1. กำหนด `i = 1` และ `sum = 0`
2. ตรวจเงื่อนไข `i <= 100` เพื่อไม่ให้วนเกิน 100 รอบ
3. ใช้ตัวดำเนินการ `%` (Modulo หารเอาเศษ) ตรวจว่า `i % 2 == 0` ซึ่งถ้าเป็นจริง `i` คือเลขคู่ ให้บวกสะสม `sum = sum + i`
4. เพิ่มค่า `i` ขึ้นทีละ 1 แล้ววนกลับไปตรวจเงื่อนไข
5. เมื่อ `i` เพิ่มถึง 101 เงื่อนไขเป็นเท็จ จึงออกจากลูปและแสดงผลรวม `sum`

````python
i = 1
total = 0
while i <= 100:
    if i % 2 == 0:          # เป็นเลขคู่หรือไม่
        total = total + i   # บวกสะสม
    i = i + 1
print("ผลรวมเลขคู่ 1 ถึง 100 =", total)

# ตรวจคำตอบด้วยวิธีอื่น
print("ตรวจด้วย sum(range(2, 101, 2)) =", sum(range(2, 101, 2)))
````

#### 🔬 ตารางติดตามการทำงาน (Trace Table)
เครื่องมือสำคัญในการตรวจสอบขั้นตอนวิธีคือ **การไล่ค่าตัวแปรทีละรอบ** ด้วยมือ ด้านล่างให้โปรแกรมพิมพ์ตารางติดตามของ 10 รอบแรกให้ดู

````python
i, total = 1, 0
print(f"{'รอบ':>4} | {'i':>3} | {'i<=100':>7} | {'i%2==0':>7} | {'sum':>5}")
print("-" * 40)
while i <= 10:              # แสดงแค่ 10 รอบแรก
    even = i % 2 == 0
    if even:
        total += i
    print(f"{i:>4} | {i:>3} | {str(i <= 100):>7} | {str(even):>7} | {total:>5}")
    i += 1
````

ระบบต้องอ่านค่าความชื้นของดินแล้วเปรียบเทียบกับค่าที่กำหนด (สมมติเป็น **0.1 หน่วย**) หากความชื้น **ต่ำกว่า** ค่าที่กำหนดให้ส่งสัญญาณ **เปิดน้ำ** หากความชื้น **มากกว่าหรือเท่ากับ** ค่าที่กำหนดให้ส่งสัญญาณ **ปิดน้ำ**

**ขั้นตอนวิธีการตัดสินใจ (ครั้งเดียว)**

- **กำหนดปัญหา:** ควบคุมการเปิด–ปิดน้ำของเครื่องรดน้ำต้นไม้
- **ข้อมูลเข้า:** ค่าความชื้นของดิน
- **ข้อมูลออก:** สัญญาณเปิด–ปิดน้ำ

**ขั้นตอนวิธีที่เพิ่มการทำซ้ำ**
1. ทำซ้ำทุก ๆ 1 วินาที
   - 1.1 อ่านค่าความชื้นของดิน
   - 1.2 ให้ H แทนค่าความชื้นดังกล่าว
   - 1.3 ถ้า H < 0.1 แล้ว
     - 1.3.1 ส่งสัญญาณเปิดน้ำ แต่ถ้าเงื่อนไขไม่เป็นจริง ให้

### 🔬 ทดลอง: จำลองระบบรดน้ำต้นไม้
เนื่องจากไม่มีเซนเซอร์จริง เราจะ **สุ่มค่าความชื้น** แทนการอ่านจากเซนเซอร์ และจำลองเพียง 10 รอบ (ระบบจริงจะทำงานไม่หยุดด้วย `while True:`)

````python
import random, time

THRESHOLD = 0.1

def read_soil_moisture():
    """จำลองการอ่านค่าจากเซนเซอร์ความชื้น (สุ่มค่า 0.00 - 0.25)"""
    return round(random.uniform(0, 0.25), 3)

for second in range(1, 11):          # ระบบจริงใช้ while True:
    H = read_soil_moisture()         # 1.1-1.2 อ่านค่าความชื้น
    if H < THRESHOLD:                # 1.3 ตัดสินใจ
        signal = "💧 เปิดน้ำ"
    else:
        signal = "⛔ ปิดน้ำ"
    print(f"วินาทีที่ {second:>2}: H = {H:.3f} -> {signal}")
    time.sleep(0.3)                  # รอ (ย่อเวลาจาก 1 วินาทีเพื่อความรวดเร็ว)
````

### 2.5.2 การหาค่าสูงสุดจากชุดข้อมูลตัวเลข
ถ้าข้อมูลมีน้อย เราหาค่าสูงสุดได้ทันทีด้วยการกวาดสายตา แต่ถ้าข้อมูลมีจำนวนมากและไม่ได้เรียงลำดับ การหาค่าสูงสุดให้แม่นยำจะยากขึ้นและใช้เวลานาน จึงต้องใช้ **การวนซ้ำ** และ **การเปรียบเทียบ**

- **กำหนดปัญหา:** ค้นหาค่าสูงสุดจากชุดข้อมูล
- **ข้อมูลเข้า:** รายการข้อมูล
- **ข้อมูลออก:** ค่าสูงสุดของข้อมูล

**ขั้นตอนวิธี**
1. รับชุดข้อมูล
2. กำหนดให้ `max` เท่ากับข้อมูลตัวแรกของชุดข้อมูล
3. พิจารณาข้อมูลตัวถัดไป
   - 3.1 เรียกข้อมูลที่กำลังพิจารณาว่า `x`
   - 3.2 เปรียบเทียบ `max` กับ `x` โดยถ้า `max < x` ให้กำหนด `max = x`
4. ทำข้อ 3 ซ้ำจนครบข้อมูลทุกตัว
5. แสดงค่า `max` ทางหน้าจอ
6. จบการทำงาน

<p align="center"><img src="images/บทที่02/f2.9.png" width="440"><br><b>ภาพที่ 2.9</b> ผังงานการหาค่าสูงสุดจากชุดข้อมูล</p>

````python
data = [34, 12, 78, 45, 91, 23, 67, 91, 5]

max_value = data[0]                       # ขั้นที่ 2 (ไม่ตั้งชื่อว่า max เพราะซ้ำกับฟังก์ชัน max())
count = 0
print(f"{'count':>5} | {'data[count]':>11} | {'max < data[count]':>17} | {'max':>4}")
print("-" * 48)
while count < len(data):                  # ขั้นที่ 4 วนจนครบทุกตัว
    x = data[count]                       # ขั้นที่ 3.1
    bigger = max_value < x
    if bigger:                            # ขั้นที่ 3.2
        max_value = x
    print(f"{count:>5} | {x:>11} | {str(bigger):>17} | {max_value:>4}")
    count = count + 1

print("\nค่าสูงสุด =", max_value)            # ขั้นที่ 5
print("ตรวจสอบด้วย max():", max(data))
````

### ✍️ ลงมือทำ 2: หาค่าต่ำสุดและค่าเฉลี่ย

````python
def min_and_avg(data):
    min_value = data[0]
    total = 0
    # ---- เขียนโค้ดของคุณตรงนี้ (ใช้ for x in data: ...) ----


    # --------------------------------------------------------
    avg = total / len(data)
    return min_value, avg

print(min_and_avg([34, 12, 78, 45]))
````

````python
# ✅ เซลล์ตรวจคำตอบ
def _check_min_avg():
    tests = [[34, 12, 78, 45], [5], [-3, -10, 4], [7, 7, 7, 7], [100, 0, 50]]
    ok = True
    for t in tests:
        exp = (min(t), sum(t) / len(t))
        try:
            got = min_and_avg(list(t))
        except Exception as e:
            got = f"ข้อผิดพลาด: {e}"
        good = isinstance(got, tuple) and got[0] == exp[0] and abs(got[1] - exp[1]) < 1e-9
        ok &= good
        print(f"{'✅' if good else '❌'} min_and_avg({t}) = {got}  (ควรได้ {exp})")
    print("\n🎉 ผ่านทุกกรณี!" if ok else "\nยังไม่ผ่านทุกกรณี ลองตรวจลูปอีกครั้ง")
_check_min_avg()
````

````python
# ✅ เซลล์ตรวจคำตอบ คำถามท้ายบท
def _run(name, func, cases):
    passed = 0
    for args, exp in cases:
        try:
            got = func(*args)
        except Exception as e:
            got = f"ข้อผิดพลาด: {e}"
        good = got == exp if not isinstance(exp, float) else (isinstance(got, (int, float)) and abs(got - exp) < 1e-6)
        passed += good
        if not good:
            print(f"   ❌ {name}{args} = {got!r}  (ควรได้ {exp!r})")
    print(f"{'✅' if passed == len(cases) else '⚠️'} {name}: ผ่าน {passed}/{len(cases)} กรณี")

_run("in_range", in_range, [((5,), True), ((0,), True), ((10,), True), ((-1,), False), ((10.5,), False)])
_run("sum_until_negative", sum_until_negative,
     [(([3, 5, 2, -1],), 10), (([-5],), 0), (([4, -2, 9],), 4), (([0, 1, 2, 3, -9],), 6)])
_run("grade", grade, [((95,), "A"), ((80,), "A"), ((79,), "B"), ((70,), "B"), ((65,), "C"),
                      ((50,), "D"), ((49,), "F"), ((0,), "F")])
_run("commission", commission, [((30000,), 1500.0), ((25000,), 500.0), ((25001,), 1250.05),
                                ((2000,), 40.0), ((10000,), 200.0), ((1999,), 0.0)])
````

#### 🔬 ตัวอย่างโปรแกรมเต็มของข้อ 7 (ใช้ while loop วนรับค่า)
รันได้หลังจากเขียนฟังก์ชัน `commission()` ในเซลล์ด้านบนแล้ว ป้อนยอดขายทีละคน และป้อน 0 เมื่อต้องการหยุด

````python
total_commission = 0
emp = 1
sales = float(input(f"ยอดขายพนักงานคนที่ {emp} (0 = จบ): "))
while sales != 0:
    c = commission(sales)
    print(f"   คอมมิชชั่น = {c:,.2f} บาท")
    total_commission += c
    emp += 1
    sales = float(input(f"ยอดขายพนักงานคนที่ {emp} (0 = จบ): "))
print(f"ผลรวมคอมมิชชั่นทั้งกลุ่ม = {total_commission:,.2f} บาท")
````
