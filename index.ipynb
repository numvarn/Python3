## 🛠️ ตรวจสอบสถานะไฟล์ทั้งหมดในโปรเจกต์
สามารถรันเซลล์โค้ดด้านล่าง (`Shift + Enter`) เพื่อตรวจสอบความครบถ้วนของไฟล์บทเรียน แบบฝึกหัด และเฉลยทั้งหมดในโฟลเดอร์นี้

```python
import json
from pathlib import Path

base_dir = Path(".")
chapter_files = sorted(base_dir.glob("บทที่*.ipynb"))
exercise_files = sorted((base_dir / "แบบฝึกหัด").glob("แบบฝึกหัด*.ipynb"))
solution_files = sorted((base_dir / "แบบฝึกหัด").glob("เฉลย*.ipynb"))

def inspect_nb(path: Path):
    with path.open("r", encoding="utf-8") as f:
        nb = json.load(f)
    cells = nb.get("cells", [])
    md_count = sum(1 for c in cells if c.get("cell_type") == "markdown")
    code_count = sum(1 for c in cells if c.get("cell_type") == "code")
    size_kb = path.stat().st_size / 1024
    return len(cells), md_count, code_count, size_kb

print(f"📁 ไดเรกทอรีปัจจุบัน: {base_dir.resolve()}")
print(f"📘 ไฟล์บทเรียน: {len(chapter_files)} ไฟล์ | 📝 แบบฝึกหัด: {len(exercise_files)} ไฟล์ | ✅ เฉลย: {len(solution_files)} ไฟล์\n")
print(f"{'ลำดับ':<6} {'ชื่อไฟล์':<62} {'เซลล์ทั้งหมด':>10} {'Markdown':>10} {'Code':>8} {'ขนาด (KB)':>12}")
print("-" * 112)

for idx, p in enumerate(chapter_files + exercise_files + solution_files, 1):
    rel = str(p.relative_to(base_dir))
    total_c, md_c, code_c, kb = inspect_nb(p)
    print(f"{idx:<6} {rel:<62} {total_c:>10} {md_c:>10} {code_c:>8} {kb:>12.1f}")
```
